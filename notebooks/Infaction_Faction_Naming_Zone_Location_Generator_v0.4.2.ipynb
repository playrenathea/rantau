```python
def generate_slots(rng):
    slots = [{'id': f'F{i}', 'types': []} for i in range(1,8)]
    slots[0]['types'] = ['Government']
    slots[1]['types'] = ['Revolutionary']
    # Stage 2: two independent, distinct recipient slots.
    elite_index, poor_index = rng.sample(range(1,7),2)
    slots[elite_index]['types'].append('Elite')
    slots[poor_index]['types'].append('Poor')
    # Stage 3: only one Outsider, any F2..F7 that has room.
    eligible = [i for i in range(1,7) if len(slots[i]['types']) <= 1]
    slots[rng.choice(eligible)]['types'].append('Outsider')
    # Stage 4: fill empty slots in randomized order, with unique leftover Types.
    remaining = [t for t in TYPES if t not in {'Government','Revolutionary','Elite','Poor','Outsider'}]
    assert len(remaining)==8
    rng.shuffle(remaining)
    empty = [i for i,s in enumerate(slots) if not s['types']]
    rng.shuffle(empty)
    for i in empty:
        slots[i]['types'].append(remaining.pop())
    # Stage 5: give each remaining Type to one random single-Type slot.
    while remaining:
        open_slots = [i for i,s in enumerate(slots) if len(s['types']) == 1]
        slots[rng.choice(open_slots)]['types'].append(remaining.pop())
    validate_slots(slots)
    return slots

def validate_slots(slots):
    assert len(slots) == 7
    assert slots[0]['types'][0] == 'Government'
    assert slots[1]['types'][0] == 'Revolutionary'
    assert all(1 <= len(s['types']) <= 2 for s in slots)
    assert all('Government' not in s['types'] or not ({'Elite','Poor','Revolutionary','Outsider'} & set(s['types'])) for s in slots)
    loc={t:i for i,s in enumerate(slots) for t in s['types']}
    assert len(loc) == 13 and set(loc) == set(TYPES)
    assert loc['Elite'] != loc['Poor']
    assert all(s['id'] == f'F{i}' for i,s in enumerate(slots,1))

def naming_score(x,y,rng):
    # Average of (arithmetic mean, geometric mean, uniform random 0..100).
    return ((x+y)/2 + math.sqrt(x*y) + rng.uniform(0,100))/3
```

```python
def candidate_ranking(types, catalog, rng, item_type):
    results=[]
    for label, data in catalog.items():
        aff = data if item_type == 'Animal' else data['affinity']
        x=aff[types[0]]
        y=aff[types[1]] if len(types)==2 else x
        score=naming_score(x,y,rng)
        results.append({'label':label,'score':score,'x':x,'y':y})
    # Lexicographic stable tie-break, no extra random calls.
    results.sort(key=lambda z:(-z['score'],z['label']))
    return results

def choose_grammar(base, rng):
    weights=BASE[base]['grammar']
    labels=[g for g in ('M','O','P','S') if weights[g]>0]
    assert labels,base
    return rng.choices(labels,weights=[weights[g] for g in labels],k=1)[0]

# Singular/plural wording for Animal grammar. Book1 supplies labels/affinity only,
# so regular plurals are defaults, with common irregular spellings handled explicitly.
IRREGULAR={'Wolf':'Wolves','Fox':'Foxes','Sheep':'Sheep','Butterfly':'Butterflies',
           'Swallow':'Swallows','Octopus':'Octopuses','Phoenix':'Phoenixes'}
def plural(animal):
    if animal in IRREGULAR: return IRREGULAR[animal]
    if animal.endswith(('s','x','z','ch','sh')): return animal+'es'
    if animal.endswith('y') and animal[-2].lower() not in 'aeiou': return animal[:-1]+'ies'
    return animal+'s'

def faction_name(base, animal, grammar):
    if grammar=='M': return f'{animal} {base}'
    if grammar=='O': return f'{base} of the {animal}'
    if grammar=='P': return f'{base} of the {plural(animal)}'
    if grammar=='S': return f"{animal}'s {base}"
    raise ValueError(grammar)

def generate_world(seed=20261120, retain_rankings=False):
    rng=random.Random(seed)
    slots=generate_slots(rng)
    factions=slots[:6]  # F7 is discarded permanently before any naming occurs.
    used_base=set(); used_animal=set()
    for faction in factions:
        types=faction['types']
        base_rank=candidate_ranking(types, BASE, rng, 'Base')
        animal_rank=candidate_ranking(types, ANIMAL, rng, 'Animal')
        best_base=next(c for c in base_rank if c['label'] not in used_base)
        best_animal=next(c for c in animal_rank if c['label'] not in used_animal)
        used_base.add(best_base['label']); used_animal.add(best_animal['label'])
        grammar=choose_grammar(best_base['label'],rng)
        faction.update({
            'base': best_base['label'], 'base_score': round(best_base['score'],5),
            'base_affinities': [best_base['x'],best_base['y']],
            'animal': best_animal['label'], 'animal_score': round(best_animal['score'],5),
            'animal_affinities': [best_animal['x'],best_animal['y']],
            'grammar': grammar,
            'name': faction_name(best_base['label'],best_animal['label'],grammar)
        })
        if retain_rankings:
            faction['base_candidates']=base_rank
            faction['animal_candidates']=animal_rank
    assert len(factions)==6
    assert len({f['base'] for f in factions})==6
    assert len({f['animal'] for f in factions})==6
    assert len({f['name'] for f in factions})==6
    return {'version':VERSION,'seed':seed,'source':BOOK1['source'],
            'factions':factions}
```

## Zone Selection — locked v0.4.1

Generate 7–9 unique Zone Types; F1–F6 each selects one sequentially, then remaining 1–3 without Faction fit. Source: Book1.xlsx snapshot (36 Scale 4–5 Zone Types).

Faction base = AVG(AVG(Type1, Type2, Base), CUBERT(Type1 × Type2 × Base), Random[0,100]); extra Zone base = Random[0,100]. For single Type, use it twice.

Define SOFT(S)=AVG(AVG(S), MAX(S)), or 0 when S is empty. Candidate score = base + 0.75×SOFT(Functions under minimum) − SOFT(Functions at/over maximum AND affinity >=75) − SOFT(Functions over maximum AND affinity >=75) + Zone Type modifier. Bonus looks at all raw Function affinities; penalties only count Function affinities >=75. Over-maximum Function is penalized twice. After selection increment ALL Functions with affinity >=75 and recompute.

Modifiers are additive points, not percentage multipliers: Caravan Camp +7; Estate +2; Military Base −3; Mining Settlement +4; Rebel Camp −4; Cultural District −1; Entertainment District −4; Market District −2; Royal Quarter −7. All 27 other Zone Types = 0.

No fixed F1 Government/F2 Revolutionary bias is added in this locked configuration; related ideas were discussed but not calibrated as part of the approved 5–40% dataset. Greedy selection, unique Types, and soft quotas remain as before.


```python
ZONE_DATA_JSON = '{"source":"Book1.xlsx (2026-11-20, Zone Scale 4-5 rows)","function_quotas":{"Authority":{"min":1,"max":1},"Defense":{"min":1,"max":2},"Religion":{"min":0,"max":1},"Knowledge":{"min":0,"max":1},"Commerce":{"min":1,"max":2},"Production":{"min":0,"max":2},"Logistics":{"min":0,"max":2},"Resources":{"min":0,"max":2},"Residence":{"min":1,"max":2},"Social":{"min":1,"max":2},"Covert":{"min":0,"max":1},"Migration":{"min":0,"max":1},"Care":{"min":0,"max":1},"Exploration":{"min":0,"max":2}},"zones":{"Caravan Camp":{"scale":4,"function":{"Authority":15.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":50.0,"Production":0.0,"Logistics":100.0,"Resources":0.0,"Residence":75.0,"Social":0.0,"Covert":0.0,"Migration":100.0,"Care":0.0,"Exploration":25.0},"type":{"Government":10.0,"Military":15.0,"Religious":10.0,"Criminal":50.0,"Revolutionary":20.0,"Conspiracy":10.0,"Merchant":85.0,"Clan":65.0,"Cultural":10.0,"Outsider":90.0,"Elite":20.0,"Poor":60.0,"Academic":10.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":15.0,"Dominion":15.0,"Council":15.0,"Court":20.0,"Commonwealth":15.0,"Directorate":15.0,"Ministry":15.0,"Authority":15.0,"Legion":20.0,"Guard":20.0,"Army":20.0,"Militia":20.0,"Vanguard":20.0,"Watch":20.0,"Corps":20.0,"Wardens":20.0,"Rangers":45.0,"Order":20.0,"Covenant":20.0,"Faith":15.0,"Sect":15.0,"Brotherhood":65.0,"Fellowship":65.0,"Chapter":50.0,"Disciples":15.0,"Cult":15.0,"Guild":45.0,"Company":45.0,"Consortium":70.0,"Syndicate":70.0,"Cartel":70.0,"Cooperative":45.0,"Corporation":70.0,"Chamber":70.0,"Trust":45.0,"Clan":65.0,"House":20.0,"Dynasty":40.0,"Family":65.0,"Tribe":75.0,"People":35.0,"Community":40.0,"Nation":35.0,"Kindred":65.0,"Lineage":40.0,"Resistance":20.0,"Movement":20.0,"Front":20.0,"Union":40.0,"Collective":40.0,"Alliance":55.0,"League":55.0,"Pact":55.0,"Liberation":20.0,"Academy":20.0,"Institute":20.0,"School":10.0,"Society":50.0,"Circle":50.0,"Cabal":50.0,"Conclave":50.0,"Scholars":15.0,"Collegium":15.0,"Keepers":15.0,"Caravan":100.0,"Fleet":75.0,"Band":100.0,"Enclave":55.0,"Commune":70.0,"Network":75.0,"Expedition":100.0,"Nomads":100.0,"Diaspora":75.0,"Confederation":45.0,"Assembly":45.0,"Coalition":45.0,"Horde":45.0,"Mercenaries":45.0,"Parliament":45.0,"Senate":45.0,"Association":45.0,"Compact":70.0}},"Cavern":{"scale":4,"function":{"Authority":15.0,"Defense":0.0,"Religion":0.0,"Knowledge":25.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":25.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":100.0},"type":{"Government":10.0,"Military":10.0,"Religious":10.0,"Criminal":65.0,"Revolutionary":45.0,"Conspiracy":70.0,"Merchant":10.0,"Clan":35.0,"Cultural":10.0,"Outsider":80.0,"Elite":10.0,"Poor":70.0,"Academic":60.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":15.0,"Dominion":15.0,"Council":15.0,"Court":15.0,"Commonwealth":15.0,"Directorate":15.0,"Ministry":15.0,"Authority":15.0,"Legion":20.0,"Guard":15.0,"Army":20.0,"Militia":35.0,"Vanguard":35.0,"Watch":15.0,"Corps":15.0,"Wardens":15.0,"Rangers":45.0,"Order":20.0,"Covenant":20.0,"Faith":15.0,"Sect":50.0,"Brotherhood":35.0,"Fellowship":35.0,"Chapter":15.0,"Disciples":50.0,"Cult":70.0,"Guild":20.0,"Company":20.0,"Consortium":15.0,"Syndicate":65.0,"Cartel":65.0,"Cooperative":40.0,"Corporation":15.0,"Chamber":15.0,"Trust":15.0,"Clan":35.0,"House":10.0,"Dynasty":35.0,"Family":35.0,"Tribe":45.0,"People":35.0,"Community":40.0,"Nation":30.0,"Kindred":35.0,"Lineage":35.0,"Resistance":60.0,"Movement":35.0,"Front":60.0,"Union":40.0,"Collective":40.0,"Alliance":35.0,"League":35.0,"Pact":40.0,"Liberation":60.0,"Academy":40.0,"Institute":40.0,"School":15.0,"Society":40.0,"Circle":40.0,"Cabal":70.0,"Conclave":70.0,"Scholars":40.0,"Collegium":40.0,"Keepers":40.0,"Caravan":45.0,"Fleet":45.0,"Band":70.0,"Enclave":5.0,"Commune":40.0,"Network":70.0,"Expedition":90.0,"Nomads":45.0,"Diaspora":45.0,"Confederation":40.0,"Assembly":40.0,"Coalition":40.0,"Horde":45.0,"Mercenaries":45.0,"Parliament":45.0,"Senate":45.0,"Association":45.0,"Compact":40.0}},"Citadel":{"scale":4,"function":{"Authority":95.0,"Defense":100.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":25.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":85.0,"Military":90.0,"Religious":10.0,"Criminal":10.0,"Revolutionary":20.0,"Conspiracy":10.0,"Merchant":10.0,"Clan":65.0,"Cultural":10.0,"Outsider":10.0,"Elite":60.0,"Poor":10.0,"Academic":10.0},"base":{"Empire":95.0,"Kingdom":95.0,"Republic":70.0,"Dominion":95.0,"Council":95.0,"Court":70.0,"Commonwealth":70.0,"Directorate":95.0,"Ministry":95.0,"Authority":95.0,"Legion":100.0,"Guard":100.0,"Army":100.0,"Militia":75.0,"Vanguard":75.0,"Watch":100.0,"Corps":100.0,"Wardens":95.0,"Rangers":70.0,"Order":100.0,"Covenant":75.0,"Faith":75.0,"Sect":15.0,"Brotherhood":35.0,"Fellowship":35.0,"Chapter":15.0,"Disciples":15.0,"Cult":15.0,"Guild":20.0,"Company":20.0,"Consortium":15.0,"Syndicate":20.0,"Cartel":20.0,"Cooperative":15.0,"Corporation":15.0,"Chamber":45.0,"Trust":65.0,"Clan":65.0,"House":15.0,"Dynasty":65.0,"Family":35.0,"Tribe":60.0,"People":35.0,"Community":35.0,"Nation":30.0,"Kindred":65.0,"Lineage":60.0,"Resistance":55.0,"Movement":55.0,"Front":55.0,"Union":20.0,"Collective":20.0,"Alliance":55.0,"League":55.0,"Pact":55.0,"Liberation":55.0,"Academy":20.0,"Institute":20.0,"School":10.0,"Society":15.0,"Circle":20.0,"Cabal":20.0,"Conclave":15.0,"Scholars":15.0,"Collegium":15.0,"Keepers":50.0,"Caravan":15.0,"Fleet":45.0,"Band":50.0,"Enclave":5.0,"Commune":15.0,"Network":15.0,"Expedition":50.0,"Nomads":20.0,"Diaspora":20.0,"Confederation":70.0,"Assembly":70.0,"Coalition":70.0,"Horde":70.0,"Mercenaries":70.0,"Parliament":95.0,"Senate":95.0,"Association":15.0,"Compact":45.0}},"Enclave":{"scale":4,"function":{"Authority":15.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":100.0,"Social":0.0,"Covert":0.0,"Migration":75.0,"Care":0.0,"Exploration":0.0},"type":{"Government":10.0,"Military":10.0,"Religious":45.0,"Criminal":10.0,"Revolutionary":55.0,"Conspiracy":65.0,"Merchant":10.0,"Clan":95.0,"Cultural":10.0,"Outsider":85.0,"Elite":20.0,"Poor":25.0,"Academic":10.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":15.0,"Dominion":15.0,"Council":15.0,"Court":20.0,"Commonwealth":15.0,"Directorate":15.0,"Ministry":15.0,"Authority":15.0,"Legion":20.0,"Guard":15.0,"Army":20.0,"Militia":35.0,"Vanguard":35.0,"Watch":15.0,"Corps":15.0,"Wardens":15.0,"Rangers":45.0,"Order":35.0,"Covenant":35.0,"Faith":35.0,"Sect":35.0,"Brotherhood":95.0,"Fellowship":95.0,"Chapter":85.0,"Disciples":35.0,"Cult":40.0,"Guild":20.0,"Company":20.0,"Consortium":15.0,"Syndicate":20.0,"Cartel":20.0,"Cooperative":30.0,"Corporation":15.0,"Chamber":15.0,"Trust":20.0,"Clan":95.0,"House":75.0,"Dynasty":45.0,"Family":95.0,"Tribe":95.0,"People":45.0,"Community":45.0,"Nation":40.0,"Kindred":95.0,"Lineage":45.0,"Resistance":40.0,"Movement":40.0,"Front":40.0,"Union":40.0,"Collective":40.0,"Alliance":90.0,"League":90.0,"Pact":90.0,"Liberation":35.0,"Academy":20.0,"Institute":20.0,"School":15.0,"Society":75.0,"Circle":90.0,"Cabal":55.0,"Conclave":45.0,"Scholars":15.0,"Collegium":15.0,"Keepers":35.0,"Caravan":70.0,"Fleet":45.0,"Band":70.0,"Enclave":100.0,"Commune":95.0,"Network":45.0,"Expedition":70.0,"Nomads":95.0,"Diaspora":95.0,"Confederation":45.0,"Assembly":45.0,"Coalition":45.0,"Horde":45.0,"Mercenaries":45.0,"Parliament":45.0,"Senate":45.0,"Association":45.0,"Compact":95.0}},"Estate":{"scale":4,"function":{"Authority":55.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":50.0,"Residence":100.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":15.0,"Military":10.0,"Religious":45.0,"Criminal":10.0,"Revolutionary":15.0,"Conspiracy":10.0,"Merchant":50.0,"Clan":45.0,"Cultural":10.0,"Outsider":20.0,"Elite":80.0,"Poor":25.0,"Academic":10.0},"base":{"Empire":55.0,"Kingdom":80.0,"Republic":20.0,"Dominion":55.0,"Council":20.0,"Court":70.0,"Commonwealth":20.0,"Directorate":20.0,"Ministry":50.0,"Authority":55.0,"Legion":20.0,"Guard":20.0,"Army":20.0,"Militia":20.0,"Vanguard":20.0,"Watch":20.0,"Corps":15.0,"Wardens":20.0,"Rangers":20.0,"Order":35.0,"Covenant":35.0,"Faith":35.0,"Sect":35.0,"Brotherhood":45.0,"Fellowship":45.0,"Chapter":35.0,"Disciples":35.0,"Cult":35.0,"Guild":35.0,"Company":35.0,"Consortium":35.0,"Syndicate":35.0,"Cartel":35.0,"Cooperative":35.0,"Corporation":35.0,"Chamber":35.0,"Trust":70.0,"Clan":45.0,"House":75.0,"Dynasty":95.0,"Family":95.0,"Tribe":45.0,"People":45.0,"Community":45.0,"Nation":40.0,"Kindred":45.0,"Lineage":95.0,"Resistance":20.0,"Movement":20.0,"Front":20.0,"Union":30.0,"Collective":30.0,"Alliance":20.0,"League":20.0,"Pact":20.0,"Liberation":20.0,"Academy":20.0,"Institute":20.0,"School":15.0,"Society":50.0,"Circle":50.0,"Cabal":20.0,"Conclave":15.0,"Scholars":15.0,"Collegium":15.0,"Keepers":35.0,"Caravan":35.0,"Fleet":35.0,"Band":20.0,"Enclave":55.0,"Commune":15.0,"Network":20.0,"Expedition":20.0,"Nomads":20.0,"Diaspora":20.0,"Confederation":45.0,"Assembly":20.0,"Coalition":15.0,"Horde":20.0,"Mercenaries":35.0,"Parliament":55.0,"Senate":70.0,"Association":35.0,"Compact":35.0}},"Village":{"scale":4,"function":{"Authority":15.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":25.0,"Production":0.0,"Logistics":0.0,"Resources":75.0,"Residence":100.0,"Social":75.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":10.0,"Military":10.0,"Religious":50.0,"Criminal":10.0,"Revolutionary":15.0,"Conspiracy":10.0,"Merchant":75.0,"Clan":70.0,"Cultural":25.0,"Outsider":55.0,"Elite":20.0,"Poor":85.0,"Academic":10.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":50.0,"Dominion":15.0,"Council":15.0,"Court":20.0,"Commonwealth":50.0,"Directorate":15.0,"Ministry":15.0,"Authority":15.0,"Legion":20.0,"Guard":15.0,"Army":20.0,"Militia":20.0,"Vanguard":20.0,"Watch":15.0,"Corps":15.0,"Wardens":15.0,"Rangers":35.0,"Order":35.0,"Covenant":35.0,"Faith":35.0,"Sect":35.0,"Brotherhood":45.0,"Fellowship":45.0,"Chapter":35.0,"Disciples":35.0,"Cult":35.0,"Guild":45.0,"Company":45.0,"Consortium":45.0,"Syndicate":70.0,"Cartel":70.0,"Cooperative":95.0,"Corporation":45.0,"Chamber":40.0,"Trust":40.0,"Clan":70.0,"House":90.0,"Dynasty":45.0,"Family":90.0,"Tribe":90.0,"People":95.0,"Community":95.0,"Nation":90.0,"Kindred":70.0,"Lineage":45.0,"Resistance":20.0,"Movement":55.0,"Front":20.0,"Union":70.0,"Collective":70.0,"Alliance":20.0,"League":20.0,"Pact":20.0,"Liberation":20.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":30.0,"Circle":20.0,"Cabal":50.0,"Conclave":50.0,"Scholars":15.0,"Collegium":15.0,"Keepers":35.0,"Caravan":45.0,"Fleet":40.0,"Band":35.0,"Enclave":60.0,"Commune":85.0,"Network":35.0,"Expedition":65.0,"Nomads":40.0,"Diaspora":40.0,"Confederation":45.0,"Assembly":35.0,"Coalition":35.0,"Horde":35.0,"Mercenaries":40.0,"Parliament":35.0,"Senate":35.0,"Association":40.0,"Compact":40.0}},"Hunting Grounds":{"scale":4,"function":{"Authority":25.0,"Defense":0.0,"Religion":0.0,"Knowledge":25.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":100.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":50.0},"type":{"Government":10.0,"Military":30.0,"Religious":35.0,"Criminal":25.0,"Revolutionary":40.0,"Conspiracy":10.0,"Merchant":45.0,"Clan":50.0,"Cultural":10.0,"Outsider":65.0,"Elite":10.0,"Poor":75.0,"Academic":15.0},"base":{"Empire":30.0,"Kingdom":30.0,"Republic":30.0,"Dominion":25.0,"Council":25.0,"Court":25.0,"Commonwealth":30.0,"Directorate":25.0,"Ministry":25.0,"Authority":25.0,"Legion":40.0,"Guard":35.0,"Army":40.0,"Militia":50.0,"Vanguard":40.0,"Watch":35.0,"Corps":40.0,"Wardens":35.0,"Rangers":90.0,"Order":35.0,"Covenant":35.0,"Faith":35.0,"Sect":35.0,"Brotherhood":50.0,"Fellowship":45.0,"Chapter":35.0,"Disciples":35.0,"Cult":35.0,"Guild":35.0,"Company":40.0,"Consortium":35.0,"Syndicate":40.0,"Cartel":40.0,"Cooperative":70.0,"Corporation":35.0,"Chamber":35.0,"Trust":35.0,"Clan":50.0,"House":0.0,"Dynasty":45.0,"Family":50.0,"Tribe":50.0,"People":65.0,"Community":70.0,"Nation":60.0,"Kindred":50.0,"Lineage":45.0,"Resistance":50.0,"Movement":50.0,"Front":50.0,"Union":60.0,"Collective":60.0,"Alliance":50.0,"League":50.0,"Pact":50.0,"Liberation":50.0,"Academy":30.0,"Institute":30.0,"School":0.0,"Society":30.0,"Circle":30.0,"Cabal":30.0,"Conclave":30.0,"Scholars":30.0,"Collegium":30.0,"Keepers":35.0,"Caravan":55.0,"Fleet":45.0,"Band":55.0,"Enclave":5.0,"Commune":65.0,"Network":40.0,"Expedition":55.0,"Nomads":55.0,"Diaspora":55.0,"Confederation":40.0,"Assembly":45.0,"Coalition":40.0,"Horde":55.0,"Mercenaries":55.0,"Parliament":40.0,"Senate":40.0,"Association":40.0,"Compact":40.0}},"Military Base":{"scale":4,"function":{"Authority":70.0,"Defense":100.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":75.0,"Resources":0.0,"Residence":50.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":80.0,"Military":100.0,"Religious":10.0,"Criminal":10.0,"Revolutionary":65.0,"Conspiracy":10.0,"Merchant":20.0,"Clan":65.0,"Cultural":10.0,"Outsider":35.0,"Elite":15.0,"Poor":30.0,"Academic":10.0},"base":{"Empire":70.0,"Kingdom":70.0,"Republic":50.0,"Dominion":70.0,"Council":70.0,"Court":50.0,"Commonwealth":50.0,"Directorate":70.0,"Ministry":70.0,"Authority":70.0,"Legion":100.0,"Guard":90.0,"Army":100.0,"Militia":90.0,"Vanguard":90.0,"Watch":90.0,"Corps":100.0,"Wardens":90.0,"Rangers":100.0,"Order":75.0,"Covenant":50.0,"Faith":50.0,"Sect":25.0,"Brotherhood":40.0,"Fellowship":40.0,"Chapter":25.0,"Disciples":25.0,"Cult":25.0,"Guild":20.0,"Company":20.0,"Consortium":20.0,"Syndicate":35.0,"Cartel":35.0,"Cooperative":35.0,"Corporation":20.0,"Chamber":45.0,"Trust":30.0,"Clan":65.0,"House":20.0,"Dynasty":40.0,"Family":45.0,"Tribe":65.0,"People":40.0,"Community":40.0,"Nation":35.0,"Kindred":65.0,"Lineage":40.0,"Resistance":65.0,"Movement":65.0,"Front":90.0,"Union":45.0,"Collective":45.0,"Alliance":65.0,"League":65.0,"Pact":65.0,"Liberation":65.0,"Academy":30.0,"Institute":30.0,"School":15.0,"Society":25.0,"Circle":30.0,"Cabal":30.0,"Conclave":30.0,"Scholars":25.0,"Collegium":25.0,"Keepers":45.0,"Caravan":30.0,"Fleet":55.0,"Band":55.0,"Enclave":30.0,"Commune":40.0,"Network":30.0,"Expedition":60.0,"Nomads":35.0,"Diaspora":35.0,"Confederation":70.0,"Assembly":70.0,"Coalition":70.0,"Horde":90.0,"Mercenaries":90.0,"Parliament":70.0,"Senate":70.0,"Association":30.0,"Compact":45.0}},"Mining Settlement":{"scale":4,"function":{"Authority":15.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":25.0,"Logistics":0.0,"Resources":100.0,"Residence":100.0,"Social":0.0,"Covert":0.0,"Migration":25.0,"Care":0.0,"Exploration":0.0},"type":{"Government":10.0,"Military":45.0,"Religious":10.0,"Criminal":10.0,"Revolutionary":15.0,"Conspiracy":10.0,"Merchant":75.0,"Clan":60.0,"Cultural":10.0,"Outsider":60.0,"Elite":20.0,"Poor":85.0,"Academic":10.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":50.0,"Dominion":15.0,"Council":15.0,"Court":20.0,"Commonwealth":50.0,"Directorate":15.0,"Ministry":15.0,"Authority":15.0,"Legion":35.0,"Guard":35.0,"Army":35.0,"Militia":35.0,"Vanguard":35.0,"Watch":35.0,"Corps":35.0,"Wardens":35.0,"Rangers":40.0,"Order":35.0,"Covenant":20.0,"Faith":15.0,"Sect":15.0,"Brotherhood":30.0,"Fellowship":30.0,"Chapter":15.0,"Disciples":15.0,"Cult":15.0,"Guild":45.0,"Company":45.0,"Consortium":45.0,"Syndicate":70.0,"Cartel":70.0,"Cooperative":70.0,"Corporation":45.0,"Chamber":40.0,"Trust":40.0,"Clan":60.0,"House":80.0,"Dynasty":30.0,"Family":55.0,"Tribe":65.0,"People":55.0,"Community":70.0,"Nation":50.0,"Kindred":55.0,"Lineage":30.0,"Resistance":20.0,"Movement":55.0,"Front":20.0,"Union":70.0,"Collective":70.0,"Alliance":20.0,"League":20.0,"Pact":20.0,"Liberation":20.0,"Academy":20.0,"Institute":20.0,"School":15.0,"Society":15.0,"Circle":20.0,"Cabal":50.0,"Conclave":50.0,"Scholars":15.0,"Collegium":15.0,"Keepers":15.0,"Caravan":45.0,"Fleet":40.0,"Band":40.0,"Enclave":60.0,"Commune":60.0,"Network":40.0,"Expedition":90.0,"Nomads":40.0,"Diaspora":40.0,"Confederation":35.0,"Assembly":35.0,"Coalition":35.0,"Horde":40.0,"Mercenaries":40.0,"Parliament":40.0,"Senate":40.0,"Association":40.0,"Compact":40.0}},"Monastery":{"scale":4,"function":{"Authority":15.0,"Defense":0.0,"Religion":100.0,"Knowledge":75.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":75.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":25.0,"Exploration":0.0},"type":{"Government":10.0,"Military":10.0,"Religious":90.0,"Criminal":10.0,"Revolutionary":15.0,"Conspiracy":55.0,"Merchant":10.0,"Clan":40.0,"Cultural":55.0,"Outsider":55.0,"Elite":20.0,"Poor":20.0,"Academic":85.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":15.0,"Dominion":15.0,"Council":50.0,"Court":20.0,"Commonwealth":15.0,"Directorate":50.0,"Ministry":15.0,"Authority":15.0,"Legion":20.0,"Guard":15.0,"Army":20.0,"Militia":20.0,"Vanguard":20.0,"Watch":15.0,"Corps":15.0,"Wardens":15.0,"Rangers":35.0,"Order":100.0,"Covenant":100.0,"Faith":100.0,"Sect":100.0,"Brotherhood":100.0,"Fellowship":95.0,"Chapter":95.0,"Disciples":100.0,"Cult":100.0,"Guild":20.0,"Company":20.0,"Consortium":15.0,"Syndicate":20.0,"Cartel":20.0,"Cooperative":20.0,"Corporation":15.0,"Chamber":15.0,"Trust":20.0,"Clan":40.0,"House":30.0,"Dynasty":40.0,"Family":40.0,"Tribe":40.0,"People":35.0,"Community":40.0,"Nation":35.0,"Kindred":40.0,"Lineage":40.0,"Resistance":20.0,"Movement":20.0,"Front":20.0,"Union":20.0,"Collective":20.0,"Alliance":20.0,"League":20.0,"Pact":35.0,"Liberation":20.0,"Academy":95.0,"Institute":95.0,"School":45.0,"Society":70.0,"Circle":70.0,"Cabal":45.0,"Conclave":45.0,"Scholars":95.0,"Collegium":95.0,"Keepers":95.0,"Caravan":40.0,"Fleet":35.0,"Band":35.0,"Enclave":30.0,"Commune":35.0,"Network":35.0,"Expedition":45.0,"Nomads":40.0,"Diaspora":40.0,"Confederation":35.0,"Assembly":35.0,"Coalition":35.0,"Horde":35.0,"Mercenaries":35.0,"Parliament":35.0,"Senate":35.0,"Association":65.0,"Compact":35.0}},"Necropolis":{"scale":4,"function":{"Authority":15.0,"Defense":0.0,"Religion":75.0,"Knowledge":25.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":25.0,"Migration":0.0,"Care":0.0,"Exploration":75.0},"type":{"Government":10.0,"Military":10.0,"Religious":85.0,"Criminal":15.0,"Revolutionary":10.0,"Conspiracy":70.0,"Merchant":10.0,"Clan":40.0,"Cultural":55.0,"Outsider":20.0,"Elite":10.0,"Poor":10.0,"Academic":55.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":15.0,"Dominion":15.0,"Council":15.0,"Court":15.0,"Commonwealth":15.0,"Directorate":15.0,"Ministry":15.0,"Authority":15.0,"Legion":20.0,"Guard":15.0,"Army":20.0,"Militia":15.0,"Vanguard":15.0,"Watch":15.0,"Corps":15.0,"Wardens":15.0,"Rangers":20.0,"Order":70.0,"Covenant":70.0,"Faith":70.0,"Sect":70.0,"Brotherhood":70.0,"Fellowship":70.0,"Chapter":70.0,"Disciples":70.0,"Cult":70.0,"Guild":20.0,"Company":20.0,"Consortium":15.0,"Syndicate":20.0,"Cartel":20.0,"Cooperative":15.0,"Corporation":15.0,"Chamber":15.0,"Trust":15.0,"Clan":40.0,"House":10.0,"Dynasty":40.0,"Family":40.0,"Tribe":40.0,"People":40.0,"Community":40.0,"Nation":35.0,"Kindred":40.0,"Lineage":40.0,"Resistance":20.0,"Movement":20.0,"Front":15.0,"Union":20.0,"Collective":20.0,"Alliance":15.0,"League":15.0,"Pact":40.0,"Liberation":15.0,"Academy":65.0,"Institute":65.0,"School":25.0,"Society":35.0,"Circle":40.0,"Cabal":40.0,"Conclave":40.0,"Scholars":65.0,"Collegium":65.0,"Keepers":70.0,"Caravan":20.0,"Fleet":20.0,"Band":20.0,"Enclave":5.0,"Commune":15.0,"Network":40.0,"Expedition":35.0,"Nomads":20.0,"Diaspora":20.0,"Confederation":40.0,"Assembly":20.0,"Coalition":15.0,"Horde":20.0,"Mercenaries":20.0,"Parliament":20.0,"Senate":20.0,"Association":35.0,"Compact":15.0}},"Rebel Camp":{"scale":4,"function":{"Authority":50.0,"Defense":100.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":75.0,"Social":0.0,"Covert":75.0,"Migration":25.0,"Care":0.0,"Exploration":25.0},"type":{"Government":0.0,"Military":85.0,"Religious":10.0,"Criminal":30.0,"Revolutionary":100.0,"Conspiracy":65.0,"Merchant":10.0,"Clan":55.0,"Cultural":10.0,"Outsider":55.0,"Elite":20.0,"Poor":60.0,"Academic":10.0},"base":{"Empire":50.0,"Kingdom":50.0,"Republic":15.0,"Dominion":50.0,"Council":50.0,"Court":20.0,"Commonwealth":15.0,"Directorate":50.0,"Ministry":50.0,"Authority":50.0,"Legion":70.0,"Guard":70.0,"Army":70.0,"Militia":100.0,"Vanguard":100.0,"Watch":70.0,"Corps":70.0,"Wardens":70.0,"Rangers":70.0,"Order":70.0,"Covenant":50.0,"Faith":50.0,"Sect":15.0,"Brotherhood":20.0,"Fellowship":20.0,"Chapter":15.0,"Disciples":15.0,"Cult":40.0,"Guild":20.0,"Company":20.0,"Consortium":15.0,"Syndicate":40.0,"Cartel":40.0,"Cooperative":40.0,"Corporation":15.0,"Chamber":15.0,"Trust":20.0,"Clan":55.0,"House":25.0,"Dynasty":20.0,"Family":20.0,"Tribe":65.0,"People":20.0,"Community":40.0,"Nation":15.0,"Kindred":55.0,"Lineage":20.0,"Resistance":100.0,"Movement":100.0,"Front":100.0,"Union":50.0,"Collective":50.0,"Alliance":75.0,"League":75.0,"Pact":75.0,"Liberation":100.0,"Academy":20.0,"Institute":20.0,"School":15.0,"Society":15.0,"Circle":40.0,"Cabal":40.0,"Conclave":40.0,"Scholars":15.0,"Collegium":15.0,"Keepers":50.0,"Caravan":40.0,"Fleet":45.0,"Band":90.0,"Enclave":35.0,"Commune":45.0,"Network":40.0,"Expedition":65.0,"Nomads":40.0,"Diaspora":40.0,"Confederation":60.0,"Assembly":60.0,"Coalition":60.0,"Horde":95.0,"Mercenaries":95.0,"Parliament":65.0,"Senate":65.0,"Association":35.0,"Compact":35.0}},"Refugee Camp":{"scale":4,"function":{"Authority":15.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":100.0,"Social":25.0,"Covert":0.0,"Migration":100.0,"Care":75.0,"Exploration":0.0},"type":{"Government":10.0,"Military":10.0,"Religious":50.0,"Criminal":10.0,"Revolutionary":60.0,"Conspiracy":10.0,"Merchant":10.0,"Clan":65.0,"Cultural":15.0,"Outsider":100.0,"Elite":20.0,"Poor":90.0,"Academic":10.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":15.0,"Dominion":15.0,"Council":15.0,"Court":20.0,"Commonwealth":15.0,"Directorate":15.0,"Ministry":15.0,"Authority":15.0,"Legion":20.0,"Guard":15.0,"Army":20.0,"Militia":40.0,"Vanguard":40.0,"Watch":15.0,"Corps":15.0,"Wardens":15.0,"Rangers":50.0,"Order":35.0,"Covenant":35.0,"Faith":35.0,"Sect":35.0,"Brotherhood":65.0,"Fellowship":65.0,"Chapter":60.0,"Disciples":35.0,"Cult":35.0,"Guild":20.0,"Company":20.0,"Consortium":15.0,"Syndicate":20.0,"Cartel":20.0,"Cooperative":45.0,"Corporation":15.0,"Chamber":15.0,"Trust":20.0,"Clan":65.0,"House":75.0,"Dynasty":40.0,"Family":65.0,"Tribe":75.0,"People":65.0,"Community":70.0,"Nation":60.0,"Kindred":65.0,"Lineage":40.0,"Resistance":40.0,"Movement":65.0,"Front":40.0,"Union":75.0,"Collective":75.0,"Alliance":65.0,"League":65.0,"Pact":65.0,"Liberation":90.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":55.0,"Circle":50.0,"Cabal":50.0,"Conclave":50.0,"Scholars":15.0,"Collegium":15.0,"Keepers":35.0,"Caravan":75.0,"Fleet":50.0,"Band":75.0,"Enclave":70.0,"Commune":70.0,"Network":50.0,"Expedition":75.0,"Nomads":75.0,"Diaspora":100.0,"Confederation":50.0,"Assembly":45.0,"Coalition":45.0,"Horde":50.0,"Mercenaries":50.0,"Parliament":50.0,"Senate":50.0,"Association":50.0,"Compact":75.0}},"Slum":{"scale":4,"function":{"Authority":15.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":100.0,"Social":50.0,"Covert":25.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":5.0,"Military":10.0,"Religious":10.0,"Criminal":55.0,"Revolutionary":75.0,"Conspiracy":15.0,"Merchant":10.0,"Clan":30.0,"Cultural":20.0,"Outsider":55.0,"Elite":0.0,"Poor":100.0,"Academic":10.0},"base":{"Empire":15.0,"Kingdom":15.0,"Republic":15.0,"Dominion":15.0,"Council":15.0,"Court":15.0,"Commonwealth":15.0,"Directorate":15.0,"Ministry":15.0,"Authority":15.0,"Legion":20.0,"Guard":15.0,"Army":20.0,"Militia":45.0,"Vanguard":45.0,"Watch":15.0,"Corps":15.0,"Wardens":15.0,"Rangers":35.0,"Order":20.0,"Covenant":20.0,"Faith":15.0,"Sect":15.0,"Brotherhood":30.0,"Fellowship":30.0,"Chapter":15.0,"Disciples":15.0,"Cult":20.0,"Guild":20.0,"Company":20.0,"Consortium":15.0,"Syndicate":40.0,"Cartel":40.0,"Cooperative":50.0,"Corporation":15.0,"Chamber":15.0,"Trust":15.0,"Clan":30.0,"House":85.0,"Dynasty":10.0,"Family":30.0,"Tribe":35.0,"People":55.0,"Community":90.0,"Nation":50.0,"Kindred":30.0,"Lineage":30.0,"Resistance":90.0,"Movement":70.0,"Front":45.0,"Union":90.0,"Collective":75.0,"Alliance":40.0,"League":40.0,"Pact":40.0,"Liberation":90.0,"Academy":20.0,"Institute":20.0,"School":10.0,"Society":20.0,"Circle":20.0,"Cabal":20.0,"Conclave":20.0,"Scholars":15.0,"Collegium":15.0,"Keepers":15.0,"Caravan":40.0,"Fleet":35.0,"Band":35.0,"Enclave":55.0,"Commune":40.0,"Network":35.0,"Expedition":35.0,"Nomads":40.0,"Diaspora":65.0,"Confederation":35.0,"Assembly":35.0,"Coalition":35.0,"Horde":35.0,"Mercenaries":35.0,"Parliament":35.0,"Senate":35.0,"Association":35.0,"Compact":35.0}},"University":{"scale":4,"function":{"Authority":45.0,"Defense":0.0,"Religion":0.0,"Knowledge":100.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":50.0,"Social":50.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":50.0,"Military":20.0,"Religious":40.0,"Criminal":15.0,"Revolutionary":25.0,"Conspiracy":35.0,"Merchant":20.0,"Clan":40.0,"Cultural":50.0,"Outsider":20.0,"Elite":35.0,"Poor":25.0,"Academic":90.0},"base":{"Empire":45.0,"Kingdom":45.0,"Republic":45.0,"Dominion":45.0,"Council":60.0,"Court":45.0,"Commonwealth":45.0,"Directorate":65.0,"Ministry":45.0,"Authority":45.0,"Legion":30.0,"Guard":40.0,"Army":30.0,"Militia":35.0,"Vanguard":30.0,"Watch":40.0,"Corps":30.0,"Wardens":40.0,"Rangers":30.0,"Order":65.0,"Covenant":65.0,"Faith":65.0,"Sect":60.0,"Brotherhood":60.0,"Fellowship":60.0,"Chapter":65.0,"Disciples":60.0,"Cult":65.0,"Guild":35.0,"Company":35.0,"Consortium":35.0,"Syndicate":30.0,"Cartel":30.0,"Cooperative":40.0,"Corporation":35.0,"Chamber":45.0,"Trust":35.0,"Clan":40.0,"House":50.0,"Dynasty":40.0,"Family":35.0,"Tribe":40.0,"People":35.0,"Community":40.0,"Nation":35.0,"Kindred":40.0,"Lineage":40.0,"Resistance":35.0,"Movement":35.0,"Front":35.0,"Union":35.0,"Collective":35.0,"Alliance":30.0,"League":30.0,"Pact":35.0,"Liberation":35.0,"Academy":100.0,"Institute":100.0,"School":80.0,"Society":90.0,"Circle":80.0,"Cabal":60.0,"Conclave":60.0,"Scholars":100.0,"Collegium":100.0,"Keepers":75.0,"Caravan":30.0,"Fleet":30.0,"Band":30.0,"Enclave":45.0,"Commune":30.0,"Network":35.0,"Expedition":50.0,"Nomads":30.0,"Diaspora":30.0,"Confederation":40.0,"Assembly":40.0,"Coalition":35.0,"Horde":30.0,"Mercenaries":30.0,"Parliament":45.0,"Senate":45.0,"Association":60.0,"Compact":35.0}},"Academic District":{"scale":5,"function":{"Authority":35.0,"Defense":0.0,"Religion":0.0,"Knowledge":100.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":50.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":50.0,"Military":10.0,"Religious":70.0,"Criminal":10.0,"Revolutionary":10.0,"Conspiracy":75.0,"Merchant":10.0,"Clan":20.0,"Cultural":25.0,"Outsider":10.0,"Elite":10.0,"Poor":10.0,"Academic":100.0},"base":{"Empire":35.0,"Kingdom":35.0,"Republic":35.0,"Dominion":35.0,"Council":85.0,"Court":35.0,"Commonwealth":35.0,"Directorate":85.0,"Ministry":35.0,"Authority":35.0,"Legion":20.0,"Guard":35.0,"Army":20.0,"Militia":15.0,"Vanguard":15.0,"Watch":35.0,"Corps":15.0,"Wardens":35.0,"Rangers":15.0,"Order":70.0,"Covenant":70.0,"Faith":70.0,"Sect":70.0,"Brotherhood":65.0,"Fellowship":65.0,"Chapter":65.0,"Disciples":65.0,"Cult":70.0,"Guild":20.0,"Company":20.0,"Consortium":15.0,"Syndicate":20.0,"Cartel":20.0,"Cooperative":15.0,"Corporation":15.0,"Chamber":35.0,"Trust":15.0,"Clan":20.0,"House":15.0,"Dynasty":20.0,"Family":20.0,"Tribe":20.0,"People":15.0,"Community":20.0,"Nation":15.0,"Kindred":20.0,"Lineage":20.0,"Resistance":20.0,"Movement":20.0,"Front":15.0,"Union":20.0,"Collective":20.0,"Alliance":15.0,"League":15.0,"Pact":40.0,"Liberation":15.0,"Academy":100.0,"Institute":100.0,"School":100.0,"Society":100.0,"Circle":100.0,"Cabal":50.0,"Conclave":50.0,"Scholars":100.0,"Collegium":100.0,"Keepers":75.0,"Caravan":15.0,"Fleet":15.0,"Band":15.0,"Enclave":5.0,"Commune":15.0,"Network":40.0,"Expedition":50.0,"Nomads":20.0,"Diaspora":20.0,"Confederation":35.0,"Assembly":35.0,"Coalition":35.0,"Horde":15.0,"Mercenaries":15.0,"Parliament":35.0,"Senate":35.0,"Association":80.0,"Compact":35.0}},"Government District":{"scale":5,"function":{"Authority":100.0,"Defense":25.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":25.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":100.0,"Military":55.0,"Religious":10.0,"Criminal":0.0,"Revolutionary":0.0,"Conspiracy":15.0,"Merchant":10.0,"Clan":20.0,"Cultural":15.0,"Outsider":10.0,"Elite":65.0,"Poor":10.0,"Academic":45.0},"base":{"Empire":75.0,"Kingdom":75.0,"Republic":100.0,"Dominion":75.0,"Council":100.0,"Court":100.0,"Commonwealth":100.0,"Directorate":100.0,"Ministry":100.0,"Authority":100.0,"Legion":65.0,"Guard":75.0,"Army":65.0,"Militia":35.0,"Vanguard":35.0,"Watch":75.0,"Corps":65.0,"Wardens":75.0,"Rangers":35.0,"Order":65.0,"Covenant":50.0,"Faith":50.0,"Sect":15.0,"Brotherhood":15.0,"Fellowship":15.0,"Chapter":15.0,"Disciples":15.0,"Cult":20.0,"Guild":20.0,"Company":20.0,"Consortium":15.0,"Syndicate":20.0,"Cartel":20.0,"Cooperative":15.0,"Corporation":15.0,"Chamber":50.0,"Trust":90.0,"Clan":20.0,"House":10.0,"Dynasty":65.0,"Family":15.0,"Tribe":15.0,"People":15.0,"Community":15.0,"Nation":10.0,"Kindred":15.0,"Lineage":50.0,"Resistance":15.0,"Movement":15.0,"Front":15.0,"Union":20.0,"Collective":20.0,"Alliance":15.0,"League":15.0,"Pact":20.0,"Liberation":15.0,"Academy":35.0,"Institute":35.0,"School":10.0,"Society":35.0,"Circle":35.0,"Cabal":35.0,"Conclave":35.0,"Scholars":35.0,"Collegium":35.0,"Keepers":35.0,"Caravan":15.0,"Fleet":35.0,"Band":15.0,"Enclave":0.0,"Commune":15.0,"Network":20.0,"Expedition":35.0,"Nomads":20.0,"Diaspora":20.0,"Confederation":100.0,"Assembly":100.0,"Coalition":100.0,"Horde":35.0,"Mercenaries":35.0,"Parliament":100.0,"Senate":100.0,"Association":20.0,"Compact":45.0}},"Artisan Quarter":{"scale":5,"function":{"Authority":15.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":100.0,"Production":100.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":25.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":10.0,"Military":10.0,"Religious":10.0,"Criminal":15.0,"Revolutionary":45.0,"Conspiracy":10.0,"Merchant":95.0,"Clan":35.0,"Cultural":70.0,"Outsider":10.0,"Elite":20.0,"Poor":65.0,"Academic":10.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":50.0,"Dominion":15.0,"Council":15.0,"Court":20.0,"Commonwealth":50.0,"Directorate":15.0,"Ministry":15.0,"Authority":15.0,"Legion":20.0,"Guard":15.0,"Army":20.0,"Militia":35.0,"Vanguard":35.0,"Watch":15.0,"Corps":15.0,"Wardens":15.0,"Rangers":15.0,"Order":20.0,"Covenant":20.0,"Faith":15.0,"Sect":15.0,"Brotherhood":85.0,"Fellowship":85.0,"Chapter":75.0,"Disciples":15.0,"Cult":15.0,"Guild":100.0,"Company":100.0,"Consortium":75.0,"Syndicate":75.0,"Cartel":75.0,"Cooperative":100.0,"Corporation":75.0,"Chamber":75.0,"Trust":75.0,"Clan":35.0,"House":15.0,"Dynasty":35.0,"Family":35.0,"Tribe":35.0,"People":35.0,"Community":40.0,"Nation":30.0,"Kindred":35.0,"Lineage":35.0,"Resistance":35.0,"Movement":35.0,"Front":35.0,"Union":90.0,"Collective":90.0,"Alliance":60.0,"League":60.0,"Pact":60.0,"Liberation":35.0,"Academy":20.0,"Institute":20.0,"School":10.0,"Society":40.0,"Circle":20.0,"Cabal":20.0,"Conclave":15.0,"Scholars":15.0,"Collegium":15.0,"Keepers":15.0,"Caravan":75.0,"Fleet":75.0,"Band":20.0,"Enclave":5.0,"Commune":30.0,"Network":75.0,"Expedition":15.0,"Nomads":50.0,"Diaspora":20.0,"Confederation":60.0,"Assembly":50.0,"Coalition":50.0,"Horde":15.0,"Mercenaries":45.0,"Parliament":15.0,"Senate":20.0,"Association":100.0,"Compact":70.0}},"Cultural District":{"scale":5,"function":{"Authority":15.0,"Defense":0.0,"Religion":0.0,"Knowledge":25.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":100.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":10.0,"Military":10.0,"Religious":50.0,"Criminal":15.0,"Revolutionary":45.0,"Conspiracy":10.0,"Merchant":10.0,"Clan":40.0,"Cultural":100.0,"Outsider":10.0,"Elite":65.0,"Poor":10.0,"Academic":70.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":15.0,"Dominion":15.0,"Council":15.0,"Court":40.0,"Commonwealth":15.0,"Directorate":15.0,"Ministry":15.0,"Authority":15.0,"Legion":20.0,"Guard":15.0,"Army":20.0,"Militia":35.0,"Vanguard":35.0,"Watch":15.0,"Corps":15.0,"Wardens":15.0,"Rangers":15.0,"Order":35.0,"Covenant":35.0,"Faith":35.0,"Sect":35.0,"Brotherhood":35.0,"Fellowship":35.0,"Chapter":35.0,"Disciples":35.0,"Cult":35.0,"Guild":20.0,"Company":20.0,"Consortium":15.0,"Syndicate":20.0,"Cartel":20.0,"Cooperative":15.0,"Corporation":15.0,"Chamber":15.0,"Trust":40.0,"Clan":40.0,"House":20.0,"Dynasty":40.0,"Family":40.0,"Tribe":35.0,"People":35.0,"Community":35.0,"Nation":30.0,"Kindred":35.0,"Lineage":35.0,"Resistance":35.0,"Movement":85.0,"Front":35.0,"Union":35.0,"Collective":35.0,"Alliance":35.0,"League":35.0,"Pact":35.0,"Liberation":35.0,"Academy":40.0,"Institute":40.0,"School":30.0,"Society":50.0,"Circle":40.0,"Cabal":40.0,"Conclave":40.0,"Scholars":40.0,"Collegium":40.0,"Keepers":40.0,"Caravan":15.0,"Fleet":15.0,"Band":20.0,"Enclave":5.0,"Commune":30.0,"Network":20.0,"Expedition":40.0,"Nomads":20.0,"Diaspora":20.0,"Confederation":35.0,"Assembly":15.0,"Coalition":15.0,"Horde":15.0,"Mercenaries":15.0,"Parliament":15.0,"Senate":40.0,"Association":50.0,"Compact":15.0}},"Entertainment District":{"scale":5,"function":{"Authority":15.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":75.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":100.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":10.0,"Military":10.0,"Religious":15.0,"Criminal":75.0,"Revolutionary":10.0,"Conspiracy":10.0,"Merchant":85.0,"Clan":20.0,"Cultural":90.0,"Outsider":45.0,"Elite":55.0,"Poor":45.0,"Academic":10.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":50.0,"Dominion":15.0,"Council":15.0,"Court":40.0,"Commonwealth":50.0,"Directorate":15.0,"Ministry":15.0,"Authority":15.0,"Legion":20.0,"Guard":15.0,"Army":20.0,"Militia":15.0,"Vanguard":15.0,"Watch":15.0,"Corps":15.0,"Wardens":15.0,"Rangers":35.0,"Order":20.0,"Covenant":20.0,"Faith":20.0,"Sect":20.0,"Brotherhood":55.0,"Fellowship":55.0,"Chapter":50.0,"Disciples":20.0,"Cult":20.0,"Guild":70.0,"Company":70.0,"Consortium":70.0,"Syndicate":70.0,"Cartel":70.0,"Cooperative":70.0,"Corporation":70.0,"Chamber":70.0,"Trust":70.0,"Clan":20.0,"House":20.0,"Dynasty":40.0,"Family":20.0,"Tribe":35.0,"People":20.0,"Community":35.0,"Nation":15.0,"Kindred":20.0,"Lineage":20.0,"Resistance":20.0,"Movement":50.0,"Front":15.0,"Union":60.0,"Collective":60.0,"Alliance":50.0,"League":50.0,"Pact":50.0,"Liberation":15.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":45.0,"Circle":20.0,"Cabal":20.0,"Conclave":15.0,"Scholars":15.0,"Collegium":15.0,"Keepers":15.0,"Caravan":70.0,"Fleet":70.0,"Band":45.0,"Enclave":0.0,"Commune":30.0,"Network":70.0,"Expedition":35.0,"Nomads":60.0,"Diaspora":35.0,"Confederation":60.0,"Assembly":60.0,"Coalition":60.0,"Horde":35.0,"Mercenaries":45.0,"Parliament":35.0,"Senate":35.0,"Association":70.0,"Compact":70.0}},"Farmland":{"scale":5,"function":{"Authority":15.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":25.0,"Logistics":0.0,"Resources":100.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":10.0,"Military":10.0,"Religious":45.0,"Criminal":10.0,"Revolutionary":10.0,"Conspiracy":10.0,"Merchant":75.0,"Clan":45.0,"Cultural":10.0,"Outsider":45.0,"Elite":10.0,"Poor":75.0,"Academic":10.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":15.0,"Dominion":15.0,"Council":15.0,"Court":15.0,"Commonwealth":15.0,"Directorate":15.0,"Ministry":15.0,"Authority":15.0,"Legion":20.0,"Guard":15.0,"Army":20.0,"Militia":15.0,"Vanguard":15.0,"Watch":15.0,"Corps":15.0,"Wardens":15.0,"Rangers":35.0,"Order":35.0,"Covenant":35.0,"Faith":35.0,"Sect":35.0,"Brotherhood":40.0,"Fellowship":40.0,"Chapter":35.0,"Disciples":35.0,"Cult":35.0,"Guild":45.0,"Company":45.0,"Consortium":45.0,"Syndicate":45.0,"Cartel":45.0,"Cooperative":70.0,"Corporation":45.0,"Chamber":40.0,"Trust":40.0,"Clan":45.0,"House":10.0,"Dynasty":45.0,"Family":45.0,"Tribe":40.0,"People":65.0,"Community":70.0,"Nation":65.0,"Kindred":45.0,"Lineage":45.0,"Resistance":20.0,"Movement":20.0,"Front":15.0,"Union":45.0,"Collective":45.0,"Alliance":15.0,"League":15.0,"Pact":15.0,"Liberation":15.0,"Academy":20.0,"Institute":20.0,"School":10.0,"Society":15.0,"Circle":20.0,"Cabal":20.0,"Conclave":15.0,"Scholars":15.0,"Collegium":15.0,"Keepers":35.0,"Caravan":45.0,"Fleet":40.0,"Band":35.0,"Enclave":5.0,"Commune":55.0,"Network":35.0,"Expedition":35.0,"Nomads":35.0,"Diaspora":35.0,"Confederation":40.0,"Assembly":30.0,"Coalition":30.0,"Horde":35.0,"Mercenaries":40.0,"Parliament":35.0,"Senate":35.0,"Association":40.0,"Compact":40.0}},"Foreign Quarter":{"scale":5,"function":{"Authority":15.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":50.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":100.0,"Social":50.0,"Covert":0.0,"Migration":100.0,"Care":0.0,"Exploration":0.0},"type":{"Government":10.0,"Military":10.0,"Religious":10.0,"Criminal":15.0,"Revolutionary":20.0,"Conspiracy":10.0,"Merchant":65.0,"Clan":65.0,"Cultural":60.0,"Outsider":100.0,"Elite":25.0,"Poor":65.0,"Academic":10.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":15.0,"Dominion":15.0,"Council":15.0,"Court":30.0,"Commonwealth":15.0,"Directorate":15.0,"Ministry":15.0,"Authority":15.0,"Legion":20.0,"Guard":15.0,"Army":20.0,"Militia":20.0,"Vanguard":20.0,"Watch":15.0,"Corps":15.0,"Wardens":15.0,"Rangers":50.0,"Order":20.0,"Covenant":20.0,"Faith":15.0,"Sect":15.0,"Brotherhood":65.0,"Fellowship":65.0,"Chapter":50.0,"Disciples":15.0,"Cult":15.0,"Guild":40.0,"Company":40.0,"Consortium":40.0,"Syndicate":40.0,"Cartel":40.0,"Cooperative":40.0,"Corporation":40.0,"Chamber":40.0,"Trust":40.0,"Clan":65.0,"House":80.0,"Dynasty":40.0,"Family":65.0,"Tribe":75.0,"People":40.0,"Community":40.0,"Nation":35.0,"Kindred":65.0,"Lineage":40.0,"Resistance":20.0,"Movement":20.0,"Front":20.0,"Union":40.0,"Collective":40.0,"Alliance":55.0,"League":55.0,"Pact":55.0,"Liberation":20.0,"Academy":20.0,"Institute":20.0,"School":10.0,"Society":65.0,"Circle":50.0,"Cabal":50.0,"Conclave":50.0,"Scholars":15.0,"Collegium":15.0,"Keepers":15.0,"Caravan":100.0,"Fleet":50.0,"Band":75.0,"Enclave":100.0,"Commune":70.0,"Network":50.0,"Expedition":75.0,"Nomads":75.0,"Diaspora":100.0,"Confederation":50.0,"Assembly":45.0,"Coalition":45.0,"Horde":50.0,"Mercenaries":50.0,"Parliament":50.0,"Senate":50.0,"Association":50.0,"Compact":75.0}},"Wilderness":{"scale":5,"function":{"Authority":15.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":50.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":75.0},"type":{"Government":5.0,"Military":10.0,"Religious":45.0,"Criminal":20.0,"Revolutionary":45.0,"Conspiracy":10.0,"Merchant":10.0,"Clan":40.0,"Cultural":10.0,"Outsider":90.0,"Elite":10.0,"Poor":30.0,"Academic":15.0},"base":{"Empire":15.0,"Kingdom":15.0,"Republic":15.0,"Dominion":15.0,"Council":15.0,"Court":15.0,"Commonwealth":15.0,"Directorate":15.0,"Ministry":15.0,"Authority":15.0,"Legion":20.0,"Guard":15.0,"Army":20.0,"Militia":35.0,"Vanguard":35.0,"Watch":15.0,"Corps":15.0,"Wardens":15.0,"Rangers":90.0,"Order":35.0,"Covenant":35.0,"Faith":35.0,"Sect":35.0,"Brotherhood":40.0,"Fellowship":40.0,"Chapter":35.0,"Disciples":35.0,"Cult":35.0,"Guild":20.0,"Company":20.0,"Consortium":15.0,"Syndicate":35.0,"Cartel":35.0,"Cooperative":70.0,"Corporation":15.0,"Chamber":15.0,"Trust":15.0,"Clan":40.0,"House":10.0,"Dynasty":40.0,"Family":40.0,"Tribe":90.0,"People":65.0,"Community":65.0,"Nation":60.0,"Kindred":40.0,"Lineage":40.0,"Resistance":35.0,"Movement":35.0,"Front":35.0,"Union":40.0,"Collective":40.0,"Alliance":35.0,"League":35.0,"Pact":35.0,"Liberation":35.0,"Academy":20.0,"Institute":20.0,"School":10.0,"Society":20.0,"Circle":20.0,"Cabal":20.0,"Conclave":20.0,"Scholars":20.0,"Collegium":20.0,"Keepers":35.0,"Caravan":45.0,"Fleet":45.0,"Band":45.0,"Enclave":5.0,"Commune":70.0,"Network":45.0,"Expedition":90.0,"Nomads":90.0,"Diaspora":45.0,"Confederation":45.0,"Assembly":45.0,"Coalition":45.0,"Horde":90.0,"Mercenaries":45.0,"Parliament":45.0,"Senate":45.0,"Association":45.0,"Compact":45.0}},"Harbor":{"scale":5,"function":{"Authority":35.0,"Defense":25.0,"Religion":0.0,"Knowledge":0.0,"Commerce":100.0,"Production":0.0,"Logistics":100.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":75.0,"Care":0.0,"Exploration":0.0},"type":{"Government":50.0,"Military":55.0,"Religious":10.0,"Criminal":55.0,"Revolutionary":15.0,"Conspiracy":10.0,"Merchant":95.0,"Clan":20.0,"Cultural":10.0,"Outsider":85.0,"Elite":20.0,"Poor":15.0,"Academic":10.0},"base":{"Empire":35.0,"Kingdom":35.0,"Republic":65.0,"Dominion":35.0,"Council":35.0,"Court":35.0,"Commonwealth":60.0,"Directorate":35.0,"Ministry":35.0,"Authority":35.0,"Legion":40.0,"Guard":40.0,"Army":40.0,"Militia":35.0,"Vanguard":35.0,"Watch":35.0,"Corps":35.0,"Wardens":35.0,"Rangers":45.0,"Order":40.0,"Covenant":20.0,"Faith":15.0,"Sect":15.0,"Brotherhood":50.0,"Fellowship":50.0,"Chapter":50.0,"Disciples":15.0,"Cult":15.0,"Guild":75.0,"Company":75.0,"Consortium":100.0,"Syndicate":75.0,"Cartel":90.0,"Cooperative":75.0,"Corporation":100.0,"Chamber":100.0,"Trust":75.0,"Clan":20.0,"House":10.0,"Dynasty":20.0,"Family":15.0,"Tribe":45.0,"People":15.0,"Community":20.0,"Nation":10.0,"Kindred":15.0,"Lineage":15.0,"Resistance":20.0,"Movement":20.0,"Front":20.0,"Union":55.0,"Collective":55.0,"Alliance":55.0,"League":50.0,"Pact":50.0,"Liberation":20.0,"Academy":20.0,"Institute":20.0,"School":10.0,"Society":15.0,"Circle":20.0,"Cabal":20.0,"Conclave":15.0,"Scholars":15.0,"Collegium":15.0,"Keepers":15.0,"Caravan":75.0,"Fleet":100.0,"Band":70.0,"Enclave":35.0,"Commune":40.0,"Network":70.0,"Expedition":70.0,"Nomads":70.0,"Diaspora":45.0,"Confederation":70.0,"Assembly":70.0,"Coalition":70.0,"Horde":45.0,"Mercenaries":45.0,"Parliament":45.0,"Senate":45.0,"Association":75.0,"Compact":70.0}},"Industrial District":{"scale":5,"function":{"Authority":15.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":100.0,"Logistics":50.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":10.0,"Military":50.0,"Religious":10.0,"Criminal":10.0,"Revolutionary":45.0,"Conspiracy":10.0,"Merchant":80.0,"Clan":20.0,"Cultural":45.0,"Outsider":10.0,"Elite":10.0,"Poor":75.0,"Academic":10.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":15.0,"Dominion":15.0,"Council":15.0,"Court":15.0,"Commonwealth":15.0,"Directorate":15.0,"Ministry":15.0,"Authority":15.0,"Legion":35.0,"Guard":35.0,"Army":35.0,"Militia":35.0,"Vanguard":35.0,"Watch":35.0,"Corps":35.0,"Wardens":35.0,"Rangers":35.0,"Order":35.0,"Covenant":20.0,"Faith":15.0,"Sect":15.0,"Brotherhood":50.0,"Fellowship":50.0,"Chapter":50.0,"Disciples":15.0,"Cult":15.0,"Guild":95.0,"Company":95.0,"Consortium":95.0,"Syndicate":45.0,"Cartel":45.0,"Cooperative":95.0,"Corporation":95.0,"Chamber":95.0,"Trust":45.0,"Clan":20.0,"House":10.0,"Dynasty":20.0,"Family":15.0,"Tribe":15.0,"People":15.0,"Community":40.0,"Nation":10.0,"Kindred":15.0,"Lineage":15.0,"Resistance":35.0,"Movement":35.0,"Front":35.0,"Union":95.0,"Collective":95.0,"Alliance":35.0,"League":35.0,"Pact":35.0,"Liberation":35.0,"Academy":20.0,"Institute":20.0,"School":10.0,"Society":35.0,"Circle":20.0,"Cabal":20.0,"Conclave":15.0,"Scholars":15.0,"Collegium":15.0,"Keepers":15.0,"Caravan":45.0,"Fleet":45.0,"Band":15.0,"Enclave":0.0,"Commune":30.0,"Network":50.0,"Expedition":15.0,"Nomads":20.0,"Diaspora":20.0,"Confederation":15.0,"Assembly":15.0,"Coalition":15.0,"Horde":35.0,"Mercenaries":45.0,"Parliament":15.0,"Senate":15.0,"Association":70.0,"Compact":40.0}},"Market District":{"scale":5,"function":{"Authority":15.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":100.0,"Production":0.0,"Logistics":25.0,"Resources":0.0,"Residence":0.0,"Social":75.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":10.0,"Military":10.0,"Religious":10.0,"Criminal":75.0,"Revolutionary":10.0,"Conspiracy":10.0,"Merchant":100.0,"Clan":20.0,"Cultural":25.0,"Outsider":45.0,"Elite":55.0,"Poor":10.0,"Academic":10.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":75.0,"Dominion":15.0,"Council":15.0,"Court":40.0,"Commonwealth":75.0,"Directorate":15.0,"Ministry":15.0,"Authority":15.0,"Legion":20.0,"Guard":15.0,"Army":20.0,"Militia":15.0,"Vanguard":15.0,"Watch":15.0,"Corps":15.0,"Wardens":15.0,"Rangers":35.0,"Order":20.0,"Covenant":20.0,"Faith":15.0,"Sect":15.0,"Brotherhood":50.0,"Fellowship":50.0,"Chapter":50.0,"Disciples":15.0,"Cult":15.0,"Guild":100.0,"Company":100.0,"Consortium":100.0,"Syndicate":75.0,"Cartel":75.0,"Cooperative":75.0,"Corporation":100.0,"Chamber":100.0,"Trust":100.0,"Clan":20.0,"House":15.0,"Dynasty":40.0,"Family":20.0,"Tribe":35.0,"People":15.0,"Community":20.0,"Nation":15.0,"Kindred":20.0,"Lineage":20.0,"Resistance":20.0,"Movement":20.0,"Front":15.0,"Union":50.0,"Collective":50.0,"Alliance":75.0,"League":75.0,"Pact":75.0,"Liberation":15.0,"Academy":20.0,"Institute":20.0,"School":10.0,"Society":30.0,"Circle":20.0,"Cabal":20.0,"Conclave":15.0,"Scholars":15.0,"Collegium":15.0,"Keepers":15.0,"Caravan":100.0,"Fleet":75.0,"Band":45.0,"Enclave":0.0,"Commune":30.0,"Network":70.0,"Expedition":35.0,"Nomads":85.0,"Diaspora":35.0,"Confederation":85.0,"Assembly":85.0,"Coalition":85.0,"Horde":35.0,"Mercenaries":50.0,"Parliament":35.0,"Senate":35.0,"Association":100.0,"Compact":100.0}},"Military District":{"scale":5,"function":{"Authority":70.0,"Defense":100.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":50.0,"Resources":0.0,"Residence":25.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":75.0,"Military":90.0,"Religious":10.0,"Criminal":10.0,"Revolutionary":80.0,"Conspiracy":10.0,"Merchant":15.0,"Clan":65.0,"Cultural":10.0,"Outsider":10.0,"Elite":10.0,"Poor":10.0,"Academic":10.0},"base":{"Empire":70.0,"Kingdom":70.0,"Republic":45.0,"Dominion":70.0,"Council":70.0,"Court":45.0,"Commonwealth":45.0,"Directorate":70.0,"Ministry":70.0,"Authority":70.0,"Legion":100.0,"Guard":100.0,"Army":100.0,"Militia":75.0,"Vanguard":75.0,"Watch":100.0,"Corps":100.0,"Wardens":95.0,"Rangers":95.0,"Order":75.0,"Covenant":50.0,"Faith":50.0,"Sect":15.0,"Brotherhood":35.0,"Fellowship":35.0,"Chapter":15.0,"Disciples":15.0,"Cult":15.0,"Guild":20.0,"Company":20.0,"Consortium":20.0,"Syndicate":20.0,"Cartel":20.0,"Cooperative":20.0,"Corporation":20.0,"Chamber":40.0,"Trust":20.0,"Clan":65.0,"House":15.0,"Dynasty":35.0,"Family":35.0,"Tribe":60.0,"People":35.0,"Community":35.0,"Nation":30.0,"Kindred":65.0,"Lineage":35.0,"Resistance":70.0,"Movement":70.0,"Front":70.0,"Union":45.0,"Collective":45.0,"Alliance":70.0,"League":70.0,"Pact":70.0,"Liberation":70.0,"Academy":20.0,"Institute":20.0,"School":10.0,"Society":15.0,"Circle":20.0,"Cabal":20.0,"Conclave":15.0,"Scholars":15.0,"Collegium":15.0,"Keepers":50.0,"Caravan":20.0,"Fleet":45.0,"Band":50.0,"Enclave":5.0,"Commune":40.0,"Network":15.0,"Expedition":75.0,"Nomads":20.0,"Diaspora":20.0,"Confederation":65.0,"Assembly":65.0,"Coalition":65.0,"Horde":70.0,"Mercenaries":70.0,"Parliament":70.0,"Senate":70.0,"Association":20.0,"Compact":40.0}},"Mining Region":{"scale":5,"function":{"Authority":15.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":25.0,"Logistics":0.0,"Resources":100.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":25.0},"type":{"Government":10.0,"Military":45.0,"Religious":10.0,"Criminal":10.0,"Revolutionary":10.0,"Conspiracy":10.0,"Merchant":75.0,"Clan":20.0,"Cultural":10.0,"Outsider":50.0,"Elite":10.0,"Poor":85.0,"Academic":10.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":15.0,"Dominion":15.0,"Council":15.0,"Court":15.0,"Commonwealth":15.0,"Directorate":15.0,"Ministry":15.0,"Authority":15.0,"Legion":35.0,"Guard":35.0,"Army":35.0,"Militia":35.0,"Vanguard":35.0,"Watch":35.0,"Corps":35.0,"Wardens":35.0,"Rangers":35.0,"Order":35.0,"Covenant":20.0,"Faith":15.0,"Sect":15.0,"Brotherhood":20.0,"Fellowship":20.0,"Chapter":15.0,"Disciples":15.0,"Cult":15.0,"Guild":45.0,"Company":45.0,"Consortium":45.0,"Syndicate":45.0,"Cartel":45.0,"Cooperative":45.0,"Corporation":45.0,"Chamber":40.0,"Trust":40.0,"Clan":20.0,"House":10.0,"Dynasty":20.0,"Family":20.0,"Tribe":35.0,"People":20.0,"Community":45.0,"Nation":15.0,"Kindred":20.0,"Lineage":20.0,"Resistance":20.0,"Movement":20.0,"Front":15.0,"Union":45.0,"Collective":45.0,"Alliance":15.0,"League":15.0,"Pact":15.0,"Liberation":15.0,"Academy":20.0,"Institute":20.0,"School":10.0,"Society":15.0,"Circle":20.0,"Cabal":20.0,"Conclave":15.0,"Scholars":15.0,"Collegium":15.0,"Keepers":15.0,"Caravan":45.0,"Fleet":40.0,"Band":35.0,"Enclave":5.0,"Commune":30.0,"Network":35.0,"Expedition":65.0,"Nomads":35.0,"Diaspora":35.0,"Confederation":35.0,"Assembly":35.0,"Coalition":35.0,"Horde":35.0,"Mercenaries":40.0,"Parliament":35.0,"Senate":35.0,"Association":40.0,"Compact":40.0}},"Noble Quarter":{"scale":5,"function":{"Authority":65.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":100.0,"Social":25.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":60.0,"Military":10.0,"Religious":10.0,"Criminal":10.0,"Revolutionary":15.0,"Conspiracy":70.0,"Merchant":10.0,"Clan":40.0,"Cultural":15.0,"Outsider":20.0,"Elite":100.0,"Poor":0.0,"Academic":10.0},"base":{"Empire":65.0,"Kingdom":65.0,"Republic":40.0,"Dominion":65.0,"Council":40.0,"Court":75.0,"Commonwealth":40.0,"Directorate":40.0,"Ministry":65.0,"Authority":65.0,"Legion":20.0,"Guard":40.0,"Army":20.0,"Militia":20.0,"Vanguard":20.0,"Watch":40.0,"Corps":15.0,"Wardens":40.0,"Rangers":20.0,"Order":20.0,"Covenant":20.0,"Faith":15.0,"Sect":15.0,"Brotherhood":40.0,"Fellowship":40.0,"Chapter":15.0,"Disciples":15.0,"Cult":40.0,"Guild":20.0,"Company":20.0,"Consortium":15.0,"Syndicate":20.0,"Cartel":20.0,"Cooperative":15.0,"Corporation":15.0,"Chamber":40.0,"Trust":75.0,"Clan":40.0,"House":70.0,"Dynasty":90.0,"Family":65.0,"Tribe":40.0,"People":40.0,"Community":40.0,"Nation":35.0,"Kindred":40.0,"Lineage":65.0,"Resistance":20.0,"Movement":20.0,"Front":20.0,"Union":20.0,"Collective":20.0,"Alliance":20.0,"League":20.0,"Pact":40.0,"Liberation":20.0,"Academy":20.0,"Institute":20.0,"School":10.0,"Society":55.0,"Circle":70.0,"Cabal":40.0,"Conclave":40.0,"Scholars":15.0,"Collegium":15.0,"Keepers":15.0,"Caravan":20.0,"Fleet":20.0,"Band":20.0,"Enclave":55.0,"Commune":15.0,"Network":40.0,"Expedition":20.0,"Nomads":20.0,"Diaspora":20.0,"Confederation":40.0,"Assembly":35.0,"Coalition":35.0,"Horde":20.0,"Mercenaries":20.0,"Parliament":65.0,"Senate":75.0,"Association":20.0,"Compact":35.0}},"Royal Quarter":{"scale":5,"function":{"Authority":100.0,"Defense":25.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":75.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":95.0,"Military":15.0,"Religious":45.0,"Criminal":10.0,"Revolutionary":15.0,"Conspiracy":50.0,"Merchant":10.0,"Clan":45.0,"Cultural":45.0,"Outsider":15.0,"Elite":100.0,"Poor":0.0,"Academic":10.0},"base":{"Empire":100.0,"Kingdom":100.0,"Republic":75.0,"Dominion":100.0,"Council":75.0,"Court":100.0,"Commonwealth":75.0,"Directorate":75.0,"Ministry":100.0,"Authority":100.0,"Legion":55.0,"Guard":75.0,"Army":55.0,"Militia":20.0,"Vanguard":20.0,"Watch":75.0,"Corps":55.0,"Wardens":75.0,"Rangers":20.0,"Order":60.0,"Covenant":60.0,"Faith":60.0,"Sect":35.0,"Brotherhood":45.0,"Fellowship":45.0,"Chapter":35.0,"Disciples":35.0,"Cult":35.0,"Guild":20.0,"Company":20.0,"Consortium":15.0,"Syndicate":20.0,"Cartel":20.0,"Cooperative":15.0,"Corporation":15.0,"Chamber":50.0,"Trust":75.0,"Clan":45.0,"House":55.0,"Dynasty":100.0,"Family":70.0,"Tribe":45.0,"People":40.0,"Community":45.0,"Nation":40.0,"Kindred":45.0,"Lineage":95.0,"Resistance":20.0,"Movement":20.0,"Front":20.0,"Union":20.0,"Collective":20.0,"Alliance":20.0,"League":20.0,"Pact":35.0,"Liberation":20.0,"Academy":20.0,"Institute":20.0,"School":10.0,"Society":60.0,"Circle":65.0,"Cabal":35.0,"Conclave":35.0,"Scholars":15.0,"Collegium":15.0,"Keepers":35.0,"Caravan":20.0,"Fleet":20.0,"Band":20.0,"Enclave":45.0,"Commune":15.0,"Network":35.0,"Expedition":20.0,"Nomads":20.0,"Diaspora":20.0,"Confederation":75.0,"Assembly":70.0,"Coalition":70.0,"Horde":15.0,"Mercenaries":20.0,"Parliament":100.0,"Senate":100.0,"Association":35.0,"Compact":45.0}},"Residential District":{"scale":5,"function":{"Authority":15.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":100.0,"Social":50.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":10.0,"Military":10.0,"Religious":10.0,"Criminal":10.0,"Revolutionary":15.0,"Conspiracy":10.0,"Merchant":10.0,"Clan":95.0,"Cultural":75.0,"Outsider":55.0,"Elite":60.0,"Poor":80.0,"Academic":10.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":75.0,"Dominion":15.0,"Council":15.0,"Court":40.0,"Commonwealth":75.0,"Directorate":15.0,"Ministry":15.0,"Authority":15.0,"Legion":20.0,"Guard":15.0,"Army":20.0,"Militia":20.0,"Vanguard":20.0,"Watch":15.0,"Corps":15.0,"Wardens":15.0,"Rangers":35.0,"Order":20.0,"Covenant":20.0,"Faith":15.0,"Sect":15.0,"Brotherhood":45.0,"Fellowship":45.0,"Chapter":15.0,"Disciples":15.0,"Cult":15.0,"Guild":20.0,"Company":20.0,"Consortium":15.0,"Syndicate":50.0,"Cartel":50.0,"Cooperative":70.0,"Corporation":15.0,"Chamber":15.0,"Trust":40.0,"Clan":95.0,"House":100.0,"Dynasty":45.0,"Family":95.0,"Tribe":95.0,"People":95.0,"Community":95.0,"Nation":90.0,"Kindred":95.0,"Lineage":45.0,"Resistance":20.0,"Movement":55.0,"Front":20.0,"Union":70.0,"Collective":70.0,"Alliance":20.0,"League":20.0,"Pact":20.0,"Liberation":20.0,"Academy":20.0,"Institute":20.0,"School":35.0,"Society":40.0,"Circle":20.0,"Cabal":50.0,"Conclave":50.0,"Scholars":15.0,"Collegium":15.0,"Keepers":15.0,"Caravan":40.0,"Fleet":35.0,"Band":35.0,"Enclave":60.0,"Commune":85.0,"Network":35.0,"Expedition":65.0,"Nomads":40.0,"Diaspora":40.0,"Confederation":45.0,"Assembly":35.0,"Coalition":35.0,"Horde":35.0,"Mercenaries":35.0,"Parliament":35.0,"Senate":40.0,"Association":40.0,"Compact":35.0}},"Ruined Quarter":{"scale":5,"function":{"Authority":15.0,"Defense":0.0,"Religion":0.0,"Knowledge":25.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":25.0,"Social":0.0,"Covert":25.0,"Migration":0.0,"Care":0.0,"Exploration":100.0},"type":{"Government":10.0,"Military":10.0,"Religious":45.0,"Criminal":55.0,"Revolutionary":50.0,"Conspiracy":70.0,"Merchant":10.0,"Clan":20.0,"Cultural":10.0,"Outsider":65.0,"Elite":10.0,"Poor":50.0,"Academic":75.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":15.0,"Dominion":15.0,"Council":15.0,"Court":15.0,"Commonwealth":15.0,"Directorate":15.0,"Ministry":15.0,"Authority":15.0,"Legion":20.0,"Guard":15.0,"Army":20.0,"Militia":35.0,"Vanguard":35.0,"Watch":15.0,"Corps":15.0,"Wardens":15.0,"Rangers":40.0,"Order":35.0,"Covenant":35.0,"Faith":35.0,"Sect":60.0,"Brotherhood":35.0,"Fellowship":35.0,"Chapter":35.0,"Disciples":60.0,"Cult":70.0,"Guild":20.0,"Company":20.0,"Consortium":15.0,"Syndicate":65.0,"Cartel":65.0,"Cooperative":35.0,"Corporation":15.0,"Chamber":15.0,"Trust":15.0,"Clan":20.0,"House":70.0,"Dynasty":20.0,"Family":15.0,"Tribe":40.0,"People":15.0,"Community":35.0,"Nation":10.0,"Kindred":15.0,"Lineage":15.0,"Resistance":90.0,"Movement":35.0,"Front":65.0,"Union":35.0,"Collective":35.0,"Alliance":35.0,"League":35.0,"Pact":40.0,"Liberation":65.0,"Academy":45.0,"Institute":45.0,"School":25.0,"Society":40.0,"Circle":45.0,"Cabal":70.0,"Conclave":70.0,"Scholars":45.0,"Collegium":45.0,"Keepers":40.0,"Caravan":40.0,"Fleet":40.0,"Band":65.0,"Enclave":10.0,"Commune":35.0,"Network":70.0,"Expedition":45.0,"Nomads":40.0,"Diaspora":40.0,"Confederation":40.0,"Assembly":40.0,"Coalition":35.0,"Horde":40.0,"Mercenaries":40.0,"Parliament":40.0,"Senate":40.0,"Association":40.0,"Compact":35.0}},"Temple District":{"scale":5,"function":{"Authority":15.0,"Defense":0.0,"Religion":100.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":75.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":10.0,"Military":10.0,"Religious":100.0,"Criminal":10.0,"Revolutionary":10.0,"Conspiracy":45.0,"Merchant":10.0,"Clan":40.0,"Cultural":65.0,"Outsider":10.0,"Elite":10.0,"Poor":10.0,"Academic":10.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":15.0,"Dominion":15.0,"Council":15.0,"Court":15.0,"Commonwealth":15.0,"Directorate":15.0,"Ministry":15.0,"Authority":15.0,"Legion":20.0,"Guard":15.0,"Army":20.0,"Militia":15.0,"Vanguard":15.0,"Watch":15.0,"Corps":15.0,"Wardens":15.0,"Rangers":15.0,"Order":100.0,"Covenant":100.0,"Faith":100.0,"Sect":100.0,"Brotherhood":75.0,"Fellowship":75.0,"Chapter":75.0,"Disciples":100.0,"Cult":100.0,"Guild":20.0,"Company":20.0,"Consortium":15.0,"Syndicate":20.0,"Cartel":20.0,"Cooperative":15.0,"Corporation":15.0,"Chamber":15.0,"Trust":15.0,"Clan":40.0,"House":15.0,"Dynasty":40.0,"Family":40.0,"Tribe":35.0,"People":35.0,"Community":35.0,"Nation":30.0,"Kindred":35.0,"Lineage":35.0,"Resistance":20.0,"Movement":20.0,"Front":15.0,"Union":20.0,"Collective":20.0,"Alliance":15.0,"League":15.0,"Pact":35.0,"Liberation":15.0,"Academy":50.0,"Institute":50.0,"School":40.0,"Society":40.0,"Circle":35.0,"Cabal":35.0,"Conclave":35.0,"Scholars":50.0,"Collegium":50.0,"Keepers":100.0,"Caravan":15.0,"Fleet":15.0,"Band":15.0,"Enclave":5.0,"Commune":15.0,"Network":35.0,"Expedition":15.0,"Nomads":20.0,"Diaspora":20.0,"Confederation":35.0,"Assembly":15.0,"Coalition":15.0,"Horde":15.0,"Mercenaries":15.0,"Parliament":15.0,"Senate":15.0,"Association":40.0,"Compact":15.0}},"Underground City":{"scale":5,"function":{"Authority":15.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":50.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":100.0,"Social":0.0,"Covert":50.0,"Migration":25.0,"Care":0.0,"Exploration":50.0},"type":{"Government":10.0,"Military":10.0,"Religious":10.0,"Criminal":80.0,"Revolutionary":55.0,"Conspiracy":75.0,"Merchant":60.0,"Clan":55.0,"Cultural":10.0,"Outsider":85.0,"Elite":25.0,"Poor":80.0,"Academic":15.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":50.0,"Dominion":15.0,"Council":15.0,"Court":30.0,"Commonwealth":50.0,"Directorate":15.0,"Ministry":15.0,"Authority":15.0,"Legion":20.0,"Guard":15.0,"Army":20.0,"Militia":35.0,"Vanguard":35.0,"Watch":15.0,"Corps":15.0,"Wardens":15.0,"Rangers":45.0,"Order":20.0,"Covenant":20.0,"Faith":15.0,"Sect":50.0,"Brotherhood":30.0,"Fellowship":30.0,"Chapter":15.0,"Disciples":50.0,"Cult":70.0,"Guild":40.0,"Company":40.0,"Consortium":40.0,"Syndicate":95.0,"Cartel":95.0,"Cooperative":70.0,"Corporation":40.0,"Chamber":40.0,"Trust":40.0,"Clan":55.0,"House":70.0,"Dynasty":30.0,"Family":55.0,"Tribe":70.0,"People":55.0,"Community":70.0,"Nation":50.0,"Kindred":55.0,"Lineage":30.0,"Resistance":65.0,"Movement":65.0,"Front":65.0,"Union":70.0,"Collective":70.0,"Alliance":35.0,"League":35.0,"Pact":40.0,"Liberation":65.0,"Academy":20.0,"Institute":20.0,"School":15.0,"Society":20.0,"Circle":45.0,"Cabal":40.0,"Conclave":40.0,"Scholars":20.0,"Collegium":20.0,"Keepers":15.0,"Caravan":45.0,"Fleet":45.0,"Band":70.0,"Enclave":60.0,"Commune":70.0,"Network":70.0,"Expedition":70.0,"Nomads":45.0,"Diaspora":45.0,"Confederation":45.0,"Assembly":45.0,"Coalition":45.0,"Horde":45.0,"Mercenaries":45.0,"Parliament":45.0,"Senate":45.0,"Association":45.0,"Compact":45.0}},"Underworld":{"scale":5,"function":{"Authority":20.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":75.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":25.0,"Covert":100.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":15.0,"Military":10.0,"Religious":10.0,"Criminal":100.0,"Revolutionary":50.0,"Conspiracy":90.0,"Merchant":40.0,"Clan":20.0,"Cultural":15.0,"Outsider":10.0,"Elite":55.0,"Poor":45.0,"Academic":10.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":20.0,"Dominion":20.0,"Council":20.0,"Court":40.0,"Commonwealth":20.0,"Directorate":20.0,"Ministry":15.0,"Authority":20.0,"Legion":20.0,"Guard":20.0,"Army":20.0,"Militia":45.0,"Vanguard":45.0,"Watch":20.0,"Corps":15.0,"Wardens":20.0,"Rangers":15.0,"Order":20.0,"Covenant":20.0,"Faith":15.0,"Sect":75.0,"Brotherhood":15.0,"Fellowship":15.0,"Chapter":15.0,"Disciples":75.0,"Cult":45.0,"Guild":35.0,"Company":35.0,"Consortium":35.0,"Syndicate":100.0,"Cartel":100.0,"Cooperative":35.0,"Corporation":35.0,"Chamber":35.0,"Trust":35.0,"Clan":20.0,"House":10.0,"Dynasty":40.0,"Family":15.0,"Tribe":15.0,"People":15.0,"Community":35.0,"Nation":10.0,"Kindred":15.0,"Lineage":15.0,"Resistance":95.0,"Movement":45.0,"Front":95.0,"Union":45.0,"Collective":45.0,"Alliance":45.0,"League":45.0,"Pact":45.0,"Liberation":95.0,"Academy":20.0,"Institute":20.0,"School":10.0,"Society":20.0,"Circle":45.0,"Cabal":100.0,"Conclave":45.0,"Scholars":15.0,"Collegium":15.0,"Keepers":15.0,"Caravan":35.0,"Fleet":35.0,"Band":100.0,"Enclave":5.0,"Commune":40.0,"Network":100.0,"Expedition":15.0,"Nomads":20.0,"Diaspora":20.0,"Confederation":15.0,"Assembly":15.0,"Coalition":15.0,"Horde":15.0,"Mercenaries":30.0,"Parliament":20.0,"Senate":35.0,"Association":30.0,"Compact":30.0}},"Warehouse District":{"scale":5,"function":{"Authority":15.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":25.0,"Production":25.0,"Logistics":100.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":10.0,"Military":75.0,"Religious":10.0,"Criminal":50.0,"Revolutionary":10.0,"Conspiracy":10.0,"Merchant":90.0,"Clan":20.0,"Cultural":10.0,"Outsider":65.0,"Elite":10.0,"Poor":10.0,"Academic":10.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":15.0,"Dominion":15.0,"Council":15.0,"Court":15.0,"Commonwealth":15.0,"Directorate":15.0,"Ministry":15.0,"Authority":15.0,"Legion":45.0,"Guard":45.0,"Army":45.0,"Militia":45.0,"Vanguard":45.0,"Watch":40.0,"Corps":45.0,"Wardens":40.0,"Rangers":40.0,"Order":45.0,"Covenant":20.0,"Faith":15.0,"Sect":15.0,"Brotherhood":15.0,"Fellowship":15.0,"Chapter":15.0,"Disciples":15.0,"Cult":15.0,"Guild":45.0,"Company":45.0,"Consortium":75.0,"Syndicate":90.0,"Cartel":75.0,"Cooperative":45.0,"Corporation":75.0,"Chamber":75.0,"Trust":45.0,"Clan":20.0,"House":10.0,"Dynasty":20.0,"Family":15.0,"Tribe":40.0,"People":15.0,"Community":15.0,"Nation":10.0,"Kindred":15.0,"Lineage":15.0,"Resistance":20.0,"Movement":20.0,"Front":15.0,"Union":20.0,"Collective":20.0,"Alliance":15.0,"League":15.0,"Pact":15.0,"Liberation":15.0,"Academy":20.0,"Institute":20.0,"School":10.0,"Society":15.0,"Circle":20.0,"Cabal":20.0,"Conclave":15.0,"Scholars":15.0,"Collegium":15.0,"Keepers":15.0,"Caravan":75.0,"Fleet":100.0,"Band":65.0,"Enclave":0.0,"Commune":35.0,"Network":65.0,"Expedition":65.0,"Nomads":65.0,"Diaspora":40.0,"Confederation":40.0,"Assembly":40.0,"Coalition":35.0,"Horde":40.0,"Mercenaries":45.0,"Parliament":40.0,"Senate":40.0,"Association":45.0,"Compact":70.0}}}}'
# Book1.xlsx Zone source snapshot: exact Scale 4–5 places, 14 Function minima/maxima,
# and Place→Faction Type / Base affinities. No Place Priority and no proximity.
ZONE_DATA = json.loads(ZONE_DATA_JSON)
ZONE_CANDIDATES = ZONE_DATA['zones']
FUNCTION_QUOTAS = ZONE_DATA['function_quotas']
assert len(ZONE_CANDIDATES)==36 and len(FUNCTION_QUOTAS)==14

ZONE_BONUS_WEIGHT = 0.75
ZONE_PENALTY_WEIGHT = 1.0
ZONE_FUNCTION_THRESHOLD = 75
# Locked calibration: 10,000 independent holdout seeds (14,900–24,899),
# all 36 Zone Types appeared between 5% and 40%. Additive SCORE POINTS,
# not per-world quotas, not affinity-percent multipliers.
ZONE_SCORE_MODIFIERS = {'Caravan Camp': 7, 'Estate': 2, 'Military Base': -3, 'Mining Settlement': 4, 'Rebel Camp': -4, 'Cultural District': -1, 'Entertainment District': -4, 'Market District': -2, 'Royal Quarter': -7}
assert set(ZONE_SCORE_MODIFIERS).issubset(ZONE_CANDIDATES)

def _zone_functions(name):
    return {k for k,v in ZONE_CANDIDATES[name]['function'].items()
            if v >= ZONE_FUNCTION_THRESHOLD}

def _soft_max(vals):
    vals=list(vals)
    return (max(vals)+sum(vals)/len(vals))/2 if vals else 0.0

def _zone_coverage_score(place, counts):
    # Bonus under minimum uses ALL raw affinities, including 0/25/50.
    # Penalties only apply if this Zone would increment that Function (>=75).
    # Once the existing count exceeds maximum, it appears in both penalties.
    relevant=ZONE_CANDIDATES[place]['function']
    under=_soft_max(v for k,v in relevant.items()
                    if counts[k] < FUNCTION_QUOTAS[k]['min'])
    at_or_over=_soft_max(v for k,v in relevant.items()
                         if v >= ZONE_FUNCTION_THRESHOLD and counts[k] >= FUNCTION_QUOTAS[k]['max'])
    over=_soft_max(v for k,v in relevant.items()
                   if v >= ZONE_FUNCTION_THRESHOLD and counts[k] > FUNCTION_QUOTAS[k]['max'])
    return under,at_or_over,over

def _faction_zone_score(place, faction, rand, counts):
    aff=ZONE_CANDIDATES[place]
    x=aff['type'][faction['types'][0]]
    y=aff['type'][faction['types'][1]] if len(faction['types'])==2 else x
    b=aff['base'][faction['base']]
    arithmetic=(x+y+b)/3
    geometric=(x*y*b)**(1/3) if x*y*b else 0.0
    under,at,over=_zone_coverage_score(place,counts)
    base_score=(arithmetic+geometric+rand)/3
    modifier=ZONE_SCORE_MODIFIERS.get(place,0)
    return base_score+ZONE_BONUS_WEIGHT*under-ZONE_PENALTY_WEIGHT*(at+over)+modifier,{'arithmetic':arithmetic,'geometric':geometric,
                  'random':rand,'under_min':under,'at_or_over_max':at,'over_max':over,'modifier':modifier}

def _extra_zone_score(place,rand,counts):
    under,at,over=_zone_coverage_score(place,counts)
    modifier=ZONE_SCORE_MODIFIERS.get(place,0)
    return rand+ZONE_BONUS_WEIGHT*under-ZONE_PENALTY_WEIGHT*(at+over)+modifier,{'random':rand,'under_min':under,
                               'at_or_over_max':at,'over_max':over,'modifier':modifier}

def generate_zones(world, seed=None, desired_count=None):
    # Separate deterministic RNG; avoids perturbing existing Faction naming RNG.
    rng=random.Random(world['seed'] if seed is None else seed)
    if desired_count is None:
        desired_count=rng.randint(7,9)
    if desired_count not in (7,8,9): raise ValueError('Zone count must be 7–9')
    counts={k:0 for k in FUNCTION_QUOTAS}
    zones=[]; used=set()
    for i in range(desired_count):
        faction=world['factions'][i] if i<6 else None
        candidates=[]
        for place in ZONE_CANDIDATES:
            rand=rng.uniform(0,100)
            if place in used:continue
            if faction is not None:
                score,parts=_faction_zone_score(place,faction,rand,counts)
            else:
                score,parts=_extra_zone_score(place,rand,counts)
            candidates.append((score,place,parts))
        score,place,parts=max(candidates,key=lambda t:(t[0],t[1]))
        coverage=sorted(_zone_functions(place))
        for k in coverage:counts[k]+=1
        used.add(place)
        zones.append({'sequence':i+1,'place':place,'scale':ZONE_CANDIDATES[place]['scale'],
                      'selected_by':faction['id'] if faction else None,
                      'faction_name':faction['name'] if faction else None,
                      'score':round(score,5),'components':{k:round(v,5) for k,v in parts.items()},
                      'covered_functions':coverage})
    missing={k:max(0,q['min']-counts[k]) for k,q in FUNCTION_QUOTAS.items()}
    excess={k:max(0,counts[k]-q['max']) for k,q in FUNCTION_QUOTAS.items()}
    return {'target':desired_count,'zones':zones,'function_counts':counts,
            'missing_minimum':{k:v for k,v in missing.items() if v},
            'excess_maximum':{k:v for k,v in excess.items() if v}}

def generate_world_with_zones(seed=20261120,desired_count=None):
    world=generate_world(seed)
    world['zone_generation']=generate_zones(world,seed=seed+1000003,desired_count=desired_count)
    return world

def zone_qa(n=1000):
    from collections import Counter
    counts=Counter();miss=Counter();over=Counter();places=Counter();compositions=set()
    examples=[]
    for seed in range(n):
        w=generate_world_with_zones(seed)
        z=w['zone_generation'];zs=z['zones']
        assert len(zs) in (7,8,9)
        assert len({x['place'] for x in zs})==len(zs)
        assert [x['selected_by'] for x in zs[:6]]==[f'F{i}' for i in range(1,7)]
        assert all(x['selected_by'] is None for x in zs[6:])
        assert all(x['place'] in ZONE_CANDIDATES for x in zs)
        recount={k:sum(k in x['covered_functions'] for x in zs) for k in FUNCTION_QUOTAS}
        assert recount==z['function_counts']
        if not z['missing_minimum']:counts['met_all_minimum']+=1
        if not z['excess_maximum']:counts['met_all_maximum']+=1
        for k,v in z['missing_minimum'].items():miss[k]+=1
        for k,v in z['excess_maximum'].items():over[k]+=1
        places.update(x['place'] for x in zs)
        compositions.add(tuple(sorted(x['place'] for x in zs)))
        counts['worlds']+=1
        counts[f"count_{len(zs)}"]+=1
        counts['total_excess_instances']+=sum(z['excess_maximum'].values())
        if len(examples)<5:examples.append(w)
    return {'worlds':n,'met_all_minimum':counts['met_all_minimum'],
            'met_all_maximum':counts['met_all_maximum'],
            'zone_count_distribution':{k:counts[k] for k in ('count_7','count_8','count_9')},
            'different_zone_sets':len(compositions),'unique_zone_types':len(places),
            'top_places':places.most_common(15),'bottom_places':places.most_common()[-12:],
            'missing_minimum_worlds_by_function':dict(miss),
            'over_maximum_worlds_by_function':dict(over),
            'total_excess_instances':counts['total_excess_instances'],
            'examples':examples}
```

```python
LOCATION_DATA_JSON = '{"source":"Book1.xlsx / Place Affinity / Scale 1–2","location_count":74,"locations":{"Apothecary":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":50.0,"Commerce":100.0,"Production":25.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":100.0,"Exploration":0.0},"type":{"Government":20.0,"Military":15.0,"Religious":70.0,"Criminal":55.0,"Revolutionary":15.0,"Conspiracy":20.0,"Merchant":85.0,"Clan":15.0,"Cultural":20.0,"Outsider":15.0,"Elite":25.0,"Poor":50.0,"Academic":65.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":45.0,"Dominion":20.0,"Council":45.0,"Court":30.0,"Commonwealth":45.0,"Directorate":45.0,"Ministry":45.0,"Authority":20.0,"Legion":20.0,"Guard":20.0,"Army":20.0,"Militia":20.0,"Vanguard":20.0,"Watch":20.0,"Corps":20.0,"Wardens":20.0,"Rangers":20.0,"Order":70.0,"Covenant":70.0,"Faith":70.0,"Sect":70.0,"Brotherhood":65.0,"Fellowship":65.0,"Chapter":65.0,"Disciples":65.0,"Cult":70.0,"Guild":70.0,"Company":70.0,"Consortium":70.0,"Syndicate":65.0,"Cartel":60.0,"Cooperative":70.0,"Corporation":70.0,"Chamber":70.0,"Trust":70.0,"Clan":45.0,"House":30.0,"Dynasty":30.0,"Family":55.0,"Tribe":45.0,"People":50.0,"Community":60.0,"Nation":35.0,"Kindred":45.0,"Lineage":20.0,"Resistance":35.0,"Movement":35.0,"Front":20.0,"Union":65.0,"Collective":65.0,"Alliance":45.0,"League":50.0,"Pact":45.0,"Liberation":35.0,"Academy":65.0,"Institute":65.0,"School":65.0,"Society":65.0,"Circle":65.0,"Cabal":45.0,"Conclave":55.0,"Scholars":65.0,"Collegium":65.0,"Keepers":65.0,"Caravan":70.0,"Fleet":60.0,"Band":45.0,"Enclave":20.0,"Commune":50.0,"Network":65.0,"Expedition":45.0,"Nomads":45.0,"Diaspora":20.0,"Confederation":45.0,"Assembly":50.0,"Coalition":40.0,"Horde":30.0,"Mercenaries":50.0,"Parliament":20.0,"Senate":30.0,"Association":70.0,"Compact":60.0},"place":{"Apothecary":100.0,"Archive":45.0,"Armoury":15.0,"Auction House":50.0,"Bank":55.0,"Bathhouse":30.0,"Bookshop":70.0,"Brothel":50.0,"Clinic":60.0,"Club":35.0,"Courthouse":15.0,"Bridge":15.0,"Crypt":20.0,"Dormitory":20.0,"Forge":25.0,"Gallery":20.0,"Gambling Den":50.0,"House":0.0,"Laboratory":50.0,"Library":50.0,"Mill":25.0,"Monument":0.0,"Mortuary":25.0,"Museum":45.0,"Observatory":45.0,"Office":50.0,"Pawnshop":55.0,"Post Office":20.0,"Printing House":50.0,"Prison":0.0,"School":45.0,"Sewer":0.0,"Shelter":40.0,"Ship":20.0,"Shop":60.0,"Shrine":0.0,"Stable":0.0,"Tailor":55.0,"Tavern":45.0,"Theater":20.0,"Tower":20.0,"Training Ground":15.0,"Treasury":25.0,"Tunnel":0.0,"Vault":25.0,"Guardhouse":15.0,"Well":0.0,"Arena":25.0,"Barracks":5.0,"Black Market":50.0,"Brewery":40.0,"Camp":5.0,"Cemetery":5.0,"Dock":30.0,"Embassy":5.0,"Factory":25.0,"Farm":25.0,"Fortress":0.0,"Garden":5.0,"Grove":5.0,"Meeting Hall":5.0,"Hideout":5.0,"Inn":30.0,"Manor":5.0,"Market":70.0,"Mine":25.0,"Outpost":5.0,"Palace":5.0,"Research Facility":50.0,"Square":30.0,"Temple":5.0,"Trading Post":50.0,"Warehouse":25.0,"Workshop":40.0,"Catacombs":5.0,"Town Hall":5.0,"Excavation Site":30.0,"Festival Grounds":30.0,"Hospital":80.0,"Ruins":25.0,"Sanctuary":40.0,"Shipyard":30.0,"Caravan Camp":25.0,"Cavern":15.0,"Citadel":10.0,"Enclave":10.0,"Estate":10.0,"Village":15.0,"Hunting Grounds":0.0,"Military Base":0.0,"Mining Settlement":10.0,"Monastery":30.0,"Necropolis":20.0,"Rebel Camp":10.0,"Refugee Camp":40.0,"Slum":10.0,"University":50.0,"Academic District":75.0,"Government District":10.0,"Artisan Quarter":50.0,"Cultural District":10.0,"Entertainment District":35.0,"Farmland":15.0,"Foreign Quarter":45.0,"Wilderness":10.0,"Harbor":65.0,"Industrial District":20.0,"Market District":80.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":10.0,"Royal Quarter":10.0,"Residential District":35.0,"Ruined Quarter":20.0,"Temple District":35.0,"Underground City":25.0,"Underworld":20.0,"Warehouse District":15.0},"score":70.0},"Archive":{"function":{"Authority":50.0,"Defense":0.0,"Religion":0.0,"Knowledge":100.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":25.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":80.0,"Military":20.0,"Religious":45.0,"Criminal":20.0,"Revolutionary":20.0,"Conspiracy":75.0,"Merchant":45.0,"Clan":15.0,"Cultural":25.0,"Outsider":15.0,"Elite":55.0,"Poor":15.0,"Academic":90.0},"base":{"Empire":70.0,"Kingdom":70.0,"Republic":95.0,"Dominion":70.0,"Council":95.0,"Court":70.0,"Commonwealth":95.0,"Directorate":95.0,"Ministry":95.0,"Authority":95.0,"Legion":45.0,"Guard":60.0,"Army":45.0,"Militia":35.0,"Vanguard":35.0,"Watch":60.0,"Corps":45.0,"Wardens":60.0,"Rangers":35.0,"Order":60.0,"Covenant":60.0,"Faith":60.0,"Sect":60.0,"Brotherhood":60.0,"Fellowship":60.0,"Chapter":60.0,"Disciples":60.0,"Cult":70.0,"Guild":40.0,"Company":40.0,"Consortium":40.0,"Syndicate":40.0,"Cartel":40.0,"Cooperative":40.0,"Corporation":40.0,"Chamber":50.0,"Trust":65.0,"Clan":35.0,"House":65.0,"Dynasty":65.0,"Family":20.0,"Tribe":35.0,"People":15.0,"Community":20.0,"Nation":15.0,"Kindred":35.0,"Lineage":55.0,"Resistance":35.0,"Movement":35.0,"Front":35.0,"Union":20.0,"Collective":20.0,"Alliance":35.0,"League":35.0,"Pact":50.0,"Liberation":35.0,"Academy":100.0,"Institute":100.0,"School":100.0,"Society":100.0,"Circle":100.0,"Cabal":45.0,"Conclave":45.0,"Scholars":100.0,"Collegium":100.0,"Keepers":95.0,"Caravan":40.0,"Fleet":40.0,"Band":35.0,"Enclave":20.0,"Commune":15.0,"Network":60.0,"Expedition":45.0,"Nomads":20.0,"Diaspora":20.0,"Confederation":95.0,"Assembly":95.0,"Coalition":90.0,"Horde":35.0,"Mercenaries":40.0,"Parliament":95.0,"Senate":95.0,"Association":85.0,"Compact":90.0},"place":{"Apothecary":45.0,"Archive":100.0,"Armoury":0.0,"Auction House":0.0,"Bank":20.0,"Bathhouse":0.0,"Bookshop":50.0,"Brothel":15.0,"Clinic":50.0,"Club":20.0,"Courthouse":55.0,"Bridge":0.0,"Crypt":20.0,"Dormitory":20.0,"Forge":0.0,"Gallery":5.0,"Gambling Den":20.0,"House":0.0,"Laboratory":65.0,"Library":65.0,"Mill":0.0,"Monument":40.0,"Mortuary":30.0,"Museum":55.0,"Observatory":60.0,"Office":55.0,"Pawnshop":15.0,"Post Office":45.0,"Printing House":50.0,"Prison":50.0,"School":60.0,"Sewer":20.0,"Shelter":0.0,"Ship":0.0,"Shop":0.0,"Shrine":0.0,"Stable":0.0,"Tailor":0.0,"Tavern":0.0,"Theater":0.0,"Tower":25.0,"Training Ground":20.0,"Treasury":50.0,"Tunnel":15.0,"Vault":20.0,"Guardhouse":45.0,"Well":0.0,"Arena":5.0,"Barracks":20.0,"Black Market":25.0,"Brewery":5.0,"Camp":5.0,"Cemetery":20.0,"Dock":5.0,"Embassy":65.0,"Factory":5.0,"Farm":5.0,"Fortress":25.0,"Garden":5.0,"Grove":5.0,"Meeting Hall":45.0,"Hideout":25.0,"Inn":5.0,"Manor":25.0,"Market":5.0,"Mine":5.0,"Outpost":5.0,"Palace":45.0,"Research Facility":80.0,"Square":25.0,"Temple":40.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":25.0,"Catacombs":30.0,"Town Hall":50.0,"Excavation Site":35.0,"Festival Grounds":5.0,"Hospital":45.0,"Ruins":30.0,"Sanctuary":5.0,"Shipyard":5.0,"Caravan Camp":10.0,"Cavern":20.0,"Citadel":55.0,"Enclave":10.0,"Estate":10.0,"Village":10.0,"Hunting Grounds":0.0,"Military Base":0.0,"Mining Settlement":10.0,"Monastery":45.0,"Necropolis":25.0,"Rebel Camp":20.0,"Refugee Camp":10.0,"Slum":10.0,"University":75.0,"Academic District":85.0,"Government District":90.0,"Artisan Quarter":10.0,"Cultural District":15.0,"Entertainment District":10.0,"Farmland":10.0,"Foreign Quarter":10.0,"Wilderness":15.0,"Harbor":10.0,"Industrial District":10.0,"Market District":10.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":15.0,"Royal Quarter":75.0,"Residential District":10.0,"Ruined Quarter":25.0,"Temple District":10.0,"Underground City":20.0,"Underworld":25.0,"Warehouse District":10.0},"score":68.33333333333333},"Armoury":{"function":{"Authority":0.0,"Defense":100.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":50.0,"Logistics":75.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":75.0,"Military":90.0,"Religious":15.0,"Criminal":45.0,"Revolutionary":80.0,"Conspiracy":15.0,"Merchant":60.0,"Clan":15.0,"Cultural":15.0,"Outsider":15.0,"Elite":15.0,"Poor":15.0,"Academic":15.0},"base":{"Empire":70.0,"Kingdom":70.0,"Republic":60.0,"Dominion":70.0,"Council":70.0,"Court":60.0,"Commonwealth":60.0,"Directorate":70.0,"Ministry":70.0,"Authority":70.0,"Legion":100.0,"Guard":100.0,"Army":100.0,"Militia":100.0,"Vanguard":100.0,"Watch":100.0,"Corps":100.0,"Wardens":95.0,"Rangers":95.0,"Order":65.0,"Covenant":45.0,"Faith":45.0,"Sect":20.0,"Brotherhood":35.0,"Fellowship":35.0,"Chapter":30.0,"Disciples":20.0,"Cult":20.0,"Guild":45.0,"Company":45.0,"Consortium":55.0,"Syndicate":55.0,"Cartel":55.0,"Cooperative":45.0,"Corporation":55.0,"Chamber":60.0,"Trust":45.0,"Clan":45.0,"House":20.0,"Dynasty":20.0,"Family":35.0,"Tribe":45.0,"People":15.0,"Community":20.0,"Nation":15.0,"Kindred":45.0,"Lineage":20.0,"Resistance":70.0,"Movement":70.0,"Front":70.0,"Union":60.0,"Collective":60.0,"Alliance":70.0,"League":70.0,"Pact":70.0,"Liberation":70.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":20.0,"Circle":35.0,"Cabal":35.0,"Conclave":20.0,"Scholars":20.0,"Collegium":20.0,"Keepers":40.0,"Caravan":55.0,"Fleet":65.0,"Band":60.0,"Enclave":20.0,"Commune":40.0,"Network":50.0,"Expedition":45.0,"Nomads":45.0,"Diaspora":20.0,"Confederation":65.0,"Assembly":65.0,"Coalition":65.0,"Horde":95.0,"Mercenaries":95.0,"Parliament":70.0,"Senate":70.0,"Association":45.0,"Compact":55.0},"place":{"Apothecary":15.0,"Archive":0.0,"Armoury":100.0,"Auction House":0.0,"Bank":5.0,"Bathhouse":0.0,"Bookshop":0.0,"Brothel":0.0,"Clinic":0.0,"Club":0.0,"Courthouse":0.0,"Bridge":35.0,"Crypt":0.0,"Dormitory":15.0,"Forge":30.0,"Gallery":0.0,"Gambling Den":0.0,"House":0.0,"Laboratory":20.0,"Library":0.0,"Mill":20.0,"Monument":0.0,"Mortuary":0.0,"Museum":0.0,"Observatory":20.0,"Office":0.0,"Pawnshop":0.0,"Post Office":30.0,"Printing House":25.0,"Prison":50.0,"School":0.0,"Sewer":25.0,"Shelter":0.0,"Ship":35.0,"Shop":20.0,"Shrine":0.0,"Stable":30.0,"Tailor":20.0,"Tavern":0.0,"Theater":0.0,"Tower":55.0,"Training Ground":55.0,"Treasury":5.0,"Tunnel":30.0,"Vault":25.0,"Guardhouse":55.0,"Well":0.0,"Arena":35.0,"Barracks":100.0,"Black Market":25.0,"Brewery":25.0,"Camp":25.0,"Cemetery":0.0,"Dock":30.0,"Embassy":5.0,"Factory":30.0,"Farm":25.0,"Fortress":75.0,"Garden":5.0,"Grove":5.0,"Meeting Hall":5.0,"Hideout":45.0,"Inn":5.0,"Manor":5.0,"Market":5.0,"Mine":25.0,"Outpost":80.0,"Palace":10.0,"Research Facility":5.0,"Square":5.0,"Temple":5.0,"Trading Post":30.0,"Warehouse":40.0,"Workshop":30.0,"Catacombs":5.0,"Town Hall":5.0,"Excavation Site":5.0,"Festival Grounds":5.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":5.0,"Shipyard":40.0,"Caravan Camp":30.0,"Cavern":10.0,"Citadel":70.0,"Enclave":10.0,"Estate":10.0,"Village":10.0,"Hunting Grounds":0.0,"Military Base":80.0,"Mining Settlement":10.0,"Monastery":10.0,"Necropolis":10.0,"Rebel Camp":90.0,"Refugee Camp":10.0,"Slum":10.0,"University":0.0,"Academic District":10.0,"Government District":15.0,"Artisan Quarter":25.0,"Cultural District":10.0,"Entertainment District":10.0,"Farmland":10.0,"Foreign Quarter":10.0,"Wilderness":10.0,"Harbor":35.0,"Industrial District":30.0,"Market District":15.0,"Military District":100.0,"Mining Region":10.0,"Noble Quarter":10.0,"Royal Quarter":15.0,"Residential District":10.0,"Ruined Quarter":10.0,"Temple District":10.0,"Underground City":10.0,"Underworld":10.0,"Warehouse District":40.0},"score":71.66666666666667},"Auction House":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":100.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":50.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":15.0,"Military":15.0,"Religious":20.0,"Criminal":55.0,"Revolutionary":15.0,"Conspiracy":15.0,"Merchant":90.0,"Clan":20.0,"Cultural":60.0,"Outsider":15.0,"Elite":75.0,"Poor":15.0,"Academic":45.0},"base":{"Empire":45.0,"Kingdom":45.0,"Republic":45.0,"Dominion":45.0,"Council":45.0,"Court":70.0,"Commonwealth":45.0,"Directorate":45.0,"Ministry":40.0,"Authority":45.0,"Legion":20.0,"Guard":20.0,"Army":20.0,"Militia":20.0,"Vanguard":20.0,"Watch":20.0,"Corps":20.0,"Wardens":20.0,"Rangers":20.0,"Order":35.0,"Covenant":20.0,"Faith":20.0,"Sect":20.0,"Brotherhood":45.0,"Fellowship":45.0,"Chapter":45.0,"Disciples":20.0,"Cult":20.0,"Guild":75.0,"Company":75.0,"Consortium":75.0,"Syndicate":75.0,"Cartel":75.0,"Cooperative":65.0,"Corporation":75.0,"Chamber":75.0,"Trust":75.0,"Clan":35.0,"House":70.0,"Dynasty":70.0,"Family":45.0,"Tribe":35.0,"People":35.0,"Community":35.0,"Nation":30.0,"Kindred":35.0,"Lineage":55.0,"Resistance":20.0,"Movement":45.0,"Front":20.0,"Union":45.0,"Collective":45.0,"Alliance":45.0,"League":50.0,"Pact":40.0,"Liberation":20.0,"Academy":40.0,"Institute":40.0,"School":40.0,"Society":65.0,"Circle":60.0,"Cabal":40.0,"Conclave":50.0,"Scholars":40.0,"Collegium":40.0,"Keepers":40.0,"Caravan":75.0,"Fleet":65.0,"Band":45.0,"Enclave":35.0,"Commune":15.0,"Network":65.0,"Expedition":40.0,"Nomads":45.0,"Diaspora":35.0,"Confederation":55.0,"Assembly":50.0,"Coalition":40.0,"Horde":30.0,"Mercenaries":55.0,"Parliament":45.0,"Senate":60.0,"Association":70.0,"Compact":60.0},"place":{"Apothecary":50.0,"Archive":0.0,"Armoury":0.0,"Auction House":100.0,"Bank":70.0,"Bathhouse":35.0,"Bookshop":60.0,"Brothel":65.0,"Clinic":0.0,"Club":60.0,"Courthouse":15.0,"Bridge":15.0,"Crypt":15.0,"Dormitory":0.0,"Forge":25.0,"Gallery":60.0,"Gambling Den":65.0,"House":20.0,"Laboratory":0.0,"Library":20.0,"Mill":20.0,"Monument":25.0,"Mortuary":0.0,"Museum":25.0,"Observatory":0.0,"Office":50.0,"Pawnshop":60.0,"Post Office":20.0,"Printing House":25.0,"Prison":0.0,"School":20.0,"Sewer":0.0,"Shelter":0.0,"Ship":20.0,"Shop":65.0,"Shrine":20.0,"Stable":0.0,"Tailor":55.0,"Tavern":60.0,"Theater":35.0,"Tower":0.0,"Training Ground":15.0,"Treasury":50.0,"Tunnel":0.0,"Vault":55.0,"Guardhouse":0.0,"Well":0.0,"Arena":35.0,"Barracks":0.0,"Black Market":50.0,"Brewery":40.0,"Camp":5.0,"Cemetery":25.0,"Dock":35.0,"Embassy":25.0,"Factory":20.0,"Farm":5.0,"Fortress":5.0,"Garden":35.0,"Grove":10.0,"Meeting Hall":30.0,"Hideout":5.0,"Inn":35.0,"Manor":40.0,"Market":80.0,"Mine":5.0,"Outpost":5.0,"Palace":40.0,"Research Facility":5.0,"Square":40.0,"Temple":25.0,"Trading Post":55.0,"Warehouse":25.0,"Workshop":30.0,"Catacombs":5.0,"Town Hall":30.0,"Excavation Site":5.0,"Festival Grounds":45.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":5.0,"Shipyard":25.0,"Caravan Camp":25.0,"Cavern":10.0,"Citadel":10.0,"Enclave":10.0,"Estate":50.0,"Village":20.0,"Hunting Grounds":0.0,"Military Base":0.0,"Mining Settlement":10.0,"Monastery":10.0,"Necropolis":10.0,"Rebel Camp":10.0,"Refugee Camp":10.0,"Slum":15.0,"University":0.0,"Academic District":15.0,"Government District":10.0,"Artisan Quarter":40.0,"Cultural District":45.0,"Entertainment District":45.0,"Farmland":10.0,"Foreign Quarter":50.0,"Wilderness":10.0,"Harbor":70.0,"Industrial District":10.0,"Market District":90.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":65.0,"Royal Quarter":55.0,"Residential District":15.0,"Ruined Quarter":10.0,"Temple District":15.0,"Underground City":25.0,"Underworld":25.0,"Warehouse District":15.0},"score":50.0},"Bank":{"function":{"Authority":25.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":100.0,"Production":0.0,"Logistics":25.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":50.0,"Military":20.0,"Religious":15.0,"Criminal":55.0,"Revolutionary":15.0,"Conspiracy":45.0,"Merchant":90.0,"Clan":15.0,"Cultural":15.0,"Outsider":15.0,"Elite":75.0,"Poor":15.0,"Academic":15.0},"base":{"Empire":65.0,"Kingdom":65.0,"Republic":65.0,"Dominion":60.0,"Council":60.0,"Court":70.0,"Commonwealth":60.0,"Directorate":65.0,"Ministry":60.0,"Authority":60.0,"Legion":35.0,"Guard":45.0,"Army":35.0,"Militia":35.0,"Vanguard":35.0,"Watch":40.0,"Corps":35.0,"Wardens":40.0,"Rangers":35.0,"Order":35.0,"Covenant":35.0,"Faith":35.0,"Sect":20.0,"Brotherhood":45.0,"Fellowship":45.0,"Chapter":40.0,"Disciples":20.0,"Cult":40.0,"Guild":75.0,"Company":75.0,"Consortium":75.0,"Syndicate":75.0,"Cartel":75.0,"Cooperative":65.0,"Corporation":90.0,"Chamber":75.0,"Trust":95.0,"Clan":35.0,"House":70.0,"Dynasty":70.0,"Family":45.0,"Tribe":35.0,"People":15.0,"Community":20.0,"Nation":15.0,"Kindred":35.0,"Lineage":55.0,"Resistance":35.0,"Movement":20.0,"Front":35.0,"Union":45.0,"Collective":45.0,"Alliance":45.0,"League":50.0,"Pact":50.0,"Liberation":35.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":55.0,"Circle":60.0,"Cabal":40.0,"Conclave":50.0,"Scholars":20.0,"Collegium":20.0,"Keepers":30.0,"Caravan":75.0,"Fleet":65.0,"Band":45.0,"Enclave":20.0,"Commune":15.0,"Network":65.0,"Expedition":35.0,"Nomads":45.0,"Diaspora":20.0,"Confederation":60.0,"Assembly":60.0,"Coalition":50.0,"Horde":35.0,"Mercenaries":55.0,"Parliament":60.0,"Senate":70.0,"Association":70.0,"Compact":60.0},"place":{"Apothecary":55.0,"Archive":20.0,"Armoury":5.0,"Auction House":70.0,"Bank":100.0,"Bathhouse":25.0,"Bookshop":55.0,"Brothel":55.0,"Clinic":0.0,"Club":45.0,"Courthouse":25.0,"Bridge":20.0,"Crypt":15.0,"Dormitory":0.0,"Forge":25.0,"Gallery":45.0,"Gambling Den":55.0,"House":15.0,"Laboratory":0.0,"Library":0.0,"Mill":20.0,"Monument":15.0,"Mortuary":15.0,"Museum":15.0,"Observatory":0.0,"Office":55.0,"Pawnshop":60.0,"Post Office":30.0,"Printing House":20.0,"Prison":20.0,"School":0.0,"Sewer":5.0,"Shelter":0.0,"Ship":25.0,"Shop":65.0,"Shrine":0.0,"Stable":5.0,"Tailor":55.0,"Tavern":50.0,"Theater":20.0,"Tower":0.0,"Training Ground":0.0,"Treasury":60.0,"Tunnel":5.0,"Vault":100.0,"Guardhouse":20.0,"Well":0.0,"Arena":25.0,"Barracks":20.0,"Black Market":55.0,"Brewery":35.0,"Camp":5.0,"Cemetery":0.0,"Dock":40.0,"Embassy":25.0,"Factory":20.0,"Farm":5.0,"Fortress":25.0,"Garden":20.0,"Grove":5.0,"Meeting Hall":20.0,"Hideout":5.0,"Inn":30.0,"Manor":45.0,"Market":75.0,"Mine":5.0,"Outpost":5.0,"Palace":45.0,"Research Facility":5.0,"Square":30.0,"Temple":20.0,"Trading Post":60.0,"Warehouse":35.0,"Workshop":30.0,"Catacombs":5.0,"Town Hall":25.0,"Excavation Site":5.0,"Festival Grounds":35.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":5.0,"Shipyard":30.0,"Caravan Camp":25.0,"Cavern":10.0,"Citadel":20.0,"Enclave":10.0,"Estate":50.0,"Village":15.0,"Hunting Grounds":0.0,"Military Base":0.0,"Mining Settlement":10.0,"Monastery":10.0,"Necropolis":10.0,"Rebel Camp":10.0,"Refugee Camp":10.0,"Slum":10.0,"University":0.0,"Academic District":10.0,"Government District":40.0,"Artisan Quarter":40.0,"Cultural District":10.0,"Entertainment District":40.0,"Farmland":10.0,"Foreign Quarter":50.0,"Wilderness":10.0,"Harbor":75.0,"Industrial District":15.0,"Market District":90.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":70.0,"Royal Quarter":70.0,"Residential District":10.0,"Ruined Quarter":10.0,"Temple District":10.0,"Underground City":25.0,"Underworld":25.0,"Warehouse District":20.0},"score":71.66666666666667},"Bathhouse":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":50.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":100.0,"Covert":0.0,"Migration":0.0,"Care":50.0,"Exploration":0.0},"type":{"Government":15.0,"Military":15.0,"Religious":25.0,"Criminal":20.0,"Revolutionary":15.0,"Conspiracy":15.0,"Merchant":75.0,"Clan":55.0,"Cultural":90.0,"Outsider":15.0,"Elite":70.0,"Poor":50.0,"Academic":20.0},"base":{"Empire":35.0,"Kingdom":45.0,"Republic":45.0,"Dominion":35.0,"Council":45.0,"Court":70.0,"Commonwealth":45.0,"Directorate":45.0,"Ministry":15.0,"Authority":20.0,"Legion":35.0,"Guard":35.0,"Army":35.0,"Militia":35.0,"Vanguard":35.0,"Watch":35.0,"Corps":35.0,"Wardens":35.0,"Rangers":30.0,"Order":35.0,"Covenant":35.0,"Faith":35.0,"Sect":35.0,"Brotherhood":45.0,"Fellowship":55.0,"Chapter":35.0,"Disciples":35.0,"Cult":35.0,"Guild":70.0,"Company":70.0,"Consortium":70.0,"Syndicate":60.0,"Cartel":60.0,"Cooperative":60.0,"Corporation":70.0,"Chamber":70.0,"Trust":70.0,"Clan":65.0,"House":70.0,"Dynasty":70.0,"Family":65.0,"Tribe":65.0,"People":60.0,"Community":65.0,"Nation":60.0,"Kindred":65.0,"Lineage":65.0,"Resistance":35.0,"Movement":55.0,"Front":35.0,"Union":65.0,"Collective":65.0,"Alliance":45.0,"League":50.0,"Pact":40.0,"Liberation":35.0,"Academy":35.0,"Institute":35.0,"School":35.0,"Society":65.0,"Circle":45.0,"Cabal":45.0,"Conclave":45.0,"Scholars":35.0,"Collegium":35.0,"Keepers":35.0,"Caravan":70.0,"Fleet":50.0,"Band":35.0,"Enclave":55.0,"Commune":50.0,"Network":35.0,"Expedition":45.0,"Nomads":45.0,"Diaspora":45.0,"Confederation":60.0,"Assembly":50.0,"Coalition":40.0,"Horde":30.0,"Mercenaries":50.0,"Parliament":45.0,"Senate":60.0,"Association":70.0,"Compact":55.0},"place":{"Apothecary":30.0,"Archive":0.0,"Armoury":0.0,"Auction House":35.0,"Bank":25.0,"Bathhouse":100.0,"Bookshop":30.0,"Brothel":60.0,"Clinic":25.0,"Club":55.0,"Courthouse":0.0,"Bridge":15.0,"Crypt":0.0,"Dormitory":35.0,"Forge":20.0,"Gallery":65.0,"Gambling Den":60.0,"House":50.0,"Laboratory":0.0,"Library":20.0,"Mill":20.0,"Monument":55.0,"Mortuary":0.0,"Museum":50.0,"Observatory":0.0,"Office":20.0,"Pawnshop":25.0,"Post Office":20.0,"Printing House":25.0,"Prison":0.0,"School":20.0,"Sewer":0.0,"Shelter":40.0,"Ship":20.0,"Shop":30.0,"Shrine":20.0,"Stable":0.0,"Tailor":25.0,"Tavern":60.0,"Theater":65.0,"Tower":0.0,"Training Ground":15.0,"Treasury":20.0,"Tunnel":0.0,"Vault":20.0,"Guardhouse":0.0,"Well":20.0,"Arena":60.0,"Barracks":35.0,"Black Market":25.0,"Brewery":30.0,"Camp":35.0,"Cemetery":30.0,"Dock":25.0,"Embassy":10.0,"Factory":20.0,"Farm":20.0,"Fortress":20.0,"Garden":65.0,"Grove":25.0,"Meeting Hall":55.0,"Hideout":20.0,"Inn":55.0,"Manor":45.0,"Market":40.0,"Mine":5.0,"Outpost":5.0,"Palace":30.0,"Research Facility":5.0,"Square":70.0,"Temple":35.0,"Trading Post":30.0,"Warehouse":25.0,"Workshop":25.0,"Catacombs":5.0,"Town Hall":35.0,"Excavation Site":5.0,"Festival Grounds":65.0,"Hospital":25.0,"Ruins":5.0,"Sanctuary":30.0,"Shipyard":25.0,"Caravan Camp":20.0,"Cavern":10.0,"Citadel":10.0,"Enclave":40.0,"Estate":40.0,"Village":65.0,"Hunting Grounds":0.0,"Military Base":15.0,"Mining Settlement":55.0,"Monastery":20.0,"Necropolis":10.0,"Rebel Camp":15.0,"Refugee Camp":45.0,"Slum":50.0,"University":45.0,"Academic District":15.0,"Government District":15.0,"Artisan Quarter":30.0,"Cultural District":85.0,"Entertainment District":85.0,"Farmland":10.0,"Foreign Quarter":50.0,"Wilderness":10.0,"Harbor":25.0,"Industrial District":10.0,"Market District":55.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":45.0,"Royal Quarter":35.0,"Residential District":75.0,"Ruined Quarter":10.0,"Temple District":20.0,"Underground City":60.0,"Underworld":20.0,"Warehouse District":10.0},"score":61.666666666666664},"Bookshop":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":75.0,"Commerce":100.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":25.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":20.0,"Military":15.0,"Religious":45.0,"Criminal":25.0,"Revolutionary":15.0,"Conspiracy":55.0,"Merchant":90.0,"Clan":20.0,"Cultural":75.0,"Outsider":15.0,"Elite":25.0,"Poor":15.0,"Academic":85.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":45.0,"Dominion":20.0,"Council":45.0,"Court":30.0,"Commonwealth":45.0,"Directorate":45.0,"Ministry":35.0,"Authority":20.0,"Legion":20.0,"Guard":20.0,"Army":20.0,"Militia":20.0,"Vanguard":20.0,"Watch":20.0,"Corps":20.0,"Wardens":20.0,"Rangers":20.0,"Order":50.0,"Covenant":50.0,"Faith":50.0,"Sect":50.0,"Brotherhood":60.0,"Fellowship":60.0,"Chapter":60.0,"Disciples":50.0,"Cult":55.0,"Guild":75.0,"Company":75.0,"Consortium":75.0,"Syndicate":65.0,"Cartel":65.0,"Cooperative":75.0,"Corporation":75.0,"Chamber":75.0,"Trust":75.0,"Clan":45.0,"House":30.0,"Dynasty":30.0,"Family":20.0,"Tribe":45.0,"People":45.0,"Community":45.0,"Nation":40.0,"Kindred":45.0,"Lineage":45.0,"Resistance":20.0,"Movement":45.0,"Front":20.0,"Union":45.0,"Collective":45.0,"Alliance":45.0,"League":50.0,"Pact":55.0,"Liberation":20.0,"Academy":70.0,"Institute":70.0,"School":70.0,"Society":70.0,"Circle":70.0,"Cabal":50.0,"Conclave":50.0,"Scholars":70.0,"Collegium":70.0,"Keepers":60.0,"Caravan":75.0,"Fleet":65.0,"Band":35.0,"Enclave":45.0,"Commune":15.0,"Network":65.0,"Expedition":50.0,"Nomads":45.0,"Diaspora":45.0,"Confederation":55.0,"Assembly":50.0,"Coalition":40.0,"Horde":15.0,"Mercenaries":45.0,"Parliament":20.0,"Senate":30.0,"Association":70.0,"Compact":60.0},"place":{"Apothecary":70.0,"Archive":50.0,"Armoury":0.0,"Auction House":60.0,"Bank":55.0,"Bathhouse":30.0,"Bookshop":100.0,"Brothel":55.0,"Clinic":45.0,"Club":45.0,"Courthouse":20.0,"Bridge":15.0,"Crypt":20.0,"Dormitory":20.0,"Forge":25.0,"Gallery":30.0,"Gambling Den":55.0,"House":0.0,"Laboratory":50.0,"Library":60.0,"Mill":20.0,"Monument":20.0,"Mortuary":20.0,"Museum":55.0,"Observatory":50.0,"Office":50.0,"Pawnshop":55.0,"Post Office":20.0,"Printing House":55.0,"Prison":0.0,"School":55.0,"Sewer":0.0,"Shelter":0.0,"Ship":20.0,"Shop":60.0,"Shrine":15.0,"Stable":0.0,"Tailor":55.0,"Tavern":55.0,"Theater":25.0,"Tower":20.0,"Training Ground":20.0,"Treasury":25.0,"Tunnel":0.0,"Vault":25.0,"Guardhouse":15.0,"Well":0.0,"Arena":30.0,"Barracks":0.0,"Black Market":50.0,"Brewery":35.0,"Camp":5.0,"Cemetery":25.0,"Dock":30.0,"Embassy":5.0,"Factory":20.0,"Farm":5.0,"Fortress":0.0,"Garden":25.0,"Grove":5.0,"Meeting Hall":25.0,"Hideout":5.0,"Inn":30.0,"Manor":5.0,"Market":75.0,"Mine":5.0,"Outpost":5.0,"Palace":5.0,"Research Facility":55.0,"Square":35.0,"Temple":25.0,"Trading Post":50.0,"Warehouse":25.0,"Workshop":35.0,"Catacombs":10.0,"Town Hall":25.0,"Excavation Site":30.0,"Festival Grounds":40.0,"Hospital":45.0,"Ruins":25.0,"Sanctuary":5.0,"Shipyard":25.0,"Caravan Camp":20.0,"Cavern":15.0,"Citadel":10.0,"Enclave":10.0,"Estate":10.0,"Village":15.0,"Hunting Grounds":0.0,"Military Base":0.0,"Mining Settlement":10.0,"Monastery":30.0,"Necropolis":20.0,"Rebel Camp":10.0,"Refugee Camp":10.0,"Slum":10.0,"University":55.0,"Academic District":80.0,"Government District":10.0,"Artisan Quarter":50.0,"Cultural District":45.0,"Entertainment District":40.0,"Farmland":10.0,"Foreign Quarter":50.0,"Wilderness":10.0,"Harbor":65.0,"Industrial District":10.0,"Market District":85.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":10.0,"Royal Quarter":10.0,"Residential District":10.0,"Ruined Quarter":20.0,"Temple District":15.0,"Underground City":25.0,"Underworld":20.0,"Warehouse District":15.0},"score":63.333333333333336},"Brothel":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":100.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":100.0,"Covert":25.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":15.0,"Military":15.0,"Religious":20.0,"Criminal":75.0,"Revolutionary":20.0,"Conspiracy":50.0,"Merchant":90.0,"Clan":25.0,"Cultural":40.0,"Outsider":45.0,"Elite":55.0,"Poor":45.0,"Academic":15.0},"base":{"Empire":20.0,"Kingdom":35.0,"Republic":45.0,"Dominion":20.0,"Council":35.0,"Court":45.0,"Commonwealth":45.0,"Directorate":35.0,"Ministry":15.0,"Authority":20.0,"Legion":20.0,"Guard":20.0,"Army":20.0,"Militia":35.0,"Vanguard":35.0,"Watch":20.0,"Corps":20.0,"Wardens":20.0,"Rangers":35.0,"Order":35.0,"Covenant":20.0,"Faith":20.0,"Sect":45.0,"Brotherhood":45.0,"Fellowship":45.0,"Chapter":45.0,"Disciples":45.0,"Cult":60.0,"Guild":75.0,"Company":75.0,"Consortium":75.0,"Syndicate":75.0,"Cartel":75.0,"Cooperative":65.0,"Corporation":75.0,"Chamber":75.0,"Trust":75.0,"Clan":30.0,"House":45.0,"Dynasty":45.0,"Family":45.0,"Tribe":35.0,"People":55.0,"Community":60.0,"Nation":40.0,"Kindred":30.0,"Lineage":35.0,"Resistance":55.0,"Movement":55.0,"Front":45.0,"Union":60.0,"Collective":60.0,"Alliance":45.0,"League":55.0,"Pact":50.0,"Liberation":55.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":40.0,"Circle":55.0,"Cabal":65.0,"Conclave":65.0,"Scholars":20.0,"Collegium":20.0,"Keepers":35.0,"Caravan":75.0,"Fleet":75.0,"Band":70.0,"Enclave":40.0,"Commune":40.0,"Network":70.0,"Expedition":40.0,"Nomads":60.0,"Diaspora":40.0,"Confederation":50.0,"Assembly":60.0,"Coalition":50.0,"Horde":50.0,"Mercenaries":60.0,"Parliament":40.0,"Senate":45.0,"Association":70.0,"Compact":60.0},"place":{"Apothecary":50.0,"Archive":15.0,"Armoury":0.0,"Auction House":65.0,"Bank":55.0,"Bathhouse":60.0,"Bookshop":55.0,"Brothel":100.0,"Clinic":0.0,"Club":65.0,"Courthouse":0.0,"Bridge":15.0,"Crypt":15.0,"Dormitory":0.0,"Forge":25.0,"Gallery":60.0,"Gambling Den":75.0,"House":20.0,"Laboratory":15.0,"Library":20.0,"Mill":20.0,"Monument":50.0,"Mortuary":0.0,"Museum":45.0,"Observatory":0.0,"Office":45.0,"Pawnshop":60.0,"Post Office":20.0,"Printing House":25.0,"Prison":15.0,"School":20.0,"Sewer":20.0,"Shelter":0.0,"Ship":20.0,"Shop":60.0,"Shrine":20.0,"Stable":0.0,"Tailor":55.0,"Tavern":70.0,"Theater":60.0,"Tower":0.0,"Training Ground":15.0,"Treasury":25.0,"Tunnel":20.0,"Vault":30.0,"Guardhouse":0.0,"Well":5.0,"Arena":55.0,"Barracks":5.0,"Black Market":55.0,"Brewery":35.0,"Camp":5.0,"Cemetery":25.0,"Dock":30.0,"Embassy":10.0,"Factory":20.0,"Farm":5.0,"Fortress":5.0,"Garden":55.0,"Grove":25.0,"Meeting Hall":55.0,"Hideout":25.0,"Inn":40.0,"Manor":10.0,"Market":65.0,"Mine":5.0,"Outpost":5.0,"Palace":20.0,"Research Facility":5.0,"Square":60.0,"Temple":30.0,"Trading Post":50.0,"Warehouse":25.0,"Workshop":30.0,"Catacombs":25.0,"Town Hall":30.0,"Excavation Site":5.0,"Festival Grounds":70.0,"Hospital":5.0,"Ruins":20.0,"Sanctuary":5.0,"Shipyard":25.0,"Caravan Camp":25.0,"Cavern":20.0,"Citadel":10.0,"Enclave":10.0,"Estate":10.0,"Village":20.0,"Hunting Grounds":0.0,"Military Base":0.0,"Mining Settlement":10.0,"Monastery":10.0,"Necropolis":10.0,"Rebel Camp":20.0,"Refugee Camp":15.0,"Slum":20.0,"University":15.0,"Academic District":15.0,"Government District":15.0,"Artisan Quarter":40.0,"Cultural District":70.0,"Entertainment District":90.0,"Farmland":10.0,"Foreign Quarter":55.0,"Wilderness":15.0,"Harbor":75.0,"Industrial District":10.0,"Market District":90.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":15.0,"Royal Quarter":10.0,"Residential District":20.0,"Ruined Quarter":20.0,"Temple District":20.0,"Underground City":25.0,"Underworld":50.0,"Warehouse District":15.0},"score":60.0},"Clinic":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":50.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":100.0,"Exploration":0.0},"type":{"Government":50.0,"Military":45.0,"Religious":70.0,"Criminal":15.0,"Revolutionary":15.0,"Conspiracy":20.0,"Merchant":45.0,"Clan":15.0,"Cultural":20.0,"Outsider":15.0,"Elite":15.0,"Poor":70.0,"Academic":80.0},"base":{"Empire":45.0,"Kingdom":45.0,"Republic":65.0,"Dominion":45.0,"Council":60.0,"Court":45.0,"Commonwealth":60.0,"Directorate":65.0,"Ministry":50.0,"Authority":40.0,"Legion":40.0,"Guard":45.0,"Army":40.0,"Militia":40.0,"Vanguard":40.0,"Watch":40.0,"Corps":40.0,"Wardens":40.0,"Rangers":40.0,"Order":70.0,"Covenant":70.0,"Faith":70.0,"Sect":70.0,"Brotherhood":55.0,"Fellowship":55.0,"Chapter":55.0,"Disciples":65.0,"Cult":70.0,"Guild":40.0,"Company":40.0,"Consortium":40.0,"Syndicate":40.0,"Cartel":40.0,"Cooperative":60.0,"Corporation":40.0,"Chamber":40.0,"Trust":40.0,"Clan":45.0,"House":35.0,"Dynasty":35.0,"Family":45.0,"Tribe":45.0,"People":50.0,"Community":65.0,"Nation":35.0,"Kindred":45.0,"Lineage":35.0,"Resistance":45.0,"Movement":55.0,"Front":35.0,"Union":70.0,"Collective":70.0,"Alliance":35.0,"League":30.0,"Pact":35.0,"Liberation":45.0,"Academy":70.0,"Institute":70.0,"School":70.0,"Society":70.0,"Circle":70.0,"Cabal":50.0,"Conclave":60.0,"Scholars":70.0,"Collegium":70.0,"Keepers":70.0,"Caravan":40.0,"Fleet":40.0,"Band":35.0,"Enclave":45.0,"Commune":50.0,"Network":20.0,"Expedition":50.0,"Nomads":35.0,"Diaspora":45.0,"Confederation":40.0,"Assembly":50.0,"Coalition":40.0,"Horde":40.0,"Mercenaries":40.0,"Parliament":40.0,"Senate":40.0,"Association":60.0,"Compact":40.0},"place":{"Apothecary":60.0,"Archive":50.0,"Armoury":0.0,"Auction House":0.0,"Bank":0.0,"Bathhouse":25.0,"Bookshop":45.0,"Brothel":0.0,"Clinic":100.0,"Club":0.0,"Courthouse":20.0,"Bridge":0.0,"Crypt":20.0,"Dormitory":20.0,"Forge":0.0,"Gallery":0.0,"Gambling Den":0.0,"House":15.0,"Laboratory":50.0,"Library":50.0,"Mill":0.0,"Monument":0.0,"Mortuary":30.0,"Museum":45.0,"Observatory":50.0,"Office":20.0,"Pawnshop":15.0,"Post Office":15.0,"Printing House":40.0,"Prison":0.0,"School":55.0,"Sewer":0.0,"Shelter":50.0,"Ship":0.0,"Shop":0.0,"Shrine":0.0,"Stable":0.0,"Tailor":0.0,"Tavern":15.0,"Theater":0.0,"Tower":20.0,"Training Ground":20.0,"Treasury":0.0,"Tunnel":0.0,"Vault":0.0,"Guardhouse":15.0,"Well":20.0,"Arena":40.0,"Barracks":20.0,"Black Market":5.0,"Brewery":5.0,"Camp":20.0,"Cemetery":25.0,"Dock":5.0,"Embassy":5.0,"Factory":20.0,"Farm":20.0,"Fortress":20.0,"Garden":20.0,"Grove":5.0,"Meeting Hall":20.0,"Hideout":20.0,"Inn":20.0,"Manor":20.0,"Market":5.0,"Mine":25.0,"Outpost":5.0,"Palace":20.0,"Research Facility":55.0,"Square":20.0,"Temple":10.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":25.0,"Catacombs":10.0,"Town Hall":5.0,"Excavation Site":30.0,"Festival Grounds":5.0,"Hospital":100.0,"Ruins":25.0,"Sanctuary":50.0,"Shipyard":5.0,"Caravan Camp":10.0,"Cavern":15.0,"Citadel":10.0,"Enclave":15.0,"Estate":15.0,"Village":20.0,"Hunting Grounds":0.0,"Military Base":15.0,"Mining Settlement":20.0,"Monastery":35.0,"Necropolis":20.0,"Rebel Camp":15.0,"Refugee Camp":65.0,"Slum":30.0,"University":55.0,"Academic District":75.0,"Government District":10.0,"Artisan Quarter":10.0,"Cultural District":15.0,"Entertainment District":10.0,"Farmland":10.0,"Foreign Quarter":10.0,"Wilderness":10.0,"Harbor":10.0,"Industrial District":10.0,"Market District":10.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":10.0,"Royal Quarter":10.0,"Residential District":35.0,"Ruined Quarter":20.0,"Temple District":35.0,"Underground City":15.0,"Underworld":10.0,"Warehouse District":10.0},"score":68.33333333333333},"Club":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":100.0,"Covert":25.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":15.0,"Military":15.0,"Religious":20.0,"Criminal":55.0,"Revolutionary":20.0,"Conspiracy":70.0,"Merchant":45.0,"Clan":25.0,"Cultural":90.0,"Outsider":15.0,"Elite":65.0,"Poor":15.0,"Academic":45.0},"base":{"Empire":45.0,"Kingdom":55.0,"Republic":20.0,"Dominion":45.0,"Council":45.0,"Court":65.0,"Commonwealth":20.0,"Directorate":45.0,"Ministry":40.0,"Authority":45.0,"Legion":20.0,"Guard":20.0,"Army":20.0,"Militia":35.0,"Vanguard":35.0,"Watch":20.0,"Corps":20.0,"Wardens":20.0,"Rangers":20.0,"Order":45.0,"Covenant":20.0,"Faith":20.0,"Sect":35.0,"Brotherhood":30.0,"Fellowship":45.0,"Chapter":45.0,"Disciples":35.0,"Cult":60.0,"Guild":40.0,"Company":40.0,"Consortium":40.0,"Syndicate":45.0,"Cartel":45.0,"Cooperative":40.0,"Corporation":40.0,"Chamber":50.0,"Trust":65.0,"Clan":45.0,"House":65.0,"Dynasty":65.0,"Family":45.0,"Tribe":45.0,"People":45.0,"Community":45.0,"Nation":40.0,"Kindred":45.0,"Lineage":55.0,"Resistance":35.0,"Movement":45.0,"Front":35.0,"Union":20.0,"Collective":20.0,"Alliance":20.0,"League":35.0,"Pact":40.0,"Liberation":35.0,"Academy":40.0,"Institute":40.0,"School":40.0,"Society":100.0,"Circle":95.0,"Cabal":95.0,"Conclave":95.0,"Scholars":40.0,"Collegium":40.0,"Keepers":85.0,"Caravan":40.0,"Fleet":40.0,"Band":45.0,"Enclave":45.0,"Commune":15.0,"Network":60.0,"Expedition":35.0,"Nomads":20.0,"Diaspora":45.0,"Confederation":45.0,"Assembly":40.0,"Coalition":15.0,"Horde":30.0,"Mercenaries":40.0,"Parliament":50.0,"Senate":65.0,"Association":95.0,"Compact":85.0},"place":{"Apothecary":35.0,"Archive":20.0,"Armoury":0.0,"Auction House":60.0,"Bank":45.0,"Bathhouse":55.0,"Bookshop":45.0,"Brothel":65.0,"Clinic":0.0,"Club":100.0,"Courthouse":15.0,"Bridge":0.0,"Crypt":15.0,"Dormitory":0.0,"Forge":0.0,"Gallery":75.0,"Gambling Den":65.0,"House":25.0,"Laboratory":20.0,"Library":25.0,"Mill":0.0,"Monument":65.0,"Mortuary":0.0,"Museum":55.0,"Observatory":0.0,"Office":35.0,"Pawnshop":40.0,"Post Office":0.0,"Printing House":25.0,"Prison":15.0,"School":20.0,"Sewer":20.0,"Shelter":0.0,"Ship":0.0,"Shop":35.0,"Shrine":20.0,"Stable":0.0,"Tailor":40.0,"Tavern":80.0,"Theater":65.0,"Tower":0.0,"Training Ground":15.0,"Treasury":40.0,"Tunnel":20.0,"Vault":45.0,"Guardhouse":0.0,"Well":5.0,"Arena":55.0,"Barracks":5.0,"Black Market":40.0,"Brewery":20.0,"Camp":5.0,"Cemetery":30.0,"Dock":5.0,"Embassy":25.0,"Factory":5.0,"Farm":5.0,"Fortress":5.0,"Garden":65.0,"Grove":25.0,"Meeting Hall":60.0,"Hideout":25.0,"Inn":25.0,"Manor":45.0,"Market":50.0,"Mine":5.0,"Outpost":5.0,"Palace":45.0,"Research Facility":5.0,"Square":55.0,"Temple":35.0,"Trading Post":30.0,"Warehouse":5.0,"Workshop":5.0,"Catacombs":25.0,"Town Hall":35.0,"Excavation Site":5.0,"Festival Grounds":60.0,"Hospital":5.0,"Ruins":20.0,"Sanctuary":5.0,"Shipyard":5.0,"Caravan Camp":10.0,"Cavern":20.0,"Citadel":10.0,"Enclave":10.0,"Estate":45.0,"Village":20.0,"Hunting Grounds":0.0,"Military Base":0.0,"Mining Settlement":10.0,"Monastery":10.0,"Necropolis":15.0,"Rebel Camp":20.0,"Refugee Camp":10.0,"Slum":20.0,"University":20.0,"Academic District":20.0,"Government District":15.0,"Artisan Quarter":15.0,"Cultural District":90.0,"Entertainment District":85.0,"Farmland":10.0,"Foreign Quarter":50.0,"Wilderness":15.0,"Harbor":40.0,"Industrial District":10.0,"Market District":55.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":70.0,"Royal Quarter":65.0,"Residential District":20.0,"Ruined Quarter":20.0,"Temple District":25.0,"Underground City":20.0,"Underworld":25.0,"Warehouse District":10.0},"score":60.0},"Courthouse":{"function":{"Authority":100.0,"Defense":0.0,"Religion":0.0,"Knowledge":25.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":100.0,"Military":50.0,"Religious":15.0,"Criminal":5.0,"Revolutionary":15.0,"Conspiracy":25.0,"Merchant":45.0,"Clan":45.0,"Cultural":20.0,"Outsider":15.0,"Elite":80.0,"Poor":15.0,"Academic":20.0},"base":{"Empire":100.0,"Kingdom":100.0,"Republic":100.0,"Dominion":100.0,"Council":100.0,"Court":100.0,"Commonwealth":100.0,"Directorate":100.0,"Ministry":100.0,"Authority":100.0,"Legion":65.0,"Guard":75.0,"Army":65.0,"Militia":45.0,"Vanguard":40.0,"Watch":75.0,"Corps":65.0,"Wardens":75.0,"Rangers":40.0,"Order":55.0,"Covenant":55.0,"Faith":45.0,"Sect":35.0,"Brotherhood":40.0,"Fellowship":40.0,"Chapter":30.0,"Disciples":35.0,"Cult":35.0,"Guild":40.0,"Company":40.0,"Consortium":40.0,"Syndicate":40.0,"Cartel":35.0,"Cooperative":40.0,"Corporation":40.0,"Chamber":65.0,"Trust":70.0,"Clan":20.0,"House":40.0,"Dynasty":30.0,"Family":40.0,"Tribe":40.0,"People":40.0,"Community":40.0,"Nation":30.0,"Kindred":40.0,"Lineage":20.0,"Resistance":35.0,"Movement":20.0,"Front":35.0,"Union":20.0,"Collective":20.0,"Alliance":35.0,"League":30.0,"Pact":35.0,"Liberation":35.0,"Academy":35.0,"Institute":35.0,"School":35.0,"Society":35.0,"Circle":35.0,"Cabal":30.0,"Conclave":45.0,"Scholars":35.0,"Collegium":35.0,"Keepers":35.0,"Caravan":40.0,"Fleet":40.0,"Band":20.0,"Enclave":35.0,"Commune":15.0,"Network":30.0,"Expedition":20.0,"Nomads":20.0,"Diaspora":20.0,"Confederation":100.0,"Assembly":100.0,"Coalition":100.0,"Horde":40.0,"Mercenaries":40.0,"Parliament":100.0,"Senate":100.0,"Association":40.0,"Compact":45.0},"place":{"Apothecary":15.0,"Archive":55.0,"Armoury":0.0,"Auction House":15.0,"Bank":25.0,"Bathhouse":0.0,"Bookshop":20.0,"Brothel":0.0,"Clinic":20.0,"Club":15.0,"Courthouse":100.0,"Bridge":0.0,"Crypt":15.0,"Dormitory":15.0,"Forge":0.0,"Gallery":20.0,"Gambling Den":15.0,"House":15.0,"Laboratory":20.0,"Library":20.0,"Mill":0.0,"Monument":45.0,"Mortuary":20.0,"Museum":20.0,"Observatory":20.0,"Office":65.0,"Pawnshop":0.0,"Post Office":45.0,"Printing House":20.0,"Prison":90.0,"School":20.0,"Sewer":0.0,"Shelter":0.0,"Ship":0.0,"Shop":0.0,"Shrine":0.0,"Stable":0.0,"Tailor":15.0,"Tavern":0.0,"Theater":15.0,"Tower":20.0,"Training Ground":20.0,"Treasury":60.0,"Tunnel":0.0,"Vault":25.0,"Guardhouse":55.0,"Well":0.0,"Arena":5.0,"Barracks":25.0,"Black Market":5.0,"Brewery":5.0,"Camp":5.0,"Cemetery":5.0,"Dock":5.0,"Embassy":60.0,"Factory":5.0,"Farm":5.0,"Fortress":35.0,"Garden":20.0,"Grove":5.0,"Meeting Hall":45.0,"Hideout":0.0,"Inn":5.0,"Manor":30.0,"Market":5.0,"Mine":5.0,"Outpost":5.0,"Palace":60.0,"Research Facility":25.0,"Square":25.0,"Temple":25.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":20.0,"Catacombs":5.0,"Town Hall":60.0,"Excavation Site":25.0,"Festival Grounds":5.0,"Hospital":25.0,"Ruins":20.0,"Sanctuary":5.0,"Shipyard":5.0,"Caravan Camp":10.0,"Cavern":10.0,"Citadel":75.0,"Enclave":10.0,"Estate":25.0,"Village":10.0,"Hunting Grounds":0.0,"Military Base":0.0,"Mining Settlement":10.0,"Monastery":20.0,"Necropolis":15.0,"Rebel Camp":10.0,"Refugee Camp":10.0,"Slum":10.0,"University":20.0,"Academic District":20.0,"Government District":100.0,"Artisan Quarter":10.0,"Cultural District":10.0,"Entertainment District":10.0,"Farmland":10.0,"Foreign Quarter":10.0,"Wilderness":10.0,"Harbor":10.0,"Industrial District":10.0,"Market District":10.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":45.0,"Royal Quarter":85.0,"Residential District":10.0,"Ruined Quarter":15.0,"Temple District":10.0,"Underground City":10.0,"Underworld":15.0,"Warehouse District":10.0},"score":80.0},"Bridge":{"function":{"Authority":0.0,"Defense":25.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":100.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":50.0,"Care":0.0,"Exploration":0.0},"type":{"Government":50.0,"Military":75.0,"Religious":15.0,"Criminal":15.0,"Revolutionary":20.0,"Conspiracy":15.0,"Merchant":80.0,"Clan":45.0,"Cultural":15.0,"Outsider":75.0,"Elite":15.0,"Poor":50.0,"Academic":15.0},"base":{"Empire":45.0,"Kingdom":45.0,"Republic":45.0,"Dominion":45.0,"Council":40.0,"Court":45.0,"Commonwealth":45.0,"Directorate":45.0,"Ministry":40.0,"Authority":40.0,"Legion":70.0,"Guard":60.0,"Army":70.0,"Militia":70.0,"Vanguard":70.0,"Watch":60.0,"Corps":70.0,"Wardens":60.0,"Rangers":70.0,"Order":50.0,"Covenant":35.0,"Faith":35.0,"Sect":20.0,"Brotherhood":50.0,"Fellowship":50.0,"Chapter":40.0,"Disciples":20.0,"Cult":20.0,"Guild":60.0,"Company":60.0,"Consortium":70.0,"Syndicate":70.0,"Cartel":60.0,"Cooperative":70.0,"Corporation":70.0,"Chamber":70.0,"Trust":60.0,"Clan":50.0,"House":40.0,"Dynasty":40.0,"Family":50.0,"Tribe":60.0,"People":60.0,"Community":60.0,"Nation":45.0,"Kindred":50.0,"Lineage":40.0,"Resistance":35.0,"Movement":35.0,"Front":35.0,"Union":45.0,"Collective":45.0,"Alliance":45.0,"League":45.0,"Pact":45.0,"Liberation":35.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":45.0,"Circle":45.0,"Cabal":45.0,"Conclave":45.0,"Scholars":20.0,"Collegium":20.0,"Keepers":30.0,"Caravan":95.0,"Fleet":70.0,"Band":95.0,"Enclave":95.0,"Commune":65.0,"Network":60.0,"Expedition":95.0,"Nomads":95.0,"Diaspora":95.0,"Confederation":50.0,"Assembly":45.0,"Coalition":45.0,"Horde":70.0,"Mercenaries":70.0,"Parliament":50.0,"Senate":50.0,"Association":60.0,"Compact":60.0},"place":{"Apothecary":15.0,"Archive":0.0,"Armoury":35.0,"Auction House":15.0,"Bank":20.0,"Bathhouse":15.0,"Bookshop":15.0,"Brothel":15.0,"Clinic":0.0,"Club":0.0,"Courthouse":0.0,"Bridge":100.0,"Crypt":0.0,"Dormitory":20.0,"Forge":20.0,"Gallery":15.0,"Gambling Den":15.0,"House":0.0,"Laboratory":0.0,"Library":0.0,"Mill":40.0,"Monument":0.0,"Mortuary":0.0,"Museum":0.0,"Observatory":15.0,"Office":15.0,"Pawnshop":15.0,"Post Office":65.0,"Printing House":0.0,"Prison":20.0,"School":0.0,"Sewer":25.0,"Shelter":20.0,"Ship":75.0,"Shop":20.0,"Shrine":0.0,"Stable":65.0,"Tailor":20.0,"Tavern":20.0,"Theater":15.0,"Tower":20.0,"Training Ground":20.0,"Treasury":20.0,"Tunnel":55.0,"Vault":25.0,"Guardhouse":20.0,"Well":35.0,"Arena":25.0,"Barracks":25.0,"Black Market":25.0,"Brewery":20.0,"Camp":35.0,"Cemetery":5.0,"Dock":70.0,"Embassy":30.0,"Factory":25.0,"Farm":5.0,"Fortress":25.0,"Garden":5.0,"Grove":5.0,"Meeting Hall":5.0,"Hideout":5.0,"Inn":30.0,"Manor":5.0,"Market":25.0,"Mine":5.0,"Outpost":75.0,"Palace":5.0,"Research Facility":5.0,"Square":20.0,"Temple":5.0,"Trading Post":75.0,"Warehouse":70.0,"Workshop":25.0,"Catacombs":5.0,"Town Hall":5.0,"Excavation Site":5.0,"Festival Grounds":20.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":5.0,"Shipyard":65.0,"Caravan Camp":75.0,"Cavern":10.0,"Citadel":20.0,"Enclave":30.0,"Estate":10.0,"Village":35.0,"Hunting Grounds":35.0,"Military Base":30.0,"Mining Settlement":15.0,"Monastery":10.0,"Necropolis":10.0,"Rebel Camp":25.0,"Refugee Camp":40.0,"Slum":10.0,"University":0.0,"Academic District":10.0,"Government District":10.0,"Artisan Quarter":20.0,"Cultural District":10.0,"Entertainment District":20.0,"Farmland":10.0,"Foreign Quarter":30.0,"Wilderness":35.0,"Harbor":85.0,"Industrial District":20.0,"Market District":20.0,"Military District":30.0,"Mining Region":15.0,"Noble Quarter":10.0,"Royal Quarter":10.0,"Residential District":10.0,"Ruined Quarter":10.0,"Temple District":10.0,"Underground City":20.0,"Underworld":10.0,"Warehouse District":75.0},"score":73.33333333333333},"Crypt":{"function":{"Authority":0.0,"Defense":0.0,"Religion":75.0,"Knowledge":25.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":75.0},"type":{"Government":15.0,"Military":15.0,"Religious":85.0,"Criminal":45.0,"Revolutionary":15.0,"Conspiracy":70.0,"Merchant":15.0,"Clan":65.0,"Cultural":25.0,"Outsider":25.0,"Elite":45.0,"Poor":15.0,"Academic":55.0},"base":{"Empire":45.0,"Kingdom":55.0,"Republic":20.0,"Dominion":45.0,"Council":35.0,"Court":60.0,"Commonwealth":20.0,"Directorate":35.0,"Ministry":50.0,"Authority":45.0,"Legion":20.0,"Guard":20.0,"Army":20.0,"Militia":35.0,"Vanguard":35.0,"Watch":20.0,"Corps":20.0,"Wardens":20.0,"Rangers":30.0,"Order":95.0,"Covenant":95.0,"Faith":95.0,"Sect":95.0,"Brotherhood":95.0,"Fellowship":95.0,"Chapter":95.0,"Disciples":95.0,"Cult":95.0,"Guild":20.0,"Company":35.0,"Consortium":20.0,"Syndicate":60.0,"Cartel":60.0,"Cooperative":20.0,"Corporation":20.0,"Chamber":35.0,"Trust":40.0,"Clan":55.0,"House":65.0,"Dynasty":65.0,"Family":55.0,"Tribe":55.0,"People":55.0,"Community":55.0,"Nation":35.0,"Kindred":55.0,"Lineage":65.0,"Resistance":45.0,"Movement":35.0,"Front":45.0,"Union":20.0,"Collective":20.0,"Alliance":20.0,"League":20.0,"Pact":40.0,"Liberation":45.0,"Academy":65.0,"Institute":65.0,"School":65.0,"Society":55.0,"Circle":60.0,"Cabal":90.0,"Conclave":70.0,"Scholars":65.0,"Collegium":65.0,"Keepers":70.0,"Caravan":30.0,"Fleet":35.0,"Band":60.0,"Enclave":40.0,"Commune":25.0,"Network":70.0,"Expedition":35.0,"Nomads":30.0,"Diaspora":30.0,"Confederation":40.0,"Assembly":25.0,"Coalition":25.0,"Horde":35.0,"Mercenaries":35.0,"Parliament":55.0,"Senate":60.0,"Association":35.0,"Compact":35.0},"place":{"Apothecary":20.0,"Archive":20.0,"Armoury":0.0,"Auction House":15.0,"Bank":15.0,"Bathhouse":0.0,"Bookshop":20.0,"Brothel":15.0,"Clinic":20.0,"Club":15.0,"Courthouse":15.0,"Bridge":0.0,"Crypt":100.0,"Dormitory":20.0,"Forge":0.0,"Gallery":20.0,"Gambling Den":15.0,"House":15.0,"Laboratory":20.0,"Library":25.0,"Mill":0.0,"Monument":20.0,"Mortuary":55.0,"Museum":20.0,"Observatory":25.0,"Office":20.0,"Pawnshop":15.0,"Post Office":15.0,"Printing House":20.0,"Prison":20.0,"School":25.0,"Sewer":50.0,"Shelter":0.0,"Ship":0.0,"Shop":0.0,"Shrine":55.0,"Stable":0.0,"Tailor":15.0,"Tavern":0.0,"Theater":15.0,"Tower":25.0,"Training Ground":15.0,"Treasury":15.0,"Tunnel":45.0,"Vault":20.0,"Guardhouse":15.0,"Well":0.0,"Arena":5.0,"Barracks":5.0,"Black Market":20.0,"Brewery":5.0,"Camp":5.0,"Cemetery":100.0,"Dock":5.0,"Embassy":20.0,"Factory":5.0,"Farm":5.0,"Fortress":5.0,"Garden":20.0,"Grove":40.0,"Meeting Hall":5.0,"Hideout":25.0,"Inn":5.0,"Manor":20.0,"Market":5.0,"Mine":5.0,"Outpost":5.0,"Palace":20.0,"Research Facility":30.0,"Square":5.0,"Temple":70.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":20.0,"Catacombs":100.0,"Town Hall":5.0,"Excavation Site":50.0,"Festival Grounds":10.0,"Hospital":25.0,"Ruins":75.0,"Sanctuary":50.0,"Shipyard":5.0,"Caravan Camp":10.0,"Cavern":75.0,"Citadel":10.0,"Enclave":10.0,"Estate":20.0,"Village":10.0,"Hunting Grounds":40.0,"Military Base":0.0,"Mining Settlement":10.0,"Monastery":75.0,"Necropolis":85.0,"Rebel Camp":20.0,"Refugee Camp":10.0,"Slum":10.0,"University":20.0,"Academic District":25.0,"Government District":10.0,"Artisan Quarter":10.0,"Cultural District":15.0,"Entertainment District":10.0,"Farmland":10.0,"Foreign Quarter":10.0,"Wilderness":60.0,"Harbor":10.0,"Industrial District":10.0,"Market District":10.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":20.0,"Royal Quarter":35.0,"Residential District":10.0,"Ruined Quarter":80.0,"Temple District":80.0,"Underground City":70.0,"Underworld":35.0,"Warehouse District":10.0},"score":70.0},"Dormitory":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":25.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":100.0,"Social":0.0,"Covert":0.0,"Migration":25.0,"Care":0.0,"Exploration":0.0},"type":{"Government":15.0,"Military":65.0,"Religious":45.0,"Criminal":15.0,"Revolutionary":25.0,"Conspiracy":20.0,"Merchant":15.0,"Clan":65.0,"Cultural":20.0,"Outsider":80.0,"Elite":30.0,"Poor":80.0,"Academic":70.0},"base":{"Empire":35.0,"Kingdom":35.0,"Republic":45.0,"Dominion":35.0,"Council":45.0,"Court":35.0,"Commonwealth":45.0,"Directorate":45.0,"Ministry":30.0,"Authority":35.0,"Legion":65.0,"Guard":65.0,"Army":65.0,"Militia":65.0,"Vanguard":65.0,"Watch":65.0,"Corps":65.0,"Wardens":65.0,"Rangers":70.0,"Order":65.0,"Covenant":60.0,"Faith":60.0,"Sect":60.0,"Brotherhood":65.0,"Fellowship":65.0,"Chapter":60.0,"Disciples":60.0,"Cult":60.0,"Guild":20.0,"Company":45.0,"Consortium":20.0,"Syndicate":45.0,"Cartel":45.0,"Cooperative":60.0,"Corporation":20.0,"Chamber":20.0,"Trust":35.0,"Clan":55.0,"House":45.0,"Dynasty":45.0,"Family":55.0,"Tribe":60.0,"People":55.0,"Community":60.0,"Nation":50.0,"Kindred":55.0,"Lineage":45.0,"Resistance":45.0,"Movement":45.0,"Front":35.0,"Union":60.0,"Collective":60.0,"Alliance":35.0,"League":35.0,"Pact":35.0,"Liberation":45.0,"Academy":70.0,"Institute":70.0,"School":70.0,"Society":65.0,"Circle":70.0,"Cabal":60.0,"Conclave":60.0,"Scholars":70.0,"Collegium":70.0,"Keepers":65.0,"Caravan":60.0,"Fleet":60.0,"Band":60.0,"Enclave":70.0,"Commune":55.0,"Network":45.0,"Expedition":70.0,"Nomads":60.0,"Diaspora":70.0,"Confederation":50.0,"Assembly":60.0,"Coalition":50.0,"Horde":70.0,"Mercenaries":70.0,"Parliament":50.0,"Senate":50.0,"Association":70.0,"Compact":50.0},"place":{"Apothecary":20.0,"Archive":20.0,"Armoury":15.0,"Auction House":0.0,"Bank":0.0,"Bathhouse":35.0,"Bookshop":20.0,"Brothel":0.0,"Clinic":20.0,"Club":0.0,"Courthouse":15.0,"Bridge":20.0,"Crypt":20.0,"Dormitory":100.0,"Forge":15.0,"Gallery":0.0,"Gambling Den":0.0,"House":60.0,"Laboratory":20.0,"Library":20.0,"Mill":0.0,"Monument":0.0,"Mortuary":20.0,"Museum":20.0,"Observatory":20.0,"Office":15.0,"Pawnshop":0.0,"Post Office":20.0,"Printing House":20.0,"Prison":20.0,"School":20.0,"Sewer":0.0,"Shelter":60.0,"Ship":20.0,"Shop":0.0,"Shrine":0.0,"Stable":20.0,"Tailor":0.0,"Tavern":20.0,"Theater":0.0,"Tower":20.0,"Training Ground":20.0,"Treasury":0.0,"Tunnel":15.0,"Vault":0.0,"Guardhouse":15.0,"Well":20.0,"Arena":20.0,"Barracks":55.0,"Black Market":5.0,"Brewery":5.0,"Camp":65.0,"Cemetery":25.0,"Dock":25.0,"Embassy":25.0,"Factory":5.0,"Farm":30.0,"Fortress":25.0,"Garden":35.0,"Grove":5.0,"Meeting Hall":20.0,"Hideout":25.0,"Inn":55.0,"Manor":60.0,"Market":25.0,"Mine":5.0,"Outpost":25.0,"Palace":30.0,"Research Facility":25.0,"Square":35.0,"Temple":5.0,"Trading Post":25.0,"Warehouse":5.0,"Workshop":25.0,"Catacombs":5.0,"Town Hall":5.0,"Excavation Site":25.0,"Festival Grounds":5.0,"Hospital":25.0,"Ruins":25.0,"Sanctuary":30.0,"Shipyard":5.0,"Caravan Camp":30.0,"Cavern":15.0,"Citadel":20.0,"Enclave":60.0,"Estate":60.0,"Village":70.0,"Hunting Grounds":0.0,"Military Base":25.0,"Mining Settlement":70.0,"Monastery":50.0,"Necropolis":20.0,"Rebel Camp":30.0,"Refugee Camp":75.0,"Slum":60.0,"University":70.0,"Academic District":90.0,"Government District":10.0,"Artisan Quarter":10.0,"Cultural District":10.0,"Entertainment District":10.0,"Farmland":10.0,"Foreign Quarter":65.0,"Wilderness":10.0,"Harbor":15.0,"Industrial District":10.0,"Market District":10.0,"Military District":40.0,"Mining Region":10.0,"Noble Quarter":60.0,"Royal Quarter":45.0,"Residential District":100.0,"Ruined Quarter":25.0,"Temple District":10.0,"Underground City":75.0,"Underworld":10.0,"Warehouse District":10.0},"score":60.0},"Forge":{"function":{"Authority":0.0,"Defense":25.0,"Religion":0.0,"Knowledge":0.0,"Commerce":50.0,"Production":100.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":20.0,"Military":70.0,"Religious":15.0,"Criminal":20.0,"Revolutionary":50.0,"Conspiracy":15.0,"Merchant":80.0,"Clan":45.0,"Cultural":15.0,"Outsider":15.0,"Elite":20.0,"Poor":65.0,"Academic":15.0},"base":{"Empire":45.0,"Kingdom":35.0,"Republic":35.0,"Dominion":45.0,"Council":35.0,"Court":20.0,"Commonwealth":35.0,"Directorate":35.0,"Ministry":35.0,"Authority":35.0,"Legion":70.0,"Guard":70.0,"Army":70.0,"Militia":70.0,"Vanguard":70.0,"Watch":65.0,"Corps":70.0,"Wardens":65.0,"Rangers":65.0,"Order":50.0,"Covenant":35.0,"Faith":35.0,"Sect":20.0,"Brotherhood":60.0,"Fellowship":50.0,"Chapter":40.0,"Disciples":20.0,"Cult":20.0,"Guild":70.0,"Company":70.0,"Consortium":70.0,"Syndicate":60.0,"Cartel":50.0,"Cooperative":70.0,"Corporation":70.0,"Chamber":70.0,"Trust":60.0,"Clan":60.0,"House":40.0,"Dynasty":40.0,"Family":40.0,"Tribe":60.0,"People":50.0,"Community":55.0,"Nation":30.0,"Kindred":60.0,"Lineage":40.0,"Resistance":65.0,"Movement":55.0,"Front":65.0,"Union":65.0,"Collective":65.0,"Alliance":60.0,"League":60.0,"Pact":60.0,"Liberation":65.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":20.0,"Circle":20.0,"Cabal":35.0,"Conclave":35.0,"Scholars":20.0,"Collegium":20.0,"Keepers":40.0,"Caravan":60.0,"Fleet":60.0,"Band":35.0,"Enclave":35.0,"Commune":50.0,"Network":45.0,"Expedition":45.0,"Nomads":35.0,"Diaspora":20.0,"Confederation":50.0,"Assembly":50.0,"Coalition":50.0,"Horde":65.0,"Mercenaries":70.0,"Parliament":35.0,"Senate":35.0,"Association":70.0,"Compact":50.0},"place":{"Apothecary":25.0,"Archive":0.0,"Armoury":30.0,"Auction House":25.0,"Bank":25.0,"Bathhouse":20.0,"Bookshop":25.0,"Brothel":25.0,"Clinic":0.0,"Club":0.0,"Courthouse":0.0,"Bridge":20.0,"Crypt":0.0,"Dormitory":15.0,"Forge":100.0,"Gallery":20.0,"Gambling Den":25.0,"House":0.0,"Laboratory":20.0,"Library":0.0,"Mill":60.0,"Monument":0.0,"Mortuary":0.0,"Museum":0.0,"Observatory":15.0,"Office":20.0,"Pawnshop":25.0,"Post Office":20.0,"Printing House":55.0,"Prison":20.0,"School":0.0,"Sewer":0.0,"Shelter":0.0,"Ship":20.0,"Shop":35.0,"Shrine":0.0,"Stable":20.0,"Tailor":65.0,"Tavern":25.0,"Theater":20.0,"Tower":20.0,"Training Ground":20.0,"Treasury":20.0,"Tunnel":15.0,"Vault":20.0,"Guardhouse":20.0,"Well":0.0,"Arena":25.0,"Barracks":25.0,"Black Market":30.0,"Brewery":65.0,"Camp":25.0,"Cemetery":5.0,"Dock":30.0,"Embassy":5.0,"Factory":80.0,"Farm":35.0,"Fortress":25.0,"Garden":5.0,"Grove":5.0,"Meeting Hall":5.0,"Hideout":5.0,"Inn":25.0,"Manor":5.0,"Market":30.0,"Mine":30.0,"Outpost":25.0,"Palace":5.0,"Research Facility":5.0,"Square":25.0,"Temple":5.0,"Trading Post":30.0,"Warehouse":35.0,"Workshop":80.0,"Catacombs":5.0,"Town Hall":5.0,"Excavation Site":10.0,"Festival Grounds":30.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":5.0,"Shipyard":75.0,"Caravan Camp":20.0,"Cavern":10.0,"Citadel":20.0,"Enclave":10.0,"Estate":10.0,"Village":15.0,"Hunting Grounds":0.0,"Military Base":20.0,"Mining Settlement":40.0,"Monastery":10.0,"Necropolis":10.0,"Rebel Camp":20.0,"Refugee Camp":10.0,"Slum":10.0,"University":0.0,"Academic District":10.0,"Government District":10.0,"Artisan Quarter":100.0,"Cultural District":10.0,"Entertainment District":25.0,"Farmland":15.0,"Foreign Quarter":15.0,"Wilderness":10.0,"Harbor":25.0,"Industrial District":85.0,"Market District":25.0,"Military District":40.0,"Mining Region":20.0,"Noble Quarter":10.0,"Royal Quarter":10.0,"Residential District":10.0,"Ruined Quarter":10.0,"Temple District":10.0,"Underground City":20.0,"Underworld":15.0,"Warehouse District":20.0},"score":85.0},"Gallery":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":25.0,"Commerce":25.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":100.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":45.0,"Military":15.0,"Religious":20.0,"Criminal":20.0,"Revolutionary":15.0,"Conspiracy":20.0,"Merchant":50.0,"Clan":25.0,"Cultural":90.0,"Outsider":15.0,"Elite":70.0,"Poor":15.0,"Academic":50.0},"base":{"Empire":60.0,"Kingdom":60.0,"Republic":40.0,"Dominion":60.0,"Council":50.0,"Court":95.0,"Commonwealth":40.0,"Directorate":50.0,"Ministry":60.0,"Authority":60.0,"Legion":20.0,"Guard":35.0,"Army":20.0,"Militia":20.0,"Vanguard":20.0,"Watch":35.0,"Corps":20.0,"Wardens":35.0,"Rangers":20.0,"Order":35.0,"Covenant":20.0,"Faith":20.0,"Sect":20.0,"Brotherhood":35.0,"Fellowship":45.0,"Chapter":35.0,"Disciples":20.0,"Cult":20.0,"Guild":45.0,"Company":45.0,"Consortium":45.0,"Syndicate":45.0,"Cartel":45.0,"Cooperative":45.0,"Corporation":45.0,"Chamber":50.0,"Trust":65.0,"Clan":45.0,"House":70.0,"Dynasty":70.0,"Family":45.0,"Tribe":45.0,"People":45.0,"Community":45.0,"Nation":40.0,"Kindred":45.0,"Lineage":55.0,"Resistance":20.0,"Movement":45.0,"Front":20.0,"Union":35.0,"Collective":35.0,"Alliance":35.0,"League":30.0,"Pact":35.0,"Liberation":20.0,"Academy":45.0,"Institute":45.0,"School":45.0,"Society":90.0,"Circle":65.0,"Cabal":45.0,"Conclave":55.0,"Scholars":45.0,"Collegium":45.0,"Keepers":40.0,"Caravan":45.0,"Fleet":40.0,"Band":35.0,"Enclave":45.0,"Commune":15.0,"Network":35.0,"Expedition":45.0,"Nomads":35.0,"Diaspora":45.0,"Confederation":50.0,"Assembly":50.0,"Coalition":40.0,"Horde":15.0,"Mercenaries":35.0,"Parliament":60.0,"Senate":70.0,"Association":55.0,"Compact":40.0},"place":{"Apothecary":20.0,"Archive":5.0,"Armoury":0.0,"Auction House":60.0,"Bank":45.0,"Bathhouse":65.0,"Bookshop":30.0,"Brothel":60.0,"Clinic":0.0,"Club":75.0,"Courthouse":20.0,"Bridge":15.0,"Crypt":20.0,"Dormitory":0.0,"Forge":20.0,"Gallery":100.0,"Gambling Den":60.0,"House":25.0,"Laboratory":5.0,"Library":30.0,"Mill":15.0,"Monument":65.0,"Mortuary":0.0,"Museum":60.0,"Observatory":5.0,"Office":20.0,"Pawnshop":20.0,"Post Office":20.0,"Printing House":25.0,"Prison":0.0,"School":30.0,"Sewer":0.0,"Shelter":0.0,"Ship":15.0,"Shop":25.0,"Shrine":20.0,"Stable":0.0,"Tailor":25.0,"Tavern":60.0,"Theater":70.0,"Tower":0.0,"Training Ground":15.0,"Treasury":45.0,"Tunnel":0.0,"Vault":40.0,"Guardhouse":0.0,"Well":5.0,"Arena":60.0,"Barracks":5.0,"Black Market":25.0,"Brewery":25.0,"Camp":5.0,"Cemetery":30.0,"Dock":25.0,"Embassy":25.0,"Factory":20.0,"Farm":5.0,"Fortress":5.0,"Garden":65.0,"Grove":25.0,"Meeting Hall":60.0,"Hideout":0.0,"Inn":35.0,"Manor":45.0,"Market":40.0,"Mine":5.0,"Outpost":5.0,"Palace":45.0,"Research Facility":10.0,"Square":65.0,"Temple":35.0,"Trading Post":25.0,"Warehouse":20.0,"Workshop":25.0,"Catacombs":5.0,"Town Hall":35.0,"Excavation Site":5.0,"Festival Grounds":65.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":5.0,"Shipyard":20.0,"Caravan Camp":20.0,"Cavern":10.0,"Citadel":10.0,"Enclave":10.0,"Estate":50.0,"Village":20.0,"Hunting Grounds":0.0,"Military Base":0.0,"Mining Settlement":10.0,"Monastery":15.0,"Necropolis":10.0,"Rebel Camp":10.0,"Refugee Camp":10.0,"Slum":20.0,"University":25.0,"Academic District":25.0,"Government District":15.0,"Artisan Quarter":25.0,"Cultural District":100.0,"Entertainment District":75.0,"Farmland":10.0,"Foreign Quarter":20.0,"Wilderness":10.0,"Harbor":20.0,"Industrial District":10.0,"Market District":35.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":70.0,"Royal Quarter":65.0,"Residential District":20.0,"Ruined Quarter":10.0,"Temple District":25.0,"Underground City":20.0,"Underworld":15.0,"Warehouse District":10.0},"score":46.666666666666664},"Gambling Den":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":100.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":100.0,"Covert":50.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":15.0,"Military":15.0,"Religious":20.0,"Criminal":80.0,"Revolutionary":25.0,"Conspiracy":60.0,"Merchant":90.0,"Clan":25.0,"Cultural":70.0,"Outsider":15.0,"Elite":75.0,"Poor":45.0,"Academic":15.0},"base":{"Empire":35.0,"Kingdom":45.0,"Republic":45.0,"Dominion":35.0,"Council":45.0,"Court":70.0,"Commonwealth":45.0,"Directorate":45.0,"Ministry":30.0,"Authority":35.0,"Legion":20.0,"Guard":20.0,"Army":20.0,"Militia":35.0,"Vanguard":35.0,"Watch":20.0,"Corps":20.0,"Wardens":20.0,"Rangers":20.0,"Order":35.0,"Covenant":20.0,"Faith":20.0,"Sect":45.0,"Brotherhood":45.0,"Fellowship":45.0,"Chapter":45.0,"Disciples":45.0,"Cult":65.0,"Guild":75.0,"Company":75.0,"Consortium":75.0,"Syndicate":100.0,"Cartel":100.0,"Cooperative":65.0,"Corporation":75.0,"Chamber":75.0,"Trust":75.0,"Clan":35.0,"House":70.0,"Dynasty":70.0,"Family":45.0,"Tribe":35.0,"People":35.0,"Community":40.0,"Nation":30.0,"Kindred":35.0,"Lineage":55.0,"Resistance":55.0,"Movement":55.0,"Front":45.0,"Union":50.0,"Collective":50.0,"Alliance":45.0,"League":55.0,"Pact":55.0,"Liberation":55.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":65.0,"Circle":65.0,"Cabal":65.0,"Conclave":65.0,"Scholars":20.0,"Collegium":20.0,"Keepers":35.0,"Caravan":75.0,"Fleet":65.0,"Band":70.0,"Enclave":35.0,"Commune":30.0,"Network":95.0,"Expedition":35.0,"Nomads":45.0,"Diaspora":35.0,"Confederation":55.0,"Assembly":50.0,"Coalition":40.0,"Horde":40.0,"Mercenaries":60.0,"Parliament":45.0,"Senate":60.0,"Association":70.0,"Compact":60.0},"place":{"Apothecary":50.0,"Archive":20.0,"Armoury":0.0,"Auction House":65.0,"Bank":55.0,"Bathhouse":60.0,"Bookshop":55.0,"Brothel":75.0,"Clinic":0.0,"Club":65.0,"Courthouse":15.0,"Bridge":15.0,"Crypt":15.0,"Dormitory":0.0,"Forge":25.0,"Gallery":60.0,"Gambling Den":100.0,"House":25.0,"Laboratory":20.0,"Library":20.0,"Mill":20.0,"Monument":55.0,"Mortuary":0.0,"Museum":50.0,"Observatory":0.0,"Office":45.0,"Pawnshop":60.0,"Post Office":20.0,"Printing House":25.0,"Prison":15.0,"School":20.0,"Sewer":40.0,"Shelter":0.0,"Ship":20.0,"Shop":55.0,"Shrine":20.0,"Stable":0.0,"Tailor":55.0,"Tavern":85.0,"Theater":65.0,"Tower":0.0,"Training Ground":15.0,"Treasury":30.0,"Tunnel":40.0,"Vault":30.0,"Guardhouse":0.0,"Well":0.0,"Arena":70.0,"Barracks":5.0,"Black Market":65.0,"Brewery":35.0,"Camp":5.0,"Cemetery":25.0,"Dock":30.0,"Embassy":25.0,"Factory":20.0,"Farm":5.0,"Fortress":5.0,"Garden":55.0,"Grove":20.0,"Meeting Hall":50.0,"Hideout":45.0,"Inn":35.0,"Manor":25.0,"Market":65.0,"Mine":5.0,"Outpost":5.0,"Palace":25.0,"Research Facility":5.0,"Square":60.0,"Temple":30.0,"Trading Post":50.0,"Warehouse":25.0,"Workshop":30.0,"Catacombs":25.0,"Town Hall":30.0,"Excavation Site":5.0,"Festival Grounds":65.0,"Hospital":5.0,"Ruins":20.0,"Sanctuary":5.0,"Shipyard":25.0,"Caravan Camp":25.0,"Cavern":15.0,"Citadel":10.0,"Enclave":10.0,"Estate":20.0,"Village":20.0,"Hunting Grounds":0.0,"Military Base":0.0,"Mining Settlement":10.0,"Monastery":10.0,"Necropolis":10.0,"Rebel Camp":45.0,"Refugee Camp":10.0,"Slum":20.0,"University":15.0,"Academic District":15.0,"Government District":15.0,"Artisan Quarter":40.0,"Cultural District":80.0,"Entertainment District":100.0,"Farmland":10.0,"Foreign Quarter":50.0,"Wilderness":15.0,"Harbor":65.0,"Industrial District":10.0,"Market District":90.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":20.0,"Royal Quarter":15.0,"Residential District":20.0,"Ruined Quarter":20.0,"Temple District":20.0,"Underground City":25.0,"Underworld":90.0,"Warehouse District":15.0},"score":60.0},"House":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":100.0,"Social":50.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":15.0,"Military":15.0,"Religious":20.0,"Criminal":20.0,"Revolutionary":25.0,"Conspiracy":15.0,"Merchant":15.0,"Clan":95.0,"Cultural":60.0,"Outsider":55.0,"Elite":80.0,"Poor":80.0,"Academic":15.0},"base":{"Empire":35.0,"Kingdom":45.0,"Republic":45.0,"Dominion":35.0,"Council":45.0,"Court":70.0,"Commonwealth":45.0,"Directorate":45.0,"Ministry":30.0,"Authority":35.0,"Legion":35.0,"Guard":35.0,"Army":35.0,"Militia":35.0,"Vanguard":35.0,"Watch":35.0,"Corps":35.0,"Wardens":35.0,"Rangers":45.0,"Order":35.0,"Covenant":45.0,"Faith":35.0,"Sect":35.0,"Brotherhood":70.0,"Fellowship":70.0,"Chapter":45.0,"Disciples":35.0,"Cult":35.0,"Guild":20.0,"Company":35.0,"Consortium":20.0,"Syndicate":45.0,"Cartel":45.0,"Cooperative":70.0,"Corporation":20.0,"Chamber":45.0,"Trust":70.0,"Clan":95.0,"House":70.0,"Dynasty":70.0,"Family":95.0,"Tribe":55.0,"People":95.0,"Community":95.0,"Nation":90.0,"Kindred":95.0,"Lineage":70.0,"Resistance":45.0,"Movement":55.0,"Front":35.0,"Union":70.0,"Collective":70.0,"Alliance":45.0,"League":45.0,"Pact":45.0,"Liberation":45.0,"Academy":35.0,"Institute":35.0,"School":35.0,"Society":55.0,"Circle":45.0,"Cabal":45.0,"Conclave":55.0,"Scholars":35.0,"Collegium":35.0,"Keepers":35.0,"Caravan":45.0,"Fleet":45.0,"Band":45.0,"Enclave":70.0,"Commune":60.0,"Network":35.0,"Expedition":55.0,"Nomads":65.0,"Diaspora":65.0,"Confederation":50.0,"Assembly":50.0,"Coalition":40.0,"Horde":45.0,"Mercenaries":45.0,"Parliament":55.0,"Senate":60.0,"Association":40.0,"Compact":50.0},"place":{"Apothecary":0.0,"Archive":0.0,"Armoury":0.0,"Auction House":20.0,"Bank":15.0,"Bathhouse":50.0,"Bookshop":0.0,"Brothel":20.0,"Clinic":15.0,"Club":25.0,"Courthouse":15.0,"Bridge":0.0,"Crypt":15.0,"Dormitory":60.0,"Forge":0.0,"Gallery":25.0,"Gambling Den":25.0,"House":100.0,"Laboratory":0.0,"Library":0.0,"Mill":0.0,"Monument":25.0,"Mortuary":0.0,"Museum":20.0,"Observatory":0.0,"Office":0.0,"Pawnshop":30.0,"Post Office":0.0,"Printing House":0.0,"Prison":0.0,"School":20.0,"Sewer":30.0,"Shelter":65.0,"Ship":0.0,"Shop":0.0,"Shrine":5.0,"Stable":0.0,"Tailor":15.0,"Tavern":30.0,"Theater":25.0,"Tower":0.0,"Training Ground":0.0,"Treasury":15.0,"Tunnel":0.0,"Vault":15.0,"Guardhouse":0.0,"Well":25.0,"Arena":25.0,"Barracks":55.0,"Black Market":30.0,"Brewery":5.0,"Camp":60.0,"Cemetery":30.0,"Dock":5.0,"Embassy":25.0,"Factory":5.0,"Farm":50.0,"Fortress":25.0,"Garden":50.0,"Grove":10.0,"Meeting Hall":30.0,"Hideout":40.0,"Inn":60.0,"Manor":80.0,"Market":25.0,"Mine":5.0,"Outpost":5.0,"Palace":40.0,"Research Facility":0.0,"Square":45.0,"Temple":25.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":5.0,"Catacombs":5.0,"Town Hall":25.0,"Excavation Site":5.0,"Festival Grounds":25.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":30.0,"Shipyard":5.0,"Caravan Camp":20.0,"Cavern":10.0,"Citadel":15.0,"Enclave":75.0,"Estate":75.0,"Village":90.0,"Hunting Grounds":0.0,"Military Base":20.0,"Mining Settlement":80.0,"Monastery":30.0,"Necropolis":10.0,"Rebel Camp":25.0,"Refugee Camp":75.0,"Slum":85.0,"University":50.0,"Academic District":15.0,"Government District":10.0,"Artisan Quarter":15.0,"Cultural District":20.0,"Entertainment District":20.0,"Farmland":10.0,"Foreign Quarter":80.0,"Wilderness":10.0,"Harbor":10.0,"Industrial District":10.0,"Market District":15.0,"Military District":15.0,"Mining Region":10.0,"Noble Quarter":70.0,"Royal Quarter":55.0,"Residential District":100.0,"Ruined Quarter":70.0,"Temple District":15.0,"Underground City":70.0,"Underworld":10.0,"Warehouse District":10.0},"score":89.66666666666667},"Laboratory":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":100.0,"Commerce":0.0,"Production":25.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":25.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":50.0,"Military":45.0,"Religious":15.0,"Criminal":20.0,"Revolutionary":20.0,"Conspiracy":60.0,"Merchant":50.0,"Clan":15.0,"Cultural":25.0,"Outsider":15.0,"Elite":15.0,"Poor":15.0,"Academic":90.0},"base":{"Empire":45.0,"Kingdom":45.0,"Republic":45.0,"Dominion":45.0,"Council":60.0,"Court":45.0,"Commonwealth":45.0,"Directorate":65.0,"Ministry":40.0,"Authority":40.0,"Legion":40.0,"Guard":45.0,"Army":40.0,"Militia":40.0,"Vanguard":40.0,"Watch":40.0,"Corps":40.0,"Wardens":40.0,"Rangers":40.0,"Order":60.0,"Covenant":45.0,"Faith":45.0,"Sect":45.0,"Brotherhood":45.0,"Fellowship":45.0,"Chapter":50.0,"Disciples":45.0,"Cult":65.0,"Guild":65.0,"Company":65.0,"Consortium":65.0,"Syndicate":45.0,"Cartel":45.0,"Cooperative":60.0,"Corporation":65.0,"Chamber":60.0,"Trust":40.0,"Clan":20.0,"House":20.0,"Dynasty":20.0,"Family":20.0,"Tribe":20.0,"People":15.0,"Community":20.0,"Nation":15.0,"Kindred":20.0,"Lineage":20.0,"Resistance":35.0,"Movement":35.0,"Front":35.0,"Union":45.0,"Collective":45.0,"Alliance":20.0,"League":35.0,"Pact":40.0,"Liberation":35.0,"Academy":100.0,"Institute":100.0,"School":100.0,"Society":75.0,"Circle":75.0,"Cabal":55.0,"Conclave":55.0,"Scholars":100.0,"Collegium":100.0,"Keepers":70.0,"Caravan":45.0,"Fleet":40.0,"Band":35.0,"Enclave":20.0,"Commune":15.0,"Network":45.0,"Expedition":45.0,"Nomads":20.0,"Diaspora":20.0,"Confederation":40.0,"Assembly":40.0,"Coalition":35.0,"Horde":40.0,"Mercenaries":40.0,"Parliament":40.0,"Senate":40.0,"Association":60.0,"Compact":40.0},"place":{"Apothecary":50.0,"Archive":65.0,"Armoury":20.0,"Auction House":0.0,"Bank":0.0,"Bathhouse":0.0,"Bookshop":50.0,"Brothel":15.0,"Clinic":50.0,"Club":20.0,"Courthouse":20.0,"Bridge":0.0,"Crypt":20.0,"Dormitory":20.0,"Forge":20.0,"Gallery":5.0,"Gambling Den":20.0,"House":0.0,"Laboratory":100.0,"Library":65.0,"Mill":20.0,"Monument":0.0,"Mortuary":25.0,"Museum":55.0,"Observatory":65.0,"Office":25.0,"Pawnshop":15.0,"Post Office":20.0,"Printing House":55.0,"Prison":15.0,"School":65.0,"Sewer":20.0,"Shelter":0.0,"Ship":0.0,"Shop":15.0,"Shrine":0.0,"Stable":0.0,"Tailor":20.0,"Tavern":0.0,"Theater":0.0,"Tower":25.0,"Training Ground":20.0,"Treasury":15.0,"Tunnel":20.0,"Vault":20.0,"Guardhouse":15.0,"Well":0.0,"Arena":5.0,"Barracks":5.0,"Black Market":25.0,"Brewery":25.0,"Camp":5.0,"Cemetery":20.0,"Dock":5.0,"Embassy":5.0,"Factory":25.0,"Farm":25.0,"Fortress":5.0,"Garden":5.0,"Grove":5.0,"Meeting Hall":5.0,"Hideout":25.0,"Inn":5.0,"Manor":5.0,"Market":5.0,"Mine":25.0,"Outpost":5.0,"Palace":5.0,"Research Facility":90.0,"Square":5.0,"Temple":5.0,"Trading Post":5.0,"Warehouse":25.0,"Workshop":35.0,"Catacombs":30.0,"Town Hall":5.0,"Excavation Site":35.0,"Festival Grounds":5.0,"Hospital":50.0,"Ruins":30.0,"Sanctuary":5.0,"Shipyard":25.0,"Caravan Camp":10.0,"Cavern":25.0,"Citadel":10.0,"Enclave":10.0,"Estate":10.0,"Village":10.0,"Hunting Grounds":0.0,"Military Base":0.0,"Mining Settlement":10.0,"Monastery":45.0,"Necropolis":25.0,"Rebel Camp":20.0,"Refugee Camp":10.0,"Slum":10.0,"University":75.0,"Academic District":90.0,"Government District":10.0,"Artisan Quarter":20.0,"Cultural District":15.0,"Entertainment District":10.0,"Farmland":10.0,"Foreign Quarter":10.0,"Wilderness":15.0,"Harbor":10.0,"Industrial District":40.0,"Market District":10.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":10.0,"Royal Quarter":10.0,"Residential District":10.0,"Ruined Quarter":25.0,"Temple District":10.0,"Underground City":20.0,"Underworld":20.0,"Warehouse District":10.0},"score":66.66666666666667},"Library":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":100.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":25.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":50.0,"Military":15.0,"Religious":45.0,"Criminal":15.0,"Revolutionary":15.0,"Conspiracy":55.0,"Merchant":15.0,"Clan":20.0,"Cultural":55.0,"Outsider":15.0,"Elite":45.0,"Poor":15.0,"Academic":90.0},"base":{"Empire":65.0,"Kingdom":65.0,"Republic":45.0,"Dominion":60.0,"Council":60.0,"Court":65.0,"Commonwealth":45.0,"Directorate":65.0,"Ministry":60.0,"Authority":60.0,"Legion":20.0,"Guard":35.0,"Army":20.0,"Militia":20.0,"Vanguard":20.0,"Watch":35.0,"Corps":20.0,"Wardens":35.0,"Rangers":20.0,"Order":85.0,"Covenant":85.0,"Faith":85.0,"Sect":85.0,"Brotherhood":85.0,"Fellowship":85.0,"Chapter":85.0,"Disciples":85.0,"Cult":25.0,"Guild":20.0,"Company":20.0,"Consortium":20.0,"Syndicate":20.0,"Cartel":20.0,"Cooperative":20.0,"Corporation":20.0,"Chamber":40.0,"Trust":40.0,"Clan":35.0,"House":60.0,"Dynasty":60.0,"Family":20.0,"Tribe":35.0,"People":35.0,"Community":35.0,"Nation":30.0,"Kindred":35.0,"Lineage":55.0,"Resistance":20.0,"Movement":45.0,"Front":20.0,"Union":20.0,"Collective":20.0,"Alliance":20.0,"League":20.0,"Pact":35.0,"Liberation":20.0,"Academy":100.0,"Institute":100.0,"School":100.0,"Society":100.0,"Circle":100.0,"Cabal":45.0,"Conclave":40.0,"Scholars":100.0,"Collegium":100.0,"Keepers":95.0,"Caravan":20.0,"Fleet":20.0,"Band":20.0,"Enclave":35.0,"Commune":15.0,"Network":45.0,"Expedition":45.0,"Nomads":20.0,"Diaspora":35.0,"Confederation":40.0,"Assembly":40.0,"Coalition":35.0,"Horde":15.0,"Mercenaries":20.0,"Parliament":60.0,"Senate":60.0,"Association":65.0,"Compact":35.0},"place":{"Apothecary":50.0,"Archive":65.0,"Armoury":0.0,"Auction House":20.0,"Bank":0.0,"Bathhouse":20.0,"Bookshop":60.0,"Brothel":20.0,"Clinic":50.0,"Club":25.0,"Courthouse":20.0,"Bridge":0.0,"Crypt":25.0,"Dormitory":20.0,"Forge":0.0,"Gallery":30.0,"Gambling Den":20.0,"House":0.0,"Laboratory":65.0,"Library":100.0,"Mill":0.0,"Monument":20.0,"Mortuary":30.0,"Museum":70.0,"Observatory":65.0,"Office":25.0,"Pawnshop":0.0,"Post Office":15.0,"Printing House":55.0,"Prison":0.0,"School":70.0,"Sewer":0.0,"Shelter":0.0,"Ship":0.0,"Shop":0.0,"Shrine":15.0,"Stable":0.0,"Tailor":15.0,"Tavern":20.0,"Theater":20.0,"Tower":30.0,"Training Ground":25.0,"Treasury":0.0,"Tunnel":0.0,"Vault":0.0,"Guardhouse":15.0,"Well":0.0,"Arena":25.0,"Barracks":5.0,"Black Market":5.0,"Brewery":20.0,"Camp":0.0,"Cemetery":30.0,"Dock":0.0,"Embassy":5.0,"Factory":5.0,"Farm":5.0,"Fortress":5.0,"Garden":30.0,"Grove":5.0,"Meeting Hall":30.0,"Hideout":5.0,"Inn":5.0,"Manor":5.0,"Market":25.0,"Mine":5.0,"Outpost":5.0,"Palace":5.0,"Research Facility":80.0,"Square":25.0,"Temple":45.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":25.0,"Catacombs":10.0,"Town Hall":25.0,"Excavation Site":35.0,"Festival Grounds":25.0,"Hospital":50.0,"Ruins":30.0,"Sanctuary":5.0,"Shipyard":5.0,"Caravan Camp":10.0,"Cavern":15.0,"Citadel":10.0,"Enclave":10.0,"Estate":10.0,"Village":10.0,"Hunting Grounds":0.0,"Military Base":0.0,"Mining Settlement":10.0,"Monastery":45.0,"Necropolis":25.0,"Rebel Camp":10.0,"Refugee Camp":10.0,"Slum":10.0,"University":90.0,"Academic District":100.0,"Government District":15.0,"Artisan Quarter":10.0,"Cultural District":45.0,"Entertainment District":20.0,"Farmland":10.0,"Foreign Quarter":10.0,"Wilderness":10.0,"Harbor":10.0,"Industrial District":10.0,"Market District":10.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":10.0,"Royal Quarter":35.0,"Residential District":10.0,"Ruined Quarter":25.0,"Temple District":15.0,"Underground City":10.0,"Underworld":10.0,"Warehouse District":10.0},"score":78.33333333333333},"Mill":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":25.0,"Production":100.0,"Logistics":0.0,"Resources":75.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":15.0,"Military":15.0,"Religious":15.0,"Criminal":20.0,"Revolutionary":15.0,"Conspiracy":15.0,"Merchant":75.0,"Clan":75.0,"Cultural":15.0,"Outsider":45.0,"Elite":20.0,"Poor":75.0,"Academic":15.0},"base":{"Empire":20.0,"Kingdom":45.0,"Republic":35.0,"Dominion":20.0,"Council":20.0,"Court":20.0,"Commonwealth":35.0,"Directorate":20.0,"Ministry":15.0,"Authority":20.0,"Legion":20.0,"Guard":20.0,"Army":20.0,"Militia":20.0,"Vanguard":20.0,"Watch":20.0,"Corps":20.0,"Wardens":20.0,"Rangers":35.0,"Order":20.0,"Covenant":45.0,"Faith":20.0,"Sect":20.0,"Brotherhood":70.0,"Fellowship":60.0,"Chapter":40.0,"Disciples":20.0,"Cult":20.0,"Guild":70.0,"Company":70.0,"Consortium":70.0,"Syndicate":60.0,"Cartel":50.0,"Cooperative":95.0,"Corporation":70.0,"Chamber":70.0,"Trust":60.0,"Clan":60.0,"House":70.0,"Dynasty":70.0,"Family":70.0,"Tribe":60.0,"People":65.0,"Community":70.0,"Nation":55.0,"Kindred":60.0,"Lineage":70.0,"Resistance":45.0,"Movement":45.0,"Front":20.0,"Union":70.0,"Collective":70.0,"Alliance":35.0,"League":40.0,"Pact":30.0,"Liberation":45.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":20.0,"Circle":20.0,"Cabal":35.0,"Conclave":35.0,"Scholars":20.0,"Collegium":20.0,"Keepers":15.0,"Caravan":60.0,"Fleet":50.0,"Band":40.0,"Enclave":50.0,"Commune":80.0,"Network":50.0,"Expedition":40.0,"Nomads":40.0,"Diaspora":40.0,"Confederation":50.0,"Assembly":50.0,"Coalition":40.0,"Horde":35.0,"Mercenaries":40.0,"Parliament":35.0,"Senate":35.0,"Association":70.0,"Compact":45.0},"place":{"Apothecary":25.0,"Archive":0.0,"Armoury":20.0,"Auction House":20.0,"Bank":20.0,"Bathhouse":20.0,"Bookshop":20.0,"Brothel":20.0,"Clinic":0.0,"Club":0.0,"Courthouse":0.0,"Bridge":40.0,"Crypt":0.0,"Dormitory":0.0,"Forge":60.0,"Gallery":15.0,"Gambling Den":20.0,"House":0.0,"Laboratory":20.0,"Library":0.0,"Mill":100.0,"Monument":0.0,"Mortuary":0.0,"Museum":0.0,"Observatory":0.0,"Office":20.0,"Pawnshop":20.0,"Post Office":20.0,"Printing House":55.0,"Prison":0.0,"School":0.0,"Sewer":0.0,"Shelter":0.0,"Ship":40.0,"Shop":30.0,"Shrine":0.0,"Stable":45.0,"Tailor":60.0,"Tavern":20.0,"Theater":15.0,"Tower":0.0,"Training Ground":0.0,"Treasury":20.0,"Tunnel":0.0,"Vault":0.0,"Guardhouse":0.0,"Well":60.0,"Arena":20.0,"Barracks":5.0,"Black Market":25.0,"Brewery":70.0,"Camp":5.0,"Cemetery":5.0,"Dock":40.0,"Embassy":5.0,"Factory":60.0,"Farm":80.0,"Fortress":5.0,"Garden":30.0,"Grove":45.0,"Meeting Hall":5.0,"Hideout":5.0,"Inn":25.0,"Manor":5.0,"Market":25.0,"Mine":45.0,"Outpost":5.0,"Palace":0.0,"Research Facility":5.0,"Square":25.0,"Temple":5.0,"Trading Post":25.0,"Warehouse":45.0,"Workshop":60.0,"Catacombs":5.0,"Town Hall":5.0,"Excavation Site":30.0,"Festival Grounds":25.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":5.0,"Shipyard":60.0,"Caravan Camp":20.0,"Cavern":15.0,"Citadel":10.0,"Enclave":10.0,"Estate":65.0,"Village":75.0,"Hunting Grounds":50.0,"Military Base":0.0,"Mining Settlement":25.0,"Monastery":10.0,"Necropolis":10.0,"Rebel Camp":10.0,"Refugee Camp":10.0,"Slum":10.0,"University":0.0,"Academic District":10.0,"Government District":10.0,"Artisan Quarter":85.0,"Cultural District":10.0,"Entertainment District":20.0,"Farmland":85.0,"Foreign Quarter":15.0,"Wilderness":65.0,"Harbor":50.0,"Industrial District":80.0,"Market District":20.0,"Military District":10.0,"Mining Region":30.0,"Noble Quarter":10.0,"Royal Quarter":10.0,"Residential District":10.0,"Ruined Quarter":10.0,"Temple District":10.0,"Underground City":20.0,"Underworld":15.0,"Warehouse District":40.0},"score":66.66666666666667},"Monument":{"function":{"Authority":25.0,"Defense":0.0,"Religion":25.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":100.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":70.0,"Military":20.0,"Religious":55.0,"Criminal":20.0,"Revolutionary":45.0,"Conspiracy":20.0,"Merchant":15.0,"Clan":75.0,"Cultural":90.0,"Outsider":15.0,"Elite":50.0,"Poor":15.0,"Academic":15.0},"base":{"Empire":95.0,"Kingdom":95.0,"Republic":70.0,"Dominion":95.0,"Council":70.0,"Court":70.0,"Commonwealth":70.0,"Directorate":70.0,"Ministry":65.0,"Authority":65.0,"Legion":45.0,"Guard":60.0,"Army":45.0,"Militia":40.0,"Vanguard":40.0,"Watch":60.0,"Corps":45.0,"Wardens":55.0,"Rangers":35.0,"Order":65.0,"Covenant":65.0,"Faith":65.0,"Sect":65.0,"Brotherhood":60.0,"Fellowship":60.0,"Chapter":45.0,"Disciples":65.0,"Cult":65.0,"Guild":20.0,"Company":20.0,"Consortium":20.0,"Syndicate":20.0,"Cartel":20.0,"Cooperative":20.0,"Corporation":20.0,"Chamber":50.0,"Trust":60.0,"Clan":30.0,"House":45.0,"Dynasty":40.0,"Family":30.0,"Tribe":15.0,"People":65.0,"Community":70.0,"Nation":65.0,"Kindred":70.0,"Lineage":95.0,"Resistance":40.0,"Movement":60.0,"Front":40.0,"Union":40.0,"Collective":40.0,"Alliance":40.0,"League":40.0,"Pact":40.0,"Liberation":40.0,"Academy":35.0,"Institute":35.0,"School":35.0,"Society":65.0,"Circle":45.0,"Cabal":20.0,"Conclave":35.0,"Scholars":35.0,"Collegium":35.0,"Keepers":65.0,"Caravan":20.0,"Fleet":20.0,"Band":20.0,"Enclave":60.0,"Commune":50.0,"Network":20.0,"Expedition":20.0,"Nomads":20.0,"Diaspora":45.0,"Confederation":65.0,"Assembly":65.0,"Coalition":65.0,"Horde":35.0,"Mercenaries":35.0,"Parliament":65.0,"Senate":70.0,"Association":45.0,"Compact":40.0},"place":{"Apothecary":0.0,"Archive":40.0,"Armoury":0.0,"Auction House":25.0,"Bank":15.0,"Bathhouse":55.0,"Bookshop":20.0,"Brothel":50.0,"Clinic":0.0,"Club":65.0,"Courthouse":45.0,"Bridge":0.0,"Crypt":20.0,"Dormitory":0.0,"Forge":0.0,"Gallery":65.0,"Gambling Den":55.0,"House":25.0,"Laboratory":0.0,"Library":20.0,"Mill":0.0,"Monument":100.0,"Mortuary":20.0,"Museum":55.0,"Observatory":15.0,"Office":40.0,"Pawnshop":0.0,"Post Office":40.0,"Printing House":20.0,"Prison":40.0,"School":25.0,"Sewer":0.0,"Shelter":0.0,"Ship":0.0,"Shop":0.0,"Shrine":30.0,"Stable":0.0,"Tailor":15.0,"Tavern":55.0,"Theater":65.0,"Tower":0.0,"Training Ground":15.0,"Treasury":45.0,"Tunnel":0.0,"Vault":15.0,"Guardhouse":40.0,"Well":5.0,"Arena":55.0,"Barracks":20.0,"Black Market":5.0,"Brewery":20.0,"Camp":5.0,"Cemetery":50.0,"Dock":0.0,"Embassy":50.0,"Factory":5.0,"Farm":5.0,"Fortress":25.0,"Garden":80.0,"Grove":50.0,"Meeting Hall":80.0,"Hideout":5.0,"Inn":25.0,"Manor":30.0,"Market":35.0,"Mine":5.0,"Outpost":5.0,"Palace":55.0,"Research Facility":5.0,"Square":80.0,"Temple":45.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":5.0,"Catacombs":25.0,"Town Hall":75.0,"Excavation Site":5.0,"Festival Grounds":75.0,"Hospital":20.0,"Ruins":40.0,"Sanctuary":25.0,"Shipyard":5.0,"Caravan Camp":10.0,"Cavern":10.0,"Citadel":50.0,"Enclave":10.0,"Estate":20.0,"Village":20.0,"Hunting Grounds":0.0,"Military Base":0.0,"Mining Settlement":10.0,"Monastery":20.0,"Necropolis":35.0,"Rebel Camp":10.0,"Refugee Camp":15.0,"Slum":20.0,"University":20.0,"Academic District":20.0,"Government District":75.0,"Artisan Quarter":15.0,"Cultural District":90.0,"Entertainment District":75.0,"Farmland":10.0,"Foreign Quarter":20.0,"Wilderness":10.0,"Harbor":10.0,"Industrial District":10.0,"Market District":20.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":25.0,"Royal Quarter":70.0,"Residential District":45.0,"Ruined Quarter":30.0,"Temple District":55.0,"Underground City":10.0,"Underworld":15.0,"Warehouse District":10.0},"score":71.66666666666667},"Mortuary":{"function":{"Authority":0.0,"Defense":0.0,"Religion":50.0,"Knowledge":25.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":25.0,"Exploration":0.0},"type":{"Government":45.0,"Military":15.0,"Religious":80.0,"Criminal":45.0,"Revolutionary":15.0,"Conspiracy":20.0,"Merchant":15.0,"Clan":15.0,"Cultural":20.0,"Outsider":15.0,"Elite":15.0,"Poor":45.0,"Academic":70.0},"base":{"Empire":40.0,"Kingdom":40.0,"Republic":60.0,"Dominion":40.0,"Council":60.0,"Court":60.0,"Commonwealth":60.0,"Directorate":60.0,"Ministry":60.0,"Authority":60.0,"Legion":35.0,"Guard":40.0,"Army":35.0,"Militia":35.0,"Vanguard":35.0,"Watch":40.0,"Corps":35.0,"Wardens":40.0,"Rangers":30.0,"Order":70.0,"Covenant":70.0,"Faith":70.0,"Sect":70.0,"Brotherhood":60.0,"Fellowship":60.0,"Chapter":60.0,"Disciples":70.0,"Cult":70.0,"Guild":20.0,"Company":35.0,"Consortium":20.0,"Syndicate":40.0,"Cartel":40.0,"Cooperative":40.0,"Corporation":20.0,"Chamber":35.0,"Trust":45.0,"Clan":45.0,"House":35.0,"Dynasty":35.0,"Family":55.0,"Tribe":45.0,"People":50.0,"Community":60.0,"Nation":35.0,"Kindred":45.0,"Lineage":35.0,"Resistance":35.0,"Movement":55.0,"Front":35.0,"Union":60.0,"Collective":60.0,"Alliance":35.0,"League":30.0,"Pact":35.0,"Liberation":35.0,"Academy":70.0,"Institute":70.0,"School":70.0,"Society":65.0,"Circle":70.0,"Cabal":50.0,"Conclave":60.0,"Scholars":70.0,"Collegium":70.0,"Keepers":70.0,"Caravan":20.0,"Fleet":20.0,"Band":40.0,"Enclave":20.0,"Commune":50.0,"Network":40.0,"Expedition":40.0,"Nomads":20.0,"Diaspora":20.0,"Confederation":60.0,"Assembly":60.0,"Coalition":50.0,"Horde":30.0,"Mercenaries":30.0,"Parliament":60.0,"Senate":60.0,"Association":55.0,"Compact":30.0},"place":{"Apothecary":25.0,"Archive":30.0,"Armoury":0.0,"Auction House":0.0,"Bank":15.0,"Bathhouse":0.0,"Bookshop":20.0,"Brothel":0.0,"Clinic":30.0,"Club":0.0,"Courthouse":20.0,"Bridge":0.0,"Crypt":55.0,"Dormitory":20.0,"Forge":0.0,"Gallery":0.0,"Gambling Den":0.0,"House":0.0,"Laboratory":25.0,"Library":30.0,"Mill":0.0,"Monument":20.0,"Mortuary":100.0,"Museum":25.0,"Observatory":30.0,"Office":20.0,"Pawnshop":0.0,"Post Office":15.0,"Printing House":25.0,"Prison":15.0,"School":25.0,"Sewer":0.0,"Shelter":5.0,"Ship":0.0,"Shop":0.0,"Shrine":60.0,"Stable":0.0,"Tailor":0.0,"Tavern":0.0,"Theater":0.0,"Tower":20.0,"Training Ground":20.0,"Treasury":15.0,"Tunnel":0.0,"Vault":15.0,"Guardhouse":15.0,"Well":0.0,"Arena":5.0,"Barracks":20.0,"Black Market":5.0,"Brewery":5.0,"Camp":5.0,"Cemetery":90.0,"Dock":5.0,"Embassy":20.0,"Factory":5.0,"Farm":5.0,"Fortress":20.0,"Garden":5.0,"Grove":30.0,"Meeting Hall":20.0,"Hideout":5.0,"Inn":5.0,"Manor":20.0,"Market":5.0,"Mine":5.0,"Outpost":5.0,"Palace":20.0,"Research Facility":35.0,"Square":20.0,"Temple":60.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":25.0,"Catacombs":50.0,"Town Hall":20.0,"Excavation Site":30.0,"Festival Grounds":10.0,"Hospital":55.0,"Ruins":25.0,"Sanctuary":60.0,"Shipyard":5.0,"Caravan Camp":10.0,"Cavern":15.0,"Citadel":15.0,"Enclave":10.0,"Estate":10.0,"Village":10.0,"Hunting Grounds":0.0,"Military Base":0.0,"Mining Settlement":10.0,"Monastery":70.0,"Necropolis":70.0,"Rebel Camp":10.0,"Refugee Camp":15.0,"Slum":10.0,"University":25.0,"Academic District":45.0,"Government District":35.0,"Artisan Quarter":10.0,"Cultural District":15.0,"Entertainment District":10.0,"Farmland":10.0,"Foreign Quarter":10.0,"Wilderness":10.0,"Harbor":10.0,"Industrial District":10.0,"Market District":10.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":10.0,"Royal Quarter":15.0,"Residential District":35.0,"Ruined Quarter":20.0,"Temple District":80.0,"Underground City":10.0,"Underworld":10.0,"Warehouse District":10.0},"score":53.333333333333336},"Museum":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":100.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":75.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":50.0,"Military":15.0,"Religious":50.0,"Criminal":20.0,"Revolutionary":15.0,"Conspiracy":25.0,"Merchant":15.0,"Clan":25.0,"Cultural":85.0,"Outsider":15.0,"Elite":65.0,"Poor":15.0,"Academic":90.0},"base":{"Empire":65.0,"Kingdom":65.0,"Republic":45.0,"Dominion":60.0,"Council":60.0,"Court":65.0,"Commonwealth":45.0,"Directorate":65.0,"Ministry":60.0,"Authority":60.0,"Legion":20.0,"Guard":35.0,"Army":20.0,"Militia":20.0,"Vanguard":20.0,"Watch":35.0,"Corps":20.0,"Wardens":35.0,"Rangers":20.0,"Order":55.0,"Covenant":55.0,"Faith":55.0,"Sect":55.0,"Brotherhood":50.0,"Fellowship":50.0,"Chapter":50.0,"Disciples":50.0,"Cult":50.0,"Guild":20.0,"Company":20.0,"Consortium":20.0,"Syndicate":20.0,"Cartel":20.0,"Cooperative":20.0,"Corporation":20.0,"Chamber":50.0,"Trust":65.0,"Clan":45.0,"House":65.0,"Dynasty":65.0,"Family":35.0,"Tribe":45.0,"People":45.0,"Community":45.0,"Nation":40.0,"Kindred":45.0,"Lineage":55.0,"Resistance":20.0,"Movement":45.0,"Front":20.0,"Union":20.0,"Collective":20.0,"Alliance":20.0,"League":20.0,"Pact":30.0,"Liberation":20.0,"Academy":75.0,"Institute":75.0,"School":75.0,"Society":90.0,"Circle":75.0,"Cabal":45.0,"Conclave":65.0,"Scholars":90.0,"Collegium":75.0,"Keepers":60.0,"Caravan":20.0,"Fleet":20.0,"Band":20.0,"Enclave":45.0,"Commune":15.0,"Network":30.0,"Expedition":45.0,"Nomads":20.0,"Diaspora":45.0,"Confederation":50.0,"Assembly":50.0,"Coalition":35.0,"Horde":15.0,"Mercenaries":20.0,"Parliament":60.0,"Senate":65.0,"Association":70.0,"Compact":35.0},"place":{"Apothecary":45.0,"Archive":55.0,"Armoury":0.0,"Auction House":25.0,"Bank":15.0,"Bathhouse":50.0,"Bookshop":55.0,"Brothel":45.0,"Clinic":45.0,"Club":55.0,"Courthouse":20.0,"Bridge":0.0,"Crypt":20.0,"Dormitory":20.0,"Forge":0.0,"Gallery":60.0,"Gambling Den":50.0,"House":20.0,"Laboratory":55.0,"Library":70.0,"Mill":0.0,"Monument":55.0,"Mortuary":25.0,"Museum":100.0,"Observatory":55.0,"Office":20.0,"Pawnshop":0.0,"Post Office":15.0,"Printing House":55.0,"Prison":0.0,"School":65.0,"Sewer":0.0,"Shelter":0.0,"Ship":0.0,"Shop":0.0,"Shrine":20.0,"Stable":0.0,"Tailor":15.0,"Tavern":50.0,"Theater":55.0,"Tower":20.0,"Training Ground":20.0,"Treasury":15.0,"Tunnel":0.0,"Vault":15.0,"Guardhouse":15.0,"Well":0.0,"Arena":50.0,"Barracks":5.0,"Black Market":0.0,"Brewery":20.0,"Camp":0.0,"Cemetery":30.0,"Dock":0.0,"Embassy":25.0,"Factory":0.0,"Farm":5.0,"Fortress":5.0,"Garden":55.0,"Grove":10.0,"Meeting Hall":50.0,"Hideout":0.0,"Inn":20.0,"Manor":25.0,"Market":30.0,"Mine":5.0,"Outpost":5.0,"Palace":25.0,"Research Facility":60.0,"Square":50.0,"Temple":30.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":25.0,"Catacombs":10.0,"Town Hall":30.0,"Excavation Site":30.0,"Festival Grounds":55.0,"Hospital":45.0,"Ruins":25.0,"Sanctuary":5.0,"Shipyard":5.0,"Caravan Camp":10.0,"Cavern":15.0,"Citadel":10.0,"Enclave":10.0,"Estate":15.0,"Village":15.0,"Hunting Grounds":0.0,"Military Base":0.0,"Mining Settlement":10.0,"Monastery":35.0,"Necropolis":25.0,"Rebel Camp":10.0,"Refugee Camp":10.0,"Slum":15.0,"University":65.0,"Academic District":90.0,"Government District":15.0,"Artisan Quarter":10.0,"Cultural District":100.0,"Entertainment District":65.0,"Farmland":10.0,"Foreign Quarter":15.0,"Wilderness":10.0,"Harbor":10.0,"Industrial District":10.0,"Market District":20.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":20.0,"Royal Quarter":35.0,"Residential District":15.0,"Ruined Quarter":20.0,"Temple District":20.0,"Underground City":10.0,"Underworld":10.0,"Warehouse District":10.0},"score":50.0},"Observatory":{"function":{"Authority":0.0,"Defense":25.0,"Religion":0.0,"Knowledge":100.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":25.0},"type":{"Government":50.0,"Military":50.0,"Religious":45.0,"Criminal":15.0,"Revolutionary":20.0,"Conspiracy":25.0,"Merchant":15.0,"Clan":15.0,"Cultural":25.0,"Outsider":20.0,"Elite":15.0,"Poor":15.0,"Academic":90.0},"base":{"Empire":65.0,"Kingdom":45.0,"Republic":45.0,"Dominion":60.0,"Council":60.0,"Court":45.0,"Commonwealth":45.0,"Directorate":65.0,"Ministry":40.0,"Authority":40.0,"Legion":45.0,"Guard":65.0,"Army":45.0,"Militia":60.0,"Vanguard":60.0,"Watch":60.0,"Corps":45.0,"Wardens":60.0,"Rangers":60.0,"Order":65.0,"Covenant":60.0,"Faith":60.0,"Sect":60.0,"Brotherhood":60.0,"Fellowship":60.0,"Chapter":60.0,"Disciples":60.0,"Cult":60.0,"Guild":20.0,"Company":20.0,"Consortium":20.0,"Syndicate":20.0,"Cartel":20.0,"Cooperative":20.0,"Corporation":20.0,"Chamber":35.0,"Trust":20.0,"Clan":45.0,"House":20.0,"Dynasty":20.0,"Family":20.0,"Tribe":45.0,"People":15.0,"Community":20.0,"Nation":15.0,"Kindred":45.0,"Lineage":20.0,"Resistance":45.0,"Movement":35.0,"Front":45.0,"Union":20.0,"Collective":20.0,"Alliance":45.0,"League":45.0,"Pact":45.0,"Liberation":45.0,"Academy":100.0,"Institute":100.0,"School":100.0,"Society":75.0,"Circle":75.0,"Cabal":45.0,"Conclave":55.0,"Scholars":100.0,"Collegium":100.0,"Keepers":70.0,"Caravan":20.0,"Fleet":40.0,"Band":35.0,"Enclave":20.0,"Commune":15.0,"Network":30.0,"Expedition":55.0,"Nomads":20.0,"Diaspora":20.0,"Confederation":60.0,"Assembly":60.0,"Coalition":50.0,"Horde":60.0,"Mercenaries":60.0,"Parliament":40.0,"Senate":40.0,"Association":55.0,"Compact":35.0},"place":{"Apothecary":45.0,"Archive":60.0,"Armoury":20.0,"Auction House":0.0,"Bank":0.0,"Bathhouse":0.0,"Bookshop":50.0,"Brothel":0.0,"Clinic":50.0,"Club":0.0,"Courthouse":20.0,"Bridge":15.0,"Crypt":25.0,"Dormitory":20.0,"Forge":15.0,"Gallery":5.0,"Gambling Den":0.0,"House":0.0,"Laboratory":65.0,"Library":65.0,"Mill":0.0,"Monument":15.0,"Mortuary":30.0,"Museum":55.0,"Observatory":100.0,"Office":25.0,"Pawnshop":0.0,"Post Office":15.0,"Printing House":50.0,"Prison":20.0,"School":65.0,"Sewer":5.0,"Shelter":0.0,"Ship":20.0,"Shop":0.0,"Shrine":15.0,"Stable":15.0,"Tailor":0.0,"Tavern":0.0,"Theater":0.0,"Tower":35.0,"Training Ground":30.0,"Treasury":0.0,"Tunnel":20.0,"Vault":0.0,"Guardhouse":20.0,"Well":0.0,"Arena":25.0,"Barracks":25.0,"Black Market":5.0,"Brewery":5.0,"Camp":20.0,"Cemetery":30.0,"Dock":5.0,"Embassy":5.0,"Factory":5.0,"Farm":5.0,"Fortress":25.0,"Garden":5.0,"Grove":25.0,"Meeting Hall":5.0,"Hideout":5.0,"Inn":5.0,"Manor":5.0,"Market":5.0,"Mine":5.0,"Outpost":30.0,"Palace":5.0,"Research Facility":80.0,"Square":5.0,"Temple":20.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":25.0,"Catacombs":30.0,"Town Hall":5.0,"Excavation Site":35.0,"Festival Grounds":5.0,"Hospital":50.0,"Ruins":35.0,"Sanctuary":25.0,"Shipyard":5.0,"Caravan Camp":10.0,"Cavern":20.0,"Citadel":20.0,"Enclave":10.0,"Estate":10.0,"Village":10.0,"Hunting Grounds":15.0,"Military Base":20.0,"Mining Settlement":10.0,"Monastery":45.0,"Necropolis":25.0,"Rebel Camp":20.0,"Refugee Camp":10.0,"Slum":10.0,"University":75.0,"Academic District":85.0,"Government District":15.0,"Artisan Quarter":10.0,"Cultural District":15.0,"Entertainment District":10.0,"Farmland":10.0,"Foreign Quarter":10.0,"Wilderness":15.0,"Harbor":10.0,"Industrial District":10.0,"Market District":10.0,"Military District":20.0,"Mining Region":10.0,"Noble Quarter":10.0,"Royal Quarter":10.0,"Residential District":10.0,"Ruined Quarter":30.0,"Temple District":15.0,"Underground City":10.0,"Underworld":10.0,"Warehouse District":10.0},"score":46.666666666666664},"Office":{"function":{"Authority":75.0,"Defense":0.0,"Religion":0.0,"Knowledge":25.0,"Commerce":50.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":85.0,"Military":50.0,"Religious":15.0,"Criminal":20.0,"Revolutionary":15.0,"Conspiracy":50.0,"Merchant":75.0,"Clan":15.0,"Cultural":20.0,"Outsider":15.0,"Elite":60.0,"Poor":15.0,"Academic":70.0},"base":{"Empire":70.0,"Kingdom":70.0,"Republic":95.0,"Dominion":70.0,"Council":95.0,"Court":70.0,"Commonwealth":95.0,"Directorate":95.0,"Ministry":95.0,"Authority":95.0,"Legion":55.0,"Guard":60.0,"Army":55.0,"Militia":45.0,"Vanguard":40.0,"Watch":60.0,"Corps":55.0,"Wardens":60.0,"Rangers":40.0,"Order":55.0,"Covenant":45.0,"Faith":45.0,"Sect":35.0,"Brotherhood":35.0,"Fellowship":35.0,"Chapter":40.0,"Disciples":35.0,"Cult":45.0,"Guild":95.0,"Company":95.0,"Consortium":95.0,"Syndicate":60.0,"Cartel":50.0,"Cooperative":60.0,"Corporation":95.0,"Chamber":95.0,"Trust":95.0,"Clan":35.0,"House":65.0,"Dynasty":65.0,"Family":20.0,"Tribe":35.0,"People":15.0,"Community":20.0,"Nation":15.0,"Kindred":35.0,"Lineage":55.0,"Resistance":35.0,"Movement":20.0,"Front":35.0,"Union":35.0,"Collective":35.0,"Alliance":45.0,"League":50.0,"Pact":50.0,"Liberation":35.0,"Academy":70.0,"Institute":70.0,"School":70.0,"Society":65.0,"Circle":70.0,"Cabal":50.0,"Conclave":50.0,"Scholars":70.0,"Collegium":70.0,"Keepers":55.0,"Caravan":70.0,"Fleet":70.0,"Band":35.0,"Enclave":20.0,"Commune":15.0,"Network":40.0,"Expedition":50.0,"Nomads":45.0,"Diaspora":20.0,"Confederation":95.0,"Assembly":95.0,"Coalition":95.0,"Horde":40.0,"Mercenaries":50.0,"Parliament":95.0,"Senate":95.0,"Association":95.0,"Compact":95.0},"place":{"Apothecary":50.0,"Archive":55.0,"Armoury":0.0,"Auction House":50.0,"Bank":55.0,"Bathhouse":20.0,"Bookshop":50.0,"Brothel":45.0,"Clinic":20.0,"Club":35.0,"Courthouse":65.0,"Bridge":15.0,"Crypt":20.0,"Dormitory":15.0,"Forge":20.0,"Gallery":20.0,"Gambling Den":45.0,"House":0.0,"Laboratory":25.0,"Library":25.0,"Mill":20.0,"Monument":40.0,"Mortuary":20.0,"Museum":20.0,"Observatory":25.0,"Office":100.0,"Pawnshop":50.0,"Post Office":55.0,"Printing House":25.0,"Prison":50.0,"School":25.0,"Sewer":0.0,"Shelter":0.0,"Ship":20.0,"Shop":50.0,"Shrine":0.0,"Stable":0.0,"Tailor":45.0,"Tavern":45.0,"Theater":20.0,"Tower":20.0,"Training Ground":20.0,"Treasury":65.0,"Tunnel":0.0,"Vault":30.0,"Guardhouse":50.0,"Well":0.0,"Arena":45.0,"Barracks":40.0,"Black Market":45.0,"Brewery":30.0,"Camp":5.0,"Cemetery":5.0,"Dock":45.0,"Embassy":70.0,"Factory":40.0,"Farm":5.0,"Fortress":50.0,"Garden":5.0,"Grove":5.0,"Meeting Hall":45.0,"Hideout":5.0,"Inn":25.0,"Manor":45.0,"Market":45.0,"Mine":5.0,"Outpost":5.0,"Palace":70.0,"Research Facility":30.0,"Square":30.0,"Temple":25.0,"Trading Post":65.0,"Warehouse":45.0,"Workshop":45.0,"Catacombs":5.0,"Town Hall":55.0,"Excavation Site":25.0,"Festival Grounds":30.0,"Hospital":25.0,"Ruins":25.0,"Sanctuary":5.0,"Shipyard":25.0,"Caravan Camp":20.0,"Cavern":10.0,"Citadel":60.0,"Enclave":10.0,"Estate":15.0,"Village":10.0,"Hunting Grounds":0.0,"Military Base":0.0,"Mining Settlement":10.0,"Monastery":20.0,"Necropolis":20.0,"Rebel Camp":10.0,"Refugee Camp":10.0,"Slum":10.0,"University":20.0,"Academic District":40.0,"Government District":100.0,"Artisan Quarter":25.0,"Cultural District":10.0,"Entertainment District":25.0,"Farmland":10.0,"Foreign Quarter":45.0,"Wilderness":10.0,"Harbor":55.0,"Industrial District":10.0,"Market District":75.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":20.0,"Royal Quarter":80.0,"Residential District":10.0,"Ruined Quarter":20.0,"Temple District":10.0,"Underground City":20.0,"Underworld":20.0,"Warehouse District":35.0},"score":78.33333333333333},"Pawnshop":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":100.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":25.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":15.0,"Military":15.0,"Religious":15.0,"Criminal":75.0,"Revolutionary":20.0,"Conspiracy":20.0,"Merchant":90.0,"Clan":15.0,"Cultural":15.0,"Outsider":45.0,"Elite":25.0,"Poor":65.0,"Academic":15.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":45.0,"Dominion":20.0,"Council":20.0,"Court":30.0,"Commonwealth":45.0,"Directorate":20.0,"Ministry":15.0,"Authority":20.0,"Legion":20.0,"Guard":20.0,"Army":20.0,"Militia":35.0,"Vanguard":35.0,"Watch":20.0,"Corps":20.0,"Wardens":20.0,"Rangers":35.0,"Order":20.0,"Covenant":20.0,"Faith":20.0,"Sect":45.0,"Brotherhood":45.0,"Fellowship":45.0,"Chapter":40.0,"Disciples":45.0,"Cult":45.0,"Guild":75.0,"Company":75.0,"Consortium":75.0,"Syndicate":100.0,"Cartel":100.0,"Cooperative":65.0,"Corporation":75.0,"Chamber":75.0,"Trust":75.0,"Clan":20.0,"House":30.0,"Dynasty":30.0,"Family":45.0,"Tribe":35.0,"People":50.0,"Community":65.0,"Nation":35.0,"Kindred":20.0,"Lineage":20.0,"Resistance":55.0,"Movement":55.0,"Front":45.0,"Union":65.0,"Collective":65.0,"Alliance":45.0,"League":55.0,"Pact":45.0,"Liberation":55.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":20.0,"Circle":45.0,"Cabal":55.0,"Conclave":45.0,"Scholars":20.0,"Collegium":20.0,"Keepers":15.0,"Caravan":75.0,"Fleet":75.0,"Band":70.0,"Enclave":50.0,"Commune":50.0,"Network":95.0,"Expedition":40.0,"Nomads":60.0,"Diaspora":50.0,"Confederation":50.0,"Assembly":60.0,"Coalition":50.0,"Horde":50.0,"Mercenaries":60.0,"Parliament":35.0,"Senate":35.0,"Association":70.0,"Compact":60.0},"place":{"Apothecary":55.0,"Archive":15.0,"Armoury":0.0,"Auction House":60.0,"Bank":60.0,"Bathhouse":25.0,"Bookshop":55.0,"Brothel":60.0,"Clinic":15.0,"Club":40.0,"Courthouse":0.0,"Bridge":15.0,"Crypt":15.0,"Dormitory":0.0,"Forge":25.0,"Gallery":20.0,"Gambling Den":60.0,"House":30.0,"Laboratory":15.0,"Library":0.0,"Mill":20.0,"Monument":0.0,"Mortuary":0.0,"Museum":0.0,"Observatory":0.0,"Office":50.0,"Pawnshop":100.0,"Post Office":20.0,"Printing House":5.0,"Prison":15.0,"School":0.0,"Sewer":40.0,"Shelter":35.0,"Ship":20.0,"Shop":65.0,"Shrine":0.0,"Stable":0.0,"Tailor":55.0,"Tavern":50.0,"Theater":20.0,"Tower":0.0,"Training Ground":0.0,"Treasury":25.0,"Tunnel":20.0,"Vault":35.0,"Guardhouse":0.0,"Well":15.0,"Arena":25.0,"Barracks":5.0,"Black Market":85.0,"Brewery":35.0,"Camp":5.0,"Cemetery":5.0,"Dock":35.0,"Embassy":5.0,"Factory":25.0,"Farm":5.0,"Fortress":0.0,"Garden":5.0,"Grove":5.0,"Meeting Hall":5.0,"Hideout":45.0,"Inn":35.0,"Manor":5.0,"Market":75.0,"Mine":25.0,"Outpost":5.0,"Palace":0.0,"Research Facility":5.0,"Square":30.0,"Temple":5.0,"Trading Post":55.0,"Warehouse":30.0,"Workshop":30.0,"Catacombs":25.0,"Town Hall":5.0,"Excavation Site":5.0,"Festival Grounds":35.0,"Hospital":20.0,"Ruins":20.0,"Sanctuary":5.0,"Shipyard":25.0,"Caravan Camp":25.0,"Cavern":20.0,"Citadel":10.0,"Enclave":10.0,"Estate":10.0,"Village":15.0,"Hunting Grounds":0.0,"Military Base":0.0,"Mining Settlement":15.0,"Monastery":10.0,"Necropolis":10.0,"Rebel Camp":20.0,"Refugee Camp":45.0,"Slum":90.0,"University":0.0,"Academic District":10.0,"Government District":10.0,"Artisan Quarter":40.0,"Cultural District":10.0,"Entertainment District":40.0,"Farmland":10.0,"Foreign Quarter":50.0,"Wilderness":15.0,"Harbor":70.0,"Industrial District":10.0,"Market District":90.0,"Military District":10.0,"Mining Region":15.0,"Noble Quarter":10.0,"Royal Quarter":10.0,"Residential District":35.0,"Ruined Quarter":50.0,"Temple District":10.0,"Underground City":30.0,"Underworld":55.0,"Warehouse District":15.0},"score":56.666666666666664},"Post Office":{"function":{"Authority":25.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":25.0,"Production":0.0,"Logistics":100.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":25.0,"Care":0.0,"Exploration":0.0},"type":{"Government":70.0,"Military":55.0,"Religious":15.0,"Criminal":20.0,"Revolutionary":20.0,"Conspiracy":45.0,"Merchant":85.0,"Clan":15.0,"Cultural":15.0,"Outsider":70.0,"Elite":20.0,"Poor":20.0,"Academic":45.0},"base":{"Empire":60.0,"Kingdom":60.0,"Republic":95.0,"Dominion":60.0,"Council":90.0,"Court":70.0,"Commonwealth":95.0,"Directorate":95.0,"Ministry":90.0,"Authority":90.0,"Legion":55.0,"Guard":60.0,"Army":55.0,"Militia":45.0,"Vanguard":45.0,"Watch":60.0,"Corps":55.0,"Wardens":55.0,"Rangers":45.0,"Order":55.0,"Covenant":45.0,"Faith":45.0,"Sect":35.0,"Brotherhood":35.0,"Fellowship":35.0,"Chapter":30.0,"Disciples":35.0,"Cult":40.0,"Guild":70.0,"Company":70.0,"Consortium":70.0,"Syndicate":65.0,"Cartel":60.0,"Cooperative":60.0,"Corporation":70.0,"Chamber":70.0,"Trust":70.0,"Clan":35.0,"House":45.0,"Dynasty":45.0,"Family":20.0,"Tribe":45.0,"People":15.0,"Community":20.0,"Nation":15.0,"Kindred":35.0,"Lineage":45.0,"Resistance":35.0,"Movement":20.0,"Front":35.0,"Union":35.0,"Collective":35.0,"Alliance":45.0,"League":55.0,"Pact":50.0,"Liberation":35.0,"Academy":40.0,"Institute":40.0,"School":40.0,"Society":40.0,"Circle":40.0,"Cabal":40.0,"Conclave":40.0,"Scholars":40.0,"Collegium":40.0,"Keepers":40.0,"Caravan":70.0,"Fleet":70.0,"Band":60.0,"Enclave":70.0,"Commune":40.0,"Network":60.0,"Expedition":60.0,"Nomads":70.0,"Diaspora":70.0,"Confederation":90.0,"Assembly":90.0,"Coalition":90.0,"Horde":45.0,"Mercenaries":50.0,"Parliament":90.0,"Senate":90.0,"Association":70.0,"Compact":60.0},"place":{"Apothecary":20.0,"Archive":45.0,"Armoury":30.0,"Auction House":20.0,"Bank":30.0,"Bathhouse":20.0,"Bookshop":20.0,"Brothel":20.0,"Clinic":15.0,"Club":0.0,"Courthouse":45.0,"Bridge":65.0,"Crypt":15.0,"Dormitory":20.0,"Forge":20.0,"Gallery":20.0,"Gambling Den":20.0,"House":0.0,"Laboratory":20.0,"Library":15.0,"Mill":20.0,"Monument":40.0,"Mortuary":15.0,"Museum":15.0,"Observatory":15.0,"Office":55.0,"Pawnshop":20.0,"Post Office":100.0,"Printing House":20.0,"Prison":40.0,"School":15.0,"Sewer":30.0,"Shelter":0.0,"Ship":60.0,"Shop":20.0,"Shrine":0.0,"Stable":60.0,"Tailor":20.0,"Tavern":20.0,"Theater":20.0,"Tower":20.0,"Training Ground":15.0,"Treasury":55.0,"Tunnel":55.0,"Vault":35.0,"Guardhouse":45.0,"Well":0.0,"Arena":20.0,"Barracks":20.0,"Black Market":30.0,"Brewery":25.0,"Camp":10.0,"Cemetery":5.0,"Dock":65.0,"Embassy":45.0,"Factory":20.0,"Farm":5.0,"Fortress":25.0,"Garden":5.0,"Grove":5.0,"Meeting Hall":25.0,"Hideout":5.0,"Inn":25.0,"Manor":25.0,"Market":30.0,"Mine":5.0,"Outpost":50.0,"Palace":40.0,"Research Facility":20.0,"Square":25.0,"Temple":20.0,"Trading Post":55.0,"Warehouse":65.0,"Workshop":25.0,"Catacombs":5.0,"Town Hall":60.0,"Excavation Site":25.0,"Festival Grounds":25.0,"Hospital":20.0,"Ruins":20.0,"Sanctuary":5.0,"Shipyard":60.0,"Caravan Camp":65.0,"Cavern":10.0,"Citadel":50.0,"Enclave":15.0,"Estate":10.0,"Village":10.0,"Hunting Grounds":0.0,"Military Base":25.0,"Mining Settlement":15.0,"Monastery":15.0,"Necropolis":15.0,"Rebel Camp":10.0,"Refugee Camp":15.0,"Slum":10.0,"University":15.0,"Academic District":15.0,"Government District":70.0,"Artisan Quarter":20.0,"Cultural District":10.0,"Entertainment District":20.0,"Farmland":10.0,"Foreign Quarter":40.0,"Wilderness":10.0,"Harbor":85.0,"Industrial District":20.0,"Market District":45.0,"Military District":20.0,"Mining Region":10.0,"Noble Quarter":10.0,"Royal Quarter":60.0,"Residential District":10.0,"Ruined Quarter":20.0,"Temple District":10.0,"Underground City":20.0,"Underworld":15.0,"Warehouse District":90.0},"score":50.0},"Printing House":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":50.0,"Commerce":25.0,"Production":100.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":25.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":50.0,"Military":15.0,"Religious":20.0,"Criminal":20.0,"Revolutionary":65.0,"Conspiracy":50.0,"Merchant":60.0,"Clan":20.0,"Cultural":75.0,"Outsider":15.0,"Elite":20.0,"Poor":15.0,"Academic":75.0},"base":{"Empire":45.0,"Kingdom":45.0,"Republic":65.0,"Dominion":45.0,"Council":60.0,"Court":65.0,"Commonwealth":60.0,"Directorate":65.0,"Ministry":60.0,"Authority":60.0,"Legion":35.0,"Guard":45.0,"Army":35.0,"Militia":55.0,"Vanguard":45.0,"Watch":40.0,"Corps":35.0,"Wardens":40.0,"Rangers":30.0,"Order":45.0,"Covenant":45.0,"Faith":45.0,"Sect":45.0,"Brotherhood":45.0,"Fellowship":55.0,"Chapter":55.0,"Disciples":45.0,"Cult":50.0,"Guild":65.0,"Company":65.0,"Consortium":65.0,"Syndicate":45.0,"Cartel":40.0,"Cooperative":65.0,"Corporation":65.0,"Chamber":65.0,"Trust":65.0,"Clan":45.0,"House":35.0,"Dynasty":35.0,"Family":20.0,"Tribe":45.0,"People":55.0,"Community":55.0,"Nation":50.0,"Kindred":45.0,"Lineage":45.0,"Resistance":90.0,"Movement":90.0,"Front":90.0,"Union":90.0,"Collective":90.0,"Alliance":90.0,"League":90.0,"Pact":90.0,"Liberation":90.0,"Academy":70.0,"Institute":70.0,"School":70.0,"Society":70.0,"Circle":70.0,"Cabal":50.0,"Conclave":50.0,"Scholars":70.0,"Collegium":70.0,"Keepers":60.0,"Caravan":45.0,"Fleet":40.0,"Band":20.0,"Enclave":45.0,"Commune":35.0,"Network":60.0,"Expedition":45.0,"Nomads":20.0,"Diaspora":45.0,"Confederation":60.0,"Assembly":60.0,"Coalition":60.0,"Horde":30.0,"Mercenaries":45.0,"Parliament":60.0,"Senate":60.0,"Association":70.0,"Compact":35.0},"place":{"Apothecary":50.0,"Archive":50.0,"Armoury":25.0,"Auction House":25.0,"Bank":20.0,"Bathhouse":25.0,"Bookshop":55.0,"Brothel":25.0,"Clinic":40.0,"Club":25.0,"Courthouse":20.0,"Bridge":0.0,"Crypt":20.0,"Dormitory":20.0,"Forge":55.0,"Gallery":25.0,"Gambling Den":25.0,"House":0.0,"Laboratory":55.0,"Library":55.0,"Mill":55.0,"Monument":20.0,"Mortuary":25.0,"Museum":55.0,"Observatory":50.0,"Office":25.0,"Pawnshop":5.0,"Post Office":20.0,"Printing House":100.0,"Prison":15.0,"School":55.0,"Sewer":0.0,"Shelter":0.0,"Ship":0.0,"Shop":25.0,"Shrine":15.0,"Stable":0.0,"Tailor":60.0,"Tavern":25.0,"Theater":25.0,"Tower":20.0,"Training Ground":20.0,"Treasury":20.0,"Tunnel":0.0,"Vault":20.0,"Guardhouse":15.0,"Well":0.0,"Arena":25.0,"Barracks":20.0,"Black Market":10.0,"Brewery":60.0,"Camp":5.0,"Cemetery":25.0,"Dock":5.0,"Embassy":20.0,"Factory":55.0,"Farm":30.0,"Fortress":20.0,"Garden":25.0,"Grove":5.0,"Meeting Hall":30.0,"Hideout":5.0,"Inn":10.0,"Manor":20.0,"Market":30.0,"Mine":30.0,"Outpost":5.0,"Palace":25.0,"Research Facility":50.0,"Square":30.0,"Temple":25.0,"Trading Post":10.0,"Warehouse":25.0,"Workshop":65.0,"Catacombs":5.0,"Town Hall":25.0,"Excavation Site":30.0,"Festival Grounds":30.0,"Hospital":40.0,"Ruins":25.0,"Sanctuary":5.0,"Shipyard":50.0,"Caravan Camp":10.0,"Cavern":15.0,"Citadel":15.0,"Enclave":10.0,"Estate":10.0,"Village":15.0,"Hunting Grounds":0.0,"Military Base":0.0,"Mining Settlement":15.0,"Monastery":30.0,"Necropolis":20.0,"Rebel Camp":10.0,"Refugee Camp":10.0,"Slum":10.0,"University":50.0,"Academic District":75.0,"Government District":35.0,"Artisan Quarter":85.0,"Cultural District":45.0,"Entertainment District":25.0,"Farmland":15.0,"Foreign Quarter":15.0,"Wilderness":10.0,"Harbor":15.0,"Industrial District":85.0,"Market District":15.0,"Military District":10.0,"Mining Region":15.0,"Noble Quarter":10.0,"Royal Quarter":15.0,"Residential District":10.0,"Ruined Quarter":20.0,"Temple District":15.0,"Underground City":10.0,"Underworld":15.0,"Warehouse District":15.0},"score":51.666666666666664},"Prison":{"function":{"Authority":100.0,"Defense":100.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":25.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":25.0},"type":{"Government":100.0,"Military":90.0,"Religious":15.0,"Criminal":45.0,"Revolutionary":60.0,"Conspiracy":20.0,"Merchant":15.0,"Clan":20.0,"Cultural":15.0,"Outsider":20.0,"Elite":35.0,"Poor":20.0,"Academic":20.0},"base":{"Empire":75.0,"Kingdom":75.0,"Republic":75.0,"Dominion":75.0,"Council":75.0,"Court":75.0,"Commonwealth":75.0,"Directorate":75.0,"Ministry":75.0,"Authority":75.0,"Legion":75.0,"Guard":90.0,"Army":75.0,"Militia":75.0,"Vanguard":75.0,"Watch":90.0,"Corps":75.0,"Wardens":90.0,"Rangers":70.0,"Order":65.0,"Covenant":45.0,"Faith":45.0,"Sect":35.0,"Brotherhood":20.0,"Fellowship":20.0,"Chapter":20.0,"Disciples":35.0,"Cult":35.0,"Guild":20.0,"Company":35.0,"Consortium":20.0,"Syndicate":40.0,"Cartel":40.0,"Cooperative":20.0,"Corporation":20.0,"Chamber":50.0,"Trust":50.0,"Clan":45.0,"House":50.0,"Dynasty":50.0,"Family":35.0,"Tribe":45.0,"People":20.0,"Community":20.0,"Nation":15.0,"Kindred":45.0,"Lineage":45.0,"Resistance":65.0,"Movement":55.0,"Front":65.0,"Union":45.0,"Collective":45.0,"Alliance":65.0,"League":65.0,"Pact":65.0,"Liberation":65.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":20.0,"Circle":35.0,"Cabal":35.0,"Conclave":35.0,"Scholars":20.0,"Collegium":20.0,"Keepers":45.0,"Caravan":20.0,"Fleet":65.0,"Band":50.0,"Enclave":20.0,"Commune":35.0,"Network":40.0,"Expedition":45.0,"Nomads":20.0,"Diaspora":20.0,"Confederation":75.0,"Assembly":75.0,"Coalition":75.0,"Horde":70.0,"Mercenaries":70.0,"Parliament":75.0,"Senate":75.0,"Association":20.0,"Compact":55.0},"place":{"Apothecary":0.0,"Archive":50.0,"Armoury":50.0,"Auction House":0.0,"Bank":20.0,"Bathhouse":0.0,"Bookshop":0.0,"Brothel":15.0,"Clinic":0.0,"Club":15.0,"Courthouse":90.0,"Bridge":20.0,"Crypt":20.0,"Dormitory":20.0,"Forge":20.0,"Gallery":0.0,"Gambling Den":15.0,"House":0.0,"Laboratory":15.0,"Library":0.0,"Mill":0.0,"Monument":40.0,"Mortuary":15.0,"Museum":0.0,"Observatory":20.0,"Office":50.0,"Pawnshop":15.0,"Post Office":40.0,"Printing House":15.0,"Prison":100.0,"School":0.0,"Sewer":20.0,"Shelter":0.0,"Ship":25.0,"Shop":0.0,"Shrine":0.0,"Stable":15.0,"Tailor":0.0,"Tavern":0.0,"Theater":0.0,"Tower":55.0,"Training Ground":55.0,"Treasury":55.0,"Tunnel":20.0,"Vault":20.0,"Guardhouse":75.0,"Well":0.0,"Arena":30.0,"Barracks":60.0,"Black Market":20.0,"Brewery":5.0,"Camp":30.0,"Cemetery":5.0,"Dock":5.0,"Embassy":50.0,"Factory":5.0,"Farm":5.0,"Fortress":70.0,"Garden":5.0,"Grove":5.0,"Meeting Hall":25.0,"Hideout":25.0,"Inn":5.0,"Manor":30.0,"Market":5.0,"Mine":5.0,"Outpost":55.0,"Palace":55.0,"Research Facility":5.0,"Square":25.0,"Temple":25.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":5.0,"Catacombs":25.0,"Town Hall":55.0,"Excavation Site":5.0,"Festival Grounds":5.0,"Hospital":5.0,"Ruins":25.0,"Sanctuary":5.0,"Shipyard":5.0,"Caravan Camp":10.0,"Cavern":20.0,"Citadel":90.0,"Enclave":10.0,"Estate":15.0,"Village":10.0,"Hunting Grounds":0.0,"Military Base":55.0,"Mining Settlement":10.0,"Monastery":10.0,"Necropolis":10.0,"Rebel Camp":60.0,"Refugee Camp":10.0,"Slum":15.0,"University":0.0,"Academic District":10.0,"Government District":90.0,"Artisan Quarter":10.0,"Cultural District":10.0,"Entertainment District":10.0,"Farmland":10.0,"Foreign Quarter":10.0,"Wilderness":20.0,"Harbor":15.0,"Industrial District":10.0,"Market District":10.0,"Military District":80.0,"Mining Region":10.0,"Noble Quarter":20.0,"Royal Quarter":75.0,"Residential District":10.0,"Ruined Quarter":20.0,"Temple District":10.0,"Underground City":20.0,"Underworld":20.0,"Warehouse District":10.0},"score":90.0},"School":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":100.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":25.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":50.0,"Military":15.0,"Religious":45.0,"Criminal":15.0,"Revolutionary":45.0,"Conspiracy":25.0,"Merchant":15.0,"Clan":50.0,"Cultural":55.0,"Outsider":15.0,"Elite":15.0,"Poor":15.0,"Academic":90.0},"base":{"Empire":45.0,"Kingdom":45.0,"Republic":65.0,"Dominion":45.0,"Council":60.0,"Court":45.0,"Commonwealth":60.0,"Directorate":65.0,"Ministry":40.0,"Authority":40.0,"Legion":35.0,"Guard":45.0,"Army":35.0,"Militia":40.0,"Vanguard":40.0,"Watch":40.0,"Corps":35.0,"Wardens":40.0,"Rangers":30.0,"Order":60.0,"Covenant":60.0,"Faith":60.0,"Sect":60.0,"Brotherhood":60.0,"Fellowship":60.0,"Chapter":60.0,"Disciples":60.0,"Cult":60.0,"Guild":20.0,"Company":20.0,"Consortium":20.0,"Syndicate":35.0,"Cartel":35.0,"Cooperative":35.0,"Corporation":20.0,"Chamber":35.0,"Trust":35.0,"Clan":65.0,"House":45.0,"Dynasty":45.0,"Family":65.0,"Tribe":60.0,"People":60.0,"Community":60.0,"Nation":55.0,"Kindred":65.0,"Lineage":45.0,"Resistance":40.0,"Movement":40.0,"Front":40.0,"Union":40.0,"Collective":40.0,"Alliance":40.0,"League":40.0,"Pact":40.0,"Liberation":40.0,"Academy":100.0,"Institute":100.0,"School":100.0,"Society":75.0,"Circle":75.0,"Cabal":55.0,"Conclave":55.0,"Scholars":100.0,"Collegium":100.0,"Keepers":70.0,"Caravan":20.0,"Fleet":20.0,"Band":35.0,"Enclave":45.0,"Commune":50.0,"Network":30.0,"Expedition":55.0,"Nomads":35.0,"Diaspora":35.0,"Confederation":40.0,"Assembly":40.0,"Coalition":40.0,"Horde":30.0,"Mercenaries":30.0,"Parliament":40.0,"Senate":40.0,"Association":65.0,"Compact":40.0},"place":{"Apothecary":45.0,"Archive":60.0,"Armoury":0.0,"Auction House":20.0,"Bank":0.0,"Bathhouse":20.0,"Bookshop":55.0,"Brothel":20.0,"Clinic":55.0,"Club":20.0,"Courthouse":20.0,"Bridge":0.0,"Crypt":25.0,"Dormitory":20.0,"Forge":0.0,"Gallery":30.0,"Gambling Den":20.0,"House":20.0,"Laboratory":65.0,"Library":70.0,"Mill":0.0,"Monument":25.0,"Mortuary":25.0,"Museum":65.0,"Observatory":65.0,"Office":25.0,"Pawnshop":0.0,"Post Office":15.0,"Printing House":55.0,"Prison":0.0,"School":100.0,"Sewer":0.0,"Shelter":15.0,"Ship":0.0,"Shop":0.0,"Shrine":20.0,"Stable":0.0,"Tailor":15.0,"Tavern":20.0,"Theater":20.0,"Tower":25.0,"Training Ground":25.0,"Treasury":0.0,"Tunnel":0.0,"Vault":0.0,"Guardhouse":15.0,"Well":15.0,"Arena":25.0,"Barracks":20.0,"Black Market":5.0,"Brewery":20.0,"Camp":20.0,"Cemetery":30.0,"Dock":0.0,"Embassy":5.0,"Factory":5.0,"Farm":20.0,"Fortress":20.0,"Garden":30.0,"Grove":10.0,"Meeting Hall":30.0,"Hideout":20.0,"Inn":25.0,"Manor":25.0,"Market":25.0,"Mine":5.0,"Outpost":5.0,"Palace":25.0,"Research Facility":80.0,"Square":25.0,"Temple":25.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":25.0,"Catacombs":10.0,"Town Hall":25.0,"Excavation Site":35.0,"Festival Grounds":25.0,"Hospital":50.0,"Ruins":30.0,"Sanctuary":20.0,"Shipyard":5.0,"Caravan Camp":10.0,"Cavern":15.0,"Citadel":10.0,"Enclave":15.0,"Estate":15.0,"Village":20.0,"Hunting Grounds":0.0,"Military Base":15.0,"Mining Settlement":15.0,"Monastery":45.0,"Necropolis":25.0,"Rebel Camp":15.0,"Refugee Camp":20.0,"Slum":10.0,"University":80.0,"Academic District":100.0,"Government District":10.0,"Artisan Quarter":10.0,"Cultural District":30.0,"Entertainment District":20.0,"Farmland":10.0,"Foreign Quarter":10.0,"Wilderness":10.0,"Harbor":10.0,"Industrial District":10.0,"Market District":10.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":10.0,"Royal Quarter":10.0,"Residential District":35.0,"Ruined Quarter":25.0,"Temple District":40.0,"Underground City":15.0,"Underworld":10.0,"Warehouse District":10.0},"score":76.66666666666667},"Sewer":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":50.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":25.0,"Migration":0.0,"Care":0.0,"Exploration":75.0},"type":{"Government":45.0,"Military":20.0,"Religious":15.0,"Criminal":70.0,"Revolutionary":50.0,"Conspiracy":70.0,"Merchant":25.0,"Clan":15.0,"Cultural":15.0,"Outsider":25.0,"Elite":15.0,"Poor":45.0,"Academic":25.0},"base":{"Empire":40.0,"Kingdom":40.0,"Republic":60.0,"Dominion":40.0,"Council":40.0,"Court":40.0,"Commonwealth":60.0,"Directorate":40.0,"Ministry":40.0,"Authority":40.0,"Legion":20.0,"Guard":35.0,"Army":20.0,"Militia":45.0,"Vanguard":40.0,"Watch":35.0,"Corps":20.0,"Wardens":35.0,"Rangers":30.0,"Order":45.0,"Covenant":20.0,"Faith":20.0,"Sect":45.0,"Brotherhood":35.0,"Fellowship":20.0,"Chapter":40.0,"Disciples":45.0,"Cult":70.0,"Guild":45.0,"Company":55.0,"Consortium":45.0,"Syndicate":70.0,"Cartel":70.0,"Cooperative":60.0,"Corporation":45.0,"Chamber":50.0,"Trust":45.0,"Clan":20.0,"House":20.0,"Dynasty":20.0,"Family":45.0,"Tribe":30.0,"People":50.0,"Community":60.0,"Nation":35.0,"Kindred":20.0,"Lineage":20.0,"Resistance":65.0,"Movement":65.0,"Front":65.0,"Union":65.0,"Collective":65.0,"Alliance":60.0,"League":60.0,"Pact":65.0,"Liberation":65.0,"Academy":30.0,"Institute":30.0,"School":30.0,"Society":45.0,"Circle":60.0,"Cabal":70.0,"Conclave":70.0,"Scholars":30.0,"Collegium":30.0,"Keepers":45.0,"Caravan":45.0,"Fleet":35.0,"Band":70.0,"Enclave":30.0,"Commune":40.0,"Network":90.0,"Expedition":35.0,"Nomads":45.0,"Diaspora":30.0,"Confederation":60.0,"Assembly":60.0,"Coalition":60.0,"Horde":45.0,"Mercenaries":45.0,"Parliament":40.0,"Senate":40.0,"Association":45.0,"Compact":50.0},"place":{"Apothecary":0.0,"Archive":20.0,"Armoury":25.0,"Auction House":0.0,"Bank":5.0,"Bathhouse":0.0,"Bookshop":0.0,"Brothel":20.0,"Clinic":0.0,"Club":20.0,"Courthouse":0.0,"Bridge":25.0,"Crypt":50.0,"Dormitory":0.0,"Forge":0.0,"Gallery":0.0,"Gambling Den":40.0,"House":30.0,"Laboratory":20.0,"Library":0.0,"Mill":0.0,"Monument":0.0,"Mortuary":0.0,"Museum":0.0,"Observatory":5.0,"Office":0.0,"Pawnshop":40.0,"Post Office":30.0,"Printing House":0.0,"Prison":20.0,"School":0.0,"Sewer":100.0,"Shelter":30.0,"Ship":30.0,"Shop":0.0,"Shrine":0.0,"Stable":25.0,"Tailor":0.0,"Tavern":0.0,"Theater":0.0,"Tower":25.0,"Training Ground":0.0,"Treasury":20.0,"Tunnel":70.0,"Vault":30.0,"Guardhouse":0.0,"Well":0.0,"Arena":5.0,"Barracks":5.0,"Black Market":60.0,"Brewery":5.0,"Camp":5.0,"Cemetery":10.0,"Dock":30.0,"Embassy":5.0,"Factory":5.0,"Farm":5.0,"Fortress":5.0,"Garden":5.0,"Grove":30.0,"Meeting Hall":5.0,"Hideout":75.0,"Inn":5.0,"Manor":5.0,"Market":5.0,"Mine":10.0,"Outpost":30.0,"Palace":5.0,"Research Facility":5.0,"Square":5.0,"Temple":5.0,"Trading Post":30.0,"Warehouse":35.0,"Workshop":5.0,"Catacombs":60.0,"Town Hall":5.0,"Excavation Site":45.0,"Festival Grounds":5.0,"Hospital":5.0,"Ruins":55.0,"Sanctuary":10.0,"Shipyard":30.0,"Caravan Camp":30.0,"Cavern":20.0,"Citadel":10.0,"Enclave":10.0,"Estate":10.0,"Village":10.0,"Hunting Grounds":40.0,"Military Base":20.0,"Mining Settlement":10.0,"Monastery":10.0,"Necropolis":45.0,"Rebel Camp":50.0,"Refugee Camp":35.0,"Slum":90.0,"University":0.0,"Academic District":10.0,"Government District":10.0,"Artisan Quarter":10.0,"Cultural District":10.0,"Entertainment District":10.0,"Farmland":10.0,"Foreign Quarter":10.0,"Wilderness":15.0,"Harbor":25.0,"Industrial District":40.0,"Market District":35.0,"Military District":15.0,"Mining Region":15.0,"Noble Quarter":10.0,"Royal Quarter":10.0,"Residential District":30.0,"Ruined Quarter":90.0,"Temple District":10.0,"Underground City":90.0,"Underworld":90.0,"Warehouse District":30.0},"score":78.33333333333333},"Shelter":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":100.0,"Social":0.0,"Covert":0.0,"Migration":50.0,"Care":75.0,"Exploration":0.0},"type":{"Government":45.0,"Military":15.0,"Religious":70.0,"Criminal":15.0,"Revolutionary":55.0,"Conspiracy":15.0,"Merchant":15.0,"Clan":65.0,"Cultural":15.0,"Outsider":80.0,"Elite":30.0,"Poor":85.0,"Academic":20.0},"base":{"Empire":40.0,"Kingdom":40.0,"Republic":50.0,"Dominion":40.0,"Council":40.0,"Court":40.0,"Commonwealth":50.0,"Directorate":40.0,"Ministry":50.0,"Authority":40.0,"Legion":35.0,"Guard":40.0,"Army":35.0,"Militia":45.0,"Vanguard":45.0,"Watch":40.0,"Corps":35.0,"Wardens":40.0,"Rangers":50.0,"Order":70.0,"Covenant":70.0,"Faith":70.0,"Sect":70.0,"Brotherhood":55.0,"Fellowship":55.0,"Chapter":55.0,"Disciples":65.0,"Cult":70.0,"Guild":20.0,"Company":45.0,"Consortium":20.0,"Syndicate":45.0,"Cartel":45.0,"Cooperative":60.0,"Corporation":20.0,"Chamber":35.0,"Trust":35.0,"Clan":55.0,"House":45.0,"Dynasty":45.0,"Family":55.0,"Tribe":35.0,"People":90.0,"Community":95.0,"Nation":85.0,"Kindred":45.0,"Lineage":45.0,"Resistance":55.0,"Movement":65.0,"Front":45.0,"Union":70.0,"Collective":70.0,"Alliance":45.0,"League":45.0,"Pact":45.0,"Liberation":55.0,"Academy":35.0,"Institute":35.0,"School":35.0,"Society":35.0,"Circle":35.0,"Cabal":45.0,"Conclave":45.0,"Scholars":35.0,"Collegium":35.0,"Keepers":65.0,"Caravan":95.0,"Fleet":60.0,"Band":95.0,"Enclave":95.0,"Commune":55.0,"Network":45.0,"Expedition":95.0,"Nomads":95.0,"Diaspora":95.0,"Confederation":50.0,"Assembly":60.0,"Coalition":50.0,"Horde":50.0,"Mercenaries":50.0,"Parliament":50.0,"Senate":50.0,"Association":45.0,"Compact":50.0},"place":{"Apothecary":40.0,"Archive":0.0,"Armoury":0.0,"Auction House":0.0,"Bank":0.0,"Bathhouse":40.0,"Bookshop":0.0,"Brothel":0.0,"Clinic":50.0,"Club":0.0,"Courthouse":0.0,"Bridge":20.0,"Crypt":0.0,"Dormitory":60.0,"Forge":0.0,"Gallery":0.0,"Gambling Den":0.0,"House":65.0,"Laboratory":0.0,"Library":0.0,"Mill":0.0,"Monument":0.0,"Mortuary":5.0,"Museum":0.0,"Observatory":0.0,"Office":0.0,"Pawnshop":35.0,"Post Office":0.0,"Printing House":0.0,"Prison":0.0,"School":15.0,"Sewer":30.0,"Shelter":100.0,"Ship":20.0,"Shop":0.0,"Shrine":0.0,"Stable":0.0,"Tailor":0.0,"Tavern":20.0,"Theater":0.0,"Tower":0.0,"Training Ground":0.0,"Treasury":0.0,"Tunnel":0.0,"Vault":0.0,"Guardhouse":0.0,"Well":20.0,"Arena":5.0,"Barracks":50.0,"Black Market":30.0,"Brewery":5.0,"Camp":75.0,"Cemetery":20.0,"Dock":25.0,"Embassy":25.0,"Factory":20.0,"Farm":30.0,"Fortress":25.0,"Garden":35.0,"Grove":5.0,"Meeting Hall":20.0,"Hideout":60.0,"Inn":55.0,"Manor":55.0,"Market":25.0,"Mine":20.0,"Outpost":5.0,"Palace":30.0,"Research Facility":5.0,"Square":40.0,"Temple":5.0,"Trading Post":25.0,"Warehouse":5.0,"Workshop":5.0,"Catacombs":5.0,"Town Hall":5.0,"Excavation Site":5.0,"Festival Grounds":5.0,"Hospital":70.0,"Ruins":5.0,"Sanctuary":75.0,"Shipyard":5.0,"Caravan Camp":30.0,"Cavern":10.0,"Citadel":15.0,"Enclave":60.0,"Estate":55.0,"Village":70.0,"Hunting Grounds":0.0,"Military Base":20.0,"Mining Settlement":70.0,"Monastery":30.0,"Necropolis":10.0,"Rebel Camp":30.0,"Refugee Camp":90.0,"Slum":100.0,"University":45.0,"Academic District":10.0,"Government District":10.0,"Artisan Quarter":10.0,"Cultural District":10.0,"Entertainment District":10.0,"Farmland":10.0,"Foreign Quarter":85.0,"Wilderness":15.0,"Harbor":15.0,"Industrial District":10.0,"Market District":10.0,"Military District":15.0,"Mining Region":10.0,"Noble Quarter":55.0,"Royal Quarter":45.0,"Residential District":70.0,"Ruined Quarter":45.0,"Temple District":35.0,"Underground City":70.0,"Underworld":10.0,"Warehouse District":10.0},"score":66.66666666666667},"Ship":{"function":{"Authority":0.0,"Defense":50.0,"Religion":0.0,"Knowledge":0.0,"Commerce":25.0,"Production":0.0,"Logistics":100.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":75.0,"Care":0.0,"Exploration":25.0},"type":{"Government":20.0,"Military":80.0,"Religious":15.0,"Criminal":50.0,"Revolutionary":25.0,"Conspiracy":15.0,"Merchant":85.0,"Clan":15.0,"Cultural":15.0,"Outsider":85.0,"Elite":20.0,"Poor":55.0,"Academic":20.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":20.0,"Dominion":20.0,"Council":20.0,"Court":20.0,"Commonwealth":20.0,"Directorate":20.0,"Ministry":20.0,"Authority":20.0,"Legion":60.0,"Guard":60.0,"Army":60.0,"Militia":60.0,"Vanguard":60.0,"Watch":60.0,"Corps":60.0,"Wardens":60.0,"Rangers":60.0,"Order":45.0,"Covenant":20.0,"Faith":20.0,"Sect":20.0,"Brotherhood":35.0,"Fellowship":20.0,"Chapter":20.0,"Disciples":20.0,"Cult":20.0,"Guild":65.0,"Company":60.0,"Consortium":70.0,"Syndicate":70.0,"Cartel":70.0,"Cooperative":60.0,"Corporation":70.0,"Chamber":70.0,"Trust":60.0,"Clan":20.0,"House":20.0,"Dynasty":20.0,"Family":35.0,"Tribe":45.0,"People":30.0,"Community":45.0,"Nation":15.0,"Kindred":20.0,"Lineage":20.0,"Resistance":35.0,"Movement":35.0,"Front":30.0,"Union":45.0,"Collective":45.0,"Alliance":30.0,"League":45.0,"Pact":30.0,"Liberation":35.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":20.0,"Circle":35.0,"Cabal":35.0,"Conclave":20.0,"Scholars":20.0,"Collegium":20.0,"Keepers":20.0,"Caravan":60.0,"Fleet":95.0,"Band":60.0,"Enclave":60.0,"Commune":50.0,"Network":70.0,"Expedition":60.0,"Nomads":60.0,"Diaspora":65.0,"Confederation":45.0,"Assembly":50.0,"Coalition":45.0,"Horde":60.0,"Mercenaries":60.0,"Parliament":45.0,"Senate":45.0,"Association":60.0,"Compact":60.0},"place":{"Apothecary":20.0,"Archive":0.0,"Armoury":35.0,"Auction House":20.0,"Bank":25.0,"Bathhouse":20.0,"Bookshop":20.0,"Brothel":20.0,"Clinic":0.0,"Club":0.0,"Courthouse":0.0,"Bridge":75.0,"Crypt":0.0,"Dormitory":20.0,"Forge":20.0,"Gallery":15.0,"Gambling Den":20.0,"House":0.0,"Laboratory":0.0,"Library":0.0,"Mill":40.0,"Monument":0.0,"Mortuary":0.0,"Museum":0.0,"Observatory":20.0,"Office":20.0,"Pawnshop":20.0,"Post Office":60.0,"Printing House":0.0,"Prison":25.0,"School":0.0,"Sewer":30.0,"Shelter":20.0,"Ship":100.0,"Shop":20.0,"Shrine":0.0,"Stable":60.0,"Tailor":20.0,"Tavern":20.0,"Theater":15.0,"Tower":25.0,"Training Ground":25.0,"Treasury":20.0,"Tunnel":55.0,"Vault":30.0,"Guardhouse":25.0,"Well":30.0,"Arena":30.0,"Barracks":30.0,"Black Market":30.0,"Brewery":25.0,"Camp":50.0,"Cemetery":5.0,"Dock":90.0,"Embassy":30.0,"Factory":25.0,"Farm":5.0,"Fortress":30.0,"Garden":0.0,"Grove":5.0,"Meeting Hall":5.0,"Hideout":5.0,"Inn":50.0,"Manor":5.0,"Market":30.0,"Mine":5.0,"Outpost":55.0,"Palace":5.0,"Research Facility":5.0,"Square":25.0,"Temple":0.0,"Trading Post":65.0,"Warehouse":65.0,"Workshop":25.0,"Catacombs":10.0,"Town Hall":5.0,"Excavation Site":10.0,"Festival Grounds":25.0,"Hospital":5.0,"Ruins":10.0,"Sanctuary":5.0,"Shipyard":75.0,"Caravan Camp":50.0,"Cavern":0.0,"Citadel":10.0,"Enclave":30.0,"Estate":0.0,"Village":0.0,"Hunting Grounds":0.0,"Military Base":30.0,"Mining Settlement":0.0,"Monastery":0.0,"Necropolis":0.0,"Rebel Camp":10.0,"Refugee Camp":35.0,"Slum":0.0,"University":0.0,"Academic District":0.0,"Government District":0.0,"Artisan Quarter":0.0,"Cultural District":0.0,"Entertainment District":0.0,"Farmland":0.0,"Foreign Quarter":45.0,"Wilderness":0.0,"Harbor":100.0,"Industrial District":10.0,"Market District":10.0,"Military District":15.0,"Mining Region":0.0,"Noble Quarter":0.0,"Royal Quarter":0.0,"Residential District":0.0,"Ruined Quarter":0.0,"Temple District":0.0,"Underground City":0.0,"Underworld":0.0,"Warehouse District":50.0},"score":85.0},"Shop":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":100.0,"Production":25.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":15.0,"Military":15.0,"Religious":15.0,"Criminal":25.0,"Revolutionary":15.0,"Conspiracy":15.0,"Merchant":95.0,"Clan":45.0,"Cultural":45.0,"Outsider":45.0,"Elite":55.0,"Poor":45.0,"Academic":15.0},"base":{"Empire":20.0,"Kingdom":35.0,"Republic":45.0,"Dominion":20.0,"Council":35.0,"Court":45.0,"Commonwealth":45.0,"Directorate":35.0,"Ministry":15.0,"Authority":20.0,"Legion":20.0,"Guard":20.0,"Army":20.0,"Militia":20.0,"Vanguard":20.0,"Watch":20.0,"Corps":20.0,"Wardens":20.0,"Rangers":35.0,"Order":20.0,"Covenant":35.0,"Faith":20.0,"Sect":20.0,"Brotherhood":60.0,"Fellowship":60.0,"Chapter":40.0,"Disciples":20.0,"Cult":20.0,"Guild":100.0,"Company":100.0,"Consortium":100.0,"Syndicate":75.0,"Cartel":65.0,"Cooperative":75.0,"Corporation":100.0,"Chamber":100.0,"Trust":75.0,"Clan":60.0,"House":45.0,"Dynasty":45.0,"Family":60.0,"Tribe":60.0,"People":60.0,"Community":60.0,"Nation":55.0,"Kindred":60.0,"Lineage":40.0,"Resistance":35.0,"Movement":35.0,"Front":20.0,"Union":60.0,"Collective":60.0,"Alliance":45.0,"League":50.0,"Pact":40.0,"Liberation":35.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":35.0,"Circle":20.0,"Cabal":35.0,"Conclave":35.0,"Scholars":20.0,"Collegium":20.0,"Keepers":15.0,"Caravan":75.0,"Fleet":75.0,"Band":40.0,"Enclave":60.0,"Commune":55.0,"Network":50.0,"Expedition":40.0,"Nomads":60.0,"Diaspora":60.0,"Confederation":60.0,"Assembly":60.0,"Coalition":50.0,"Horde":35.0,"Mercenaries":45.0,"Parliament":40.0,"Senate":45.0,"Association":75.0,"Compact":60.0},"place":{"Apothecary":60.0,"Archive":0.0,"Armoury":20.0,"Auction House":65.0,"Bank":65.0,"Bathhouse":30.0,"Bookshop":60.0,"Brothel":60.0,"Clinic":0.0,"Club":35.0,"Courthouse":0.0,"Bridge":20.0,"Crypt":0.0,"Dormitory":0.0,"Forge":35.0,"Gallery":25.0,"Gambling Den":55.0,"House":0.0,"Laboratory":15.0,"Library":0.0,"Mill":30.0,"Monument":0.0,"Mortuary":0.0,"Museum":0.0,"Observatory":0.0,"Office":50.0,"Pawnshop":65.0,"Post Office":20.0,"Printing House":25.0,"Prison":0.0,"School":0.0,"Sewer":0.0,"Shelter":0.0,"Ship":20.0,"Shop":100.0,"Shrine":0.0,"Stable":0.0,"Tailor":65.0,"Tavern":50.0,"Theater":25.0,"Tower":0.0,"Training Ground":0.0,"Treasury":25.0,"Tunnel":0.0,"Vault":30.0,"Guardhouse":0.0,"Well":0.0,"Arena":25.0,"Barracks":5.0,"Black Market":70.0,"Brewery":45.0,"Camp":5.0,"Cemetery":5.0,"Dock":35.0,"Embassy":5.0,"Factory":30.0,"Farm":25.0,"Fortress":5.0,"Garden":5.0,"Grove":5.0,"Meeting Hall":5.0,"Hideout":5.0,"Inn":35.0,"Manor":5.0,"Market":100.0,"Mine":25.0,"Outpost":5.0,"Palace":5.0,"Research Facility":5.0,"Square":30.0,"Temple":5.0,"Trading Post":75.0,"Warehouse":35.0,"Workshop":55.0,"Catacombs":5.0,"Town Hall":5.0,"Excavation Site":5.0,"Festival Grounds":35.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":5.0,"Shipyard":35.0,"Caravan Camp":25.0,"Cavern":10.0,"Citadel":10.0,"Enclave":10.0,"Estate":10.0,"Village":15.0,"Hunting Grounds":0.0,"Military Base":0.0,"Mining Settlement":15.0,"Monastery":10.0,"Necropolis":10.0,"Rebel Camp":10.0,"Refugee Camp":10.0,"Slum":10.0,"University":0.0,"Academic District":10.0,"Government District":10.0,"Artisan Quarter":55.0,"Cultural District":10.0,"Entertainment District":40.0,"Farmland":15.0,"Foreign Quarter":75.0,"Wilderness":10.0,"Harbor":80.0,"Industrial District":25.0,"Market District":100.0,"Military District":10.0,"Mining Region":15.0,"Noble Quarter":10.0,"Royal Quarter":10.0,"Residential District":35.0,"Ruined Quarter":10.0,"Temple District":10.0,"Underground City":25.0,"Underworld":25.0,"Warehouse District":20.0},"score":91.66666666666667},"Shrine":{"function":{"Authority":0.0,"Defense":0.0,"Religion":100.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":25.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":15.0,"Military":15.0,"Religious":90.0,"Criminal":15.0,"Revolutionary":15.0,"Conspiracy":15.0,"Merchant":15.0,"Clan":70.0,"Cultural":55.0,"Outsider":45.0,"Elite":15.0,"Poor":45.0,"Academic":15.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":45.0,"Dominion":20.0,"Council":20.0,"Court":20.0,"Commonwealth":45.0,"Directorate":20.0,"Ministry":40.0,"Authority":20.0,"Legion":20.0,"Guard":20.0,"Army":20.0,"Militia":20.0,"Vanguard":20.0,"Watch":20.0,"Corps":20.0,"Wardens":20.0,"Rangers":35.0,"Order":100.0,"Covenant":100.0,"Faith":100.0,"Sect":100.0,"Brotherhood":100.0,"Fellowship":95.0,"Chapter":95.0,"Disciples":100.0,"Cult":100.0,"Guild":20.0,"Company":35.0,"Consortium":20.0,"Syndicate":20.0,"Cartel":20.0,"Cooperative":60.0,"Corporation":20.0,"Chamber":20.0,"Trust":20.0,"Clan":70.0,"House":60.0,"Dynasty":60.0,"Family":70.0,"Tribe":65.0,"People":65.0,"Community":65.0,"Nation":60.0,"Kindred":70.0,"Lineage":60.0,"Resistance":35.0,"Movement":35.0,"Front":20.0,"Union":40.0,"Collective":40.0,"Alliance":45.0,"League":40.0,"Pact":40.0,"Liberation":35.0,"Academy":45.0,"Institute":45.0,"School":45.0,"Society":55.0,"Circle":45.0,"Cabal":45.0,"Conclave":55.0,"Scholars":45.0,"Collegium":45.0,"Keepers":70.0,"Caravan":40.0,"Fleet":40.0,"Band":40.0,"Enclave":70.0,"Commune":55.0,"Network":35.0,"Expedition":40.0,"Nomads":60.0,"Diaspora":60.0,"Confederation":45.0,"Assembly":40.0,"Coalition":30.0,"Horde":35.0,"Mercenaries":35.0,"Parliament":35.0,"Senate":35.0,"Association":35.0,"Compact":50.0},"place":{"Apothecary":0.0,"Archive":0.0,"Armoury":0.0,"Auction House":20.0,"Bank":0.0,"Bathhouse":20.0,"Bookshop":15.0,"Brothel":20.0,"Clinic":0.0,"Club":20.0,"Courthouse":0.0,"Bridge":0.0,"Crypt":55.0,"Dormitory":0.0,"Forge":0.0,"Gallery":20.0,"Gambling Den":20.0,"House":5.0,"Laboratory":0.0,"Library":15.0,"Mill":0.0,"Monument":30.0,"Mortuary":60.0,"Museum":20.0,"Observatory":15.0,"Office":0.0,"Pawnshop":0.0,"Post Office":0.0,"Printing House":15.0,"Prison":0.0,"School":20.0,"Sewer":0.0,"Shelter":0.0,"Ship":0.0,"Shop":0.0,"Shrine":100.0,"Stable":0.0,"Tailor":15.0,"Tavern":20.0,"Theater":20.0,"Tower":0.0,"Training Ground":15.0,"Treasury":0.0,"Tunnel":0.0,"Vault":0.0,"Guardhouse":0.0,"Well":5.0,"Arena":25.0,"Barracks":5.0,"Black Market":5.0,"Brewery":20.0,"Camp":40.0,"Cemetery":80.0,"Dock":5.0,"Embassy":5.0,"Factory":5.0,"Farm":5.0,"Fortress":5.0,"Garden":45.0,"Grove":55.0,"Meeting Hall":25.0,"Hideout":5.0,"Inn":10.0,"Manor":5.0,"Market":25.0,"Mine":5.0,"Outpost":5.0,"Palace":40.0,"Research Facility":5.0,"Square":45.0,"Temple":80.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":5.0,"Catacombs":65.0,"Town Hall":25.0,"Excavation Site":5.0,"Festival Grounds":30.0,"Hospital":25.0,"Ruins":40.0,"Sanctuary":75.0,"Shipyard":5.0,"Caravan Camp":10.0,"Cavern":35.0,"Citadel":10.0,"Enclave":35.0,"Estate":10.0,"Village":40.0,"Hunting Grounds":0.0,"Military Base":0.0,"Mining Settlement":10.0,"Monastery":90.0,"Necropolis":75.0,"Rebel Camp":10.0,"Refugee Camp":15.0,"Slum":15.0,"University":0.0,"Academic District":15.0,"Government District":10.0,"Artisan Quarter":10.0,"Cultural District":25.0,"Entertainment District":20.0,"Farmland":10.0,"Foreign Quarter":15.0,"Wilderness":35.0,"Harbor":10.0,"Industrial District":10.0,"Market District":10.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":10.0,"Royal Quarter":10.0,"Residential District":40.0,"Ruined Quarter":35.0,"Temple District":100.0,"Underground City":10.0,"Underworld":10.0,"Warehouse District":10.0},"score":81.66666666666667},"Stable":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":100.0,"Resources":25.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":15.0,"Military":75.0,"Religious":15.0,"Criminal":15.0,"Revolutionary":15.0,"Conspiracy":15.0,"Merchant":80.0,"Clan":70.0,"Cultural":15.0,"Outsider":65.0,"Elite":45.0,"Poor":50.0,"Academic":15.0},"base":{"Empire":45.0,"Kingdom":55.0,"Republic":45.0,"Dominion":45.0,"Council":55.0,"Court":40.0,"Commonwealth":45.0,"Directorate":55.0,"Ministry":40.0,"Authority":45.0,"Legion":70.0,"Guard":70.0,"Army":70.0,"Militia":70.0,"Vanguard":70.0,"Watch":70.0,"Corps":70.0,"Wardens":70.0,"Rangers":70.0,"Order":60.0,"Covenant":55.0,"Faith":45.0,"Sect":20.0,"Brotherhood":55.0,"Fellowship":40.0,"Chapter":20.0,"Disciples":20.0,"Cult":20.0,"Guild":70.0,"Company":70.0,"Consortium":70.0,"Syndicate":60.0,"Cartel":60.0,"Cooperative":70.0,"Corporation":70.0,"Chamber":70.0,"Trust":70.0,"Clan":60.0,"House":70.0,"Dynasty":70.0,"Family":70.0,"Tribe":55.0,"People":65.0,"Community":65.0,"Nation":50.0,"Kindred":60.0,"Lineage":70.0,"Resistance":35.0,"Movement":35.0,"Front":35.0,"Union":45.0,"Collective":45.0,"Alliance":45.0,"League":50.0,"Pact":40.0,"Liberation":35.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":20.0,"Circle":20.0,"Cabal":20.0,"Conclave":35.0,"Scholars":20.0,"Collegium":20.0,"Keepers":30.0,"Caravan":95.0,"Fleet":60.0,"Band":90.0,"Enclave":95.0,"Commune":65.0,"Network":55.0,"Expedition":90.0,"Nomads":90.0,"Diaspora":90.0,"Confederation":55.0,"Assembly":65.0,"Coalition":55.0,"Horde":70.0,"Mercenaries":70.0,"Parliament":65.0,"Senate":65.0,"Association":70.0,"Compact":60.0},"place":{"Apothecary":0.0,"Archive":0.0,"Armoury":30.0,"Auction House":0.0,"Bank":5.0,"Bathhouse":0.0,"Bookshop":0.0,"Brothel":0.0,"Clinic":0.0,"Club":0.0,"Courthouse":0.0,"Bridge":65.0,"Crypt":0.0,"Dormitory":20.0,"Forge":20.0,"Gallery":0.0,"Gambling Den":0.0,"House":0.0,"Laboratory":0.0,"Library":0.0,"Mill":45.0,"Monument":0.0,"Mortuary":0.0,"Museum":0.0,"Observatory":15.0,"Office":0.0,"Pawnshop":0.0,"Post Office":60.0,"Printing House":0.0,"Prison":15.0,"School":0.0,"Sewer":25.0,"Shelter":0.0,"Ship":60.0,"Shop":0.0,"Shrine":0.0,"Stable":100.0,"Tailor":0.0,"Tavern":0.0,"Theater":0.0,"Tower":15.0,"Training Ground":15.0,"Treasury":5.0,"Tunnel":60.0,"Vault":30.0,"Guardhouse":15.0,"Well":45.0,"Arena":40.0,"Barracks":40.0,"Black Market":25.0,"Brewery":40.0,"Camp":40.0,"Cemetery":5.0,"Dock":60.0,"Embassy":5.0,"Factory":5.0,"Farm":60.0,"Fortress":20.0,"Garden":30.0,"Grove":40.0,"Meeting Hall":5.0,"Hideout":5.0,"Inn":40.0,"Manor":40.0,"Market":5.0,"Mine":10.0,"Outpost":50.0,"Palace":5.0,"Research Facility":5.0,"Square":5.0,"Temple":5.0,"Trading Post":70.0,"Warehouse":65.0,"Workshop":5.0,"Catacombs":5.0,"Town Hall":5.0,"Excavation Site":10.0,"Festival Grounds":5.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":5.0,"Shipyard":55.0,"Caravan Camp":70.0,"Cavern":10.0,"Citadel":35.0,"Enclave":10.0,"Estate":60.0,"Village":70.0,"Hunting Grounds":45.0,"Military Base":45.0,"Mining Settlement":15.0,"Monastery":10.0,"Necropolis":10.0,"Rebel Camp":20.0,"Refugee Camp":10.0,"Slum":10.0,"University":0.0,"Academic District":10.0,"Government District":10.0,"Artisan Quarter":10.0,"Cultural District":10.0,"Entertainment District":10.0,"Farmland":55.0,"Foreign Quarter":15.0,"Wilderness":50.0,"Harbor":70.0,"Industrial District":20.0,"Market District":40.0,"Military District":25.0,"Mining Region":20.0,"Noble Quarter":10.0,"Royal Quarter":10.0,"Residential District":10.0,"Ruined Quarter":10.0,"Temple District":10.0,"Underground City":10.0,"Underworld":10.0,"Warehouse District":80.0},"score":75.0},"Tailor":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":100.0,"Production":100.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":15.0,"Military":45.0,"Religious":15.0,"Criminal":25.0,"Revolutionary":15.0,"Conspiracy":15.0,"Merchant":95.0,"Clan":45.0,"Cultural":65.0,"Outsider":15.0,"Elite":55.0,"Poor":65.0,"Academic":15.0},"base":{"Empire":35.0,"Kingdom":35.0,"Republic":45.0,"Dominion":35.0,"Council":35.0,"Court":65.0,"Commonwealth":45.0,"Directorate":35.0,"Ministry":30.0,"Authority":35.0,"Legion":40.0,"Guard":40.0,"Army":40.0,"Militia":40.0,"Vanguard":40.0,"Watch":40.0,"Corps":40.0,"Wardens":40.0,"Rangers":40.0,"Order":35.0,"Covenant":35.0,"Faith":20.0,"Sect":20.0,"Brotherhood":60.0,"Fellowship":60.0,"Chapter":40.0,"Disciples":20.0,"Cult":20.0,"Guild":75.0,"Company":75.0,"Consortium":75.0,"Syndicate":65.0,"Cartel":65.0,"Cooperative":75.0,"Corporation":75.0,"Chamber":75.0,"Trust":75.0,"Clan":60.0,"House":65.0,"Dynasty":65.0,"Family":60.0,"Tribe":60.0,"People":60.0,"Community":65.0,"Nation":55.0,"Kindred":60.0,"Lineage":60.0,"Resistance":45.0,"Movement":55.0,"Front":20.0,"Union":65.0,"Collective":65.0,"Alliance":45.0,"League":50.0,"Pact":40.0,"Liberation":45.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":55.0,"Circle":45.0,"Cabal":35.0,"Conclave":35.0,"Scholars":20.0,"Collegium":20.0,"Keepers":15.0,"Caravan":75.0,"Fleet":65.0,"Band":35.0,"Enclave":50.0,"Commune":50.0,"Network":45.0,"Expedition":35.0,"Nomads":45.0,"Diaspora":45.0,"Confederation":60.0,"Assembly":50.0,"Coalition":40.0,"Horde":40.0,"Mercenaries":55.0,"Parliament":35.0,"Senate":45.0,"Association":75.0,"Compact":60.0},"place":{"Apothecary":55.0,"Archive":0.0,"Armoury":20.0,"Auction House":55.0,"Bank":55.0,"Bathhouse":25.0,"Bookshop":55.0,"Brothel":55.0,"Clinic":0.0,"Club":40.0,"Courthouse":15.0,"Bridge":20.0,"Crypt":15.0,"Dormitory":0.0,"Forge":65.0,"Gallery":25.0,"Gambling Den":55.0,"House":15.0,"Laboratory":20.0,"Library":15.0,"Mill":60.0,"Monument":15.0,"Mortuary":0.0,"Museum":15.0,"Observatory":0.0,"Office":45.0,"Pawnshop":55.0,"Post Office":20.0,"Printing House":60.0,"Prison":0.0,"School":15.0,"Sewer":0.0,"Shelter":0.0,"Ship":20.0,"Shop":65.0,"Shrine":15.0,"Stable":0.0,"Tailor":100.0,"Tavern":50.0,"Theater":25.0,"Tower":0.0,"Training Ground":15.0,"Treasury":25.0,"Tunnel":0.0,"Vault":30.0,"Guardhouse":0.0,"Well":0.0,"Arena":25.0,"Barracks":5.0,"Black Market":50.0,"Brewery":70.0,"Camp":5.0,"Cemetery":20.0,"Dock":30.0,"Embassy":20.0,"Factory":60.0,"Farm":30.0,"Fortress":5.0,"Garden":20.0,"Grove":5.0,"Meeting Hall":20.0,"Hideout":5.0,"Inn":30.0,"Manor":20.0,"Market":70.0,"Mine":30.0,"Outpost":5.0,"Palace":20.0,"Research Facility":5.0,"Square":30.0,"Temple":20.0,"Trading Post":50.0,"Warehouse":30.0,"Workshop":65.0,"Catacombs":5.0,"Town Hall":20.0,"Excavation Site":10.0,"Festival Grounds":35.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":5.0,"Shipyard":55.0,"Caravan Camp":25.0,"Cavern":10.0,"Citadel":10.0,"Enclave":10.0,"Estate":20.0,"Village":15.0,"Hunting Grounds":0.0,"Military Base":0.0,"Mining Settlement":15.0,"Monastery":10.0,"Necropolis":10.0,"Rebel Camp":10.0,"Refugee Camp":10.0,"Slum":10.0,"University":0.0,"Academic District":10.0,"Government District":10.0,"Artisan Quarter":90.0,"Cultural District":20.0,"Entertainment District":35.0,"Farmland":15.0,"Foreign Quarter":50.0,"Wilderness":10.0,"Harbor":65.0,"Industrial District":70.0,"Market District":80.0,"Military District":10.0,"Mining Region":15.0,"Noble Quarter":15.0,"Royal Quarter":15.0,"Residential District":35.0,"Ruined Quarter":10.0,"Temple District":10.0,"Underground City":25.0,"Underworld":20.0,"Warehouse District":20.0},"score":56.666666666666664},"Tavern":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":75.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":25.0,"Social":100.0,"Covert":0.0,"Migration":25.0,"Care":0.0,"Exploration":0.0},"type":{"Government":15.0,"Military":15.0,"Religious":20.0,"Criminal":75.0,"Revolutionary":50.0,"Conspiracy":15.0,"Merchant":85.0,"Clan":55.0,"Cultural":90.0,"Outsider":55.0,"Elite":25.0,"Poor":70.0,"Academic":15.0},"base":{"Empire":35.0,"Kingdom":35.0,"Republic":45.0,"Dominion":35.0,"Council":20.0,"Court":35.0,"Commonwealth":45.0,"Directorate":20.0,"Ministry":15.0,"Authority":20.0,"Legion":35.0,"Guard":35.0,"Army":35.0,"Militia":45.0,"Vanguard":40.0,"Watch":35.0,"Corps":35.0,"Wardens":35.0,"Rangers":45.0,"Order":35.0,"Covenant":35.0,"Faith":35.0,"Sect":35.0,"Brotherhood":65.0,"Fellowship":65.0,"Chapter":45.0,"Disciples":35.0,"Cult":35.0,"Guild":70.0,"Company":70.0,"Consortium":70.0,"Syndicate":70.0,"Cartel":70.0,"Cooperative":70.0,"Corporation":70.0,"Chamber":70.0,"Trust":70.0,"Clan":65.0,"House":45.0,"Dynasty":45.0,"Family":65.0,"Tribe":65.0,"People":60.0,"Community":65.0,"Nation":60.0,"Kindred":65.0,"Lineage":55.0,"Resistance":55.0,"Movement":65.0,"Front":45.0,"Union":70.0,"Collective":70.0,"Alliance":60.0,"League":60.0,"Pact":60.0,"Liberation":55.0,"Academy":35.0,"Institute":35.0,"School":35.0,"Society":55.0,"Circle":45.0,"Cabal":45.0,"Conclave":35.0,"Scholars":35.0,"Collegium":35.0,"Keepers":35.0,"Caravan":70.0,"Fleet":70.0,"Band":60.0,"Enclave":65.0,"Commune":60.0,"Network":70.0,"Expedition":45.0,"Nomads":65.0,"Diaspora":65.0,"Confederation":60.0,"Assembly":60.0,"Coalition":60.0,"Horde":55.0,"Mercenaries":90.0,"Parliament":35.0,"Senate":35.0,"Association":70.0,"Compact":60.0},"place":{"Apothecary":45.0,"Archive":0.0,"Armoury":0.0,"Auction House":60.0,"Bank":50.0,"Bathhouse":60.0,"Bookshop":55.0,"Brothel":70.0,"Clinic":15.0,"Club":80.0,"Courthouse":0.0,"Bridge":20.0,"Crypt":0.0,"Dormitory":20.0,"Forge":25.0,"Gallery":60.0,"Gambling Den":85.0,"House":30.0,"Laboratory":0.0,"Library":20.0,"Mill":20.0,"Monument":55.0,"Mortuary":0.0,"Museum":50.0,"Observatory":0.0,"Office":45.0,"Pawnshop":50.0,"Post Office":20.0,"Printing House":25.0,"Prison":0.0,"School":20.0,"Sewer":0.0,"Shelter":20.0,"Ship":20.0,"Shop":50.0,"Shrine":20.0,"Stable":0.0,"Tailor":50.0,"Tavern":100.0,"Theater":65.0,"Tower":0.0,"Training Ground":15.0,"Treasury":20.0,"Tunnel":0.0,"Vault":20.0,"Guardhouse":0.0,"Well":20.0,"Arena":60.0,"Barracks":25.0,"Black Market":50.0,"Brewery":50.0,"Camp":30.0,"Cemetery":25.0,"Dock":30.0,"Embassy":25.0,"Factory":25.0,"Farm":25.0,"Fortress":20.0,"Garden":60.0,"Grove":25.0,"Meeting Hall":55.0,"Hideout":25.0,"Inn":65.0,"Manor":30.0,"Market":80.0,"Mine":5.0,"Outpost":5.0,"Palace":30.0,"Research Facility":5.0,"Square":65.0,"Temple":30.0,"Trading Post":55.0,"Warehouse":25.0,"Workshop":30.0,"Catacombs":5.0,"Town Hall":35.0,"Excavation Site":5.0,"Festival Grounds":80.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":25.0,"Shipyard":25.0,"Caravan Camp":30.0,"Cavern":10.0,"Citadel":10.0,"Enclave":25.0,"Estate":20.0,"Village":35.0,"Hunting Grounds":0.0,"Military Base":15.0,"Mining Settlement":25.0,"Monastery":20.0,"Necropolis":10.0,"Rebel Camp":20.0,"Refugee Camp":25.0,"Slum":90.0,"University":25.0,"Academic District":15.0,"Government District":10.0,"Artisan Quarter":50.0,"Cultural District":75.0,"Entertainment District":90.0,"Farmland":10.0,"Foreign Quarter":85.0,"Wilderness":10.0,"Harbor":90.0,"Industrial District":10.0,"Market District":90.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":15.0,"Royal Quarter":10.0,"Residential District":50.0,"Ruined Quarter":10.0,"Temple District":20.0,"Underground City":30.0,"Underworld":20.0,"Warehouse District":15.0},"score":95.0},"Theater":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":25.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":100.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":45.0,"Military":15.0,"Religious":20.0,"Criminal":20.0,"Revolutionary":45.0,"Conspiracy":15.0,"Merchant":50.0,"Clan":25.0,"Cultural":90.0,"Outsider":15.0,"Elite":50.0,"Poor":15.0,"Academic":15.0},"base":{"Empire":60.0,"Kingdom":60.0,"Republic":60.0,"Dominion":60.0,"Council":40.0,"Court":65.0,"Commonwealth":60.0,"Directorate":40.0,"Ministry":60.0,"Authority":60.0,"Legion":20.0,"Guard":35.0,"Army":20.0,"Militia":40.0,"Vanguard":35.0,"Watch":35.0,"Corps":20.0,"Wardens":35.0,"Rangers":20.0,"Order":20.0,"Covenant":20.0,"Faith":20.0,"Sect":20.0,"Brotherhood":35.0,"Fellowship":45.0,"Chapter":35.0,"Disciples":20.0,"Cult":20.0,"Guild":65.0,"Company":65.0,"Consortium":65.0,"Syndicate":45.0,"Cartel":45.0,"Cooperative":45.0,"Corporation":65.0,"Chamber":60.0,"Trust":60.0,"Clan":45.0,"House":65.0,"Dynasty":65.0,"Family":35.0,"Tribe":45.0,"People":45.0,"Community":45.0,"Nation":40.0,"Kindred":45.0,"Lineage":55.0,"Resistance":40.0,"Movement":60.0,"Front":40.0,"Union":40.0,"Collective":40.0,"Alliance":60.0,"League":60.0,"Pact":60.0,"Liberation":40.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":65.0,"Circle":45.0,"Cabal":35.0,"Conclave":35.0,"Scholars":20.0,"Collegium":20.0,"Keepers":20.0,"Caravan":65.0,"Fleet":40.0,"Band":35.0,"Enclave":45.0,"Commune":30.0,"Network":35.0,"Expedition":35.0,"Nomads":45.0,"Diaspora":45.0,"Confederation":60.0,"Assembly":60.0,"Coalition":60.0,"Horde":15.0,"Mercenaries":35.0,"Parliament":60.0,"Senate":60.0,"Association":70.0,"Compact":50.0},"place":{"Apothecary":20.0,"Archive":0.0,"Armoury":0.0,"Auction House":35.0,"Bank":20.0,"Bathhouse":65.0,"Bookshop":25.0,"Brothel":60.0,"Clinic":0.0,"Club":65.0,"Courthouse":15.0,"Bridge":15.0,"Crypt":15.0,"Dormitory":0.0,"Forge":20.0,"Gallery":70.0,"Gambling Den":65.0,"House":25.0,"Laboratory":0.0,"Library":20.0,"Mill":15.0,"Monument":65.0,"Mortuary":0.0,"Museum":55.0,"Observatory":0.0,"Office":20.0,"Pawnshop":20.0,"Post Office":20.0,"Printing House":25.0,"Prison":0.0,"School":20.0,"Sewer":0.0,"Shelter":0.0,"Ship":15.0,"Shop":25.0,"Shrine":20.0,"Stable":0.0,"Tailor":25.0,"Tavern":65.0,"Theater":100.0,"Tower":0.0,"Training Ground":15.0,"Treasury":20.0,"Tunnel":0.0,"Vault":20.0,"Guardhouse":0.0,"Well":5.0,"Arena":60.0,"Barracks":5.0,"Black Market":25.0,"Brewery":25.0,"Camp":5.0,"Cemetery":30.0,"Dock":25.0,"Embassy":25.0,"Factory":20.0,"Farm":5.0,"Fortress":5.0,"Garden":70.0,"Grove":25.0,"Meeting Hall":60.0,"Hideout":5.0,"Inn":35.0,"Manor":25.0,"Market":45.0,"Mine":5.0,"Outpost":5.0,"Palace":30.0,"Research Facility":5.0,"Square":65.0,"Temple":35.0,"Trading Post":25.0,"Warehouse":20.0,"Workshop":25.0,"Catacombs":5.0,"Town Hall":35.0,"Excavation Site":5.0,"Festival Grounds":80.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":5.0,"Shipyard":20.0,"Caravan Camp":20.0,"Cavern":10.0,"Citadel":10.0,"Enclave":10.0,"Estate":20.0,"Village":20.0,"Hunting Grounds":0.0,"Military Base":0.0,"Mining Settlement":10.0,"Monastery":10.0,"Necropolis":10.0,"Rebel Camp":10.0,"Refugee Camp":15.0,"Slum":20.0,"University":20.0,"Academic District":20.0,"Government District":15.0,"Artisan Quarter":25.0,"Cultural District":85.0,"Entertainment District":100.0,"Farmland":10.0,"Foreign Quarter":20.0,"Wilderness":10.0,"Harbor":20.0,"Industrial District":10.0,"Market District":50.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":20.0,"Royal Quarter":35.0,"Residential District":20.0,"Ruined Quarter":10.0,"Temple District":25.0,"Underground City":20.0,"Underworld":20.0,"Warehouse District":10.0},"score":55.0},"Tower":{"function":{"Authority":0.0,"Defense":75.0,"Religion":0.0,"Knowledge":25.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":25.0,"Migration":0.0,"Care":0.0,"Exploration":25.0},"type":{"Government":70.0,"Military":85.0,"Religious":45.0,"Criminal":20.0,"Revolutionary":25.0,"Conspiracy":55.0,"Merchant":15.0,"Clan":15.0,"Cultural":20.0,"Outsider":20.0,"Elite":15.0,"Poor":15.0,"Academic":70.0},"base":{"Empire":70.0,"Kingdom":70.0,"Republic":60.0,"Dominion":70.0,"Council":70.0,"Court":60.0,"Commonwealth":60.0,"Directorate":70.0,"Ministry":65.0,"Authority":65.0,"Legion":95.0,"Guard":95.0,"Army":95.0,"Militia":70.0,"Vanguard":95.0,"Watch":95.0,"Corps":95.0,"Wardens":95.0,"Rangers":95.0,"Order":70.0,"Covenant":60.0,"Faith":60.0,"Sect":60.0,"Brotherhood":60.0,"Fellowship":60.0,"Chapter":60.0,"Disciples":60.0,"Cult":65.0,"Guild":20.0,"Company":20.0,"Consortium":45.0,"Syndicate":35.0,"Cartel":35.0,"Cooperative":20.0,"Corporation":45.0,"Chamber":60.0,"Trust":20.0,"Clan":45.0,"House":20.0,"Dynasty":20.0,"Family":20.0,"Tribe":45.0,"People":15.0,"Community":20.0,"Nation":15.0,"Kindred":45.0,"Lineage":20.0,"Resistance":45.0,"Movement":45.0,"Front":45.0,"Union":30.0,"Collective":30.0,"Alliance":45.0,"League":45.0,"Pact":55.0,"Liberation":45.0,"Academy":70.0,"Institute":70.0,"School":70.0,"Society":65.0,"Circle":70.0,"Cabal":50.0,"Conclave":50.0,"Scholars":70.0,"Collegium":70.0,"Keepers":65.0,"Caravan":35.0,"Fleet":95.0,"Band":45.0,"Enclave":20.0,"Commune":25.0,"Network":45.0,"Expedition":60.0,"Nomads":35.0,"Diaspora":20.0,"Confederation":65.0,"Assembly":65.0,"Coalition":55.0,"Horde":70.0,"Mercenaries":70.0,"Parliament":65.0,"Senate":70.0,"Association":55.0,"Compact":45.0},"place":{"Apothecary":20.0,"Archive":25.0,"Armoury":55.0,"Auction House":0.0,"Bank":0.0,"Bathhouse":0.0,"Bookshop":20.0,"Brothel":0.0,"Clinic":20.0,"Club":0.0,"Courthouse":20.0,"Bridge":20.0,"Crypt":25.0,"Dormitory":20.0,"Forge":20.0,"Gallery":0.0,"Gambling Den":0.0,"House":0.0,"Laboratory":25.0,"Library":30.0,"Mill":0.0,"Monument":0.0,"Mortuary":20.0,"Museum":20.0,"Observatory":35.0,"Office":20.0,"Pawnshop":0.0,"Post Office":20.0,"Printing House":20.0,"Prison":55.0,"School":25.0,"Sewer":25.0,"Shelter":0.0,"Ship":25.0,"Shop":0.0,"Shrine":0.0,"Stable":15.0,"Tailor":0.0,"Tavern":0.0,"Theater":0.0,"Tower":100.0,"Training Ground":65.0,"Treasury":0.0,"Tunnel":20.0,"Vault":20.0,"Guardhouse":60.0,"Well":0.0,"Arena":35.0,"Barracks":55.0,"Black Market":25.0,"Brewery":0.0,"Camp":25.0,"Cemetery":10.0,"Dock":20.0,"Embassy":5.0,"Factory":5.0,"Farm":5.0,"Fortress":75.0,"Garden":5.0,"Grove":10.0,"Meeting Hall":5.0,"Hideout":30.0,"Inn":0.0,"Manor":5.0,"Market":5.0,"Mine":5.0,"Outpost":75.0,"Palace":10.0,"Research Facility":35.0,"Square":5.0,"Temple":5.0,"Trading Post":20.0,"Warehouse":20.0,"Workshop":25.0,"Catacombs":25.0,"Town Hall":5.0,"Excavation Site":30.0,"Festival Grounds":5.0,"Hospital":25.0,"Ruins":50.0,"Sanctuary":5.0,"Shipyard":20.0,"Caravan Camp":15.0,"Cavern":20.0,"Citadel":75.0,"Enclave":10.0,"Estate":10.0,"Village":10.0,"Hunting Grounds":35.0,"Military Base":70.0,"Mining Settlement":10.0,"Monastery":40.0,"Necropolis":25.0,"Rebel Camp":65.0,"Refugee Camp":10.0,"Slum":10.0,"University":25.0,"Academic District":40.0,"Government District":15.0,"Artisan Quarter":10.0,"Cultural District":10.0,"Entertainment District":10.0,"Farmland":10.0,"Foreign Quarter":10.0,"Wilderness":35.0,"Harbor":35.0,"Industrial District":10.0,"Market District":10.0,"Military District":85.0,"Mining Region":10.0,"Noble Quarter":10.0,"Royal Quarter":15.0,"Residential District":10.0,"Ruined Quarter":40.0,"Temple District":10.0,"Underground City":15.0,"Underworld":15.0,"Warehouse District":15.0},"score":86.66666666666667},"Training Ground":{"function":{"Authority":0.0,"Defense":100.0,"Religion":0.0,"Knowledge":25.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":55.0,"Military":90.0,"Religious":15.0,"Criminal":15.0,"Revolutionary":80.0,"Conspiracy":20.0,"Merchant":15.0,"Clan":45.0,"Cultural":20.0,"Outsider":15.0,"Elite":15.0,"Poor":15.0,"Academic":50.0},"base":{"Empire":65.0,"Kingdom":65.0,"Republic":45.0,"Dominion":65.0,"Council":65.0,"Court":45.0,"Commonwealth":45.0,"Directorate":65.0,"Ministry":65.0,"Authority":65.0,"Legion":100.0,"Guard":100.0,"Army":100.0,"Militia":100.0,"Vanguard":100.0,"Watch":100.0,"Corps":100.0,"Wardens":95.0,"Rangers":95.0,"Order":75.0,"Covenant":55.0,"Faith":45.0,"Sect":35.0,"Brotherhood":40.0,"Fellowship":40.0,"Chapter":30.0,"Disciples":35.0,"Cult":35.0,"Guild":20.0,"Company":20.0,"Consortium":20.0,"Syndicate":20.0,"Cartel":20.0,"Cooperative":20.0,"Corporation":20.0,"Chamber":35.0,"Trust":20.0,"Clan":60.0,"House":40.0,"Dynasty":40.0,"Family":40.0,"Tribe":60.0,"People":40.0,"Community":40.0,"Nation":30.0,"Kindred":60.0,"Lineage":40.0,"Resistance":70.0,"Movement":70.0,"Front":70.0,"Union":60.0,"Collective":60.0,"Alliance":70.0,"League":70.0,"Pact":70.0,"Liberation":70.0,"Academy":65.0,"Institute":65.0,"School":65.0,"Society":40.0,"Circle":45.0,"Cabal":35.0,"Conclave":35.0,"Scholars":60.0,"Collegium":65.0,"Keepers":60.0,"Caravan":20.0,"Fleet":65.0,"Band":45.0,"Enclave":35.0,"Commune":40.0,"Network":20.0,"Expedition":55.0,"Nomads":20.0,"Diaspora":20.0,"Confederation":60.0,"Assembly":60.0,"Coalition":60.0,"Horde":95.0,"Mercenaries":95.0,"Parliament":65.0,"Senate":65.0,"Association":35.0,"Compact":35.0},"place":{"Apothecary":15.0,"Archive":20.0,"Armoury":55.0,"Auction House":15.0,"Bank":0.0,"Bathhouse":15.0,"Bookshop":20.0,"Brothel":15.0,"Clinic":20.0,"Club":15.0,"Courthouse":20.0,"Bridge":20.0,"Crypt":15.0,"Dormitory":20.0,"Forge":20.0,"Gallery":15.0,"Gambling Den":15.0,"House":0.0,"Laboratory":20.0,"Library":25.0,"Mill":0.0,"Monument":15.0,"Mortuary":20.0,"Museum":20.0,"Observatory":30.0,"Office":20.0,"Pawnshop":0.0,"Post Office":15.0,"Printing House":20.0,"Prison":55.0,"School":25.0,"Sewer":0.0,"Shelter":0.0,"Ship":25.0,"Shop":0.0,"Shrine":15.0,"Stable":15.0,"Tailor":15.0,"Tavern":15.0,"Theater":15.0,"Tower":65.0,"Training Ground":100.0,"Treasury":0.0,"Tunnel":15.0,"Vault":0.0,"Guardhouse":65.0,"Well":0.0,"Arena":40.0,"Barracks":75.0,"Black Market":5.0,"Brewery":20.0,"Camp":25.0,"Cemetery":20.0,"Dock":5.0,"Embassy":5.0,"Factory":5.0,"Farm":5.0,"Fortress":80.0,"Garden":20.0,"Grove":5.0,"Meeting Hall":25.0,"Hideout":10.0,"Inn":5.0,"Manor":5.0,"Market":20.0,"Mine":5.0,"Outpost":60.0,"Palace":10.0,"Research Facility":25.0,"Square":20.0,"Temple":20.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":20.0,"Catacombs":5.0,"Town Hall":25.0,"Excavation Site":25.0,"Festival Grounds":20.0,"Hospital":25.0,"Ruins":25.0,"Sanctuary":5.0,"Shipyard":5.0,"Caravan Camp":10.0,"Cavern":10.0,"Citadel":75.0,"Enclave":10.0,"Estate":10.0,"Village":10.0,"Hunting Grounds":0.0,"Military Base":75.0,"Mining Settlement":10.0,"Monastery":20.0,"Necropolis":20.0,"Rebel Camp":70.0,"Refugee Camp":10.0,"Slum":10.0,"University":20.0,"Academic District":20.0,"Government District":15.0,"Artisan Quarter":10.0,"Cultural District":20.0,"Entertainment District":15.0,"Farmland":10.0,"Foreign Quarter":10.0,"Wilderness":10.0,"Harbor":15.0,"Industrial District":10.0,"Market District":10.0,"Military District":100.0,"Mining Region":10.0,"Noble Quarter":10.0,"Royal Quarter":15.0,"Residential District":10.0,"Ruined Quarter":20.0,"Temple District":10.0,"Underground City":10.0,"Underworld":10.0,"Warehouse District":10.0},"score":71.66666666666667},"Treasury":{"function":{"Authority":100.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":50.0,"Production":0.0,"Logistics":25.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":90.0,"Military":55.0,"Religious":15.0,"Criminal":50.0,"Revolutionary":15.0,"Conspiracy":50.0,"Merchant":80.0,"Clan":15.0,"Cultural":15.0,"Outsider":15.0,"Elite":85.0,"Poor":15.0,"Academic":15.0},"base":{"Empire":100.0,"Kingdom":100.0,"Republic":75.0,"Dominion":100.0,"Council":75.0,"Court":100.0,"Commonwealth":75.0,"Directorate":75.0,"Ministry":70.0,"Authority":75.0,"Legion":65.0,"Guard":75.0,"Army":65.0,"Militia":45.0,"Vanguard":45.0,"Watch":75.0,"Corps":65.0,"Wardens":75.0,"Rangers":45.0,"Order":65.0,"Covenant":45.0,"Faith":45.0,"Sect":35.0,"Brotherhood":35.0,"Fellowship":35.0,"Chapter":30.0,"Disciples":35.0,"Cult":45.0,"Guild":60.0,"Company":60.0,"Consortium":60.0,"Syndicate":60.0,"Cartel":50.0,"Cooperative":60.0,"Corporation":60.0,"Chamber":65.0,"Trust":95.0,"Clan":15.0,"House":60.0,"Dynasty":55.0,"Family":45.0,"Tribe":35.0,"People":15.0,"Community":20.0,"Nation":15.0,"Kindred":35.0,"Lineage":30.0,"Resistance":35.0,"Movement":35.0,"Front":35.0,"Union":35.0,"Collective":35.0,"Alliance":35.0,"League":40.0,"Pact":40.0,"Liberation":35.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":45.0,"Circle":55.0,"Cabal":45.0,"Conclave":55.0,"Scholars":20.0,"Collegium":20.0,"Keepers":30.0,"Caravan":60.0,"Fleet":50.0,"Band":40.0,"Enclave":20.0,"Commune":15.0,"Network":40.0,"Expedition":35.0,"Nomads":35.0,"Diaspora":20.0,"Confederation":70.0,"Assembly":70.0,"Coalition":60.0,"Horde":45.0,"Mercenaries":50.0,"Parliament":75.0,"Senate":75.0,"Association":60.0,"Compact":50.0},"place":{"Apothecary":25.0,"Archive":50.0,"Armoury":5.0,"Auction House":50.0,"Bank":60.0,"Bathhouse":20.0,"Bookshop":25.0,"Brothel":25.0,"Clinic":0.0,"Club":40.0,"Courthouse":60.0,"Bridge":20.0,"Crypt":15.0,"Dormitory":0.0,"Forge":20.0,"Gallery":45.0,"Gambling Den":30.0,"House":15.0,"Laboratory":15.0,"Library":0.0,"Mill":20.0,"Monument":45.0,"Mortuary":15.0,"Museum":15.0,"Observatory":0.0,"Office":65.0,"Pawnshop":25.0,"Post Office":55.0,"Printing House":20.0,"Prison":55.0,"School":0.0,"Sewer":20.0,"Shelter":0.0,"Ship":20.0,"Shop":25.0,"Shrine":0.0,"Stable":5.0,"Tailor":25.0,"Tavern":20.0,"Theater":20.0,"Tower":0.0,"Training Ground":0.0,"Treasury":100.0,"Tunnel":20.0,"Vault":100.0,"Guardhouse":50.0,"Well":0.0,"Arena":25.0,"Barracks":20.0,"Black Market":30.0,"Brewery":25.0,"Camp":5.0,"Cemetery":0.0,"Dock":30.0,"Embassy":60.0,"Factory":20.0,"Farm":5.0,"Fortress":30.0,"Garden":20.0,"Grove":5.0,"Meeting Hall":25.0,"Hideout":20.0,"Inn":25.0,"Manor":45.0,"Market":30.0,"Mine":5.0,"Outpost":5.0,"Palace":100.0,"Research Facility":5.0,"Square":30.0,"Temple":25.0,"Trading Post":30.0,"Warehouse":30.0,"Workshop":25.0,"Catacombs":20.0,"Town Hall":55.0,"Excavation Site":5.0,"Festival Grounds":30.0,"Hospital":5.0,"Ruins":20.0,"Sanctuary":5.0,"Shipyard":25.0,"Caravan Camp":25.0,"Cavern":15.0,"Citadel":70.0,"Enclave":10.0,"Estate":50.0,"Village":10.0,"Hunting Grounds":0.0,"Military Base":0.0,"Mining Settlement":10.0,"Monastery":10.0,"Necropolis":10.0,"Rebel Camp":15.0,"Refugee Camp":10.0,"Slum":10.0,"University":0.0,"Academic District":10.0,"Government District":85.0,"Artisan Quarter":25.0,"Cultural District":10.0,"Entertainment District":25.0,"Farmland":10.0,"Foreign Quarter":15.0,"Wilderness":15.0,"Harbor":30.0,"Industrial District":15.0,"Market District":25.0,"Military District":15.0,"Mining Region":10.0,"Noble Quarter":75.0,"Royal Quarter":90.0,"Residential District":10.0,"Ruined Quarter":15.0,"Temple District":10.0,"Underground City":20.0,"Underworld":30.0,"Warehouse District":20.0},"score":73.33333333333333},"Tunnel":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":100.0,"Resources":25.0,"Residence":0.0,"Social":0.0,"Covert":25.0,"Migration":0.0,"Care":0.0,"Exploration":75.0},"type":{"Government":15.0,"Military":55.0,"Religious":15.0,"Criminal":70.0,"Revolutionary":70.0,"Conspiracy":70.0,"Merchant":30.0,"Clan":20.0,"Cultural":15.0,"Outsider":55.0,"Elite":15.0,"Poor":50.0,"Academic":25.0},"base":{"Empire":45.0,"Kingdom":45.0,"Republic":20.0,"Dominion":45.0,"Council":45.0,"Court":20.0,"Commonwealth":20.0,"Directorate":45.0,"Ministry":40.0,"Authority":45.0,"Legion":65.0,"Guard":65.0,"Army":65.0,"Militia":70.0,"Vanguard":70.0,"Watch":65.0,"Corps":65.0,"Wardens":65.0,"Rangers":65.0,"Order":65.0,"Covenant":45.0,"Faith":45.0,"Sect":45.0,"Brotherhood":35.0,"Fellowship":20.0,"Chapter":40.0,"Disciples":45.0,"Cult":70.0,"Guild":30.0,"Company":50.0,"Consortium":45.0,"Syndicate":70.0,"Cartel":70.0,"Cooperative":45.0,"Corporation":45.0,"Chamber":45.0,"Trust":30.0,"Clan":45.0,"House":20.0,"Dynasty":20.0,"Family":45.0,"Tribe":55.0,"People":35.0,"Community":40.0,"Nation":15.0,"Kindred":45.0,"Lineage":20.0,"Resistance":70.0,"Movement":60.0,"Front":70.0,"Union":60.0,"Collective":60.0,"Alliance":65.0,"League":65.0,"Pact":65.0,"Liberation":70.0,"Academy":30.0,"Institute":30.0,"School":30.0,"Society":45.0,"Circle":60.0,"Cabal":70.0,"Conclave":70.0,"Scholars":30.0,"Collegium":30.0,"Keepers":55.0,"Caravan":55.0,"Fleet":55.0,"Band":70.0,"Enclave":45.0,"Commune":45.0,"Network":90.0,"Expedition":65.0,"Nomads":55.0,"Diaspora":45.0,"Confederation":50.0,"Assembly":60.0,"Coalition":60.0,"Horde":65.0,"Mercenaries":65.0,"Parliament":55.0,"Senate":55.0,"Association":35.0,"Compact":50.0},"place":{"Apothecary":0.0,"Archive":15.0,"Armoury":30.0,"Auction House":0.0,"Bank":5.0,"Bathhouse":0.0,"Bookshop":0.0,"Brothel":20.0,"Clinic":0.0,"Club":20.0,"Courthouse":0.0,"Bridge":55.0,"Crypt":45.0,"Dormitory":15.0,"Forge":15.0,"Gallery":0.0,"Gambling Den":40.0,"House":0.0,"Laboratory":20.0,"Library":0.0,"Mill":0.0,"Monument":0.0,"Mortuary":0.0,"Museum":0.0,"Observatory":20.0,"Office":0.0,"Pawnshop":20.0,"Post Office":55.0,"Printing House":0.0,"Prison":20.0,"School":0.0,"Sewer":70.0,"Shelter":0.0,"Ship":55.0,"Shop":0.0,"Shrine":0.0,"Stable":60.0,"Tailor":0.0,"Tavern":0.0,"Theater":0.0,"Tower":20.0,"Training Ground":15.0,"Treasury":20.0,"Tunnel":100.0,"Vault":35.0,"Guardhouse":15.0,"Well":5.0,"Arena":20.0,"Barracks":20.0,"Black Market":65.0,"Brewery":5.0,"Camp":25.0,"Cemetery":5.0,"Dock":50.0,"Embassy":5.0,"Factory":5.0,"Farm":5.0,"Fortress":20.0,"Garden":5.0,"Grove":30.0,"Meeting Hall":5.0,"Hideout":65.0,"Inn":5.0,"Manor":5.0,"Market":5.0,"Mine":90.0,"Outpost":50.0,"Palace":5.0,"Research Facility":5.0,"Square":5.0,"Temple":5.0,"Trading Post":50.0,"Warehouse":55.0,"Workshop":5.0,"Catacombs":70.0,"Town Hall":5.0,"Excavation Site":75.0,"Festival Grounds":5.0,"Hospital":5.0,"Ruins":75.0,"Sanctuary":5.0,"Shipyard":50.0,"Caravan Camp":60.0,"Cavern":75.0,"Citadel":35.0,"Enclave":10.0,"Estate":10.0,"Village":10.0,"Hunting Grounds":50.0,"Military Base":25.0,"Mining Settlement":70.0,"Monastery":10.0,"Necropolis":45.0,"Rebel Camp":45.0,"Refugee Camp":10.0,"Slum":15.0,"University":0.0,"Academic District":10.0,"Government District":10.0,"Artisan Quarter":10.0,"Cultural District":10.0,"Entertainment District":10.0,"Farmland":15.0,"Foreign Quarter":10.0,"Wilderness":60.0,"Harbor":65.0,"Industrial District":20.0,"Market District":15.0,"Military District":25.0,"Mining Region":70.0,"Noble Quarter":10.0,"Royal Quarter":10.0,"Residential District":10.0,"Ruined Quarter":80.0,"Temple District":10.0,"Underground City":100.0,"Underworld":70.0,"Warehouse District":70.0},"score":85.0},"Vault":{"function":{"Authority":25.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":75.0,"Production":0.0,"Logistics":50.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":25.0,"Migration":0.0,"Care":0.0,"Exploration":25.0},"type":{"Government":70.0,"Military":50.0,"Religious":15.0,"Criminal":55.0,"Revolutionary":20.0,"Conspiracy":70.0,"Merchant":85.0,"Clan":15.0,"Cultural":15.0,"Outsider":20.0,"Elite":75.0,"Poor":15.0,"Academic":20.0},"base":{"Empire":70.0,"Kingdom":70.0,"Republic":70.0,"Dominion":70.0,"Council":70.0,"Court":70.0,"Commonwealth":70.0,"Directorate":70.0,"Ministry":65.0,"Authority":65.0,"Legion":65.0,"Guard":70.0,"Army":65.0,"Militia":45.0,"Vanguard":40.0,"Watch":65.0,"Corps":65.0,"Wardens":65.0,"Rangers":40.0,"Order":65.0,"Covenant":45.0,"Faith":45.0,"Sect":45.0,"Brotherhood":20.0,"Fellowship":20.0,"Chapter":40.0,"Disciples":45.0,"Cult":70.0,"Guild":70.0,"Company":70.0,"Consortium":70.0,"Syndicate":70.0,"Cartel":70.0,"Cooperative":60.0,"Corporation":70.0,"Chamber":70.0,"Trust":95.0,"Clan":35.0,"House":70.0,"Dynasty":70.0,"Family":45.0,"Tribe":35.0,"People":15.0,"Community":20.0,"Nation":15.0,"Kindred":35.0,"Lineage":55.0,"Resistance":45.0,"Movement":35.0,"Front":45.0,"Union":20.0,"Collective":20.0,"Alliance":45.0,"League":55.0,"Pact":55.0,"Liberation":45.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":45.0,"Circle":60.0,"Cabal":70.0,"Conclave":70.0,"Scholars":20.0,"Collegium":20.0,"Keepers":45.0,"Caravan":70.0,"Fleet":70.0,"Band":65.0,"Enclave":20.0,"Commune":15.0,"Network":70.0,"Expedition":45.0,"Nomads":45.0,"Diaspora":20.0,"Confederation":65.0,"Assembly":65.0,"Coalition":55.0,"Horde":40.0,"Mercenaries":50.0,"Parliament":65.0,"Senate":70.0,"Association":70.0,"Compact":60.0},"place":{"Apothecary":25.0,"Archive":20.0,"Armoury":25.0,"Auction House":55.0,"Bank":100.0,"Bathhouse":20.0,"Bookshop":25.0,"Brothel":30.0,"Clinic":0.0,"Club":45.0,"Courthouse":25.0,"Bridge":25.0,"Crypt":20.0,"Dormitory":0.0,"Forge":20.0,"Gallery":40.0,"Gambling Den":30.0,"House":15.0,"Laboratory":20.0,"Library":0.0,"Mill":0.0,"Monument":15.0,"Mortuary":15.0,"Museum":15.0,"Observatory":0.0,"Office":30.0,"Pawnshop":35.0,"Post Office":35.0,"Printing House":20.0,"Prison":20.0,"School":0.0,"Sewer":30.0,"Shelter":0.0,"Ship":30.0,"Shop":30.0,"Shrine":0.0,"Stable":30.0,"Tailor":30.0,"Tavern":20.0,"Theater":20.0,"Tower":20.0,"Training Ground":0.0,"Treasury":100.0,"Tunnel":35.0,"Vault":100.0,"Guardhouse":20.0,"Well":0.0,"Arena":10.0,"Barracks":20.0,"Black Market":60.0,"Brewery":25.0,"Camp":5.0,"Cemetery":5.0,"Dock":40.0,"Embassy":25.0,"Factory":5.0,"Farm":5.0,"Fortress":45.0,"Garden":20.0,"Grove":10.0,"Meeting Hall":25.0,"Hideout":45.0,"Inn":25.0,"Manor":60.0,"Market":30.0,"Mine":5.0,"Outpost":30.0,"Palace":60.0,"Research Facility":5.0,"Square":30.0,"Temple":20.0,"Trading Post":40.0,"Warehouse":55.0,"Workshop":25.0,"Catacombs":30.0,"Town Hall":25.0,"Excavation Site":10.0,"Festival Grounds":25.0,"Hospital":5.0,"Ruins":25.0,"Sanctuary":5.0,"Shipyard":35.0,"Caravan Camp":35.0,"Cavern":20.0,"Citadel":40.0,"Enclave":10.0,"Estate":50.0,"Village":15.0,"Hunting Grounds":0.0,"Military Base":20.0,"Mining Settlement":10.0,"Monastery":10.0,"Necropolis":15.0,"Rebel Camp":20.0,"Refugee Camp":10.0,"Slum":10.0,"University":0.0,"Academic District":10.0,"Government District":25.0,"Artisan Quarter":20.0,"Cultural District":10.0,"Entertainment District":20.0,"Farmland":10.0,"Foreign Quarter":15.0,"Wilderness":20.0,"Harbor":45.0,"Industrial District":15.0,"Market District":50.0,"Military District":15.0,"Mining Region":10.0,"Noble Quarter":70.0,"Royal Quarter":70.0,"Residential District":10.0,"Ruined Quarter":25.0,"Temple District":10.0,"Underground City":30.0,"Underworld":50.0,"Warehouse District":50.0},"score":75.0},"Guardhouse":{"function":{"Authority":50.0,"Defense":100.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":80.0,"Military":90.0,"Religious":15.0,"Criminal":15.0,"Revolutionary":30.0,"Conspiracy":20.0,"Merchant":45.0,"Clan":45.0,"Cultural":15.0,"Outsider":15.0,"Elite":25.0,"Poor":15.0,"Academic":15.0},"base":{"Empire":70.0,"Kingdom":70.0,"Republic":70.0,"Dominion":70.0,"Council":70.0,"Court":70.0,"Commonwealth":70.0,"Directorate":70.0,"Ministry":70.0,"Authority":70.0,"Legion":100.0,"Guard":100.0,"Army":100.0,"Militia":65.0,"Vanguard":100.0,"Watch":100.0,"Corps":100.0,"Wardens":95.0,"Rangers":95.0,"Order":65.0,"Covenant":55.0,"Faith":45.0,"Sect":35.0,"Brotherhood":40.0,"Fellowship":40.0,"Chapter":30.0,"Disciples":35.0,"Cult":35.0,"Guild":60.0,"Company":60.0,"Consortium":60.0,"Syndicate":40.0,"Cartel":35.0,"Cooperative":40.0,"Corporation":60.0,"Chamber":70.0,"Trust":60.0,"Clan":60.0,"House":50.0,"Dynasty":50.0,"Family":60.0,"Tribe":60.0,"People":60.0,"Community":60.0,"Nation":45.0,"Kindred":60.0,"Lineage":50.0,"Resistance":50.0,"Movement":50.0,"Front":45.0,"Union":30.0,"Collective":30.0,"Alliance":45.0,"League":55.0,"Pact":45.0,"Liberation":45.0,"Academy":35.0,"Institute":35.0,"School":35.0,"Society":35.0,"Circle":35.0,"Cabal":20.0,"Conclave":20.0,"Scholars":35.0,"Collegium":35.0,"Keepers":40.0,"Caravan":60.0,"Fleet":75.0,"Band":45.0,"Enclave":35.0,"Commune":45.0,"Network":20.0,"Expedition":45.0,"Nomads":45.0,"Diaspora":20.0,"Confederation":70.0,"Assembly":70.0,"Coalition":60.0,"Horde":60.0,"Mercenaries":60.0,"Parliament":70.0,"Senate":70.0,"Association":60.0,"Compact":60.0},"place":{"Apothecary":15.0,"Archive":45.0,"Armoury":55.0,"Auction House":0.0,"Bank":20.0,"Bathhouse":0.0,"Bookshop":15.0,"Brothel":0.0,"Clinic":15.0,"Club":0.0,"Courthouse":55.0,"Bridge":20.0,"Crypt":15.0,"Dormitory":15.0,"Forge":20.0,"Gallery":0.0,"Gambling Den":0.0,"House":0.0,"Laboratory":15.0,"Library":15.0,"Mill":0.0,"Monument":40.0,"Mortuary":15.0,"Museum":15.0,"Observatory":20.0,"Office":50.0,"Pawnshop":0.0,"Post Office":45.0,"Printing House":15.0,"Prison":75.0,"School":15.0,"Sewer":0.0,"Shelter":0.0,"Ship":25.0,"Shop":0.0,"Shrine":0.0,"Stable":15.0,"Tailor":0.0,"Tavern":0.0,"Theater":0.0,"Tower":60.0,"Training Ground":65.0,"Treasury":50.0,"Tunnel":15.0,"Vault":20.0,"Guardhouse":100.0,"Well":0.0,"Arena":35.0,"Barracks":60.0,"Black Market":5.0,"Brewery":5.0,"Camp":25.0,"Cemetery":5.0,"Dock":40.0,"Embassy":45.0,"Factory":5.0,"Farm":5.0,"Fortress":80.0,"Garden":5.0,"Grove":5.0,"Meeting Hall":25.0,"Hideout":10.0,"Inn":5.0,"Manor":25.0,"Market":5.0,"Mine":5.0,"Outpost":75.0,"Palace":100.0,"Research Facility":20.0,"Square":25.0,"Temple":25.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":20.0,"Catacombs":5.0,"Town Hall":50.0,"Excavation Site":20.0,"Festival Grounds":5.0,"Hospital":20.0,"Ruins":20.0,"Sanctuary":5.0,"Shipyard":5.0,"Caravan Camp":10.0,"Cavern":10.0,"Citadel":90.0,"Enclave":10.0,"Estate":15.0,"Village":10.0,"Hunting Grounds":0.0,"Military Base":75.0,"Mining Settlement":10.0,"Monastery":15.0,"Necropolis":15.0,"Rebel Camp":60.0,"Refugee Camp":10.0,"Slum":10.0,"University":15.0,"Academic District":15.0,"Government District":80.0,"Artisan Quarter":10.0,"Cultural District":10.0,"Entertainment District":10.0,"Farmland":10.0,"Foreign Quarter":10.0,"Wilderness":10.0,"Harbor":40.0,"Industrial District":10.0,"Market District":35.0,"Military District":90.0,"Mining Region":10.0,"Noble Quarter":15.0,"Royal Quarter":70.0,"Residential District":35.0,"Ruined Quarter":15.0,"Temple District":10.0,"Underground City":10.0,"Underworld":10.0,"Warehouse District":10.0},"score":71.66666666666667},"Well":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":100.0,"Residence":0.0,"Social":25.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":15.0,"Military":15.0,"Religious":45.0,"Criminal":15.0,"Revolutionary":15.0,"Conspiracy":15.0,"Merchant":25.0,"Clan":75.0,"Cultural":20.0,"Outsider":45.0,"Elite":15.0,"Poor":75.0,"Academic":15.0},"base":{"Empire":35.0,"Kingdom":35.0,"Republic":45.0,"Dominion":35.0,"Council":20.0,"Court":35.0,"Commonwealth":45.0,"Directorate":20.0,"Ministry":30.0,"Authority":20.0,"Legion":35.0,"Guard":35.0,"Army":35.0,"Militia":35.0,"Vanguard":35.0,"Watch":35.0,"Corps":35.0,"Wardens":35.0,"Rangers":40.0,"Order":40.0,"Covenant":50.0,"Faith":40.0,"Sect":40.0,"Brotherhood":70.0,"Fellowship":70.0,"Chapter":60.0,"Disciples":40.0,"Cult":40.0,"Guild":30.0,"Company":35.0,"Consortium":30.0,"Syndicate":45.0,"Cartel":45.0,"Cooperative":70.0,"Corporation":30.0,"Chamber":30.0,"Trust":35.0,"Clan":70.0,"House":60.0,"Dynasty":60.0,"Family":70.0,"Tribe":70.0,"People":65.0,"Community":70.0,"Nation":55.0,"Kindred":70.0,"Lineage":60.0,"Resistance":45.0,"Movement":55.0,"Front":35.0,"Union":70.0,"Collective":70.0,"Alliance":45.0,"League":40.0,"Pact":40.0,"Liberation":45.0,"Academy":35.0,"Institute":35.0,"School":35.0,"Society":45.0,"Circle":45.0,"Cabal":45.0,"Conclave":55.0,"Scholars":35.0,"Collegium":35.0,"Keepers":40.0,"Caravan":40.0,"Fleet":40.0,"Band":40.0,"Enclave":70.0,"Commune":55.0,"Network":35.0,"Expedition":50.0,"Nomads":60.0,"Diaspora":60.0,"Confederation":50.0,"Assembly":50.0,"Coalition":40.0,"Horde":40.0,"Mercenaries":40.0,"Parliament":35.0,"Senate":35.0,"Association":35.0,"Compact":50.0},"place":{"Apothecary":0.0,"Archive":0.0,"Armoury":0.0,"Auction House":0.0,"Bank":0.0,"Bathhouse":20.0,"Bookshop":0.0,"Brothel":5.0,"Clinic":20.0,"Club":5.0,"Courthouse":0.0,"Bridge":35.0,"Crypt":0.0,"Dormitory":20.0,"Forge":0.0,"Gallery":5.0,"Gambling Den":0.0,"House":25.0,"Laboratory":0.0,"Library":0.0,"Mill":60.0,"Monument":5.0,"Mortuary":0.0,"Museum":0.0,"Observatory":0.0,"Office":0.0,"Pawnshop":15.0,"Post Office":0.0,"Printing House":0.0,"Prison":0.0,"School":15.0,"Sewer":0.0,"Shelter":20.0,"Ship":30.0,"Shop":0.0,"Shrine":5.0,"Stable":45.0,"Tailor":0.0,"Tavern":20.0,"Theater":5.0,"Tower":0.0,"Training Ground":0.0,"Treasury":0.0,"Tunnel":5.0,"Vault":0.0,"Guardhouse":0.0,"Well":100.0,"Arena":5.0,"Barracks":20.0,"Black Market":5.0,"Brewery":45.0,"Camp":40.0,"Cemetery":25.0,"Dock":35.0,"Embassy":5.0,"Factory":20.0,"Farm":75.0,"Fortress":20.0,"Garden":60.0,"Grove":70.0,"Meeting Hall":25.0,"Hideout":20.0,"Inn":45.0,"Manor":25.0,"Market":10.0,"Mine":40.0,"Outpost":40.0,"Palace":20.0,"Research Facility":5.0,"Square":45.0,"Temple":10.0,"Trading Post":5.0,"Warehouse":30.0,"Workshop":5.0,"Catacombs":5.0,"Town Hall":10.0,"Excavation Site":30.0,"Festival Grounds":10.0,"Hospital":25.0,"Ruins":5.0,"Sanctuary":40.0,"Shipyard":35.0,"Caravan Camp":35.0,"Cavern":40.0,"Citadel":10.0,"Enclave":35.0,"Estate":50.0,"Village":90.0,"Hunting Grounds":80.0,"Military Base":15.0,"Mining Settlement":40.0,"Monastery":20.0,"Necropolis":10.0,"Rebel Camp":15.0,"Refugee Camp":35.0,"Slum":40.0,"University":15.0,"Academic District":10.0,"Government District":10.0,"Artisan Quarter":10.0,"Cultural District":15.0,"Entertainment District":15.0,"Farmland":90.0,"Foreign Quarter":15.0,"Wilderness":80.0,"Harbor":40.0,"Industrial District":10.0,"Market District":15.0,"Military District":10.0,"Mining Region":35.0,"Noble Quarter":10.0,"Royal Quarter":10.0,"Residential District":35.0,"Ruined Quarter":35.0,"Temple District":15.0,"Underground City":25.0,"Underworld":10.0,"Warehouse District":30.0},"score":70.0},"Arena":{"function":{"Authority":0.0,"Defense":75.0,"Religion":0.0,"Knowledge":0.0,"Commerce":25.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":100.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":20.0,"Military":85.0,"Religious":20.0,"Criminal":50.0,"Revolutionary":55.0,"Conspiracy":15.0,"Merchant":50.0,"Clan":25.0,"Cultural":90.0,"Outsider":15.0,"Elite":50.0,"Poor":15.0,"Academic":15.0},"base":{"Empire":45.0,"Kingdom":55.0,"Republic":45.0,"Dominion":45.0,"Council":55.0,"Court":45.0,"Commonwealth":45.0,"Directorate":55.0,"Ministry":45.0,"Authority":45.0,"Legion":70.0,"Guard":70.0,"Army":70.0,"Militia":60.0,"Vanguard":60.0,"Watch":70.0,"Corps":70.0,"Wardens":70.0,"Rangers":70.0,"Order":65.0,"Covenant":45.0,"Faith":45.0,"Sect":20.0,"Brotherhood":35.0,"Fellowship":45.0,"Chapter":35.0,"Disciples":20.0,"Cult":20.0,"Guild":65.0,"Company":65.0,"Consortium":65.0,"Syndicate":45.0,"Cartel":45.0,"Cooperative":45.0,"Corporation":65.0,"Chamber":60.0,"Trust":60.0,"Clan":45.0,"House":45.0,"Dynasty":45.0,"Family":35.0,"Tribe":45.0,"People":55.0,"Community":55.0,"Nation":50.0,"Kindred":45.0,"Lineage":45.0,"Resistance":55.0,"Movement":65.0,"Front":55.0,"Union":65.0,"Collective":65.0,"Alliance":65.0,"League":65.0,"Pact":65.0,"Liberation":55.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":45.0,"Circle":35.0,"Cabal":35.0,"Conclave":35.0,"Scholars":20.0,"Collegium":20.0,"Keepers":45.0,"Caravan":65.0,"Fleet":60.0,"Band":50.0,"Enclave":45.0,"Commune":35.0,"Network":40.0,"Expedition":45.0,"Nomads":45.0,"Diaspora":45.0,"Confederation":55.0,"Assembly":50.0,"Coalition":50.0,"Horde":60.0,"Mercenaries":60.0,"Parliament":55.0,"Senate":60.0,"Association":70.0,"Compact":50.0},"place":{"Apothecary":25.0,"Archive":5.0,"Armoury":35.0,"Auction House":35.0,"Bank":25.0,"Bathhouse":60.0,"Bookshop":30.0,"Brothel":55.0,"Clinic":40.0,"Club":55.0,"Courthouse":5.0,"Bridge":25.0,"Crypt":5.0,"Dormitory":20.0,"Forge":25.0,"Gallery":60.0,"Gambling Den":70.0,"House":25.0,"Laboratory":5.0,"Library":25.0,"Mill":20.0,"Monument":55.0,"Mortuary":5.0,"Museum":50.0,"Observatory":25.0,"Office":45.0,"Pawnshop":25.0,"Post Office":20.0,"Printing House":25.0,"Prison":30.0,"School":25.0,"Sewer":5.0,"Shelter":5.0,"Ship":30.0,"Shop":25.0,"Shrine":25.0,"Stable":40.0,"Tailor":25.0,"Tavern":60.0,"Theater":60.0,"Tower":35.0,"Training Ground":40.0,"Treasury":25.0,"Tunnel":20.0,"Vault":10.0,"Guardhouse":35.0,"Well":5.0,"Arena":100.0,"Barracks":35.0,"Black Market":25.0,"Brewery":25.0,"Camp":25.0,"Cemetery":25.0,"Dock":25.0,"Embassy":5.0,"Factory":20.0,"Farm":5.0,"Fortress":35.0,"Garden":55.0,"Grove":25.0,"Meeting Hall":55.0,"Hideout":5.0,"Inn":30.0,"Manor":10.0,"Market":35.0,"Mine":5.0,"Outpost":35.0,"Palace":25.0,"Research Facility":5.0,"Square":60.0,"Temple":30.0,"Trading Post":25.0,"Warehouse":25.0,"Workshop":25.0,"Catacombs":5.0,"Town Hall":35.0,"Excavation Site":5.0,"Festival Grounds":75.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":5.0,"Shipyard":25.0,"Caravan Camp":25.0,"Cavern":15.0,"Citadel":55.0,"Enclave":15.0,"Estate":15.0,"Village":25.0,"Hunting Grounds":5.0,"Military Base":30.0,"Mining Settlement":15.0,"Monastery":15.0,"Necropolis":15.0,"Rebel Camp":40.0,"Refugee Camp":20.0,"Slum":25.0,"University":25.0,"Academic District":15.0,"Government District":15.0,"Artisan Quarter":25.0,"Cultural District":80.0,"Entertainment District":85.0,"Farmland":10.0,"Foreign Quarter":20.0,"Wilderness":10.0,"Harbor":25.0,"Industrial District":10.0,"Market District":50.0,"Military District":45.0,"Mining Region":10.0,"Noble Quarter":15.0,"Royal Quarter":15.0,"Residential District":20.0,"Ruined Quarter":10.0,"Temple District":20.0,"Underground City":20.0,"Underworld":15.0,"Warehouse District":10.0},"score":76.66666666666667},"Barracks":{"function":{"Authority":0.0,"Defense":100.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":75.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":75.0,"Military":90.0,"Religious":15.0,"Criminal":15.0,"Revolutionary":60.0,"Conspiracy":15.0,"Merchant":15.0,"Clan":60.0,"Cultural":15.0,"Outsider":25.0,"Elite":25.0,"Poor":30.0,"Academic":15.0},"base":{"Empire":70.0,"Kingdom":70.0,"Republic":60.0,"Dominion":70.0,"Council":70.0,"Court":60.0,"Commonwealth":60.0,"Directorate":70.0,"Ministry":70.0,"Authority":70.0,"Legion":100.0,"Guard":100.0,"Army":100.0,"Militia":100.0,"Vanguard":100.0,"Watch":100.0,"Corps":100.0,"Wardens":95.0,"Rangers":95.0,"Order":65.0,"Covenant":55.0,"Faith":45.0,"Sect":35.0,"Brotherhood":45.0,"Fellowship":45.0,"Chapter":30.0,"Disciples":35.0,"Cult":35.0,"Guild":20.0,"Company":20.0,"Consortium":20.0,"Syndicate":45.0,"Cartel":45.0,"Cooperative":45.0,"Corporation":20.0,"Chamber":40.0,"Trust":35.0,"Clan":65.0,"House":45.0,"Dynasty":45.0,"Family":55.0,"Tribe":65.0,"People":55.0,"Community":55.0,"Nation":50.0,"Kindred":65.0,"Lineage":45.0,"Resistance":65.0,"Movement":65.0,"Front":65.0,"Union":55.0,"Collective":55.0,"Alliance":65.0,"League":65.0,"Pact":65.0,"Liberation":65.0,"Academy":35.0,"Institute":35.0,"School":35.0,"Society":35.0,"Circle":35.0,"Cabal":45.0,"Conclave":45.0,"Scholars":35.0,"Collegium":35.0,"Keepers":40.0,"Caravan":30.0,"Fleet":65.0,"Band":45.0,"Enclave":45.0,"Commune":50.0,"Network":30.0,"Expedition":45.0,"Nomads":35.0,"Diaspora":35.0,"Confederation":65.0,"Assembly":65.0,"Coalition":65.0,"Horde":95.0,"Mercenaries":95.0,"Parliament":70.0,"Senate":70.0,"Association":30.0,"Compact":45.0},"place":{"Apothecary":5.0,"Archive":20.0,"Armoury":100.0,"Auction House":0.0,"Bank":20.0,"Bathhouse":35.0,"Bookshop":0.0,"Brothel":5.0,"Clinic":20.0,"Club":5.0,"Courthouse":25.0,"Bridge":25.0,"Crypt":5.0,"Dormitory":55.0,"Forge":25.0,"Gallery":5.0,"Gambling Den":5.0,"House":55.0,"Laboratory":5.0,"Library":5.0,"Mill":5.0,"Monument":20.0,"Mortuary":20.0,"Museum":5.0,"Observatory":25.0,"Office":40.0,"Pawnshop":5.0,"Post Office":20.0,"Printing House":20.0,"Prison":60.0,"School":20.0,"Sewer":5.0,"Shelter":50.0,"Ship":30.0,"Shop":5.0,"Shrine":5.0,"Stable":40.0,"Tailor":5.0,"Tavern":25.0,"Theater":5.0,"Tower":55.0,"Training Ground":75.0,"Treasury":20.0,"Tunnel":20.0,"Vault":20.0,"Guardhouse":60.0,"Well":20.0,"Arena":35.0,"Barracks":100.0,"Black Market":5.0,"Brewery":5.0,"Camp":55.0,"Cemetery":20.0,"Dock":5.0,"Embassy":20.0,"Factory":5.0,"Farm":25.0,"Fortress":80.0,"Garden":35.0,"Grove":5.0,"Meeting Hall":20.0,"Hideout":30.0,"Inn":45.0,"Manor":55.0,"Market":5.0,"Mine":5.0,"Outpost":60.0,"Palace":35.0,"Research Facility":5.0,"Square":40.0,"Temple":20.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":5.0,"Catacombs":5.0,"Town Hall":25.0,"Excavation Site":5.0,"Festival Grounds":5.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":25.0,"Shipyard":5.0,"Caravan Camp":25.0,"Cavern":15.0,"Citadel":90.0,"Enclave":50.0,"Estate":50.0,"Village":65.0,"Hunting Grounds":5.0,"Military Base":80.0,"Mining Settlement":65.0,"Monastery":30.0,"Necropolis":15.0,"Rebel Camp":80.0,"Refugee Camp":50.0,"Slum":55.0,"University":45.0,"Academic District":10.0,"Government District":25.0,"Artisan Quarter":10.0,"Cultural District":10.0,"Entertainment District":10.0,"Farmland":10.0,"Foreign Quarter":50.0,"Wilderness":10.0,"Harbor":15.0,"Industrial District":10.0,"Market District":10.0,"Military District":100.0,"Mining Region":10.0,"Noble Quarter":55.0,"Royal Quarter":55.0,"Residential District":65.0,"Ruined Quarter":15.0,"Temple District":10.0,"Underground City":65.0,"Underworld":10.0,"Warehouse District":10.0},"score":86.66666666666667},"Black Market":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":100.0,"Production":0.0,"Logistics":25.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":100.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":15.0,"Military":20.0,"Religious":15.0,"Criminal":100.0,"Revolutionary":60.0,"Conspiracy":90.0,"Merchant":80.0,"Clan":15.0,"Cultural":15.0,"Outsider":45.0,"Elite":25.0,"Poor":45.0,"Academic":15.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":45.0,"Dominion":20.0,"Council":20.0,"Court":30.0,"Commonwealth":45.0,"Directorate":20.0,"Ministry":15.0,"Authority":20.0,"Legion":20.0,"Guard":20.0,"Army":20.0,"Militia":45.0,"Vanguard":45.0,"Watch":20.0,"Corps":20.0,"Wardens":20.0,"Rangers":35.0,"Order":45.0,"Covenant":20.0,"Faith":20.0,"Sect":45.0,"Brotherhood":45.0,"Fellowship":45.0,"Chapter":40.0,"Disciples":45.0,"Cult":75.0,"Guild":60.0,"Company":60.0,"Consortium":70.0,"Syndicate":100.0,"Cartel":100.0,"Cooperative":60.0,"Corporation":70.0,"Chamber":70.0,"Trust":60.0,"Clan":20.0,"House":30.0,"Dynasty":30.0,"Family":45.0,"Tribe":35.0,"People":50.0,"Community":60.0,"Nation":35.0,"Kindred":20.0,"Lineage":20.0,"Resistance":65.0,"Movement":65.0,"Front":65.0,"Union":65.0,"Collective":65.0,"Alliance":55.0,"League":55.0,"Pact":65.0,"Liberation":65.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":45.0,"Circle":65.0,"Cabal":75.0,"Conclave":75.0,"Scholars":20.0,"Collegium":20.0,"Keepers":40.0,"Caravan":60.0,"Fleet":70.0,"Band":75.0,"Enclave":40.0,"Commune":40.0,"Network":100.0,"Expedition":40.0,"Nomads":50.0,"Diaspora":40.0,"Confederation":50.0,"Assembly":50.0,"Coalition":50.0,"Horde":50.0,"Mercenaries":60.0,"Parliament":35.0,"Senate":35.0,"Association":60.0,"Compact":60.0},"place":{"Apothecary":50.0,"Archive":25.0,"Armoury":25.0,"Auction House":50.0,"Bank":55.0,"Bathhouse":25.0,"Bookshop":50.0,"Brothel":55.0,"Clinic":5.0,"Club":40.0,"Courthouse":5.0,"Bridge":25.0,"Crypt":20.0,"Dormitory":5.0,"Forge":30.0,"Gallery":25.0,"Gambling Den":65.0,"House":30.0,"Laboratory":25.0,"Library":5.0,"Mill":25.0,"Monument":5.0,"Mortuary":5.0,"Museum":0.0,"Observatory":5.0,"Office":45.0,"Pawnshop":85.0,"Post Office":30.0,"Printing House":10.0,"Prison":20.0,"School":5.0,"Sewer":60.0,"Shelter":30.0,"Ship":30.0,"Shop":70.0,"Shrine":5.0,"Stable":25.0,"Tailor":50.0,"Tavern":50.0,"Theater":25.0,"Tower":25.0,"Training Ground":5.0,"Treasury":30.0,"Tunnel":65.0,"Vault":60.0,"Guardhouse":5.0,"Well":5.0,"Arena":25.0,"Barracks":5.0,"Black Market":100.0,"Brewery":30.0,"Camp":5.0,"Cemetery":0.0,"Dock":35.0,"Embassy":5.0,"Factory":20.0,"Farm":5.0,"Fortress":5.0,"Garden":0.0,"Grove":5.0,"Meeting Hall":5.0,"Hideout":80.0,"Inn":30.0,"Manor":5.0,"Market":50.0,"Mine":5.0,"Outpost":20.0,"Palace":0.0,"Research Facility":5.0,"Square":30.0,"Temple":0.0,"Trading Post":55.0,"Warehouse":30.0,"Workshop":30.0,"Catacombs":30.0,"Town Hall":5.0,"Excavation Site":5.0,"Festival Grounds":30.0,"Hospital":5.0,"Ruins":20.0,"Sanctuary":5.0,"Shipyard":30.0,"Caravan Camp":35.0,"Cavern":35.0,"Citadel":15.0,"Enclave":20.0,"Estate":15.0,"Village":20.0,"Hunting Grounds":5.0,"Military Base":20.0,"Mining Settlement":20.0,"Monastery":15.0,"Necropolis":20.0,"Rebel Camp":50.0,"Refugee Camp":35.0,"Slum":90.0,"University":5.0,"Academic District":10.0,"Government District":10.0,"Artisan Quarter":35.0,"Cultural District":10.0,"Entertainment District":35.0,"Farmland":10.0,"Foreign Quarter":45.0,"Wilderness":15.0,"Harbor":80.0,"Industrial District":15.0,"Market District":70.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":10.0,"Royal Quarter":10.0,"Residential District":30.0,"Ruined Quarter":60.0,"Temple District":10.0,"Underground City":45.0,"Underworld":100.0,"Warehouse District":25.0},"score":78.33333333333333},"Brewery":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":75.0,"Production":100.0,"Logistics":0.0,"Resources":25.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":15.0,"Military":15.0,"Religious":15.0,"Criminal":50.0,"Revolutionary":15.0,"Conspiracy":15.0,"Merchant":85.0,"Clan":50.0,"Cultural":65.0,"Outsider":15.0,"Elite":25.0,"Poor":70.0,"Academic":15.0},"base":{"Empire":20.0,"Kingdom":35.0,"Republic":45.0,"Dominion":20.0,"Council":20.0,"Court":30.0,"Commonwealth":45.0,"Directorate":20.0,"Ministry":15.0,"Authority":20.0,"Legion":20.0,"Guard":20.0,"Army":20.0,"Militia":20.0,"Vanguard":20.0,"Watch":20.0,"Corps":20.0,"Wardens":20.0,"Rangers":20.0,"Order":20.0,"Covenant":35.0,"Faith":20.0,"Sect":20.0,"Brotherhood":60.0,"Fellowship":60.0,"Chapter":40.0,"Disciples":20.0,"Cult":20.0,"Guild":70.0,"Company":70.0,"Consortium":70.0,"Syndicate":65.0,"Cartel":60.0,"Cooperative":70.0,"Corporation":70.0,"Chamber":70.0,"Trust":70.0,"Clan":55.0,"House":45.0,"Dynasty":45.0,"Family":45.0,"Tribe":50.0,"People":60.0,"Community":65.0,"Nation":55.0,"Kindred":55.0,"Lineage":50.0,"Resistance":45.0,"Movement":55.0,"Front":20.0,"Union":70.0,"Collective":70.0,"Alliance":45.0,"League":50.0,"Pact":40.0,"Liberation":45.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":40.0,"Circle":35.0,"Cabal":35.0,"Conclave":35.0,"Scholars":20.0,"Collegium":20.0,"Keepers":15.0,"Caravan":70.0,"Fleet":60.0,"Band":40.0,"Enclave":50.0,"Commune":50.0,"Network":60.0,"Expedition":35.0,"Nomads":45.0,"Diaspora":45.0,"Confederation":60.0,"Assembly":50.0,"Coalition":40.0,"Horde":30.0,"Mercenaries":50.0,"Parliament":20.0,"Senate":30.0,"Association":70.0,"Compact":60.0},"place":{"Apothecary":40.0,"Archive":5.0,"Armoury":25.0,"Auction House":40.0,"Bank":35.0,"Bathhouse":30.0,"Bookshop":35.0,"Brothel":35.0,"Clinic":5.0,"Club":20.0,"Courthouse":5.0,"Bridge":20.0,"Crypt":5.0,"Dormitory":5.0,"Forge":65.0,"Gallery":25.0,"Gambling Den":35.0,"House":5.0,"Laboratory":25.0,"Library":20.0,"Mill":70.0,"Monument":20.0,"Mortuary":5.0,"Museum":20.0,"Observatory":5.0,"Office":30.0,"Pawnshop":35.0,"Post Office":25.0,"Printing House":60.0,"Prison":5.0,"School":20.0,"Sewer":5.0,"Shelter":5.0,"Ship":25.0,"Shop":45.0,"Shrine":20.0,"Stable":40.0,"Tailor":70.0,"Tavern":50.0,"Theater":25.0,"Tower":0.0,"Training Ground":20.0,"Treasury":25.0,"Tunnel":5.0,"Vault":25.0,"Guardhouse":5.0,"Well":45.0,"Arena":25.0,"Barracks":5.0,"Black Market":30.0,"Brewery":100.0,"Camp":5.0,"Cemetery":20.0,"Dock":30.0,"Embassy":5.0,"Factory":60.0,"Farm":55.0,"Fortress":5.0,"Garden":35.0,"Grove":40.0,"Meeting Hall":20.0,"Hideout":5.0,"Inn":30.0,"Manor":5.0,"Market":40.0,"Mine":35.0,"Outpost":5.0,"Palace":5.0,"Research Facility":5.0,"Square":30.0,"Temple":20.0,"Trading Post":35.0,"Warehouse":50.0,"Workshop":80.0,"Catacombs":5.0,"Town Hall":20.0,"Excavation Site":25.0,"Festival Grounds":30.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":5.0,"Shipyard":55.0,"Caravan Camp":30.0,"Cavern":20.0,"Citadel":15.0,"Enclave":15.0,"Estate":45.0,"Village":70.0,"Hunting Grounds":40.0,"Military Base":5.0,"Mining Settlement":25.0,"Monastery":15.0,"Necropolis":15.0,"Rebel Camp":15.0,"Refugee Camp":15.0,"Slum":20.0,"University":5.0,"Academic District":10.0,"Government District":10.0,"Artisan Quarter":90.0,"Cultural District":15.0,"Entertainment District":35.0,"Farmland":60.0,"Foreign Quarter":15.0,"Wilderness":50.0,"Harbor":35.0,"Industrial District":85.0,"Market District":45.0,"Military District":10.0,"Mining Region":20.0,"Noble Quarter":10.0,"Royal Quarter":10.0,"Residential District":10.0,"Ruined Quarter":10.0,"Temple District":10.0,"Underground City":25.0,"Underworld":20.0,"Warehouse District":20.0},"score":58.333333333333336},"Camp":{"function":{"Authority":0.0,"Defense":25.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":100.0,"Social":0.0,"Covert":0.0,"Migration":75.0,"Care":0.0,"Exploration":0.0},"type":{"Government":20.0,"Military":70.0,"Religious":15.0,"Criminal":15.0,"Revolutionary":75.0,"Conspiracy":15.0,"Merchant":45.0,"Clan":65.0,"Cultural":15.0,"Outsider":85.0,"Elite":30.0,"Poor":85.0,"Academic":15.0},"base":{"Empire":35.0,"Kingdom":35.0,"Republic":45.0,"Dominion":35.0,"Council":35.0,"Court":35.0,"Commonwealth":45.0,"Directorate":35.0,"Ministry":35.0,"Authority":35.0,"Legion":70.0,"Guard":60.0,"Army":70.0,"Militia":95.0,"Vanguard":70.0,"Watch":60.0,"Corps":70.0,"Wardens":55.0,"Rangers":70.0,"Order":50.0,"Covenant":35.0,"Faith":35.0,"Sect":35.0,"Brotherhood":65.0,"Fellowship":55.0,"Chapter":40.0,"Disciples":35.0,"Cult":35.0,"Guild":40.0,"Company":50.0,"Consortium":40.0,"Syndicate":50.0,"Cartel":50.0,"Cooperative":70.0,"Corporation":40.0,"Chamber":40.0,"Trust":40.0,"Clan":65.0,"House":45.0,"Dynasty":45.0,"Family":65.0,"Tribe":90.0,"People":65.0,"Community":70.0,"Nation":50.0,"Kindred":65.0,"Lineage":45.0,"Resistance":95.0,"Movement":95.0,"Front":95.0,"Union":95.0,"Collective":95.0,"Alliance":95.0,"League":95.0,"Pact":95.0,"Liberation":95.0,"Academy":35.0,"Institute":35.0,"School":35.0,"Society":45.0,"Circle":45.0,"Cabal":45.0,"Conclave":45.0,"Scholars":35.0,"Collegium":35.0,"Keepers":30.0,"Caravan":95.0,"Fleet":60.0,"Band":95.0,"Enclave":95.0,"Commune":70.0,"Network":45.0,"Expedition":95.0,"Nomads":95.0,"Diaspora":95.0,"Confederation":50.0,"Assembly":60.0,"Coalition":60.0,"Horde":95.0,"Mercenaries":95.0,"Parliament":50.0,"Senate":50.0,"Association":50.0,"Compact":60.0},"place":{"Apothecary":5.0,"Archive":5.0,"Armoury":25.0,"Auction House":5.0,"Bank":5.0,"Bathhouse":35.0,"Bookshop":5.0,"Brothel":5.0,"Clinic":20.0,"Club":5.0,"Courthouse":5.0,"Bridge":35.0,"Crypt":5.0,"Dormitory":65.0,"Forge":25.0,"Gallery":5.0,"Gambling Den":5.0,"House":60.0,"Laboratory":5.0,"Library":0.0,"Mill":5.0,"Monument":5.0,"Mortuary":5.0,"Museum":0.0,"Observatory":20.0,"Office":5.0,"Pawnshop":5.0,"Post Office":10.0,"Printing House":5.0,"Prison":30.0,"School":20.0,"Sewer":5.0,"Shelter":75.0,"Ship":50.0,"Shop":5.0,"Shrine":40.0,"Stable":40.0,"Tailor":5.0,"Tavern":30.0,"Theater":5.0,"Tower":25.0,"Training Ground":25.0,"Treasury":5.0,"Tunnel":25.0,"Vault":5.0,"Guardhouse":25.0,"Well":40.0,"Arena":25.0,"Barracks":55.0,"Black Market":5.0,"Brewery":5.0,"Camp":100.0,"Cemetery":20.0,"Dock":45.0,"Embassy":30.0,"Factory":5.0,"Farm":30.0,"Fortress":30.0,"Garden":35.0,"Grove":5.0,"Meeting Hall":20.0,"Hideout":45.0,"Inn":70.0,"Manor":55.0,"Market":25.0,"Mine":40.0,"Outpost":50.0,"Palace":30.0,"Research Facility":5.0,"Square":40.0,"Temple":5.0,"Trading Post":45.0,"Warehouse":5.0,"Workshop":5.0,"Catacombs":5.0,"Town Hall":5.0,"Excavation Site":40.0,"Festival Grounds":5.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":30.0,"Shipyard":5.0,"Caravan Camp":75.0,"Cavern":40.0,"Citadel":30.0,"Enclave":85.0,"Estate":55.0,"Village":80.0,"Hunting Grounds":40.0,"Military Base":50.0,"Mining Settlement":80.0,"Monastery":35.0,"Necropolis":15.0,"Rebel Camp":55.0,"Refugee Camp":85.0,"Slum":60.0,"University":45.0,"Academic District":10.0,"Government District":10.0,"Artisan Quarter":10.0,"Cultural District":10.0,"Entertainment District":10.0,"Farmland":10.0,"Foreign Quarter":75.0,"Wilderness":35.0,"Harbor":40.0,"Industrial District":10.0,"Market District":10.0,"Military District":25.0,"Mining Region":15.0,"Noble Quarter":55.0,"Royal Quarter":45.0,"Residential District":70.0,"Ruined Quarter":15.0,"Temple District":10.0,"Underground City":70.0,"Underworld":10.0,"Warehouse District":10.0},"score":90.0},"Cemetery":{"function":{"Authority":0.0,"Defense":0.0,"Religion":75.0,"Knowledge":25.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":25.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":25.0},"type":{"Government":15.0,"Military":15.0,"Religious":85.0,"Criminal":15.0,"Revolutionary":15.0,"Conspiracy":20.0,"Merchant":15.0,"Clan":70.0,"Cultural":55.0,"Outsider":20.0,"Elite":15.0,"Poor":45.0,"Academic":55.0},"base":{"Empire":35.0,"Kingdom":35.0,"Republic":45.0,"Dominion":35.0,"Council":20.0,"Court":35.0,"Commonwealth":45.0,"Directorate":20.0,"Ministry":40.0,"Authority":20.0,"Legion":35.0,"Guard":35.0,"Army":35.0,"Militia":35.0,"Vanguard":35.0,"Watch":35.0,"Corps":35.0,"Wardens":35.0,"Rangers":35.0,"Order":70.0,"Covenant":70.0,"Faith":70.0,"Sect":70.0,"Brotherhood":70.0,"Fellowship":70.0,"Chapter":70.0,"Disciples":70.0,"Cult":70.0,"Guild":20.0,"Company":20.0,"Consortium":20.0,"Syndicate":45.0,"Cartel":45.0,"Cooperative":60.0,"Corporation":20.0,"Chamber":20.0,"Trust":35.0,"Clan":70.0,"House":60.0,"Dynasty":60.0,"Family":70.0,"Tribe":65.0,"People":65.0,"Community":65.0,"Nation":60.0,"Kindred":70.0,"Lineage":60.0,"Resistance":35.0,"Movement":55.0,"Front":35.0,"Union":60.0,"Collective":60.0,"Alliance":35.0,"League":30.0,"Pact":35.0,"Liberation":35.0,"Academy":65.0,"Institute":65.0,"School":65.0,"Society":45.0,"Circle":45.0,"Cabal":55.0,"Conclave":55.0,"Scholars":65.0,"Collegium":65.0,"Keepers":70.0,"Caravan":20.0,"Fleet":20.0,"Band":35.0,"Enclave":50.0,"Commune":50.0,"Network":20.0,"Expedition":55.0,"Nomads":35.0,"Diaspora":35.0,"Confederation":45.0,"Assembly":30.0,"Coalition":30.0,"Horde":35.0,"Mercenaries":35.0,"Parliament":20.0,"Senate":20.0,"Association":45.0,"Compact":30.0},"place":{"Apothecary":5.0,"Archive":20.0,"Armoury":0.0,"Auction House":25.0,"Bank":0.0,"Bathhouse":30.0,"Bookshop":25.0,"Brothel":25.0,"Clinic":25.0,"Club":30.0,"Courthouse":5.0,"Bridge":5.0,"Crypt":100.0,"Dormitory":25.0,"Forge":5.0,"Gallery":30.0,"Gambling Den":25.0,"House":30.0,"Laboratory":20.0,"Library":30.0,"Mill":5.0,"Monument":50.0,"Mortuary":90.0,"Museum":30.0,"Observatory":30.0,"Office":5.0,"Pawnshop":5.0,"Post Office":5.0,"Printing House":25.0,"Prison":5.0,"School":30.0,"Sewer":10.0,"Shelter":20.0,"Ship":5.0,"Shop":5.0,"Shrine":80.0,"Stable":5.0,"Tailor":20.0,"Tavern":25.0,"Theater":30.0,"Tower":10.0,"Training Ground":20.0,"Treasury":0.0,"Tunnel":5.0,"Vault":5.0,"Guardhouse":5.0,"Well":25.0,"Arena":25.0,"Barracks":20.0,"Black Market":0.0,"Brewery":20.0,"Camp":20.0,"Cemetery":100.0,"Dock":5.0,"Embassy":5.0,"Factory":5.0,"Farm":20.0,"Fortress":20.0,"Garden":30.0,"Grove":40.0,"Meeting Hall":30.0,"Hideout":20.0,"Inn":25.0,"Manor":25.0,"Market":25.0,"Mine":5.0,"Outpost":5.0,"Palace":25.0,"Research Facility":25.0,"Square":30.0,"Temple":60.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":5.0,"Catacombs":75.0,"Town Hall":25.0,"Excavation Site":25.0,"Festival Grounds":30.0,"Hospital":25.0,"Ruins":25.0,"Sanctuary":60.0,"Shipyard":5.0,"Caravan Camp":20.0,"Cavern":25.0,"Citadel":15.0,"Enclave":25.0,"Estate":25.0,"Village":45.0,"Hunting Grounds":10.0,"Military Base":20.0,"Mining Settlement":30.0,"Monastery":75.0,"Necropolis":80.0,"Rebel Camp":25.0,"Refugee Camp":25.0,"Slum":20.0,"University":30.0,"Academic District":20.0,"Government District":10.0,"Artisan Quarter":10.0,"Cultural District":25.0,"Entertainment District":20.0,"Farmland":10.0,"Foreign Quarter":15.0,"Wilderness":15.0,"Harbor":10.0,"Industrial District":15.0,"Market District":15.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":10.0,"Royal Quarter":10.0,"Residential District":40.0,"Ruined Quarter":20.0,"Temple District":90.0,"Underground City":25.0,"Underworld":10.0,"Warehouse District":10.0},"score":81.66666666666667},"Dock":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":75.0,"Production":0.0,"Logistics":100.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":50.0,"Care":0.0,"Exploration":0.0},"type":{"Government":15.0,"Military":55.0,"Religious":15.0,"Criminal":50.0,"Revolutionary":20.0,"Conspiracy":15.0,"Merchant":90.0,"Clan":15.0,"Cultural":15.0,"Outsider":75.0,"Elite":25.0,"Poor":70.0,"Academic":15.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":20.0,"Dominion":20.0,"Council":20.0,"Court":30.0,"Commonwealth":20.0,"Directorate":20.0,"Ministry":15.0,"Authority":20.0,"Legion":45.0,"Guard":45.0,"Army":45.0,"Militia":45.0,"Vanguard":45.0,"Watch":45.0,"Corps":45.0,"Wardens":45.0,"Rangers":50.0,"Order":40.0,"Covenant":20.0,"Faith":20.0,"Sect":20.0,"Brotherhood":45.0,"Fellowship":20.0,"Chapter":20.0,"Disciples":20.0,"Cult":20.0,"Guild":65.0,"Company":65.0,"Consortium":75.0,"Syndicate":75.0,"Cartel":75.0,"Cooperative":65.0,"Corporation":75.0,"Chamber":75.0,"Trust":65.0,"Clan":20.0,"House":30.0,"Dynasty":30.0,"Family":35.0,"Tribe":40.0,"People":40.0,"Community":55.0,"Nation":15.0,"Kindred":20.0,"Lineage":20.0,"Resistance":45.0,"Movement":45.0,"Front":20.0,"Union":60.0,"Collective":60.0,"Alliance":20.0,"League":45.0,"Pact":20.0,"Liberation":45.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":20.0,"Circle":35.0,"Cabal":35.0,"Conclave":20.0,"Scholars":20.0,"Collegium":20.0,"Keepers":15.0,"Caravan":65.0,"Fleet":100.0,"Band":60.0,"Enclave":60.0,"Commune":55.0,"Network":70.0,"Expedition":60.0,"Nomads":60.0,"Diaspora":60.0,"Confederation":40.0,"Assembly":55.0,"Coalition":40.0,"Horde":50.0,"Mercenaries":55.0,"Parliament":40.0,"Senate":40.0,"Association":60.0,"Compact":60.0},"place":{"Apothecary":30.0,"Archive":5.0,"Armoury":30.0,"Auction House":35.0,"Bank":40.0,"Bathhouse":25.0,"Bookshop":30.0,"Brothel":30.0,"Clinic":5.0,"Club":5.0,"Courthouse":5.0,"Bridge":70.0,"Crypt":5.0,"Dormitory":25.0,"Forge":30.0,"Gallery":25.0,"Gambling Den":30.0,"House":5.0,"Laboratory":5.0,"Library":0.0,"Mill":40.0,"Monument":0.0,"Mortuary":5.0,"Museum":0.0,"Observatory":5.0,"Office":45.0,"Pawnshop":35.0,"Post Office":65.0,"Printing House":5.0,"Prison":5.0,"School":0.0,"Sewer":30.0,"Shelter":25.0,"Ship":90.0,"Shop":35.0,"Shrine":5.0,"Stable":60.0,"Tailor":30.0,"Tavern":30.0,"Theater":25.0,"Tower":20.0,"Training Ground":5.0,"Treasury":30.0,"Tunnel":50.0,"Vault":40.0,"Guardhouse":40.0,"Well":35.0,"Arena":25.0,"Barracks":5.0,"Black Market":35.0,"Brewery":30.0,"Camp":45.0,"Cemetery":5.0,"Dock":100.0,"Embassy":25.0,"Factory":25.0,"Farm":5.0,"Fortress":5.0,"Garden":0.0,"Grove":5.0,"Meeting Hall":5.0,"Hideout":5.0,"Inn":50.0,"Manor":5.0,"Market":40.0,"Mine":5.0,"Outpost":45.0,"Palace":5.0,"Research Facility":5.0,"Square":25.0,"Temple":0.0,"Trading Post":75.0,"Warehouse":90.0,"Workshop":30.0,"Catacombs":5.0,"Town Hall":5.0,"Excavation Site":5.0,"Festival Grounds":30.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":5.0,"Shipyard":100.0,"Caravan Camp":50.0,"Cavern":0.0,"Citadel":0.0,"Enclave":30.0,"Estate":0.0,"Village":0.0,"Hunting Grounds":5.0,"Military Base":30.0,"Mining Settlement":0.0,"Monastery":0.0,"Necropolis":0.0,"Rebel Camp":0.0,"Refugee Camp":30.0,"Slum":0.0,"University":5.0,"Academic District":0.0,"Government District":0.0,"Artisan Quarter":10.0,"Cultural District":0.0,"Entertainment District":10.0,"Farmland":0.0,"Foreign Quarter":45.0,"Wilderness":0.0,"Harbor":100.0,"Industrial District":10.0,"Market District":15.0,"Military District":10.0,"Mining Region":0.0,"Noble Quarter":0.0,"Royal Quarter":0.0,"Residential District":0.0,"Ruined Quarter":0.0,"Temple District":0.0,"Underground City":10.0,"Underworld":10.0,"Warehouse District":90.0},"score":78.33333333333333},"Embassy":{"function":{"Authority":100.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":25.0,"Covert":0.0,"Migration":75.0,"Care":0.0,"Exploration":0.0},"type":{"Government":90.0,"Military":20.0,"Religious":20.0,"Criminal":15.0,"Revolutionary":25.0,"Conspiracy":50.0,"Merchant":45.0,"Clan":50.0,"Cultural":20.0,"Outsider":85.0,"Elite":80.0,"Poor":25.0,"Academic":15.0},"base":{"Empire":75.0,"Kingdom":75.0,"Republic":75.0,"Dominion":75.0,"Council":75.0,"Court":75.0,"Commonwealth":75.0,"Directorate":75.0,"Ministry":70.0,"Authority":75.0,"Legion":45.0,"Guard":65.0,"Army":45.0,"Militia":35.0,"Vanguard":35.0,"Watch":65.0,"Corps":45.0,"Wardens":65.0,"Rangers":50.0,"Order":45.0,"Covenant":45.0,"Faith":45.0,"Sect":20.0,"Brotherhood":50.0,"Fellowship":50.0,"Chapter":45.0,"Disciples":20.0,"Cult":45.0,"Guild":40.0,"Company":50.0,"Consortium":40.0,"Syndicate":60.0,"Cartel":50.0,"Cooperative":40.0,"Corporation":40.0,"Chamber":65.0,"Trust":70.0,"Clan":55.0,"House":70.0,"Dynasty":70.0,"Family":55.0,"Tribe":60.0,"People":40.0,"Community":40.0,"Nation":30.0,"Kindred":55.0,"Lineage":60.0,"Resistance":35.0,"Movement":30.0,"Front":35.0,"Union":30.0,"Collective":30.0,"Alliance":45.0,"League":45.0,"Pact":50.0,"Liberation":35.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":55.0,"Circle":65.0,"Cabal":55.0,"Conclave":55.0,"Scholars":20.0,"Collegium":20.0,"Keepers":35.0,"Caravan":70.0,"Fleet":70.0,"Band":60.0,"Enclave":70.0,"Commune":60.0,"Network":70.0,"Expedition":60.0,"Nomads":60.0,"Diaspora":70.0,"Confederation":70.0,"Assembly":70.0,"Coalition":60.0,"Horde":50.0,"Mercenaries":50.0,"Parliament":75.0,"Senate":75.0,"Association":50.0,"Compact":60.0},"place":{"Apothecary":5.0,"Archive":65.0,"Armoury":5.0,"Auction House":25.0,"Bank":25.0,"Bathhouse":10.0,"Bookshop":5.0,"Brothel":10.0,"Clinic":5.0,"Club":25.0,"Courthouse":60.0,"Bridge":30.0,"Crypt":20.0,"Dormitory":25.0,"Forge":5.0,"Gallery":25.0,"Gambling Den":25.0,"House":25.0,"Laboratory":5.0,"Library":5.0,"Mill":5.0,"Monument":50.0,"Mortuary":20.0,"Museum":25.0,"Observatory":5.0,"Office":70.0,"Pawnshop":5.0,"Post Office":45.0,"Printing House":20.0,"Prison":50.0,"School":5.0,"Sewer":5.0,"Shelter":25.0,"Ship":30.0,"Shop":5.0,"Shrine":5.0,"Stable":5.0,"Tailor":20.0,"Tavern":25.0,"Theater":25.0,"Tower":5.0,"Training Ground":5.0,"Treasury":60.0,"Tunnel":5.0,"Vault":25.0,"Guardhouse":45.0,"Well":5.0,"Arena":5.0,"Barracks":20.0,"Black Market":5.0,"Brewery":5.0,"Camp":30.0,"Cemetery":5.0,"Dock":25.0,"Embassy":100.0,"Factory":5.0,"Farm":5.0,"Fortress":30.0,"Garden":45.0,"Grove":5.0,"Meeting Hall":30.0,"Hideout":5.0,"Inn":30.0,"Manor":45.0,"Market":30.0,"Mine":5.0,"Outpost":10.0,"Palace":55.0,"Research Facility":5.0,"Square":30.0,"Temple":30.0,"Trading Post":25.0,"Warehouse":5.0,"Workshop":5.0,"Catacombs":5.0,"Town Hall":60.0,"Excavation Site":5.0,"Festival Grounds":10.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":5.0,"Shipyard":5.0,"Caravan Camp":40.0,"Cavern":20.0,"Citadel":60.0,"Enclave":50.0,"Estate":30.0,"Village":20.0,"Hunting Grounds":5.0,"Military Base":5.0,"Mining Settlement":20.0,"Monastery":15.0,"Necropolis":15.0,"Rebel Camp":20.0,"Refugee Camp":40.0,"Slum":20.0,"University":5.0,"Academic District":10.0,"Government District":85.0,"Artisan Quarter":10.0,"Cultural District":15.0,"Entertainment District":15.0,"Farmland":10.0,"Foreign Quarter":45.0,"Wilderness":10.0,"Harbor":20.0,"Industrial District":10.0,"Market District":15.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":30.0,"Royal Quarter":80.0,"Residential District":15.0,"Ruined Quarter":10.0,"Temple District":10.0,"Underground City":15.0,"Underworld":15.0,"Warehouse District":10.0},"score":66.66666666666667},"Factory":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":100.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":15.0,"Military":45.0,"Religious":15.0,"Criminal":15.0,"Revolutionary":45.0,"Conspiracy":15.0,"Merchant":75.0,"Clan":15.0,"Cultural":15.0,"Outsider":15.0,"Elite":15.0,"Poor":65.0,"Academic":15.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":35.0,"Dominion":20.0,"Council":20.0,"Court":20.0,"Commonwealth":35.0,"Directorate":20.0,"Ministry":15.0,"Authority":20.0,"Legion":40.0,"Guard":40.0,"Army":40.0,"Militia":40.0,"Vanguard":40.0,"Watch":40.0,"Corps":40.0,"Wardens":40.0,"Rangers":40.0,"Order":35.0,"Covenant":20.0,"Faith":20.0,"Sect":20.0,"Brotherhood":50.0,"Fellowship":45.0,"Chapter":40.0,"Disciples":20.0,"Cult":20.0,"Guild":70.0,"Company":90.0,"Consortium":70.0,"Syndicate":60.0,"Cartel":50.0,"Cooperative":70.0,"Corporation":90.0,"Chamber":70.0,"Trust":60.0,"Clan":20.0,"House":20.0,"Dynasty":20.0,"Family":20.0,"Tribe":20.0,"People":40.0,"Community":55.0,"Nation":25.0,"Kindred":20.0,"Lineage":20.0,"Resistance":50.0,"Movement":50.0,"Front":40.0,"Union":90.0,"Collective":65.0,"Alliance":40.0,"League":50.0,"Pact":40.0,"Liberation":50.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":20.0,"Circle":20.0,"Cabal":35.0,"Conclave":35.0,"Scholars":20.0,"Collegium":20.0,"Keepers":15.0,"Caravan":60.0,"Fleet":70.0,"Band":35.0,"Enclave":35.0,"Commune":50.0,"Network":45.0,"Expedition":45.0,"Nomads":35.0,"Diaspora":35.0,"Confederation":30.0,"Assembly":40.0,"Coalition":30.0,"Horde":40.0,"Mercenaries":50.0,"Parliament":20.0,"Senate":20.0,"Association":70.0,"Compact":45.0},"place":{"Apothecary":25.0,"Archive":5.0,"Armoury":30.0,"Auction House":20.0,"Bank":20.0,"Bathhouse":20.0,"Bookshop":20.0,"Brothel":20.0,"Clinic":20.0,"Club":5.0,"Courthouse":5.0,"Bridge":25.0,"Crypt":5.0,"Dormitory":5.0,"Forge":80.0,"Gallery":20.0,"Gambling Den":20.0,"House":5.0,"Laboratory":25.0,"Library":5.0,"Mill":60.0,"Monument":5.0,"Mortuary":5.0,"Museum":0.0,"Observatory":5.0,"Office":40.0,"Pawnshop":25.0,"Post Office":20.0,"Printing House":55.0,"Prison":5.0,"School":5.0,"Sewer":5.0,"Shelter":20.0,"Ship":25.0,"Shop":30.0,"Shrine":5.0,"Stable":5.0,"Tailor":60.0,"Tavern":25.0,"Theater":20.0,"Tower":5.0,"Training Ground":5.0,"Treasury":20.0,"Tunnel":5.0,"Vault":5.0,"Guardhouse":5.0,"Well":20.0,"Arena":20.0,"Barracks":5.0,"Black Market":20.0,"Brewery":60.0,"Camp":5.0,"Cemetery":5.0,"Dock":25.0,"Embassy":5.0,"Factory":100.0,"Farm":35.0,"Fortress":5.0,"Garden":5.0,"Grove":5.0,"Meeting Hall":5.0,"Hideout":5.0,"Inn":20.0,"Manor":5.0,"Market":25.0,"Mine":35.0,"Outpost":5.0,"Palace":5.0,"Research Facility":5.0,"Square":25.0,"Temple":5.0,"Trading Post":25.0,"Warehouse":45.0,"Workshop":80.0,"Catacombs":5.0,"Town Hall":5.0,"Excavation Site":10.0,"Festival Grounds":20.0,"Hospital":20.0,"Ruins":5.0,"Sanctuary":5.0,"Shipyard":60.0,"Caravan Camp":25.0,"Cavern":15.0,"Citadel":15.0,"Enclave":15.0,"Estate":15.0,"Village":20.0,"Hunting Grounds":5.0,"Military Base":5.0,"Mining Settlement":50.0,"Monastery":15.0,"Necropolis":15.0,"Rebel Camp":20.0,"Refugee Camp":25.0,"Slum":25.0,"University":5.0,"Academic District":10.0,"Government District":10.0,"Artisan Quarter":85.0,"Cultural District":10.0,"Entertainment District":20.0,"Farmland":20.0,"Foreign Quarter":10.0,"Wilderness":10.0,"Harbor":20.0,"Industrial District":100.0,"Market District":20.0,"Military District":10.0,"Mining Region":20.0,"Noble Quarter":10.0,"Royal Quarter":10.0,"Residential District":10.0,"Ruined Quarter":10.0,"Temple District":20.0,"Underground City":20.0,"Underworld":10.0,"Warehouse District":40.0},"score":61.666666666666664},"Farm":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":75.0,"Logistics":0.0,"Resources":100.0,"Residence":50.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":15.0,"Military":15.0,"Religious":45.0,"Criminal":15.0,"Revolutionary":20.0,"Conspiracy":15.0,"Merchant":75.0,"Clan":80.0,"Cultural":15.0,"Outsider":50.0,"Elite":25.0,"Poor":80.0,"Academic":15.0},"base":{"Empire":20.0,"Kingdom":45.0,"Republic":20.0,"Dominion":20.0,"Council":20.0,"Court":30.0,"Commonwealth":20.0,"Directorate":20.0,"Ministry":30.0,"Authority":20.0,"Legion":20.0,"Guard":20.0,"Army":20.0,"Militia":20.0,"Vanguard":20.0,"Watch":20.0,"Corps":20.0,"Wardens":20.0,"Rangers":35.0,"Order":40.0,"Covenant":50.0,"Faith":40.0,"Sect":40.0,"Brotherhood":60.0,"Fellowship":50.0,"Chapter":40.0,"Disciples":40.0,"Cult":40.0,"Guild":60.0,"Company":60.0,"Consortium":60.0,"Syndicate":60.0,"Cartel":45.0,"Cooperative":95.0,"Corporation":60.0,"Chamber":60.0,"Trust":60.0,"Clan":60.0,"House":70.0,"Dynasty":70.0,"Family":70.0,"Tribe":60.0,"People":70.0,"Community":70.0,"Nation":55.0,"Kindred":60.0,"Lineage":70.0,"Resistance":45.0,"Movement":45.0,"Front":20.0,"Union":60.0,"Collective":60.0,"Alliance":20.0,"League":45.0,"Pact":20.0,"Liberation":45.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":20.0,"Circle":20.0,"Cabal":20.0,"Conclave":35.0,"Scholars":20.0,"Collegium":20.0,"Keepers":40.0,"Caravan":60.0,"Fleet":50.0,"Band":40.0,"Enclave":50.0,"Commune":85.0,"Network":35.0,"Expedition":45.0,"Nomads":45.0,"Diaspora":45.0,"Confederation":40.0,"Assembly":50.0,"Coalition":35.0,"Horde":35.0,"Mercenaries":40.0,"Parliament":35.0,"Senate":35.0,"Association":60.0,"Compact":40.0},"place":{"Apothecary":25.0,"Archive":5.0,"Armoury":25.0,"Auction House":5.0,"Bank":5.0,"Bathhouse":20.0,"Bookshop":5.0,"Brothel":5.0,"Clinic":20.0,"Club":5.0,"Courthouse":5.0,"Bridge":5.0,"Crypt":5.0,"Dormitory":30.0,"Forge":35.0,"Gallery":5.0,"Gambling Den":5.0,"House":50.0,"Laboratory":25.0,"Library":5.0,"Mill":80.0,"Monument":5.0,"Mortuary":5.0,"Museum":5.0,"Observatory":5.0,"Office":5.0,"Pawnshop":5.0,"Post Office":5.0,"Printing House":30.0,"Prison":5.0,"School":20.0,"Sewer":5.0,"Shelter":30.0,"Ship":5.0,"Shop":25.0,"Shrine":5.0,"Stable":60.0,"Tailor":30.0,"Tavern":25.0,"Theater":5.0,"Tower":5.0,"Training Ground":5.0,"Treasury":5.0,"Tunnel":5.0,"Vault":5.0,"Guardhouse":5.0,"Well":75.0,"Arena":5.0,"Barracks":25.0,"Black Market":5.0,"Brewery":55.0,"Camp":30.0,"Cemetery":20.0,"Dock":5.0,"Embassy":5.0,"Factory":35.0,"Farm":100.0,"Fortress":20.0,"Garden":35.0,"Grove":45.0,"Meeting Hall":20.0,"Hideout":20.0,"Inn":30.0,"Manor":30.0,"Market":5.0,"Mine":45.0,"Outpost":5.0,"Palace":25.0,"Research Facility":5.0,"Square":20.0,"Temple":5.0,"Trading Post":5.0,"Warehouse":25.0,"Workshop":55.0,"Catacombs":5.0,"Town Hall":5.0,"Excavation Site":30.0,"Festival Grounds":5.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":25.0,"Shipyard":30.0,"Caravan Camp":5.0,"Cavern":5.0,"Citadel":0.0,"Enclave":20.0,"Estate":70.0,"Village":90.0,"Hunting Grounds":55.0,"Military Base":25.0,"Mining Settlement":35.0,"Monastery":45.0,"Necropolis":0.0,"Rebel Camp":5.0,"Refugee Camp":10.0,"Slum":20.0,"University":25.0,"Academic District":0.0,"Government District":0.0,"Artisan Quarter":15.0,"Cultural District":0.0,"Entertainment District":0.0,"Farmland":100.0,"Foreign Quarter":0.0,"Wilderness":75.0,"Harbor":0.0,"Industrial District":20.0,"Market District":0.0,"Military District":0.0,"Mining Region":25.0,"Noble Quarter":10.0,"Royal Quarter":0.0,"Residential District":15.0,"Ruined Quarter":0.0,"Temple District":0.0,"Underground City":15.0,"Underworld":0.0,"Warehouse District":0.0},"score":88.33333333333333},"Fortress":{"function":{"Authority":50.0,"Defense":100.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":25.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":80.0,"Military":90.0,"Religious":15.0,"Criminal":15.0,"Revolutionary":80.0,"Conspiracy":20.0,"Merchant":15.0,"Clan":50.0,"Cultural":15.0,"Outsider":20.0,"Elite":55.0,"Poor":20.0,"Academic":15.0},"base":{"Empire":70.0,"Kingdom":70.0,"Republic":60.0,"Dominion":70.0,"Council":70.0,"Court":60.0,"Commonwealth":60.0,"Directorate":70.0,"Ministry":70.0,"Authority":70.0,"Legion":100.0,"Guard":100.0,"Army":100.0,"Militia":75.0,"Vanguard":100.0,"Watch":100.0,"Corps":100.0,"Wardens":95.0,"Rangers":95.0,"Order":65.0,"Covenant":55.0,"Faith":45.0,"Sect":35.0,"Brotherhood":40.0,"Fellowship":40.0,"Chapter":30.0,"Disciples":35.0,"Cult":35.0,"Guild":20.0,"Company":20.0,"Consortium":20.0,"Syndicate":35.0,"Cartel":35.0,"Cooperative":35.0,"Corporation":20.0,"Chamber":50.0,"Trust":55.0,"Clan":65.0,"House":55.0,"Dynasty":55.0,"Family":45.0,"Tribe":60.0,"People":40.0,"Community":40.0,"Nation":35.0,"Kindred":65.0,"Lineage":50.0,"Resistance":70.0,"Movement":70.0,"Front":70.0,"Union":60.0,"Collective":60.0,"Alliance":70.0,"League":70.0,"Pact":70.0,"Liberation":70.0,"Academy":35.0,"Institute":35.0,"School":35.0,"Society":35.0,"Circle":35.0,"Cabal":35.0,"Conclave":35.0,"Scholars":35.0,"Collegium":35.0,"Keepers":40.0,"Caravan":20.0,"Fleet":65.0,"Band":45.0,"Enclave":45.0,"Commune":45.0,"Network":20.0,"Expedition":45.0,"Nomads":35.0,"Diaspora":35.0,"Confederation":70.0,"Assembly":70.0,"Coalition":70.0,"Horde":70.0,"Mercenaries":90.0,"Parliament":70.0,"Senate":70.0,"Association":20.0,"Compact":50.0},"place":{"Apothecary":0.0,"Archive":25.0,"Armoury":75.0,"Auction House":5.0,"Bank":25.0,"Bathhouse":20.0,"Bookshop":0.0,"Brothel":5.0,"Clinic":20.0,"Club":5.0,"Courthouse":35.0,"Bridge":25.0,"Crypt":5.0,"Dormitory":25.0,"Forge":25.0,"Gallery":5.0,"Gambling Den":5.0,"House":25.0,"Laboratory":5.0,"Library":5.0,"Mill":5.0,"Monument":25.0,"Mortuary":20.0,"Museum":5.0,"Observatory":25.0,"Office":50.0,"Pawnshop":0.0,"Post Office":25.0,"Printing House":20.0,"Prison":70.0,"School":20.0,"Sewer":5.0,"Shelter":25.0,"Ship":30.0,"Shop":5.0,"Shrine":5.0,"Stable":20.0,"Tailor":5.0,"Tavern":20.0,"Theater":5.0,"Tower":75.0,"Training Ground":80.0,"Treasury":30.0,"Tunnel":20.0,"Vault":45.0,"Guardhouse":80.0,"Well":20.0,"Arena":35.0,"Barracks":80.0,"Black Market":5.0,"Brewery":5.0,"Camp":30.0,"Cemetery":20.0,"Dock":5.0,"Embassy":30.0,"Factory":5.0,"Farm":20.0,"Fortress":100.0,"Garden":20.0,"Grove":5.0,"Meeting Hall":30.0,"Hideout":25.0,"Inn":25.0,"Manor":30.0,"Market":5.0,"Mine":5.0,"Outpost":60.0,"Palace":40.0,"Research Facility":5.0,"Square":25.0,"Temple":25.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":5.0,"Catacombs":5.0,"Town Hall":35.0,"Excavation Site":5.0,"Festival Grounds":5.0,"Hospital":5.0,"Ruins":40.0,"Sanctuary":25.0,"Shipyard":5.0,"Caravan Camp":20.0,"Cavern":15.0,"Citadel":100.0,"Enclave":30.0,"Estate":30.0,"Village":25.0,"Hunting Grounds":5.0,"Military Base":75.0,"Mining Settlement":25.0,"Monastery":25.0,"Necropolis":15.0,"Rebel Camp":75.0,"Refugee Camp":25.0,"Slum":20.0,"University":25.0,"Academic District":10.0,"Government District":40.0,"Artisan Quarter":10.0,"Cultural District":10.0,"Entertainment District":10.0,"Farmland":10.0,"Foreign Quarter":10.0,"Wilderness":10.0,"Harbor":15.0,"Industrial District":10.0,"Market District":10.0,"Military District":90.0,"Mining Region":10.0,"Noble Quarter":20.0,"Royal Quarter":40.0,"Residential District":20.0,"Ruined Quarter":10.0,"Temple District":10.0,"Underground City":20.0,"Underworld":10.0,"Warehouse District":10.0},"score":91.66666666666667},"Garden":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":100.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":45.0,"Military":15.0,"Religious":50.0,"Criminal":20.0,"Revolutionary":15.0,"Conspiracy":15.0,"Merchant":15.0,"Clan":30.0,"Cultural":90.0,"Outsider":15.0,"Elite":65.0,"Poor":15.0,"Academic":45.0},"base":{"Empire":60.0,"Kingdom":60.0,"Republic":60.0,"Dominion":60.0,"Council":50.0,"Court":65.0,"Commonwealth":60.0,"Directorate":50.0,"Ministry":60.0,"Authority":60.0,"Legion":35.0,"Guard":40.0,"Army":35.0,"Militia":35.0,"Vanguard":35.0,"Watch":40.0,"Corps":35.0,"Wardens":40.0,"Rangers":30.0,"Order":65.0,"Covenant":65.0,"Faith":65.0,"Sect":65.0,"Brotherhood":70.0,"Fellowship":70.0,"Chapter":60.0,"Disciples":60.0,"Cult":60.0,"Guild":20.0,"Company":20.0,"Consortium":20.0,"Syndicate":35.0,"Cartel":35.0,"Cooperative":35.0,"Corporation":20.0,"Chamber":50.0,"Trust":65.0,"Clan":35.0,"House":70.0,"Dynasty":70.0,"Family":35.0,"Tribe":35.0,"People":35.0,"Community":40.0,"Nation":35.0,"Kindred":35.0,"Lineage":70.0,"Resistance":35.0,"Movement":45.0,"Front":35.0,"Union":35.0,"Collective":35.0,"Alliance":35.0,"League":30.0,"Pact":30.0,"Liberation":35.0,"Academy":60.0,"Institute":60.0,"School":60.0,"Society":75.0,"Circle":60.0,"Cabal":40.0,"Conclave":50.0,"Scholars":60.0,"Collegium":60.0,"Keepers":60.0,"Caravan":20.0,"Fleet":20.0,"Band":35.0,"Enclave":60.0,"Commune":40.0,"Network":20.0,"Expedition":40.0,"Nomads":35.0,"Diaspora":45.0,"Confederation":55.0,"Assembly":50.0,"Coalition":40.0,"Horde":30.0,"Mercenaries":30.0,"Parliament":60.0,"Senate":65.0,"Association":55.0,"Compact":40.0},"place":{"Apothecary":5.0,"Archive":5.0,"Armoury":5.0,"Auction House":35.0,"Bank":20.0,"Bathhouse":65.0,"Bookshop":25.0,"Brothel":55.0,"Clinic":20.0,"Club":65.0,"Courthouse":20.0,"Bridge":5.0,"Crypt":20.0,"Dormitory":35.0,"Forge":5.0,"Gallery":65.0,"Gambling Den":55.0,"House":50.0,"Laboratory":5.0,"Library":30.0,"Mill":30.0,"Monument":80.0,"Mortuary":5.0,"Museum":55.0,"Observatory":5.0,"Office":5.0,"Pawnshop":5.0,"Post Office":5.0,"Printing House":25.0,"Prison":5.0,"School":30.0,"Sewer":5.0,"Shelter":35.0,"Ship":0.0,"Shop":5.0,"Shrine":45.0,"Stable":30.0,"Tailor":20.0,"Tavern":60.0,"Theater":70.0,"Tower":5.0,"Training Ground":20.0,"Treasury":20.0,"Tunnel":5.0,"Vault":20.0,"Guardhouse":5.0,"Well":60.0,"Arena":55.0,"Barracks":35.0,"Black Market":0.0,"Brewery":35.0,"Camp":35.0,"Cemetery":30.0,"Dock":0.0,"Embassy":45.0,"Factory":5.0,"Farm":35.0,"Fortress":20.0,"Garden":100.0,"Grove":60.0,"Meeting Hall":65.0,"Hideout":20.0,"Inn":50.0,"Manor":60.0,"Market":35.0,"Mine":5.0,"Outpost":5.0,"Palace":50.0,"Research Facility":5.0,"Square":70.0,"Temple":55.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":5.0,"Catacombs":5.0,"Town Hall":55.0,"Excavation Site":5.0,"Festival Grounds":60.0,"Hospital":40.0,"Ruins":5.0,"Sanctuary":40.0,"Shipyard":5.0,"Caravan Camp":15.0,"Cavern":15.0,"Citadel":15.0,"Enclave":55.0,"Estate":60.0,"Village":60.0,"Hunting Grounds":30.0,"Military Base":20.0,"Mining Settlement":45.0,"Monastery":40.0,"Necropolis":20.0,"Rebel Camp":20.0,"Refugee Camp":45.0,"Slum":50.0,"University":65.0,"Academic District":20.0,"Government District":15.0,"Artisan Quarter":15.0,"Cultural District":80.0,"Entertainment District":75.0,"Farmland":40.0,"Foreign Quarter":50.0,"Wilderness":40.0,"Harbor":10.0,"Industrial District":10.0,"Market District":25.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":70.0,"Royal Quarter":55.0,"Residential District":75.0,"Ruined Quarter":10.0,"Temple District":50.0,"Underground City":45.0,"Underworld":15.0,"Warehouse District":10.0},"score":80.0},"Grove":{"function":{"Authority":0.0,"Defense":0.0,"Religion":25.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":25.0,"Residence":0.0,"Social":25.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":50.0},"type":{"Government":25.0,"Military":15.0,"Religious":60.0,"Criminal":25.0,"Revolutionary":25.0,"Conspiracy":15.0,"Merchant":15.0,"Clan":70.0,"Cultural":45.0,"Outsider":45.0,"Elite":30.0,"Poor":45.0,"Academic":30.0},"base":{"Empire":40.0,"Kingdom":40.0,"Republic":45.0,"Dominion":40.0,"Council":35.0,"Court":40.0,"Commonwealth":45.0,"Directorate":35.0,"Ministry":45.0,"Authority":35.0,"Legion":30.0,"Guard":30.0,"Army":30.0,"Militia":30.0,"Vanguard":30.0,"Watch":30.0,"Corps":30.0,"Wardens":30.0,"Rangers":35.0,"Order":65.0,"Covenant":65.0,"Faith":65.0,"Sect":65.0,"Brotherhood":65.0,"Fellowship":65.0,"Chapter":60.0,"Disciples":65.0,"Cult":65.0,"Guild":20.0,"Company":30.0,"Consortium":20.0,"Syndicate":30.0,"Cartel":30.0,"Cooperative":55.0,"Corporation":20.0,"Chamber":35.0,"Trust":40.0,"Clan":65.0,"House":55.0,"Dynasty":55.0,"Family":65.0,"Tribe":90.0,"People":75.0,"Community":75.0,"Nation":70.0,"Kindred":65.0,"Lineage":55.0,"Resistance":35.0,"Movement":40.0,"Front":30.0,"Union":40.0,"Collective":40.0,"Alliance":35.0,"League":35.0,"Pact":35.0,"Liberation":35.0,"Academy":45.0,"Institute":45.0,"School":45.0,"Society":50.0,"Circle":45.0,"Cabal":40.0,"Conclave":45.0,"Scholars":45.0,"Collegium":45.0,"Keepers":55.0,"Caravan":35.0,"Fleet":35.0,"Band":40.0,"Enclave":55.0,"Commune":55.0,"Network":35.0,"Expedition":40.0,"Nomads":45.0,"Diaspora":50.0,"Confederation":50.0,"Assembly":40.0,"Coalition":35.0,"Horde":35.0,"Mercenaries":35.0,"Parliament":45.0,"Senate":45.0,"Association":35.0,"Compact":40.0},"place":{"Apothecary":5.0,"Archive":5.0,"Armoury":5.0,"Auction House":10.0,"Bank":5.0,"Bathhouse":25.0,"Bookshop":5.0,"Brothel":25.0,"Clinic":5.0,"Club":25.0,"Courthouse":5.0,"Bridge":5.0,"Crypt":40.0,"Dormitory":5.0,"Forge":5.0,"Gallery":25.0,"Gambling Den":20.0,"House":10.0,"Laboratory":5.0,"Library":5.0,"Mill":45.0,"Monument":50.0,"Mortuary":30.0,"Museum":10.0,"Observatory":25.0,"Office":5.0,"Pawnshop":5.0,"Post Office":5.0,"Printing House":5.0,"Prison":5.0,"School":10.0,"Sewer":30.0,"Shelter":5.0,"Ship":5.0,"Shop":5.0,"Shrine":55.0,"Stable":40.0,"Tailor":5.0,"Tavern":25.0,"Theater":25.0,"Tower":10.0,"Training Ground":5.0,"Treasury":5.0,"Tunnel":30.0,"Vault":10.0,"Guardhouse":5.0,"Well":70.0,"Arena":25.0,"Barracks":5.0,"Black Market":5.0,"Brewery":40.0,"Camp":5.0,"Cemetery":40.0,"Dock":5.0,"Embassy":5.0,"Factory":5.0,"Farm":45.0,"Fortress":5.0,"Garden":60.0,"Grove":100.0,"Meeting Hall":25.0,"Hideout":10.0,"Inn":10.0,"Manor":5.0,"Market":10.0,"Mine":30.0,"Outpost":10.0,"Palace":10.0,"Research Facility":5.0,"Square":25.0,"Temple":35.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":5.0,"Catacombs":40.0,"Town Hall":10.0,"Excavation Site":30.0,"Festival Grounds":25.0,"Hospital":25.0,"Ruins":35.0,"Sanctuary":50.0,"Shipyard":5.0,"Caravan Camp":5.0,"Cavern":35.0,"Citadel":5.0,"Enclave":5.0,"Estate":40.0,"Village":55.0,"Hunting Grounds":75.0,"Military Base":5.0,"Mining Settlement":20.0,"Monastery":30.0,"Necropolis":40.0,"Rebel Camp":5.0,"Refugee Camp":5.0,"Slum":10.0,"University":10.0,"Academic District":0.0,"Government District":0.0,"Artisan Quarter":0.0,"Cultural District":20.0,"Entertainment District":15.0,"Farmland":60.0,"Foreign Quarter":0.0,"Wilderness":90.0,"Harbor":0.0,"Industrial District":0.0,"Market District":0.0,"Military District":0.0,"Mining Region":20.0,"Noble Quarter":0.0,"Royal Quarter":0.0,"Residential District":5.0,"Ruined Quarter":25.0,"Temple District":30.0,"Underground City":15.0,"Underworld":0.0,"Warehouse District":0.0},"score":70.0},"Meeting Hall":{"function":{"Authority":25.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":100.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":70.0,"Military":45.0,"Religious":50.0,"Criminal":20.0,"Revolutionary":65.0,"Conspiracy":20.0,"Merchant":45.0,"Clan":50.0,"Cultural":90.0,"Outsider":15.0,"Elite":20.0,"Poor":15.0,"Academic":45.0},"base":{"Empire":70.0,"Kingdom":70.0,"Republic":95.0,"Dominion":70.0,"Council":90.0,"Court":70.0,"Commonwealth":95.0,"Directorate":95.0,"Ministry":90.0,"Authority":90.0,"Legion":60.0,"Guard":70.0,"Army":60.0,"Militia":55.0,"Vanguard":45.0,"Watch":65.0,"Corps":60.0,"Wardens":65.0,"Rangers":40.0,"Order":90.0,"Covenant":90.0,"Faith":85.0,"Sect":85.0,"Brotherhood":95.0,"Fellowship":95.0,"Chapter":85.0,"Disciples":85.0,"Cult":85.0,"Guild":85.0,"Company":85.0,"Consortium":85.0,"Syndicate":40.0,"Cartel":40.0,"Cooperative":85.0,"Corporation":85.0,"Chamber":90.0,"Trust":60.0,"Clan":60.0,"House":70.0,"Dynasty":70.0,"Family":45.0,"Tribe":50.0,"People":65.0,"Community":70.0,"Nation":55.0,"Kindred":45.0,"Lineage":70.0,"Resistance":90.0,"Movement":90.0,"Front":90.0,"Union":90.0,"Collective":90.0,"Alliance":90.0,"League":90.0,"Pact":90.0,"Liberation":90.0,"Academy":60.0,"Institute":60.0,"School":60.0,"Society":100.0,"Circle":85.0,"Cabal":85.0,"Conclave":85.0,"Scholars":60.0,"Collegium":60.0,"Keepers":85.0,"Caravan":40.0,"Fleet":40.0,"Band":35.0,"Enclave":70.0,"Commune":90.0,"Network":20.0,"Expedition":40.0,"Nomads":45.0,"Diaspora":55.0,"Confederation":90.0,"Assembly":90.0,"Coalition":90.0,"Horde":40.0,"Mercenaries":40.0,"Parliament":90.0,"Senate":90.0,"Association":95.0,"Compact":90.0},"place":{"Apothecary":5.0,"Archive":45.0,"Armoury":5.0,"Auction House":30.0,"Bank":20.0,"Bathhouse":55.0,"Bookshop":25.0,"Brothel":55.0,"Clinic":20.0,"Club":60.0,"Courthouse":45.0,"Bridge":5.0,"Crypt":5.0,"Dormitory":20.0,"Forge":5.0,"Gallery":60.0,"Gambling Den":50.0,"House":30.0,"Laboratory":5.0,"Library":30.0,"Mill":5.0,"Monument":80.0,"Mortuary":20.0,"Museum":50.0,"Observatory":5.0,"Office":45.0,"Pawnshop":5.0,"Post Office":25.0,"Printing House":30.0,"Prison":25.0,"School":30.0,"Sewer":5.0,"Shelter":20.0,"Ship":5.0,"Shop":5.0,"Shrine":25.0,"Stable":5.0,"Tailor":20.0,"Tavern":55.0,"Theater":60.0,"Tower":5.0,"Training Ground":25.0,"Treasury":25.0,"Tunnel":5.0,"Vault":25.0,"Guardhouse":25.0,"Well":25.0,"Arena":55.0,"Barracks":20.0,"Black Market":5.0,"Brewery":20.0,"Camp":20.0,"Cemetery":30.0,"Dock":5.0,"Embassy":30.0,"Factory":5.0,"Farm":20.0,"Fortress":30.0,"Garden":65.0,"Grove":25.0,"Meeting Hall":100.0,"Hideout":20.0,"Inn":30.0,"Manor":30.0,"Market":35.0,"Mine":5.0,"Outpost":5.0,"Palace":40.0,"Research Facility":5.0,"Square":65.0,"Temple":40.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":5.0,"Catacombs":5.0,"Town Hall":60.0,"Excavation Site":5.0,"Festival Grounds":55.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":20.0,"Shipyard":5.0,"Caravan Camp":15.0,"Cavern":15.0,"Citadel":45.0,"Enclave":40.0,"Estate":25.0,"Village":35.0,"Hunting Grounds":5.0,"Military Base":20.0,"Mining Settlement":25.0,"Monastery":25.0,"Necropolis":20.0,"Rebel Camp":25.0,"Refugee Camp":25.0,"Slum":25.0,"University":30.0,"Academic District":45.0,"Government District":45.0,"Artisan Quarter":15.0,"Cultural District":90.0,"Entertainment District":65.0,"Farmland":10.0,"Foreign Quarter":20.0,"Wilderness":10.0,"Harbor":10.0,"Industrial District":10.0,"Market District":25.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":15.0,"Royal Quarter":40.0,"Residential District":25.0,"Ruined Quarter":10.0,"Temple District":25.0,"Underground City":15.0,"Underworld":15.0,"Warehouse District":10.0},"score":80.0},"Hideout":{"function":{"Authority":0.0,"Defense":25.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":25.0,"Social":0.0,"Covert":100.0,"Migration":0.0,"Care":0.0,"Exploration":25.0},"type":{"Government":20.0,"Military":50.0,"Religious":15.0,"Criminal":95.0,"Revolutionary":90.0,"Conspiracy":95.0,"Merchant":15.0,"Clan":20.0,"Cultural":15.0,"Outsider":50.0,"Elite":20.0,"Poor":50.0,"Academic":20.0},"base":{"Empire":35.0,"Kingdom":35.0,"Republic":35.0,"Dominion":35.0,"Council":20.0,"Court":35.0,"Commonwealth":35.0,"Directorate":20.0,"Ministry":20.0,"Authority":20.0,"Legion":45.0,"Guard":45.0,"Army":45.0,"Militia":100.0,"Vanguard":75.0,"Watch":40.0,"Corps":45.0,"Wardens":40.0,"Rangers":40.0,"Order":55.0,"Covenant":35.0,"Faith":35.0,"Sect":45.0,"Brotherhood":55.0,"Fellowship":45.0,"Chapter":50.0,"Disciples":45.0,"Cult":90.0,"Guild":20.0,"Company":45.0,"Consortium":20.0,"Syndicate":100.0,"Cartel":100.0,"Cooperative":45.0,"Corporation":20.0,"Chamber":20.0,"Trust":35.0,"Clan":45.0,"House":35.0,"Dynasty":35.0,"Family":55.0,"Tribe":50.0,"People":55.0,"Community":60.0,"Nation":40.0,"Kindred":45.0,"Lineage":35.0,"Resistance":100.0,"Movement":100.0,"Front":100.0,"Union":100.0,"Collective":100.0,"Alliance":100.0,"League":95.0,"Pact":100.0,"Liberation":100.0,"Academy":35.0,"Institute":35.0,"School":35.0,"Society":80.0,"Circle":100.0,"Cabal":100.0,"Conclave":100.0,"Scholars":35.0,"Collegium":35.0,"Keepers":80.0,"Caravan":45.0,"Fleet":40.0,"Band":75.0,"Enclave":60.0,"Commune":70.0,"Network":100.0,"Expedition":45.0,"Nomads":65.0,"Diaspora":65.0,"Confederation":40.0,"Assembly":40.0,"Coalition":50.0,"Horde":85.0,"Mercenaries":85.0,"Parliament":35.0,"Senate":35.0,"Association":35.0,"Compact":50.0},"place":{"Apothecary":5.0,"Archive":25.0,"Armoury":45.0,"Auction House":5.0,"Bank":5.0,"Bathhouse":20.0,"Bookshop":5.0,"Brothel":25.0,"Clinic":20.0,"Club":25.0,"Courthouse":0.0,"Bridge":5.0,"Crypt":25.0,"Dormitory":25.0,"Forge":5.0,"Gallery":0.0,"Gambling Den":45.0,"House":40.0,"Laboratory":25.0,"Library":5.0,"Mill":5.0,"Monument":5.0,"Mortuary":5.0,"Museum":0.0,"Observatory":5.0,"Office":5.0,"Pawnshop":45.0,"Post Office":5.0,"Printing House":5.0,"Prison":25.0,"School":20.0,"Sewer":75.0,"Shelter":60.0,"Ship":5.0,"Shop":5.0,"Shrine":5.0,"Stable":5.0,"Tailor":5.0,"Tavern":25.0,"Theater":5.0,"Tower":30.0,"Training Ground":10.0,"Treasury":20.0,"Tunnel":65.0,"Vault":45.0,"Guardhouse":10.0,"Well":20.0,"Arena":5.0,"Barracks":30.0,"Black Market":80.0,"Brewery":5.0,"Camp":45.0,"Cemetery":20.0,"Dock":5.0,"Embassy":5.0,"Factory":5.0,"Farm":20.0,"Fortress":25.0,"Garden":20.0,"Grove":10.0,"Meeting Hall":20.0,"Hideout":100.0,"Inn":25.0,"Manor":25.0,"Market":5.0,"Mine":5.0,"Outpost":10.0,"Palace":25.0,"Research Facility":5.0,"Square":20.0,"Temple":5.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":5.0,"Catacombs":35.0,"Town Hall":5.0,"Excavation Site":10.0,"Festival Grounds":5.0,"Hospital":5.0,"Ruins":45.0,"Sanctuary":25.0,"Shipyard":5.0,"Caravan Camp":20.0,"Cavern":45.0,"Citadel":20.0,"Enclave":45.0,"Estate":25.0,"Village":25.0,"Hunting Grounds":5.0,"Military Base":25.0,"Mining Settlement":25.0,"Monastery":25.0,"Necropolis":25.0,"Rebel Camp":100.0,"Refugee Camp":45.0,"Slum":75.0,"University":25.0,"Academic District":10.0,"Government District":10.0,"Artisan Quarter":10.0,"Cultural District":10.0,"Entertainment District":10.0,"Farmland":10.0,"Foreign Quarter":15.0,"Wilderness":35.0,"Harbor":10.0,"Industrial District":10.0,"Market District":10.0,"Military District":15.0,"Mining Region":10.0,"Noble Quarter":15.0,"Royal Quarter":15.0,"Residential District":45.0,"Ruined Quarter":90.0,"Temple District":10.0,"Underground City":50.0,"Underworld":100.0,"Warehouse District":10.0},"score":88.33333333333333},"Inn":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":75.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":100.0,"Social":75.0,"Covert":0.0,"Migration":75.0,"Care":0.0,"Exploration":0.0},"type":{"Government":15.0,"Military":15.0,"Religious":20.0,"Criminal":25.0,"Revolutionary":25.0,"Conspiracy":15.0,"Merchant":85.0,"Clan":65.0,"Cultural":65.0,"Outsider":85.0,"Elite":30.0,"Poor":65.0,"Academic":15.0},"base":{"Empire":35.0,"Kingdom":35.0,"Republic":45.0,"Dominion":35.0,"Council":20.0,"Court":35.0,"Commonwealth":45.0,"Directorate":20.0,"Ministry":15.0,"Authority":20.0,"Legion":35.0,"Guard":35.0,"Army":35.0,"Militia":35.0,"Vanguard":35.0,"Watch":35.0,"Corps":35.0,"Wardens":35.0,"Rangers":50.0,"Order":35.0,"Covenant":35.0,"Faith":35.0,"Sect":35.0,"Brotherhood":55.0,"Fellowship":55.0,"Chapter":45.0,"Disciples":35.0,"Cult":35.0,"Guild":70.0,"Company":70.0,"Consortium":70.0,"Syndicate":70.0,"Cartel":60.0,"Cooperative":60.0,"Corporation":70.0,"Chamber":70.0,"Trust":70.0,"Clan":65.0,"House":45.0,"Dynasty":45.0,"Family":65.0,"Tribe":70.0,"People":65.0,"Community":65.0,"Nation":60.0,"Kindred":65.0,"Lineage":45.0,"Resistance":35.0,"Movement":45.0,"Front":35.0,"Union":55.0,"Collective":55.0,"Alliance":45.0,"League":55.0,"Pact":45.0,"Liberation":35.0,"Academy":35.0,"Institute":35.0,"School":35.0,"Society":55.0,"Circle":45.0,"Cabal":45.0,"Conclave":45.0,"Scholars":35.0,"Collegium":35.0,"Keepers":35.0,"Caravan":95.0,"Fleet":70.0,"Band":95.0,"Enclave":95.0,"Commune":70.0,"Network":60.0,"Expedition":95.0,"Nomads":95.0,"Diaspora":95.0,"Confederation":70.0,"Assembly":70.0,"Coalition":60.0,"Horde":50.0,"Mercenaries":50.0,"Parliament":45.0,"Senate":45.0,"Association":70.0,"Compact":60.0},"place":{"Apothecary":30.0,"Archive":5.0,"Armoury":5.0,"Auction House":35.0,"Bank":30.0,"Bathhouse":55.0,"Bookshop":30.0,"Brothel":40.0,"Clinic":20.0,"Club":25.0,"Courthouse":5.0,"Bridge":30.0,"Crypt":5.0,"Dormitory":55.0,"Forge":25.0,"Gallery":35.0,"Gambling Den":35.0,"House":60.0,"Laboratory":5.0,"Library":5.0,"Mill":25.0,"Monument":25.0,"Mortuary":5.0,"Museum":20.0,"Observatory":5.0,"Office":25.0,"Pawnshop":35.0,"Post Office":25.0,"Printing House":10.0,"Prison":5.0,"School":25.0,"Sewer":5.0,"Shelter":55.0,"Ship":50.0,"Shop":35.0,"Shrine":10.0,"Stable":40.0,"Tailor":30.0,"Tavern":65.0,"Theater":35.0,"Tower":0.0,"Training Ground":5.0,"Treasury":25.0,"Tunnel":5.0,"Vault":25.0,"Guardhouse":5.0,"Well":45.0,"Arena":30.0,"Barracks":45.0,"Black Market":30.0,"Brewery":30.0,"Camp":70.0,"Cemetery":25.0,"Dock":50.0,"Embassy":30.0,"Factory":20.0,"Farm":30.0,"Fortress":25.0,"Garden":50.0,"Grove":10.0,"Meeting Hall":30.0,"Hideout":25.0,"Inn":100.0,"Manor":55.0,"Market":40.0,"Mine":5.0,"Outpost":5.0,"Palace":30.0,"Research Facility":5.0,"Square":55.0,"Temple":20.0,"Trading Post":70.0,"Warehouse":25.0,"Workshop":30.0,"Catacombs":5.0,"Town Hall":25.0,"Excavation Site":5.0,"Festival Grounds":40.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":30.0,"Shipyard":25.0,"Caravan Camp":75.0,"Cavern":20.0,"Citadel":20.0,"Enclave":70.0,"Estate":55.0,"Village":85.0,"Hunting Grounds":5.0,"Military Base":25.0,"Mining Settlement":70.0,"Monastery":30.0,"Necropolis":15.0,"Rebel Camp":35.0,"Refugee Camp":70.0,"Slum":60.0,"University":45.0,"Academic District":15.0,"Government District":10.0,"Artisan Quarter":30.0,"Cultural District":20.0,"Entertainment District":35.0,"Farmland":15.0,"Foreign Quarter":90.0,"Wilderness":10.0,"Harbor":70.0,"Industrial District":10.0,"Market District":50.0,"Military District":10.0,"Mining Region":15.0,"Noble Quarter":55.0,"Royal Quarter":40.0,"Residential District":85.0,"Ruined Quarter":15.0,"Temple District":15.0,"Underground City":75.0,"Underworld":20.0,"Warehouse District":15.0},"score":85.0},"Manor":{"function":{"Authority":25.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":100.0,"Social":25.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":50.0,"Military":20.0,"Religious":20.0,"Criminal":15.0,"Revolutionary":25.0,"Conspiracy":45.0,"Merchant":45.0,"Clan":95.0,"Cultural":20.0,"Outsider":25.0,"Elite":90.0,"Poor":30.0,"Academic":15.0},"base":{"Empire":90.0,"Kingdom":90.0,"Republic":55.0,"Dominion":85.0,"Council":50.0,"Court":95.0,"Commonwealth":50.0,"Directorate":55.0,"Ministry":60.0,"Authority":60.0,"Legion":35.0,"Guard":45.0,"Army":35.0,"Militia":35.0,"Vanguard":35.0,"Watch":40.0,"Corps":35.0,"Wardens":40.0,"Rangers":35.0,"Order":35.0,"Covenant":45.0,"Faith":35.0,"Sect":35.0,"Brotherhood":60.0,"Fellowship":50.0,"Chapter":35.0,"Disciples":35.0,"Cult":40.0,"Guild":40.0,"Company":40.0,"Consortium":40.0,"Syndicate":50.0,"Cartel":50.0,"Cooperative":50.0,"Corporation":40.0,"Chamber":50.0,"Trust":70.0,"Clan":65.0,"House":95.0,"Dynasty":95.0,"Family":90.0,"Tribe":60.0,"People":60.0,"Community":60.0,"Nation":55.0,"Kindred":60.0,"Lineage":95.0,"Resistance":35.0,"Movement":45.0,"Front":35.0,"Union":50.0,"Collective":50.0,"Alliance":35.0,"League":35.0,"Pact":40.0,"Liberation":35.0,"Academy":35.0,"Institute":35.0,"School":35.0,"Society":55.0,"Circle":60.0,"Cabal":50.0,"Conclave":50.0,"Scholars":35.0,"Collegium":35.0,"Keepers":35.0,"Caravan":40.0,"Fleet":35.0,"Band":35.0,"Enclave":50.0,"Commune":40.0,"Network":40.0,"Expedition":45.0,"Nomads":35.0,"Diaspora":35.0,"Confederation":50.0,"Assembly":40.0,"Coalition":40.0,"Horde":35.0,"Mercenaries":40.0,"Parliament":60.0,"Senate":70.0,"Association":40.0,"Compact":40.0},"place":{"Apothecary":5.0,"Archive":25.0,"Armoury":5.0,"Auction House":40.0,"Bank":45.0,"Bathhouse":45.0,"Bookshop":5.0,"Brothel":10.0,"Clinic":20.0,"Club":45.0,"Courthouse":30.0,"Bridge":5.0,"Crypt":20.0,"Dormitory":60.0,"Forge":5.0,"Gallery":45.0,"Gambling Den":25.0,"House":80.0,"Laboratory":5.0,"Library":5.0,"Mill":5.0,"Monument":30.0,"Mortuary":20.0,"Museum":25.0,"Observatory":5.0,"Office":45.0,"Pawnshop":5.0,"Post Office":25.0,"Printing House":20.0,"Prison":30.0,"School":25.0,"Sewer":5.0,"Shelter":55.0,"Ship":5.0,"Shop":5.0,"Shrine":5.0,"Stable":40.0,"Tailor":20.0,"Tavern":30.0,"Theater":25.0,"Tower":5.0,"Training Ground":5.0,"Treasury":45.0,"Tunnel":5.0,"Vault":60.0,"Guardhouse":25.0,"Well":25.0,"Arena":10.0,"Barracks":55.0,"Black Market":5.0,"Brewery":5.0,"Camp":55.0,"Cemetery":25.0,"Dock":5.0,"Embassy":45.0,"Factory":5.0,"Farm":30.0,"Fortress":30.0,"Garden":60.0,"Grove":5.0,"Meeting Hall":30.0,"Hideout":25.0,"Inn":55.0,"Manor":100.0,"Market":10.0,"Mine":5.0,"Outpost":5.0,"Palace":75.0,"Research Facility":5.0,"Square":45.0,"Temple":25.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":5.0,"Catacombs":5.0,"Town Hall":30.0,"Excavation Site":5.0,"Festival Grounds":10.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":30.0,"Shipyard":5.0,"Caravan Camp":30.0,"Cavern":15.0,"Citadel":35.0,"Enclave":60.0,"Estate":90.0,"Village":75.0,"Hunting Grounds":5.0,"Military Base":30.0,"Mining Settlement":70.0,"Monastery":35.0,"Necropolis":20.0,"Rebel Camp":35.0,"Refugee Camp":55.0,"Slum":60.0,"University":50.0,"Academic District":10.0,"Government District":25.0,"Artisan Quarter":10.0,"Cultural District":15.0,"Entertainment District":15.0,"Farmland":10.0,"Foreign Quarter":55.0,"Wilderness":10.0,"Harbor":10.0,"Industrial District":10.0,"Market District":15.0,"Military District":15.0,"Mining Region":10.0,"Noble Quarter":100.0,"Royal Quarter":85.0,"Residential District":80.0,"Ruined Quarter":15.0,"Temple District":15.0,"Underground City":70.0,"Underworld":15.0,"Warehouse District":10.0},"score":88.33333333333333},"Market":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":100.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":75.0,"Covert":0.0,"Migration":25.0,"Care":0.0,"Exploration":0.0},"type":{"Government":45.0,"Military":15.0,"Religious":20.0,"Criminal":55.0,"Revolutionary":20.0,"Conspiracy":15.0,"Merchant":90.0,"Clan":55.0,"Cultural":35.0,"Outsider":70.0,"Elite":25.0,"Poor":70.0,"Academic":15.0},"base":{"Empire":40.0,"Kingdom":40.0,"Republic":60.0,"Dominion":40.0,"Council":40.0,"Court":40.0,"Commonwealth":60.0,"Directorate":40.0,"Ministry":40.0,"Authority":40.0,"Legion":20.0,"Guard":35.0,"Army":20.0,"Militia":20.0,"Vanguard":20.0,"Watch":35.0,"Corps":20.0,"Wardens":35.0,"Rangers":40.0,"Order":20.0,"Covenant":35.0,"Faith":20.0,"Sect":20.0,"Brotherhood":65.0,"Fellowship":55.0,"Chapter":45.0,"Disciples":20.0,"Cult":20.0,"Guild":75.0,"Company":75.0,"Consortium":75.0,"Syndicate":75.0,"Cartel":75.0,"Cooperative":100.0,"Corporation":75.0,"Chamber":75.0,"Trust":95.0,"Clan":45.0,"House":45.0,"Dynasty":45.0,"Family":45.0,"Tribe":45.0,"People":60.0,"Community":65.0,"Nation":50.0,"Kindred":45.0,"Lineage":45.0,"Resistance":45.0,"Movement":45.0,"Front":20.0,"Union":70.0,"Collective":70.0,"Alliance":45.0,"League":55.0,"Pact":45.0,"Liberation":45.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":40.0,"Circle":35.0,"Cabal":35.0,"Conclave":35.0,"Scholars":20.0,"Collegium":20.0,"Keepers":20.0,"Caravan":75.0,"Fleet":75.0,"Band":60.0,"Enclave":70.0,"Commune":90.0,"Network":70.0,"Expedition":60.0,"Nomads":70.0,"Diaspora":70.0,"Confederation":65.0,"Assembly":65.0,"Coalition":55.0,"Horde":45.0,"Mercenaries":55.0,"Parliament":45.0,"Senate":50.0,"Association":95.0,"Compact":95.0},"place":{"Apothecary":70.0,"Archive":5.0,"Armoury":5.0,"Auction House":80.0,"Bank":75.0,"Bathhouse":40.0,"Bookshop":75.0,"Brothel":65.0,"Clinic":5.0,"Club":50.0,"Courthouse":5.0,"Bridge":25.0,"Crypt":5.0,"Dormitory":25.0,"Forge":30.0,"Gallery":40.0,"Gambling Den":65.0,"House":25.0,"Laboratory":5.0,"Library":25.0,"Mill":25.0,"Monument":35.0,"Mortuary":5.0,"Museum":30.0,"Observatory":5.0,"Office":45.0,"Pawnshop":75.0,"Post Office":30.0,"Printing House":30.0,"Prison":5.0,"School":25.0,"Sewer":5.0,"Shelter":25.0,"Ship":30.0,"Shop":100.0,"Shrine":25.0,"Stable":5.0,"Tailor":70.0,"Tavern":80.0,"Theater":45.0,"Tower":5.0,"Training Ground":20.0,"Treasury":30.0,"Tunnel":5.0,"Vault":30.0,"Guardhouse":5.0,"Well":10.0,"Arena":35.0,"Barracks":5.0,"Black Market":50.0,"Brewery":40.0,"Camp":25.0,"Cemetery":25.0,"Dock":40.0,"Embassy":30.0,"Factory":25.0,"Farm":5.0,"Fortress":5.0,"Garden":35.0,"Grove":10.0,"Meeting Hall":35.0,"Hideout":5.0,"Inn":40.0,"Manor":10.0,"Market":100.0,"Mine":5.0,"Outpost":5.0,"Palace":10.0,"Research Facility":5.0,"Square":65.0,"Temple":30.0,"Trading Post":75.0,"Warehouse":30.0,"Workshop":30.0,"Catacombs":5.0,"Town Hall":50.0,"Excavation Site":5.0,"Festival Grounds":65.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":5.0,"Shipyard":25.0,"Caravan Camp":50.0,"Cavern":20.0,"Citadel":15.0,"Enclave":45.0,"Estate":20.0,"Village":55.0,"Hunting Grounds":5.0,"Military Base":5.0,"Mining Settlement":20.0,"Monastery":15.0,"Necropolis":15.0,"Rebel Camp":20.0,"Refugee Camp":30.0,"Slum":25.0,"University":10.0,"Academic District":15.0,"Government District":10.0,"Artisan Quarter":50.0,"Cultural District":30.0,"Entertainment District":50.0,"Farmland":15.0,"Foreign Quarter":90.0,"Wilderness":10.0,"Harbor":80.0,"Industrial District":10.0,"Market District":100.0,"Military District":10.0,"Mining Region":15.0,"Noble Quarter":15.0,"Royal Quarter":10.0,"Residential District":20.0,"Ruined Quarter":10.0,"Temple District":15.0,"Underground City":25.0,"Underworld":25.0,"Warehouse District":15.0},"score":86.66666666666667},"Mine":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":50.0,"Logistics":0.0,"Resources":100.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":25.0},"type":{"Government":45.0,"Military":45.0,"Religious":15.0,"Criminal":15.0,"Revolutionary":15.0,"Conspiracy":15.0,"Merchant":75.0,"Clan":25.0,"Cultural":15.0,"Outsider":50.0,"Elite":15.0,"Poor":80.0,"Academic":20.0},"base":{"Empire":40.0,"Kingdom":40.0,"Republic":40.0,"Dominion":40.0,"Council":40.0,"Court":40.0,"Commonwealth":40.0,"Directorate":40.0,"Ministry":40.0,"Authority":40.0,"Legion":40.0,"Guard":40.0,"Army":40.0,"Militia":40.0,"Vanguard":40.0,"Watch":40.0,"Corps":40.0,"Wardens":40.0,"Rangers":40.0,"Order":35.0,"Covenant":20.0,"Faith":20.0,"Sect":20.0,"Brotherhood":45.0,"Fellowship":30.0,"Chapter":20.0,"Disciples":20.0,"Cult":20.0,"Guild":60.0,"Company":60.0,"Consortium":60.0,"Syndicate":60.0,"Cartel":45.0,"Cooperative":60.0,"Corporation":60.0,"Chamber":60.0,"Trust":60.0,"Clan":30.0,"House":30.0,"Dynasty":30.0,"Family":30.0,"Tribe":35.0,"People":45.0,"Community":60.0,"Nation":25.0,"Kindred":30.0,"Lineage":30.0,"Resistance":45.0,"Movement":45.0,"Front":20.0,"Union":60.0,"Collective":60.0,"Alliance":20.0,"League":40.0,"Pact":20.0,"Liberation":45.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":20.0,"Circle":20.0,"Cabal":20.0,"Conclave":20.0,"Scholars":20.0,"Collegium":20.0,"Keepers":20.0,"Caravan":60.0,"Fleet":50.0,"Band":40.0,"Enclave":45.0,"Commune":50.0,"Network":35.0,"Expedition":65.0,"Nomads":45.0,"Diaspora":45.0,"Confederation":40.0,"Assembly":50.0,"Coalition":35.0,"Horde":40.0,"Mercenaries":50.0,"Parliament":40.0,"Senate":40.0,"Association":60.0,"Compact":40.0},"place":{"Apothecary":25.0,"Archive":5.0,"Armoury":25.0,"Auction House":5.0,"Bank":5.0,"Bathhouse":5.0,"Bookshop":5.0,"Brothel":5.0,"Clinic":25.0,"Club":5.0,"Courthouse":5.0,"Bridge":5.0,"Crypt":5.0,"Dormitory":5.0,"Forge":30.0,"Gallery":5.0,"Gambling Den":5.0,"House":5.0,"Laboratory":25.0,"Library":5.0,"Mill":45.0,"Monument":5.0,"Mortuary":5.0,"Museum":5.0,"Observatory":5.0,"Office":5.0,"Pawnshop":25.0,"Post Office":5.0,"Printing House":30.0,"Prison":5.0,"School":5.0,"Sewer":10.0,"Shelter":20.0,"Ship":5.0,"Shop":25.0,"Shrine":5.0,"Stable":10.0,"Tailor":30.0,"Tavern":5.0,"Theater":5.0,"Tower":5.0,"Training Ground":5.0,"Treasury":5.0,"Tunnel":90.0,"Vault":5.0,"Guardhouse":5.0,"Well":40.0,"Arena":5.0,"Barracks":5.0,"Black Market":5.0,"Brewery":35.0,"Camp":40.0,"Cemetery":5.0,"Dock":5.0,"Embassy":5.0,"Factory":35.0,"Farm":45.0,"Fortress":5.0,"Garden":5.0,"Grove":30.0,"Meeting Hall":5.0,"Hideout":5.0,"Inn":5.0,"Manor":5.0,"Market":5.0,"Mine":100.0,"Outpost":5.0,"Palace":5.0,"Research Facility":5.0,"Square":5.0,"Temple":5.0,"Trading Post":5.0,"Warehouse":45.0,"Workshop":50.0,"Catacombs":10.0,"Town Hall":5.0,"Excavation Site":75.0,"Festival Grounds":5.0,"Hospital":25.0,"Ruins":35.0,"Sanctuary":5.0,"Shipyard":30.0,"Caravan Camp":5.0,"Cavern":65.0,"Citadel":0.0,"Enclave":0.0,"Estate":20.0,"Village":25.0,"Hunting Grounds":45.0,"Military Base":5.0,"Mining Settlement":80.0,"Monastery":0.0,"Necropolis":0.0,"Rebel Camp":0.0,"Refugee Camp":0.0,"Slum":0.0,"University":5.0,"Academic District":0.0,"Government District":0.0,"Artisan Quarter":15.0,"Cultural District":0.0,"Entertainment District":0.0,"Farmland":30.0,"Foreign Quarter":0.0,"Wilderness":45.0,"Harbor":0.0,"Industrial District":15.0,"Market District":0.0,"Military District":0.0,"Mining Region":100.0,"Noble Quarter":0.0,"Royal Quarter":0.0,"Residential District":0.0,"Ruined Quarter":0.0,"Temple District":0.0,"Underground City":0.0,"Underworld":0.0,"Warehouse District":0.0},"score":85.0},"Outpost":{"function":{"Authority":0.0,"Defense":100.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":50.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":25.0,"Care":0.0,"Exploration":25.0},"type":{"Government":65.0,"Military":90.0,"Religious":25.0,"Criminal":15.0,"Revolutionary":40.0,"Conspiracy":25.0,"Merchant":30.0,"Clan":35.0,"Cultural":15.0,"Outsider":40.0,"Elite":15.0,"Poor":30.0,"Academic":30.0},"base":{"Empire":65.0,"Kingdom":65.0,"Republic":55.0,"Dominion":65.0,"Council":65.0,"Court":55.0,"Commonwealth":55.0,"Directorate":65.0,"Ministry":65.0,"Authority":65.0,"Legion":100.0,"Guard":85.0,"Army":100.0,"Militia":75.0,"Vanguard":100.0,"Watch":85.0,"Corps":100.0,"Wardens":85.0,"Rangers":95.0,"Order":70.0,"Covenant":55.0,"Faith":55.0,"Sect":40.0,"Brotherhood":45.0,"Fellowship":45.0,"Chapter":40.0,"Disciples":40.0,"Cult":40.0,"Guild":40.0,"Company":40.0,"Consortium":45.0,"Syndicate":35.0,"Cartel":35.0,"Cooperative":35.0,"Corporation":45.0,"Chamber":55.0,"Trust":40.0,"Clan":55.0,"House":40.0,"Dynasty":40.0,"Family":40.0,"Tribe":55.0,"People":40.0,"Community":40.0,"Nation":30.0,"Kindred":55.0,"Lineage":35.0,"Resistance":55.0,"Movement":55.0,"Front":90.0,"Union":35.0,"Collective":35.0,"Alliance":55.0,"League":55.0,"Pact":50.0,"Liberation":55.0,"Academy":40.0,"Institute":40.0,"School":40.0,"Society":40.0,"Circle":40.0,"Cabal":35.0,"Conclave":35.0,"Scholars":40.0,"Collegium":40.0,"Keepers":55.0,"Caravan":45.0,"Fleet":70.0,"Band":55.0,"Enclave":35.0,"Commune":35.0,"Network":35.0,"Expedition":55.0,"Nomads":40.0,"Diaspora":35.0,"Confederation":65.0,"Assembly":65.0,"Coalition":55.0,"Horde":75.0,"Mercenaries":90.0,"Parliament":65.0,"Senate":65.0,"Association":50.0,"Compact":45.0},"place":{"Apothecary":5.0,"Archive":5.0,"Armoury":80.0,"Auction House":5.0,"Bank":5.0,"Bathhouse":5.0,"Bookshop":5.0,"Brothel":5.0,"Clinic":5.0,"Club":5.0,"Courthouse":5.0,"Bridge":75.0,"Crypt":5.0,"Dormitory":25.0,"Forge":25.0,"Gallery":5.0,"Gambling Den":5.0,"House":5.0,"Laboratory":5.0,"Library":5.0,"Mill":5.0,"Monument":5.0,"Mortuary":5.0,"Museum":5.0,"Observatory":30.0,"Office":5.0,"Pawnshop":5.0,"Post Office":50.0,"Printing House":5.0,"Prison":55.0,"School":5.0,"Sewer":30.0,"Shelter":5.0,"Ship":55.0,"Shop":5.0,"Shrine":5.0,"Stable":50.0,"Tailor":5.0,"Tavern":5.0,"Theater":5.0,"Tower":75.0,"Training Ground":60.0,"Treasury":5.0,"Tunnel":50.0,"Vault":30.0,"Guardhouse":75.0,"Well":40.0,"Arena":35.0,"Barracks":60.0,"Black Market":20.0,"Brewery":5.0,"Camp":50.0,"Cemetery":5.0,"Dock":45.0,"Embassy":10.0,"Factory":5.0,"Farm":5.0,"Fortress":60.0,"Garden":5.0,"Grove":10.0,"Meeting Hall":5.0,"Hideout":10.0,"Inn":5.0,"Manor":5.0,"Market":5.0,"Mine":5.0,"Outpost":100.0,"Palace":10.0,"Research Facility":5.0,"Square":5.0,"Temple":5.0,"Trading Post":45.0,"Warehouse":50.0,"Workshop":5.0,"Catacombs":10.0,"Town Hall":5.0,"Excavation Site":10.0,"Festival Grounds":5.0,"Hospital":5.0,"Ruins":10.0,"Sanctuary":5.0,"Shipyard":45.0,"Caravan Camp":45.0,"Cavern":10.0,"Citadel":60.0,"Enclave":5.0,"Estate":5.0,"Village":5.0,"Hunting Grounds":45.0,"Military Base":65.0,"Mining Settlement":5.0,"Monastery":5.0,"Necropolis":5.0,"Rebel Camp":70.0,"Refugee Camp":5.0,"Slum":5.0,"University":5.0,"Academic District":0.0,"Government District":5.0,"Artisan Quarter":0.0,"Cultural District":0.0,"Entertainment District":0.0,"Farmland":0.0,"Foreign Quarter":0.0,"Wilderness":90.0,"Harbor":55.0,"Industrial District":0.0,"Market District":0.0,"Military District":75.0,"Mining Region":0.0,"Noble Quarter":0.0,"Royal Quarter":0.0,"Residential District":0.0,"Ruined Quarter":0.0,"Temple District":0.0,"Underground City":0.0,"Underworld":0.0,"Warehouse District":55.0},"score":90.0},"Palace":{"function":{"Authority":100.0,"Defense":25.0,"Religion":0.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":75.0,"Social":50.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":100.0,"Military":55.0,"Religious":50.0,"Criminal":20.0,"Revolutionary":25.0,"Conspiracy":20.0,"Merchant":15.0,"Clan":40.0,"Cultural":30.0,"Outsider":25.0,"Elite":100.0,"Poor":0.0,"Academic":15.0},"base":{"Empire":100.0,"Kingdom":100.0,"Republic":65.0,"Dominion":100.0,"Council":75.0,"Court":100.0,"Commonwealth":65.0,"Directorate":75.0,"Ministry":75.0,"Authority":75.0,"Legion":65.0,"Guard":75.0,"Army":65.0,"Militia":45.0,"Vanguard":45.0,"Watch":75.0,"Corps":65.0,"Wardens":75.0,"Rangers":45.0,"Order":65.0,"Covenant":65.0,"Faith":65.0,"Sect":45.0,"Brotherhood":60.0,"Fellowship":50.0,"Chapter":40.0,"Disciples":40.0,"Cult":45.0,"Guild":20.0,"Company":20.0,"Consortium":20.0,"Syndicate":35.0,"Cartel":35.0,"Cooperative":35.0,"Corporation":20.0,"Chamber":65.0,"Trust":65.0,"Clan":40.0,"House":100.0,"Dynasty":100.0,"Family":60.0,"Tribe":60.0,"People":60.0,"Community":60.0,"Nation":45.0,"Kindred":60.0,"Lineage":95.0,"Resistance":35.0,"Movement":35.0,"Front":35.0,"Union":35.0,"Collective":35.0,"Alliance":35.0,"League":35.0,"Pact":35.0,"Liberation":35.0,"Academy":35.0,"Institute":35.0,"School":35.0,"Society":45.0,"Circle":45.0,"Cabal":35.0,"Conclave":45.0,"Scholars":35.0,"Collegium":35.0,"Keepers":40.0,"Caravan":30.0,"Fleet":45.0,"Band":35.0,"Enclave":50.0,"Commune":30.0,"Network":30.0,"Expedition":35.0,"Nomads":35.0,"Diaspora":35.0,"Confederation":65.0,"Assembly":65.0,"Coalition":65.0,"Horde":45.0,"Mercenaries":45.0,"Parliament":75.0,"Senate":90.0,"Association":30.0,"Compact":55.0},"place":{"Apothecary":5.0,"Archive":45.0,"Armoury":10.0,"Auction House":40.0,"Bank":45.0,"Bathhouse":30.0,"Bookshop":5.0,"Brothel":20.0,"Clinic":20.0,"Club":45.0,"Courthouse":60.0,"Bridge":5.0,"Crypt":20.0,"Dormitory":30.0,"Forge":5.0,"Gallery":45.0,"Gambling Den":25.0,"House":40.0,"Laboratory":5.0,"Library":5.0,"Mill":0.0,"Monument":55.0,"Mortuary":20.0,"Museum":25.0,"Observatory":5.0,"Office":70.0,"Pawnshop":0.0,"Post Office":40.0,"Printing House":25.0,"Prison":55.0,"School":25.0,"Sewer":5.0,"Shelter":30.0,"Ship":5.0,"Shop":5.0,"Shrine":40.0,"Stable":5.0,"Tailor":20.0,"Tavern":30.0,"Theater":30.0,"Tower":10.0,"Training Ground":10.0,"Treasury":100.0,"Tunnel":5.0,"Vault":60.0,"Guardhouse":100.0,"Well":20.0,"Arena":25.0,"Barracks":35.0,"Black Market":0.0,"Brewery":5.0,"Camp":30.0,"Cemetery":25.0,"Dock":5.0,"Embassy":55.0,"Factory":5.0,"Farm":25.0,"Fortress":40.0,"Garden":50.0,"Grove":10.0,"Meeting Hall":40.0,"Hideout":25.0,"Inn":30.0,"Manor":75.0,"Market":10.0,"Mine":5.0,"Outpost":10.0,"Palace":100.0,"Research Facility":5.0,"Square":30.0,"Temple":30.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":5.0,"Catacombs":5.0,"Town Hall":60.0,"Excavation Site":5.0,"Festival Grounds":25.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":25.0,"Shipyard":5.0,"Caravan Camp":20.0,"Cavern":15.0,"Citadel":75.0,"Enclave":35.0,"Estate":65.0,"Village":35.0,"Hunting Grounds":5.0,"Military Base":30.0,"Mining Settlement":30.0,"Monastery":30.0,"Necropolis":15.0,"Rebel Camp":5.0,"Refugee Camp":30.0,"Slum":30.0,"University":30.0,"Academic District":15.0,"Government District":75.0,"Artisan Quarter":10.0,"Cultural District":20.0,"Entertainment District":15.0,"Farmland":10.0,"Foreign Quarter":20.0,"Wilderness":5.0,"Harbor":10.0,"Industrial District":10.0,"Market District":15.0,"Military District":20.0,"Mining Region":10.0,"Noble Quarter":85.0,"Royal Quarter":100.0,"Residential District":30.0,"Ruined Quarter":10.0,"Temple District":15.0,"Underground City":25.0,"Underworld":15.0,"Warehouse District":10.0},"score":95.0},"Research Facility":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":100.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":50.0,"Military":30.0,"Religious":40.0,"Criminal":15.0,"Revolutionary":15.0,"Conspiracy":50.0,"Merchant":25.0,"Clan":15.0,"Cultural":25.0,"Outsider":15.0,"Elite":15.0,"Poor":15.0,"Academic":90.0},"base":{"Empire":50.0,"Kingdom":40.0,"Republic":40.0,"Dominion":45.0,"Council":70.0,"Court":40.0,"Commonwealth":40.0,"Directorate":70.0,"Ministry":40.0,"Authority":40.0,"Legion":35.0,"Guard":50.0,"Army":35.0,"Militia":40.0,"Vanguard":40.0,"Watch":45.0,"Corps":35.0,"Wardens":45.0,"Rangers":40.0,"Order":60.0,"Covenant":55.0,"Faith":55.0,"Sect":55.0,"Brotherhood":55.0,"Fellowship":55.0,"Chapter":60.0,"Disciples":55.0,"Cult":60.0,"Guild":40.0,"Company":40.0,"Consortium":40.0,"Syndicate":30.0,"Cartel":30.0,"Cooperative":40.0,"Corporation":40.0,"Chamber":45.0,"Trust":30.0,"Clan":30.0,"House":20.0,"Dynasty":20.0,"Family":20.0,"Tribe":30.0,"People":15.0,"Community":20.0,"Nation":15.0,"Kindred":30.0,"Lineage":20.0,"Resistance":35.0,"Movement":30.0,"Front":35.0,"Union":30.0,"Collective":30.0,"Alliance":30.0,"League":35.0,"Pact":40.0,"Liberation":35.0,"Academy":100.0,"Institute":100.0,"School":100.0,"Society":80.0,"Circle":80.0,"Cabal":50.0,"Conclave":50.0,"Scholars":100.0,"Collegium":100.0,"Keepers":70.0,"Caravan":30.0,"Fleet":35.0,"Band":30.0,"Enclave":20.0,"Commune":15.0,"Network":40.0,"Expedition":50.0,"Nomads":20.0,"Diaspora":20.0,"Confederation":45.0,"Assembly":45.0,"Coalition":40.0,"Horde":40.0,"Mercenaries":40.0,"Parliament":40.0,"Senate":40.0,"Association":60.0,"Compact":35.0},"place":{"Apothecary":50.0,"Archive":80.0,"Armoury":5.0,"Auction House":5.0,"Bank":5.0,"Bathhouse":5.0,"Bookshop":55.0,"Brothel":5.0,"Clinic":55.0,"Club":5.0,"Courthouse":25.0,"Bridge":5.0,"Crypt":30.0,"Dormitory":25.0,"Forge":5.0,"Gallery":10.0,"Gambling Den":5.0,"House":0.0,"Laboratory":90.0,"Library":80.0,"Mill":5.0,"Monument":5.0,"Mortuary":35.0,"Museum":60.0,"Observatory":80.0,"Office":30.0,"Pawnshop":5.0,"Post Office":20.0,"Printing House":50.0,"Prison":5.0,"School":80.0,"Sewer":5.0,"Shelter":5.0,"Ship":5.0,"Shop":5.0,"Shrine":5.0,"Stable":5.0,"Tailor":5.0,"Tavern":5.0,"Theater":5.0,"Tower":35.0,"Training Ground":25.0,"Treasury":5.0,"Tunnel":5.0,"Vault":5.0,"Guardhouse":20.0,"Well":5.0,"Arena":5.0,"Barracks":5.0,"Black Market":5.0,"Brewery":5.0,"Camp":5.0,"Cemetery":25.0,"Dock":5.0,"Embassy":5.0,"Factory":5.0,"Farm":5.0,"Fortress":5.0,"Garden":5.0,"Grove":5.0,"Meeting Hall":5.0,"Hideout":5.0,"Inn":5.0,"Manor":5.0,"Market":5.0,"Mine":5.0,"Outpost":5.0,"Palace":5.0,"Research Facility":100.0,"Square":5.0,"Temple":5.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":25.0,"Catacombs":10.0,"Town Hall":5.0,"Excavation Site":35.0,"Festival Grounds":5.0,"Hospital":50.0,"Ruins":35.0,"Sanctuary":5.0,"Shipyard":5.0,"Caravan Camp":0.0,"Cavern":10.0,"Citadel":5.0,"Enclave":5.0,"Estate":5.0,"Village":0.0,"Hunting Grounds":10.0,"Military Base":5.0,"Mining Settlement":0.0,"Monastery":40.0,"Necropolis":25.0,"Rebel Camp":5.0,"Refugee Camp":0.0,"Slum":0.0,"University":75.0,"Academic District":90.0,"Government District":0.0,"Artisan Quarter":0.0,"Cultural District":5.0,"Entertainment District":0.0,"Farmland":0.0,"Foreign Quarter":0.0,"Wilderness":0.0,"Harbor":0.0,"Industrial District":0.0,"Market District":0.0,"Military District":0.0,"Mining Region":0.0,"Noble Quarter":0.0,"Royal Quarter":0.0,"Residential District":0.0,"Ruined Quarter":20.0,"Temple District":0.0,"Underground City":0.0,"Underworld":0.0,"Warehouse District":0.0},"score":53.333333333333336},"Square":{"function":{"Authority":25.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":50.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":100.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":70.0,"Military":20.0,"Religious":20.0,"Criminal":20.0,"Revolutionary":65.0,"Conspiracy":20.0,"Merchant":60.0,"Clan":35.0,"Cultural":90.0,"Outsider":15.0,"Elite":20.0,"Poor":65.0,"Academic":15.0},"base":{"Empire":60.0,"Kingdom":60.0,"Republic":70.0,"Dominion":60.0,"Council":70.0,"Court":70.0,"Commonwealth":70.0,"Directorate":70.0,"Ministry":65.0,"Authority":65.0,"Legion":35.0,"Guard":50.0,"Army":35.0,"Militia":55.0,"Vanguard":45.0,"Watch":50.0,"Corps":35.0,"Wardens":45.0,"Rangers":35.0,"Order":35.0,"Covenant":35.0,"Faith":35.0,"Sect":35.0,"Brotherhood":55.0,"Fellowship":55.0,"Chapter":35.0,"Disciples":35.0,"Cult":35.0,"Guild":65.0,"Company":65.0,"Consortium":65.0,"Syndicate":55.0,"Cartel":55.0,"Cooperative":55.0,"Corporation":65.0,"Chamber":65.0,"Trust":65.0,"Clan":35.0,"House":45.0,"Dynasty":45.0,"Family":40.0,"Tribe":35.0,"People":40.0,"Community":55.0,"Nation":85.0,"Kindred":35.0,"Lineage":55.0,"Resistance":90.0,"Movement":90.0,"Front":90.0,"Union":90.0,"Collective":90.0,"Alliance":90.0,"League":90.0,"Pact":90.0,"Liberation":90.0,"Academy":35.0,"Institute":35.0,"School":35.0,"Society":55.0,"Circle":35.0,"Cabal":45.0,"Conclave":45.0,"Scholars":35.0,"Collegium":35.0,"Keepers":35.0,"Caravan":65.0,"Fleet":45.0,"Band":35.0,"Enclave":55.0,"Commune":65.0,"Network":35.0,"Expedition":45.0,"Nomads":45.0,"Diaspora":45.0,"Confederation":65.0,"Assembly":65.0,"Coalition":65.0,"Horde":35.0,"Mercenaries":45.0,"Parliament":65.0,"Senate":70.0,"Association":70.0,"Compact":55.0},"place":{"Apothecary":30.0,"Archive":25.0,"Armoury":5.0,"Auction House":40.0,"Bank":30.0,"Bathhouse":70.0,"Bookshop":35.0,"Brothel":60.0,"Clinic":20.0,"Club":55.0,"Courthouse":25.0,"Bridge":20.0,"Crypt":5.0,"Dormitory":35.0,"Forge":25.0,"Gallery":65.0,"Gambling Den":60.0,"House":45.0,"Laboratory":5.0,"Library":25.0,"Mill":25.0,"Monument":80.0,"Mortuary":20.0,"Museum":50.0,"Observatory":5.0,"Office":30.0,"Pawnshop":30.0,"Post Office":25.0,"Printing House":30.0,"Prison":25.0,"School":25.0,"Sewer":5.0,"Shelter":40.0,"Ship":25.0,"Shop":30.0,"Shrine":45.0,"Stable":5.0,"Tailor":30.0,"Tavern":65.0,"Theater":65.0,"Tower":5.0,"Training Ground":20.0,"Treasury":30.0,"Tunnel":5.0,"Vault":30.0,"Guardhouse":25.0,"Well":45.0,"Arena":60.0,"Barracks":40.0,"Black Market":30.0,"Brewery":30.0,"Camp":40.0,"Cemetery":30.0,"Dock":25.0,"Embassy":30.0,"Factory":25.0,"Farm":20.0,"Fortress":25.0,"Garden":70.0,"Grove":25.0,"Meeting Hall":65.0,"Hideout":20.0,"Inn":55.0,"Manor":45.0,"Market":65.0,"Mine":5.0,"Outpost":5.0,"Palace":30.0,"Research Facility":5.0,"Square":100.0,"Temple":35.0,"Trading Post":30.0,"Warehouse":25.0,"Workshop":25.0,"Catacombs":5.0,"Town Hall":60.0,"Excavation Site":5.0,"Festival Grounds":80.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":20.0,"Shipyard":25.0,"Caravan Camp":25.0,"Cavern":15.0,"Citadel":25.0,"Enclave":40.0,"Estate":40.0,"Village":65.0,"Hunting Grounds":5.0,"Military Base":20.0,"Mining Settlement":55.0,"Monastery":25.0,"Necropolis":15.0,"Rebel Camp":25.0,"Refugee Camp":45.0,"Slum":50.0,"University":45.0,"Academic District":20.0,"Government District":45.0,"Artisan Quarter":30.0,"Cultural District":85.0,"Entertainment District":80.0,"Farmland":10.0,"Foreign Quarter":50.0,"Wilderness":10.0,"Harbor":25.0,"Industrial District":10.0,"Market District":55.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":45.0,"Royal Quarter":45.0,"Residential District":75.0,"Ruined Quarter":10.0,"Temple District":20.0,"Underground City":60.0,"Underworld":20.0,"Warehouse District":10.0},"score":86.66666666666667},"Temple":{"function":{"Authority":25.0,"Defense":0.0,"Religion":100.0,"Knowledge":0.0,"Commerce":0.0,"Production":0.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":75.0,"Covert":0.0,"Migration":0.0,"Care":25.0,"Exploration":0.0},"type":{"Government":50.0,"Military":20.0,"Religious":90.0,"Criminal":20.0,"Revolutionary":15.0,"Conspiracy":20.0,"Merchant":15.0,"Clan":55.0,"Cultural":65.0,"Outsider":15.0,"Elite":50.0,"Poor":15.0,"Academic":20.0},"base":{"Empire":65.0,"Kingdom":65.0,"Republic":65.0,"Dominion":60.0,"Council":40.0,"Court":65.0,"Commonwealth":60.0,"Directorate":45.0,"Ministry":60.0,"Authority":60.0,"Legion":35.0,"Guard":45.0,"Army":35.0,"Militia":35.0,"Vanguard":35.0,"Watch":40.0,"Corps":35.0,"Wardens":40.0,"Rangers":35.0,"Order":100.0,"Covenant":100.0,"Faith":100.0,"Sect":100.0,"Brotherhood":100.0,"Fellowship":95.0,"Chapter":95.0,"Disciples":100.0,"Cult":100.0,"Guild":20.0,"Company":20.0,"Consortium":20.0,"Syndicate":20.0,"Cartel":20.0,"Cooperative":20.0,"Corporation":20.0,"Chamber":40.0,"Trust":40.0,"Clan":65.0,"House":65.0,"Dynasty":65.0,"Family":65.0,"Tribe":65.0,"People":60.0,"Community":65.0,"Nation":60.0,"Kindred":65.0,"Lineage":65.0,"Resistance":35.0,"Movement":35.0,"Front":35.0,"Union":20.0,"Collective":20.0,"Alliance":45.0,"League":40.0,"Pact":45.0,"Liberation":35.0,"Academy":45.0,"Institute":45.0,"School":45.0,"Society":55.0,"Circle":45.0,"Cabal":45.0,"Conclave":55.0,"Scholars":45.0,"Collegium":45.0,"Keepers":90.0,"Caravan":20.0,"Fleet":20.0,"Band":20.0,"Enclave":65.0,"Commune":40.0,"Network":20.0,"Expedition":20.0,"Nomads":45.0,"Diaspora":55.0,"Confederation":40.0,"Assembly":40.0,"Coalition":40.0,"Horde":35.0,"Mercenaries":35.0,"Parliament":60.0,"Senate":60.0,"Association":40.0,"Compact":50.0},"place":{"Apothecary":5.0,"Archive":40.0,"Armoury":5.0,"Auction House":25.0,"Bank":20.0,"Bathhouse":35.0,"Bookshop":25.0,"Brothel":30.0,"Clinic":10.0,"Club":35.0,"Courthouse":25.0,"Bridge":5.0,"Crypt":70.0,"Dormitory":5.0,"Forge":5.0,"Gallery":35.0,"Gambling Den":30.0,"House":25.0,"Laboratory":5.0,"Library":45.0,"Mill":5.0,"Monument":45.0,"Mortuary":60.0,"Museum":30.0,"Observatory":20.0,"Office":25.0,"Pawnshop":5.0,"Post Office":20.0,"Printing House":25.0,"Prison":25.0,"School":25.0,"Sewer":5.0,"Shelter":5.0,"Ship":0.0,"Shop":5.0,"Shrine":80.0,"Stable":5.0,"Tailor":20.0,"Tavern":30.0,"Theater":35.0,"Tower":5.0,"Training Ground":20.0,"Treasury":25.0,"Tunnel":5.0,"Vault":20.0,"Guardhouse":25.0,"Well":10.0,"Arena":30.0,"Barracks":20.0,"Black Market":0.0,"Brewery":20.0,"Camp":5.0,"Cemetery":60.0,"Dock":0.0,"Embassy":30.0,"Factory":5.0,"Farm":5.0,"Fortress":25.0,"Garden":55.0,"Grove":35.0,"Meeting Hall":40.0,"Hideout":5.0,"Inn":20.0,"Manor":25.0,"Market":30.0,"Mine":5.0,"Outpost":5.0,"Palace":30.0,"Research Facility":5.0,"Square":35.0,"Temple":100.0,"Trading Post":5.0,"Warehouse":5.0,"Workshop":5.0,"Catacombs":45.0,"Town Hall":40.0,"Excavation Site":5.0,"Festival Grounds":35.0,"Hospital":25.0,"Ruins":5.0,"Sanctuary":70.0,"Shipyard":5.0,"Caravan Camp":15.0,"Cavern":15.0,"Citadel":25.0,"Enclave":40.0,"Estate":20.0,"Village":25.0,"Hunting Grounds":5.0,"Military Base":5.0,"Mining Settlement":15.0,"Monastery":70.0,"Necropolis":60.0,"Rebel Camp":15.0,"Refugee Camp":20.0,"Slum":20.0,"University":10.0,"Academic District":15.0,"Government District":25.0,"Artisan Quarter":15.0,"Cultural District":30.0,"Entertainment District":30.0,"Farmland":10.0,"Foreign Quarter":15.0,"Wilderness":10.0,"Harbor":10.0,"Industrial District":10.0,"Market District":15.0,"Military District":10.0,"Mining Region":10.0,"Noble Quarter":15.0,"Royal Quarter":40.0,"Residential District":40.0,"Ruined Quarter":35.0,"Temple District":100.0,"Underground City":10.0,"Underworld":15.0,"Warehouse District":10.0},"score":93.33333333333333},"Trading Post":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":100.0,"Production":0.0,"Logistics":75.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":50.0,"Care":0.0,"Exploration":0.0},"type":{"Government":30.0,"Military":30.0,"Religious":15.0,"Criminal":45.0,"Revolutionary":20.0,"Conspiracy":15.0,"Merchant":95.0,"Clan":35.0,"Cultural":20.0,"Outsider":75.0,"Elite":25.0,"Poor":55.0,"Academic":15.0},"base":{"Empire":35.0,"Kingdom":35.0,"Republic":50.0,"Dominion":35.0,"Council":35.0,"Court":35.0,"Commonwealth":50.0,"Directorate":35.0,"Ministry":35.0,"Authority":35.0,"Legion":35.0,"Guard":35.0,"Army":35.0,"Militia":35.0,"Vanguard":35.0,"Watch":35.0,"Corps":35.0,"Wardens":35.0,"Rangers":50.0,"Order":35.0,"Covenant":30.0,"Faith":30.0,"Sect":30.0,"Brotherhood":50.0,"Fellowship":50.0,"Chapter":40.0,"Disciples":30.0,"Cult":30.0,"Guild":75.0,"Company":90.0,"Consortium":80.0,"Syndicate":80.0,"Cartel":80.0,"Cooperative":70.0,"Corporation":80.0,"Chamber":80.0,"Trust":75.0,"Clan":50.0,"House":35.0,"Dynasty":35.0,"Family":50.0,"Tribe":55.0,"People":45.0,"Community":50.0,"Nation":35.0,"Kindred":45.0,"Lineage":35.0,"Resistance":35.0,"Movement":40.0,"Front":30.0,"Union":55.0,"Collective":55.0,"Alliance":45.0,"League":55.0,"Pact":45.0,"Liberation":35.0,"Academy":30.0,"Institute":30.0,"School":30.0,"Society":40.0,"Circle":40.0,"Cabal":40.0,"Conclave":40.0,"Scholars":30.0,"Collegium":30.0,"Keepers":25.0,"Caravan":90.0,"Fleet":80.0,"Band":75.0,"Enclave":70.0,"Commune":65.0,"Network":70.0,"Expedition":75.0,"Nomads":80.0,"Diaspora":70.0,"Confederation":60.0,"Assembly":60.0,"Coalition":50.0,"Horde":50.0,"Mercenaries":55.0,"Parliament":45.0,"Senate":45.0,"Association":70.0,"Compact":70.0},"place":{"Apothecary":50.0,"Archive":5.0,"Armoury":30.0,"Auction House":55.0,"Bank":60.0,"Bathhouse":30.0,"Bookshop":50.0,"Brothel":50.0,"Clinic":5.0,"Club":30.0,"Courthouse":5.0,"Bridge":75.0,"Crypt":5.0,"Dormitory":25.0,"Forge":30.0,"Gallery":25.0,"Gambling Den":50.0,"House":5.0,"Laboratory":5.0,"Library":5.0,"Mill":25.0,"Monument":5.0,"Mortuary":5.0,"Museum":5.0,"Observatory":5.0,"Office":65.0,"Pawnshop":55.0,"Post Office":55.0,"Printing House":10.0,"Prison":5.0,"School":5.0,"Sewer":30.0,"Shelter":25.0,"Ship":65.0,"Shop":75.0,"Shrine":5.0,"Stable":70.0,"Tailor":50.0,"Tavern":55.0,"Theater":25.0,"Tower":20.0,"Training Ground":5.0,"Treasury":30.0,"Tunnel":50.0,"Vault":40.0,"Guardhouse":5.0,"Well":5.0,"Arena":25.0,"Barracks":5.0,"Black Market":55.0,"Brewery":35.0,"Camp":45.0,"Cemetery":5.0,"Dock":75.0,"Embassy":25.0,"Factory":25.0,"Farm":5.0,"Fortress":5.0,"Garden":5.0,"Grove":5.0,"Meeting Hall":5.0,"Hideout":5.0,"Inn":70.0,"Manor":5.0,"Market":75.0,"Mine":5.0,"Outpost":45.0,"Palace":5.0,"Research Facility":5.0,"Square":30.0,"Temple":5.0,"Trading Post":100.0,"Warehouse":75.0,"Workshop":30.0,"Catacombs":5.0,"Town Hall":5.0,"Excavation Site":5.0,"Festival Grounds":35.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":5.0,"Shipyard":50.0,"Caravan Camp":90.0,"Cavern":5.0,"Citadel":5.0,"Enclave":40.0,"Estate":5.0,"Village":10.0,"Hunting Grounds":5.0,"Military Base":30.0,"Mining Settlement":10.0,"Monastery":5.0,"Necropolis":0.0,"Rebel Camp":5.0,"Refugee Camp":45.0,"Slum":5.0,"University":5.0,"Academic District":0.0,"Government District":0.0,"Artisan Quarter":30.0,"Cultural District":0.0,"Entertainment District":30.0,"Farmland":0.0,"Foreign Quarter":75.0,"Wilderness":0.0,"Harbor":90.0,"Industrial District":5.0,"Market District":90.0,"Military District":0.0,"Mining Region":0.0,"Noble Quarter":0.0,"Royal Quarter":0.0,"Residential District":0.0,"Ruined Quarter":0.0,"Temple District":0.0,"Underground City":25.0,"Underworld":20.0,"Warehouse District":80.0},"score":75.0},"Warehouse":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":0.0,"Commerce":25.0,"Production":25.0,"Logistics":100.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":45.0,"Military":75.0,"Religious":15.0,"Criminal":50.0,"Revolutionary":15.0,"Conspiracy":15.0,"Merchant":85.0,"Clan":15.0,"Cultural":15.0,"Outsider":65.0,"Elite":20.0,"Poor":45.0,"Academic":15.0},"base":{"Empire":40.0,"Kingdom":40.0,"Republic":60.0,"Dominion":40.0,"Council":40.0,"Court":40.0,"Commonwealth":60.0,"Directorate":40.0,"Ministry":40.0,"Authority":40.0,"Legion":60.0,"Guard":60.0,"Army":60.0,"Militia":60.0,"Vanguard":60.0,"Watch":60.0,"Corps":60.0,"Wardens":60.0,"Rangers":60.0,"Order":45.0,"Covenant":20.0,"Faith":20.0,"Sect":20.0,"Brotherhood":35.0,"Fellowship":35.0,"Chapter":30.0,"Disciples":20.0,"Cult":20.0,"Guild":95.0,"Company":95.0,"Consortium":95.0,"Syndicate":95.0,"Cartel":95.0,"Cooperative":70.0,"Corporation":95.0,"Chamber":95.0,"Trust":70.0,"Clan":20.0,"House":20.0,"Dynasty":20.0,"Family":35.0,"Tribe":40.0,"People":30.0,"Community":40.0,"Nation":15.0,"Kindred":20.0,"Lineage":20.0,"Resistance":35.0,"Movement":35.0,"Front":20.0,"Union":60.0,"Collective":60.0,"Alliance":45.0,"League":50.0,"Pact":40.0,"Liberation":35.0,"Academy":20.0,"Institute":20.0,"School":20.0,"Society":20.0,"Circle":35.0,"Cabal":35.0,"Conclave":35.0,"Scholars":20.0,"Collegium":20.0,"Keepers":15.0,"Caravan":70.0,"Fleet":95.0,"Band":65.0,"Enclave":55.0,"Commune":45.0,"Network":90.0,"Expedition":65.0,"Nomads":65.0,"Diaspora":55.0,"Confederation":65.0,"Assembly":65.0,"Coalition":55.0,"Horde":60.0,"Mercenaries":60.0,"Parliament":45.0,"Senate":45.0,"Association":70.0,"Compact":60.0},"place":{"Apothecary":25.0,"Archive":5.0,"Armoury":40.0,"Auction House":25.0,"Bank":35.0,"Bathhouse":25.0,"Bookshop":25.0,"Brothel":25.0,"Clinic":5.0,"Club":5.0,"Courthouse":5.0,"Bridge":70.0,"Crypt":5.0,"Dormitory":5.0,"Forge":35.0,"Gallery":20.0,"Gambling Den":25.0,"House":5.0,"Laboratory":25.0,"Library":5.0,"Mill":45.0,"Monument":5.0,"Mortuary":5.0,"Museum":5.0,"Observatory":5.0,"Office":45.0,"Pawnshop":30.0,"Post Office":65.0,"Printing House":25.0,"Prison":5.0,"School":5.0,"Sewer":35.0,"Shelter":5.0,"Ship":65.0,"Shop":35.0,"Shrine":5.0,"Stable":65.0,"Tailor":30.0,"Tavern":25.0,"Theater":20.0,"Tower":20.0,"Training Ground":5.0,"Treasury":30.0,"Tunnel":55.0,"Vault":55.0,"Guardhouse":5.0,"Well":30.0,"Arena":25.0,"Barracks":5.0,"Black Market":30.0,"Brewery":50.0,"Camp":5.0,"Cemetery":5.0,"Dock":90.0,"Embassy":5.0,"Factory":45.0,"Farm":25.0,"Fortress":5.0,"Garden":5.0,"Grove":5.0,"Meeting Hall":5.0,"Hideout":5.0,"Inn":25.0,"Manor":5.0,"Market":30.0,"Mine":45.0,"Outpost":50.0,"Palace":5.0,"Research Facility":5.0,"Square":25.0,"Temple":5.0,"Trading Post":75.0,"Warehouse":100.0,"Workshop":50.0,"Catacombs":5.0,"Town Hall":5.0,"Excavation Site":45.0,"Festival Grounds":25.0,"Hospital":5.0,"Ruins":5.0,"Sanctuary":5.0,"Shipyard":85.0,"Caravan Camp":75.0,"Cavern":20.0,"Citadel":20.0,"Enclave":15.0,"Estate":15.0,"Village":20.0,"Hunting Grounds":5.0,"Military Base":35.0,"Mining Settlement":45.0,"Monastery":15.0,"Necropolis":15.0,"Rebel Camp":20.0,"Refugee Camp":15.0,"Slum":15.0,"University":5.0,"Academic District":10.0,"Government District":10.0,"Artisan Quarter":25.0,"Cultural District":10.0,"Entertainment District":20.0,"Farmland":15.0,"Foreign Quarter":15.0,"Wilderness":10.0,"Harbor":90.0,"Industrial District":90.0,"Market District":45.0,"Military District":20.0,"Mining Region":15.0,"Noble Quarter":10.0,"Royal Quarter":10.0,"Residential District":10.0,"Ruined Quarter":10.0,"Temple District":10.0,"Underground City":20.0,"Underworld":15.0,"Warehouse District":100.0},"score":81.66666666666667},"Workshop":{"function":{"Authority":0.0,"Defense":0.0,"Religion":0.0,"Knowledge":25.0,"Commerce":50.0,"Production":100.0,"Logistics":0.0,"Resources":0.0,"Residence":0.0,"Social":0.0,"Covert":0.0,"Migration":0.0,"Care":0.0,"Exploration":0.0},"type":{"Government":15.0,"Military":45.0,"Religious":15.0,"Criminal":20.0,"Revolutionary":15.0,"Conspiracy":20.0,"Merchant":80.0,"Clan":45.0,"Cultural":65.0,"Outsider":15.0,"Elite":20.0,"Poor":65.0,"Academic":50.0},"base":{"Empire":20.0,"Kingdom":20.0,"Republic":45.0,"Dominion":20.0,"Council":35.0,"Court":20.0,"Commonwealth":45.0,"Directorate":35.0,"Ministry":15.0,"Authority":20.0,"Legion":40.0,"Guard":40.0,"Army":40.0,"Militia":40.0,"Vanguard":40.0,"Watch":40.0,"Corps":40.0,"Wardens":40.0,"Rangers":40.0,"Order":40.0,"Covenant":35.0,"Faith":35.0,"Sect":35.0,"Brotherhood":60.0,"Fellowship":60.0,"Chapter":50.0,"Disciples":35.0,"Cult":35.0,"Guild":95.0,"Company":95.0,"Consortium":95.0,"Syndicate":60.0,"Cartel":50.0,"Cooperative":95.0,"Corporation":95.0,"Chamber":95.0,"Trust":70.0,"Clan":50.0,"House":40.0,"Dynasty":40.0,"Family":40.0,"Tribe":50.0,"People":60.0,"Community":65.0,"Nation":55.0,"Kindred":50.0,"Lineage":50.0,"Resistance":45.0,"Movement":55.0,"Front":20.0,"Union":90.0,"Collective":65.0,"Alliance":45.0,"League":50.0,"Pact":45.0,"Liberation":45.0,"Academy":45.0,"Institute":45.0,"School":45.0,"Society":45.0,"Circle":45.0,"Cabal":45.0,"Conclave":45.0,"Scholars":45.0,"Collegium":45.0,"Keepers":40.0,"Caravan":70.0,"Fleet":50.0,"Band":35.0,"Enclave":50.0,"Commune":75.0,"Network":45.0,"Expedition":55.0,"Nomads":45.0,"Diaspora":45.0,"Confederation":60.0,"Assembly":50.0,"Coalition":40.0,"Horde":40.0,"Mercenaries":50.0,"Parliament":20.0,"Senate":20.0,"Association":70.0,"Compact":60.0},"place":{"Apothecary":40.0,"Archive":25.0,"Armoury":30.0,"Auction House":30.0,"Bank":30.0,"Bathhouse":25.0,"Bookshop":35.0,"Brothel":30.0,"Clinic":25.0,"Club":5.0,"Courthouse":20.0,"Bridge":25.0,"Crypt":20.0,"Dormitory":25.0,"Forge":80.0,"Gallery":25.0,"Gambling Den":30.0,"House":5.0,"Laboratory":35.0,"Library":25.0,"Mill":60.0,"Monument":5.0,"Mortuary":25.0,"Museum":25.0,"Observatory":25.0,"Office":45.0,"Pawnshop":30.0,"Post Office":25.0,"Printing House":65.0,"Prison":5.0,"School":25.0,"Sewer":5.0,"Shelter":5.0,"Ship":25.0,"Shop":55.0,"Shrine":5.0,"Stable":5.0,"Tailor":65.0,"Tavern":30.0,"Theater":25.0,"Tower":25.0,"Training Ground":20.0,"Treasury":25.0,"Tunnel":5.0,"Vault":25.0,"Guardhouse":20.0,"Well":5.0,"Arena":25.0,"Barracks":5.0,"Black Market":30.0,"Brewery":80.0,"Camp":5.0,"Cemetery":5.0,"Dock":30.0,"Embassy":5.0,"Factory":80.0,"Farm":55.0,"Fortress":5.0,"Garden":5.0,"Grove":5.0,"Meeting Hall":5.0,"Hideout":5.0,"Inn":30.0,"Manor":5.0,"Market":30.0,"Mine":50.0,"Outpost":5.0,"Palace":5.0,"Research Facility":25.0,"Square":25.0,"Temple":5.0,"Trading Post":30.0,"Warehouse":50.0,"Workshop":100.0,"Catacombs":5.0,"Town Hall":5.0,"Excavation Site":50.0,"Festival Grounds":30.0,"Hospital":25.0,"Ruins":25.0,"Sanctuary":5.0,"Shipyard":75.0,"Caravan Camp":30.0,"Cavern":20.0,"Citadel":15.0,"Enclave":15.0,"Estate":20.0,"Village":45.0,"Hunting Grounds":5.0,"Military Base":5.0,"Mining Settlement":45.0,"Monastery":25.0,"Necropolis":25.0,"Rebel Camp":15.0,"Refugee Camp":15.0,"Slum":15.0,"University":25.0,"Academic District":20.0,"Government District":10.0,"Artisan Quarter":100.0,"Cultural District":10.0,"Entertainment District":25.0,"Farmland":15.0,"Foreign Quarter":15.0,"Wilderness":10.0,"Harbor":25.0,"Industrial District":85.0,"Market District":45.0,"Military District":10.0,"Mining Region":15.0,"Noble Quarter":10.0,"Royal Quarter":10.0,"Residential District":10.0,"Ruined Quarter":20.0,"Temple District":10.0,"Underground City":20.0,"Underworld":15.0,"Warehouse District":20.0},"score":85.0}},"function_quotas":{"Authority":{"min":1,"max":4},"Defense":{"min":1,"max":5},"Religion":{"min":1,"max":4},"Knowledge":{"min":1,"max":4},"Commerce":{"min":2,"max":7},"Production":{"min":1,"max":5},"Logistics":{"min":1,"max":5},"Resources":{"min":1,"max":5},"Residence":{"min":2,"max":6},"Social":{"min":2,"max":7},"Covert":{"min":0,"max":3},"Migration":{"min":1,"max":3},"Care":{"min":1,"max":3},"Exploration":{"min":1,"max":5}}}'

import json,random,math,collections,statistics,time
LOCATION_DATA = json.loads(LOCATION_DATA_JSON)
LOCATION_CANDIDATES=LOCATION_DATA['locations']
LOCATION_QUOTAS=LOCATION_DATA['function_quotas']
LOCATION_FUNCTION_THRESHOLD=75
LOCATION_ZONE_RANDOM_WEIGHT=0.20
LOCATION_FACTION_RANDOM_WEIGHT=0.60
LOCATION_FUNCTION_BONUS_WEIGHT=0.75
LOCATION_FUNCTION_PENALTY_WEIGHT=1.0
LOCATION_SCORE_MODIFIERS={'Hideout':-8,'Camp':-8,'Shelter':-2}
LOCATION_TARGET_MEAN=25
LOCATION_TARGET_SIGMA=2.5   # EXPERIMENTAL shape parameter, not asserted as user-selected

def _location_function_score(candidate,counts):
    funcs=LOCATION_CANDIDATES[candidate]['function']
    under=_soft_max(v for k,v in funcs.items() if counts[k]<LOCATION_QUOTAS[k]['min'])
    at=_soft_max(v for k,v in funcs.items() if v>=75 and counts[k]>=LOCATION_QUOTAS[k]['max'])
    over=_soft_max(v for k,v in funcs.items() if v>=75 and counts[k]>LOCATION_QUOTAS[k]['max'])
    return 0.75*under-at-over

def _location_faction_fit(name,faction):
    a=LOCATION_CANDIDATES[name]
    types=faction['types']
    x=a['type'][types[0]]
    y=a['type'][types[1]] if len(types)>1 else x
    b=a['base'][faction['base']]
    avg=(x+y+b)/3
    return avg,(x*y*b)**(1/3) if x*y*b else 0.0

def _location_world_context(name,chosen,zones,factions):
    row=LOCATION_CANDIDATES[name]['place']
    # Direction: candidate Location -> already chosen Places, as encoded in Book1.
    existing=sum(row[p] for p in chosen)/len(chosen) if chosen else 0.0
    zone=sum(row[z] for z in zones)/len(zones) if zones else 0.0
    context=(existing+zone)/2
    # Each Faction's Type+Base uses the same 3-affinity arithmetic mean.
    faction=sum(_location_faction_fit(name,f)[0] for f in factions)/len(factions)
    return context,faction

def generate_locations(world,seed=None,target=None):
    z=world['zone_generation']['zones']; zones=[item['place'] for item in z]
    factions=world['factions'];rng=random.Random(world['seed']+2_000_011 if seed is None else seed)
    if target is None:
        target=max(20,min(30,round(rng.gauss(LOCATION_TARGET_MEAN,LOCATION_TARGET_SIGMA))))
    if not 20<=target<=30:raise ValueError('Location target must be 20–30')
    chosen=[]; used=set(); counts={k:0 for k in LOCATION_QUOTAS}
    stages=[]
    def select(stage,affinity_score):
        scored=[]
        for name in LOCATION_CANDIDATES:
            r=rng.uniform(0,100)
            if name in used:continue
            score,parts=affinity_score(name,r)
            score+=LOCATION_SCORE_MODIFIERS.get(name,0)
            scored.append((score,name,parts))
        if not scored:raise ValueError('No available Location')
        value,name,parts=max(scored,key=lambda t:(t[0],t[1]))
        used.add(name);chosen.append(name)
        for k,v in LOCATION_CANDIDATES[name]['function'].items():
            if v>=75:counts[k]+=1
        stages.append({'sequence':len(stages)+1,'place':name,'stage':stage,'score':round(value,4),'modifier':LOCATION_SCORE_MODIFIERS.get(name,0),'parts':parts})
    for zone in zones:
        if len(chosen)>=target:break
        select('zone',lambda name,r,zone=zone:(LOCATION_CANDIDATES[name]['place'][zone]+LOCATION_ZONE_RANDOM_WEIGHT*r,{'zone':zone}))
    for faction in factions:
        if len(chosen)>=target:break
        def scoring(name,r,faction=faction):
            arithmetic,geometric=_location_faction_fit(name,faction)
            return arithmetic+geometric+LOCATION_FACTION_RANDOM_WEIGHT*r,{'faction':faction['id']}
        select('faction',scoring)
    while len(chosen)<target:
        def supplemental(name,r):
            context,facfit=_location_world_context(name,chosen,zones,factions)
            r1,r2=rng.random(),rng.random()
            return r+_location_function_score(name,counts)+r1*context+r2*facfit,{'r1':round(r1,4),'r2':round(r2,4)}
        select('supplemental',supplemental)
    missing={k:v['min']-counts[k] for k,v in LOCATION_QUOTAS.items() if counts[k]<v['min']}
    over={k:counts[k]-v['max'] for k,v in LOCATION_QUOTAS.items() if counts[k]>v['max']}
    return {'target':target,'locations':stages,'function_counts':counts,'missing_minimum':missing,'excess_maximum':over}

def generate_world_with_locations(seed=20261120):
    world=generate_world_with_zones(seed)
    world['location_generation']=generate_locations(world)
    world['version']='0.4.2'
    return world

def location_qa(n=3000,seed_start=3_300_000,save_examples=10):
    freq=collections.Counter();fail_min=collections.Counter();fail_max=collections.Counter()
    stage=collections.Counter();target_counts=collections.Counter();function_covert=collections.Counter()
    min_ok=max_ok=weakzone=weakfac=0
    total=zonefit=facfit=0;sets=set();examples=[];start=time.perf_counter()
    for seed in range(seed_start,seed_start+n):
        world=generate_world_with_locations(seed)
        loc=world['location_generation']; names=[x['place'] for x in loc['locations']]
        assert 20<=loc['target']<=30 and len(names)==loc['target'] and len(names)==len(set(names))
        assert all(x in LOCATION_CANDIDATES for x in names)
        assert len([x for x in loc['locations'] if x['stage']=='zone'])==len(world['zone_generation']['zones'])
        assert len([x for x in loc['locations'] if x['stage']=='faction'])==6
        recount={k:sum(LOCATION_CANDIDATES[p]['function'][k]>=75 for p in names) for k in LOCATION_QUOTAS}
        assert recount==loc['function_counts']
        freq.update(names); stage.update(x['stage'] for x in loc['locations']);target_counts[loc['target']]+=1
        sets.add(tuple(sorted(names)))
        if not loc['missing_minimum']:min_ok+=1
        else:fail_min.update(loc['missing_minimum'])
        if not loc['excess_maximum']:max_ok+=1
        else:fail_max.update(loc['excess_maximum'])
        function_covert[recount['Covert']]+=1
        zones=[z['place'] for z in world['zone_generation']['zones']]
        for p in names:
            a=LOCATION_CANDIDATES[p]
            zb=max(a['place'][z] for z in zones)
            fb=max(_location_faction_fit(p,f)[0] for f in world['factions'])
            zonefit+=zb;facfit+=fb;weakzone+=zb<50;weakfac+=fb<50
            total+=1
        if len(examples)<save_examples:
            examples.append({'seed':seed,'factions':[{'name':f['name'],'types':f['types'],'base':f['base']} for f in world['factions']],
                             'zones':zones,'locations':[{'name':v['place'],'stage':v['stage'],'context':v['parts']} for v in loc['locations']],
                             'missing':loc['missing_minimum'],'excess':loc['excess_maximum']})
    probs=[v/sum(freq.values()) for v in freq.values()]
    effective=math.exp(-sum(p*math.log(p) for p in probs))
    rates={p:round(100*freq[p]/n,3) for p in LOCATION_CANDIDATES}
    return {'version':'0.4.2','n':n,'seed_start':seed_start,'mean_locations':round(sum(k*v for k,v in target_counts.items())/n,3),
            'target_distribution':dict(sorted(target_counts.items())),'stage_selections':dict(stage),
            'unique_location_types':len(freq),'different_location_sets':len(sets),'effective_diversity':round(effective,3),
            'min_ok_percent':round(100*min_ok/n,3),'max_ok_percent':round(100*max_ok/n,3),
            'under_minimum_by_function':dict(fail_min),'over_maximum_by_function':dict(fail_max),
            'zero_covert_percent':round(100*function_covert[0]/n,3),'covert_distribution':dict(function_covert),
            'mean_best_zone_fit':round(zonefit/total,3),'mean_best_faction_fit':round(facfit/total,3),
            'weak_best_zone_percent':round(100*weakzone/total,3),'weak_best_faction_percent':round(100*weakfac/total,3),
            'number_over_60':sum(x>60 for x in rates.values()),'max_rate':max(rates.values()),'min_rate':min(rates.values()),
            'top_15':sorted(rates.items(),key=lambda kv:-kv[1])[:15], 'bottom_10':sorted(rates.items(),key=lambda kv:kv[1])[:10],
            'rates':rates,'elapsed_seconds':round(time.perf_counter()-start,3),'examples':examples}
```

```python
LOCATION_QA = location_qa(3000, 3_300_000)
print({k:v for k,v in LOCATION_QA.items() if k not in ('rates','examples')})
print('Top', LOCATION_QA['top_15'])
```

```python
# Structural QA and diversity metrics (rerunnable in Colab).
def qa(n=10000):
    stats=Counter(); distinct_types=Counter(); names=Counter(); bases=Counter(); animals=Counter()
    for seed in range(n):
        world=generate_world(seed)
        fs=world['factions']
        assert len(fs)==6 and [f['id'] for f in fs]==[f'F{i}' for i in range(1,7)]
        assert fs[0]['types'][0]=='Government' and fs[1]['types'][0]=='Revolutionary'
        assert all(1<=len(f['types'])<=2 for f in fs)
        assert all('Government' not in f['types'] or not (set(f['types']) & {'Elite','Poor','Revolutionary','Outsider'}) for f in fs)
        assert len({t for f in fs for t in f['types']}) == sum(map(lambda f:len(f['types']),fs))
        assert len({f['name'] for f in fs})==6
        assert len({f['base'] for f in fs})==6
        assert len({f['animal'] for f in fs})==6
        assert all(BASE[f['base']]['grammar'][f['grammar']]>0 for f in fs)
        # Base and Animal remain selected strictly by highest available score.
        # Full rankings can be inspected on an individual seed with retain_rankings=True.
        stats['worlds']+=1
        stats['active_11_types']+=(sum(len(f['types']) for f in fs)==11)
        stats['active_12_types']+=(sum(len(f['types']) for f in fs)==12)
        active_types=tuple(sorted(t for f in fs for t in f['types']))
        distinct_types[active_types]+=1
        for f in fs:
            names[f['name']]+=1; bases[f['base']]+=1; animals[f['animal']]+=1
    return {'worlds':n,'active_type_11':stats['active_11_types'],
            'active_type_12':stats['active_12_types'],
            'distinct_type_sets':len(distinct_types),
            'distinct_names':len(names),'distinct_bases':len(bases),
            'distinct_animals':len(animals),
            'top_bases':bases.most_common(10),'top_animals':animals.most_common(10)}
qa_result=qa(10000)
print('QA PASS',qa_result)
assert generate_world(20261120)==generate_world(20261120)
```
