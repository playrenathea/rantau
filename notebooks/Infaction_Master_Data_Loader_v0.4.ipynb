# Infaction Master v0.4.x — Book1-first data loader

**Precedence:** original unmodified Book1 → documented approved overlays → active effective generator. The 164 v0.4.1 Zone affinity changes are recorded separately, never written over Book1.

Selection v0.4.2 locked; Zone City v0.4.3 approved experimental module; Location Placement pending.

[Native master Sheet](https://docs.google.com/spreadsheets/d/1hihvEPBM9FXZxzhjHRTGEqYg9jhCXiuONV8mIKj2pwc/edit) · [Canonical JSON](https://github.com/playrenathea/rantau/blob/main/data/infaction_master_v0.4.json)


```
import json, urllib.request
MASTER_URL='https://raw.githubusercontent.com/playrenathea/rantau/main/data/infaction_master_v0.4.json'
with urllib.request.urlopen(MASTER_URL,timeout=30) as response:
    MASTER=json.load(response)
assert MASTER['selection_locked']=='0.4.2'
assert len(MASTER['raw']['zone_selection']['zones'])==36
assert len(MASTER['raw']['location_selection']['locations'])==74
assert len(MASTER['overlay']['zone_affinity_overrides'])==164
print('Loaded Book1 source + approved overlays; source is unchanged')
```

```
BOOK1=MASTER['raw']['book1_original']  # exact original Book1 tables, no tuning
ZONE_BOOK1=MASTER['raw']['zone_selection']  # raw Book1 Zone affinities
ZONE_ACTIVE=MASTER['active']['zone_selection'] # effective locked v0.4.1 Zone snapshot
LOCATION_ACTIVE=MASTER['raw']['location_selection'] # Book1 Location affinities + separate approved modifiers
NAMING=MASTER['raw']['faction_naming']
ZONE_MODIFIERS=MASTER['overlay']['zone_score_modifier']
LOCATION_MODIFIERS=MASTER['overlay']['location_score_modifier']
CITY_SCORES=MASTER['overlay']['zone_city_current']
RULES=MASTER['rules']
assert ZONE_BOOK1['zones']['Citadel']['function']['Authority'] != ZONE_ACTIVE['zones']['Citadel']['function']['Authority']
print('Book1 City:',MASTER['raw']['place_catalog']['Cavern']['inside_city_original'],'Active City:',CITY_SCORES['Cavern'])
print('Ready: use ZONE_ACTIVE for the locked generator; BOOK1/ZONE_BOOK1 for source provenance')
```
