# Infaction Master Data — v0.4.x
Load the master JSON from the repository. Original standalone generator notebooks retain their embedded snapshots; editing Google Sheets does **not** auto-update GitHub JSON or the generator.
[Google Sheets master](https://docs.google.com/spreadsheets/d/1hihvEPBM9FXZxzhjHRTGEqYg9jhCXiuONV8mIKj2pwc/edit)


```
import json, urllib.request
URL = 'https://raw.githubusercontent.com/playrenathea/rantau/main/data/infaction_master_v0.4.json'
with urllib.request.urlopen(URL, timeout=30) as res:
    MASTER=json.load(res)
assert MASTER['selection_locked']=='0.4.2'
assert len(MASTER['raw']['zone_selection']['zones'])==36
assert len(MASTER['raw']['location_selection']['locations'])==74
assert MASTER['overlay']['location_function_quota']['Covert']=={'min':0,'max':3}
print('Master loaded, 36 Zone types, 74 Location types')
print('Mandatory Outside:',[z for z,s in MASTER['overlay']['zone_city_current'].items() if s==0])
```

```
ZONES=MASTER['raw']['zone_selection']
LOCATIONS=MASTER['raw']['location_selection']
NAMING=MASTER['raw']['faction_naming']
PLACES=MASTER['raw']['place_catalog']
PARAMETERS=MASTER['overlay']['parameters']
print('Ready for subsequent Placement experiments')
```
