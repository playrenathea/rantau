# Infaction Master v0.4.x — canonical Book1-first loader

The authoritative runtime data snapshot is **`data/infaction_master_v0.4.json`**. The unmodified Book1 archive is an integrity/reference file only. Load the master once; apply approved locked overlays; verify it before generation. This loader now checks the complete 118×118 Place Affinity matrix. Selection v0.4.2, Location→Zone Placement and City Placement are locked. Site Grouping is exploratory.

[Master JSON](https://github.com/playrenathea/rantau/blob/main/data/infaction_master_v0.4.json) · [Native Sheet](https://docs.google.com/spreadsheets/d/1hihvEPBM9FXZxzhjHRTGEqYg9jhCXiuONV8mIKj2pwc/edit) · [City Colab](https://colab.research.google.com/github/playrenathea/rantau/blob/main/notebooks/Infaction_Locked_City_Placement_v0.4.6.ipynb)


```
# Canonical master data loader + mandatory integrity checks.
import json, urllib.request
MASTER_URL='https://raw.githubusercontent.com/playrenathea/rantau/main/data/infaction_master_v0.4.json'
with urllib.request.urlopen(MASTER_URL,timeout=30) as response:
    MASTER=json.load(response)
assert MASTER['selection_locked']=='0.4.2'
assert MASTER['source_file_sha256']=='a4d8a4c542b62f45445a18c9c781d87cc0e7638bc51bb65c1a3a6919edea0985'
assert len(MASTER['raw']['zone_selection']['zones'])==36
assert len(MASTER['raw']['location_selection']['locations'])==74
assert len(MASTER['overlay']['zone_affinity_overrides'])==164
pa=MASTER['raw']['book1_original']['Place Affinity']['rows']
names=[r[0] for r in pa[2:]]
assert len(names)==118 and len(set(names))==118 and names==pa[1][113:231]
for i in range(118):
    row=pa[i+2][113:231]
    assert len(row)==118 and row[i]==100
    for j in range(118):
        v=row[j]
        assert isinstance(v,(int,float)) and 0<=v<=100 and v==pa[j+2][113+i]
print('PASS: Canonical master loaded; 118x118 affinity, 36 Zones, 74 Locations, 164 overrides')
```
